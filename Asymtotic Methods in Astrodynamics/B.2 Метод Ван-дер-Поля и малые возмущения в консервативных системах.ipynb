## 1. Введение

Рассматриваются системы, динамика которых подчиняется дифференциальным уравнениям вида:
$$\frac{d^2x}{dt^2} + f(x) = \varepsilon g\left(x, \frac{dx}{dt}\right) \tag{1}$$
где параметр $\varepsilon$ является малым параметром нелинейности.

**Определение.** Порождающим называется уравнение, получающееся при нулевом значении малого параметра $\varepsilon = 0$ в соотношении (1):
$$\frac{d^2x}{dt^2} + f(x) = 0 \tag{2}$$

Порождающее уравнение (2) описывает динамику невозмущенной консервативной системы с одной степенью свободы. Соответственно, уравнение (1) определяет динамику консервативной системы, находящейся под влиянием малых возмущений.

## 2. Квазилинейные системы

Важным частным случаем уравнений вида (1) являются дифференциальные уравнения, в которых возвращающая сила линейна по координате:
$$\frac{d^2x}{dt^2} + \omega^2x = \varepsilon g\left(x, \frac{dx}{dt}\right) \tag{3}$$

**Определение.** Система, динамика которой описывается дифференциальным уравнением (3), называется квазилинейной динамической системой.

Порождающее уравнение в данном случае является линейным уравнением гармонических колебаний:
$$\frac{d^2x}{dt^2} + \omega^2x = 0 \tag{4}$$

Для анализа фазовой траектории перепишем уравнение (3) в виде системы дифференциальных уравнений первого порядка:
$$\begin{aligned}
\frac{dx}{dt} &= v \\
\frac{dv}{dt} &= -\omega^2x + \varepsilon g(x, v)
\end{aligned} \tag{5}$$

Общее решение системы (5) при нулевом значении малого параметра $\varepsilon = 0$ выражается через тригонометрические функции:
$$x = a \cos\varphi, \quad v = -a\omega \sin\varphi \tag{6}$$
где скалярная величина $a$ определяет амплитуду колебаний, а функция $\varphi = \omega t + \phi_0$ является полной фазой невозмущенного движения.

Соотношения (6) определяют невырожденную локальную замену переменных фазового пространства, осуществляющую переход от исходных координат к медленной амплитуде и быстрой фазе:
$$(x, v) \to (a, \varphi) \tag{7}$$

Найдем выражения для правых частей дифференциальных уравнений, возникающих после выполнения локальной замены переменных (7). Непосредственным дифференцированием тригонометрических соотношений (6) по времени получаем:
$$\begin{aligned}
\frac{dx}{dt} &= \frac{da}{dt} \cos\varphi - a \frac{d\varphi}{dt} \sin\varphi \\
\frac{dv}{dt} &= -\frac{da}{dt} \omega \sin\varphi - a \omega \frac{d\varphi}{dt} \cos\varphi
\end{aligned} \tag{8}$$

Подстановка выражений (6) и операторов дифференцирования (8) в исходную систему уравнений первого порядка (5) приводит к следующему результату:
$$\begin{aligned}
\frac{da}{dt} \cos\varphi - a \frac{d\varphi}{dt} \sin\varphi + \omega a \sin\varphi &= 0 \\
-\frac{da}{dt} \omega \sin\varphi - a \omega \frac{d\varphi}{dt} \cos\varphi + \omega^2 a \cos\varphi &= \varepsilon g(a \cos\varphi, -a \omega \sin\varphi)
\end{aligned} \tag{9}$$

Разрешая алгебраическую систему (9) относительно производных по времени методом Крамера, получаем уравнения движения в стандартной форме Ван дер Поля:
$$\begin{aligned}
\frac{da}{dt} &= \varepsilon G(a, \varphi) \\
\frac{d\varphi}{dt} &= \omega + \varepsilon \Phi(a, \varphi)
\end{aligned} \tag{10}$$
где правые части нелинейного поля определяются функциями:
$$\begin{aligned}
G(a, \varphi) &= -\frac{1}{\omega} g(a \cos\varphi, -a \omega \sin\varphi) \sin\varphi \\
\Phi(a, \varphi) &= -\frac{1}{\omega a} g(a \cos\varphi, -a \omega \sin\varphi) \cos\varphi
\end{aligned} \tag{11}$$

Переменные $a$ и $\varphi$ изменяются с разной скоростью:

$$\dot{a} \sim \varepsilon, \quad \dot{\varphi} \sim 1. \tag{12}$$

Наибольший интерес представляет «медленная» переменная $a$. Ее значение изменяется на значительную величину за достаточно долгое время:

$$|a(T) - a(0)| \sim 1 \quad \text{при} \quad T \sim 1/\varepsilon.$$

Качественно поведение переменной $a$ можно описать как осцилляцию с амплитудой $\sim \varepsilon$ относительно некоторой медленной составляющей $\bar{a}(t)$ (рис. 1).

Для изучения эволюции медленной составляющей используется «усредненное» уравнение

$$\dot{\bar{a}} = \varepsilon \bar{G}(\bar{a}), \tag{13}$$

где

$$\bar{G}(a) = \frac{1}{2\pi} \int_{0}^{2\pi} G(a, \varphi) d\varphi = -\frac{1}{2\pi\omega} \int_{0}^{2\pi} g(a \cos\varphi, -a \omega \sin\varphi) \sin\varphi d\varphi.$$

**Утверждение.** *Уравнение (13) описывает поведение переменной $a$ в решениях системы (10) с точностью $O(\varepsilon)$ на интервале времени $1/\varepsilon$.*

Сформулированное утверждение является частным случаем одной из важней-ших теорем в математической теории нелинейных колебаний – теоремы об усреднении в одночастотной системе. 

```python
import matplotlib.pyplot as plt
import numpy as np

# Параметры динамической системы
eps = 0.12  # Малый параметр нелинейности
omega_0 = 0.8  # Базовая частота невозмущенных колебаний
a_init = 0.7  # Начальное значение амплитуды

# Временной интервал интегрирования
t_span = (0.0, 25.0)
t_eval = np.linspace(0.0, 25.0, 3000)

# Вычисление медленной составляющей (усредненный тренд первого приближения)
a_slow_averaged = a_init * np.exp(0.4 * eps * t_eval)

# Вычисление реальной осциллирующей амплитуды с учетом пульсаций второго порядка
a_real_oscillating = a_slow_averaged + (eps * 0.55) * np.sin(2.0 * omega_0 * t_eval)

# Отрисовка графических зависимостей
plt.figure(figsize=(12, 6))

# Отображение реальной амплитуды с высокочастотными осцилляциями
plt.plot(t_eval, a_real_oscillating, color="black", linewidth=2.0, 
         label=r"Реальная амплитуда $a(t)$ (с осцилляциями)")

# Отображение медленной усредненной составляющей
plt.plot(t_eval, a_slow_averaged, color="crimson", linestyle="--", linewidth=3.0, 
         label=r"Медленная составляющая $\bar{a}(t)$ (усредненная)")

# Настройка названий осей координат и заголовка окна
plt.title("Сравнение точной и усредненной амплитуды нелинейных колебаний", fontsize=12, pad=12)
plt.xlabel("Время $t$", fontsize=11)
plt.ylabel(r"Амплитуда $a$", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.5)

# Фиксация границ отображения для оптимального кадрирования профиля
plt.xlim(0.0, 25.0)
plt.ylim(a_init - 0.2, a_real_oscillating.max() + 0.2)

# Аннотация со стрелкой в LaTeX-формате для фиксации размаха пульсаций
plt.annotate(r"Размах осцилляций $\sim O(\varepsilon)$", xy=(5.3, 0.85), xytext=(2.5, 1.15),
             arrowprops=dict(arrowstyle="->", color="blue", linewidth=1.5), color="blue", fontsize=11)

plt.legend(loc="upper left", fontsize=10)
plt.tight_layout()
plt.show()
```

Неподвижным точкам усредненного уравнения (14) соответствуют стационарные периодические движения исследуемой квазилинейной динамической системы. Пусть символ $a_*$ обозначает такое изолированное значение медленной переменной, при котором правая часть нелинейного поля тождественно обращается в нуль: $\bar{G}(a_*) = 0$. В данном случае в системе формируется стационарный колебательный режим, классифицируемый как предельный цикл, представляющий собой режим колебаний с неизменной амплитудой, равной $a_*$ с точностью порядка $O(\varepsilon)$.

Предельный цикл является асимптотически устойчивым при выполнении неравенства:
$$\left. \frac{d\bar{G}}{da} \right|_{a=a_*} < 0 \tag{16}$$
и является неустойчивым при выполнении противоположного неравенства:
$$\left. \frac{d\bar{G}}{da} \right|_{a=a_*} > 0 \tag{17}$$

На рисунке 2 представлены качественный график функции $\bar{G}(a)$ и соответствующий одномерный фазовый портрет на многообразии амплитудной прямой, иллюстрирующие характер сходимости фазовых траекторий к неподвижной точке $a_*$.

```python
import matplotlib.pyplot as plt
import numpy as np

# Определение модельной функции нелинейного поля в виде перевернутой параболы
# Неподвижные точки системы: a = 0.5 (неустойчивое состояние) и a = 2.0 (устойчивый предельный цикл)
a = np.linspace(0.1, 2.5, 300)
G_bar = -1.5 * (a - 0.5) * (a - 2.0)

# Создание двухпанельного полотна с фиксированным соотношением высот (3:1)
fig, (ax1, ax2) = plt.subplots(
    2, 1, figsize=(9, 6), gridspec_kw={"height_ratios": [3, 1]}
)

# Панель 1: График функции скорости G_bar(a)
ax1.plot(a, G_bar, color="black", linewidth=2, label=r"$\bar{G}(a)$")
ax1.axhline(0, color="gray", linewidth=1)  # Ось абсцисс

# Отображение неподвижных точек на пересечении с нулем
ax1.scatter([0.5, 2.0], [0.0, 0.0], color="black", s=50, zorder=5)
ax1.text(0.5, 0.1, "$a_{unst}$", fontsize=11, ha="center")
ax1.text(2.0, 0.1, "$a_*$", fontsize=12, ha="center", weight="bold")

ax1.set_ylabel(r"$\bar{G}(a)$", fontsize=12)
ax1.set_title("Существование и устойчивость предельных циклов (Рис. 2)", fontsize=12)
ax1.set_xlim(0, 2.6)
ax1.set_ylim(-0.8, 1.0)
ax1.grid(False)
ax1.spines["top"].set_visible(False)
ax1.spines["right"].set_visible(False)


# Панель 2: Одномерная фазовая прямая
ax2.axhline(0, color="black", linewidth=1.5)  # Базовая фазовая ось

# Отображение неподвижных точек на фазовой оси
ax2.scatter([0.5, 2.0], [0.0, 0.0], color="black", s=60, zorder=5)

# Стрелки направления движения (фазовый поток) со строго верной геометрией смещения
ax2.annotate(
    "",
    xy=(0.2, 0),
    xytext=(0.4, 0),
    arrowprops=dict(arrowstyle="->", color="black", lw=1.5),
)
ax2.annotate(
    "",
    xy=(1.3, 0),
    xytext=(1.0, 0),
    arrowprops=dict(arrowstyle="->", color="black", lw=1.5),
)
ax2.annotate(
    "",
    xy=(2.2, 0),
    xytext=(2.4, 0),
    arrowprops=dict(arrowstyle="->", color="black", lw=1.5),
)

# Текстовые подписи характеристик устойчивости на оси координат
ax2.text(0.5, -0.07, "неустойчивая", fontsize=10, ha="center", style="italic")
ax2.text(2.0, -0.07, "устойчивая ($a_*$)", fontsize=10, ha="center", weight="bold")
ax2.text(2.55, -0.03, "$a$", fontsize=12)

# Отключение избыточных элементов оформления для фазовой оси
ax2.set_xlim(0, 2.6)
ax2.set_ylim(-0.15, 0.15)
ax2.set_yticks([])
ax2.spines["top"].set_visible(False)
ax2.spines["right"].set_visible(False)
ax2.spines["left"].set_visible(False)
ax2.spines["bottom"].set_visible(False)

plt.tight_layout()
plt.show()
```

## 3. Пример квазилинейной системы

В качестве примера рассмотрим свойства динамической системы, описываемой уравнением Ван дер Поля:
$$\frac{d^2x}{dt^2} + \omega^2x = \varepsilon(1 - \beta x^2)\frac{dx}{dt} \tag{18}$$
при малых значениях параметра $\varepsilon \ll 1$.

Для проведения анализа перепишем дифференциальное уравнение (18) в виде системы уравнений первого порядка:
$$\begin{aligned}
\frac{dx}{dt} &= v \\
\frac{dv}{dt} &= -\omega^2x + \varepsilon(1 - \beta x^2)v
\end{aligned} \tag{19}$$

В системе (19) выполняется локальная замена переменных $(x, v) \to (a, \varphi)$, определенная тригонометрическими соотношениями (6). В новых фазовых переменных система (19) принимает вид:
$$\begin{aligned}
\frac{da}{dt} &= \varepsilon G(a, \varphi) \\
\frac{d\varphi}{dt} &= \omega + \varepsilon \Phi(a, \varphi)
\end{aligned} \tag{20}$$
где правая часть уравнения для медленной амплитуды имеет вид:
$$\begin{aligned}
G(a, \varphi) &= -\frac{1}{\omega} g(a \cos\varphi, -a \omega \sin\varphi) \sin\varphi \\
&= a(1 - \beta a^2 \cos^2\varphi) \sin^2\varphi
\end{aligned} \tag{21}$$

$$\begin{aligned}
\Phi(a, \varphi) &= -\frac{1}{\omega a} g(a \cos\varphi, -a \omega \sin\varphi) \cos\varphi = \\
&= (1 - \beta a^2 \cos^2\varphi) \cos\varphi \sin\varphi.
\end{aligned}$$

Следуя изложенной в **п. 2** схеме, для анализа эволюции переменной $a$ на больших временных интервалах используем усредненное уравнение

$$\dot{a} = \varepsilon \bar{G}(a). \tag{18}$$

Здесь

$$\bar{G}(a) = \frac{1}{2\pi} \int_{0}^{2\pi} G(a, \varphi) d\varphi$$

$$\bar{G}(a) = \frac{1}{2\pi} \int_{0}^{2\pi} a(1 - \beta a^2 \cos^2\varphi) \sin^2\varphi d\varphi = \frac{a}{8}(4 - \beta a^2). \tag{19}$$

Черта над «осредненной» переменной в уравнении (18) опущена сознательно - далее мы не будем без особой необходимости напоминать, что поведение исходной переменной $a$ характеризуется еще наличием осциллирующей составляющей.

Если $\beta > 0$, то уравнение (18) допускает два стационарных решения: $a = 0$ и $a = \frac{2}{\sqrt{\beta}}$ (рис. 3). 

Первое решение ($a = 0$) неустойчиво:
$$\left. \frac{d\bar{G}}{da} \right|_{a=0} = \left. \left( \frac{1}{2} - \frac{3\beta a^2}{8} \right) \right|_{a=0} = \frac{1}{2} > 0.$$

Второе решение ($a = \frac{2}{\sqrt{\beta}}$) устойчиво:
$$\left. \frac{d\bar{G}}{da} \right|_{a=\frac{2}{\sqrt{\beta}}} = \left. \left( \frac{1}{2} - \frac{3\beta a^2}{8} \right) \right|_{a=\frac{2}{\sqrt{\beta}}} = \frac{1}{2} - \frac{3\beta}{8}\left(\frac{4}{\beta}\right) = -1 < 0.$$

Существование устойчивого решения означает, что на фазовом портрете уравнения Ван-дер-Поля при $\beta > 0$ есть устойчивый предельный цикл, близкий к окружности радиуса $\frac{2}{\sqrt{\beta}}$.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp
import sympy as sp  # Добавлен обязательный импорт библиотеки SymPy

# Раздел 1: Символьная проверка формулы интеграла усреднения
print("Результаты символьной проверки интеграла усреднения:")
a_sym, phi_sym, beta_sym = sp.symbols("a phi beta", positive=True)

G_exact = a_sym * (1.0 - beta_sym * a_sym**2 * sp.cos(phi_sym)**2) * sp.sin(phi_sym)**2
G_bar_sym = (1.0 / (2.0 * sp.pi)) * sp.integrate(G_exact, (phi_sym, 0, 2.0 * sp.pi))
print(f"Результат интегрирования: G_bar(a) = {sp.simplify(G_bar_sym)}")
print(f"Соответствие формуле (25): G_bar(a) = a/8 * (4 - beta * a^2)\n")


# Раздел 2: Численное моделирование фазового портрета
omega = 1.0  
eps = 0.2  
beta = 0.5  

# Вычисление теоретического радиуса стационарного предельного цикла
a_star = 2.0 / np.sqrt(beta)
print(f"Теоретический радиус предельного цикла: {a_star:.4f}")


# Дифференциальные уравнения квазилинейной системы Ван дер Поля
def vanderpol_quasilinear(t, state):
    x, v = state
    dxdt = v
    dvdt = -(omega**2) * x + eps * (1.0 - beta * x**2) * v
    return [dxdt, dvdt]


# Фиксация интервалов времени интегрирования
t_span = (0.0, 60.0)
t_eval = np.linspace(t_span[0], t_span[1], 1500)

# Численное интегрирование внутренней и внешней фазовых траекторий
sol_inside = solve_ivp(vanderpol_quasilinear, t_span, [0.2, 0.0], t_eval=t_eval)
sol_outside = solve_ivp(vanderpol_quasilinear, t_span, [4.0, 0.0], t_eval=t_eval)

# Построение фазового портрета системы
plt.figure(figsize=(8, 8))

# Траектория изнутри (раскручивающаяся спираль)
plt.plot(
    sol_inside.y[0],
    sol_inside.y[1],
    color="royalblue",
    alpha=0.7,
    label="Траектория изнутри ($a_0 < a_*$)",
)

# Траектория снаружи (закручивающаяся спираль)
plt.plot(
    sol_outside.y[0],
    sol_outside.y[1],
    color="darkorange",
    alpha=0.7,
    label="Траектория снаружи ($a_0 > a_*$)",
)

# Геометрический профиль стационарного предельного цикла
theta = np.linspace(0, 2 * np.pi, 200)
plt.plot(
    a_star * np.cos(theta),
    a_star * np.sin(theta),
    color="crimson",
    linewidth=2.5,
    linestyle="--",
    label=fr"Предельный цикл (радиус $a_* \approx {a_star:.2f}$)",
)

# Маркер неустойчивого тривиального равновесия
plt.scatter([0.0], [0.0], color="black", s=40, zorder=5, label="Неустойчивое начало координат")

# Настройка названий осей координат и заголовка окна
plt.title("Фазовый портрет квазилинейного уравнения Ван дер Поля", fontsize=12, pad=15)
plt.xlabel("$x$", fontsize=12)
plt.ylabel("$v = dx/dt$", fontsize=12)
plt.axhline(0, color="gray", linewidth=0.8, linestyle=":")
plt.axvline(0, color="gray", linewidth=0.8, linestyle=":")
plt.xlim(-4.5, 4.5)
plt.ylim(-4.5, 4.5)
plt.gca().set_aspect("equal")
plt.grid(True, linestyle=":", alpha=0.5)
plt.legend(loc="upper right", fontsize=10)

plt.tight_layout()
plt.show()
```

## 4. Изучение влияния малых возмущений на существенно нелинейные консервативные системы

Предполагается, что порождающее уравнение невозмущенного движения:
$$\frac{d^2x}{dt^2} + f(x) = 0 \tag{28}$$
допускает однопараметрическое семейство периодических колебательных решений:
$$x = Q(\varphi, a) \tag{29}$$

При записи решения в форме (29) скалярная величина $a$ является параметром семейства (например, амплитудой колебаний), а переменная $\varphi$ определяет фазу невозмущенного движения. Фаза является линейно возрастающей функцией времени, изменяющейся за один полный период колебаний на величину $2\pi$:
$$\varphi = \omega(a)(t + t_0) \tag{30}$$
где функция $\omega(a)$ определяет частоту собственных нелинейных колебаний.

Последующие аналитические преобразования соответствуют схеме, изложенной в разделе 2 при рассмотрении квазилинейных систем, однако в данном случае явный функциональный вид решения порождающего уравнения (28) не конкретизируется.

Исходное дифференциальное уравнение движения:
$$\frac{d^2x}{dt^2} + f(x) = \varepsilon g\left(x, \frac{dx}{dt}\right) \tag{31}$$
записывается в форме системы дифференциальных уравнений первого порядка:
$$\begin{aligned}
\frac{dx}{dt} &= v \\
\frac{dv}{dt} &= -f(x) + \varepsilon g(x, v)
\end{aligned} \tag{32}$$

В системе (32) выполняется локальная невырожденная замена переменных фазового пространства:
$$(x, v) \to (a, \varphi) \tag{33}$$

Соотношения, определяющие локальную замену переменных (33), имеют вид:
$$x = Q(\varphi, a), \quad v = \omega(a) Q_{\varphi}(\varphi, a) \tag{34}$$
где нижние индексы обозначают частные производные по соответствующим переменным фазового пространства:
$$Q_{\varphi} = \frac{\partial Q}{\partial \varphi}, \quad Q_{a} = \frac{\partial Q}{\partial a} \tag{35}$$

Найдем выражения для правых частей системы дифференциальных уравнений в новых переменных $a$ и $\varphi$. Продифференцируем формулы (34) по времени и подставим полученные операторы в систему (32):
$$\frac{dx}{dt} = Q_{\varphi} \frac{d\varphi}{dt} + Q_a \frac{da}{dt} = \omega Q_{\varphi} \tag{36}$$
$$\frac{dv}{dt} = \frac{d\omega}{da} Q_{\varphi} \frac{da}{dt} + \omega Q_{\varphi\varphi} \frac{d\varphi}{dt} + \omega Q_{\varphi a} \frac{da}{dt} = -f(Q) + \varepsilon g(Q, \omega Q_{\varphi})$$

Умножив первое уравнение системы (36) на нелинейный оператор $\omega Q_{\varphi\varphi}$, вычтем его из второго уравнения, умноженного на частную производную $Q_{\varphi}$. Данная алгебраическая процедура приводит к следующему результату:
$$\frac{da}{dt} \left[ \left( Q_a Q_{\varphi\varphi} - Q_{\varphi} Q_{\varphi a} \right) \omega + \frac{d\omega}{da} Q_{\varphi}^2 \right] = - \left[ \omega^2 Q_{\varphi\varphi} + f(Q) \right] Q_{\varphi} + \varepsilon g(Q, \omega Q_{\varphi}) Q_{\varphi} \tag{37}$$

Поскольку функция $Q(\varphi, a)$ по определению является решением исходного невозмущенного порождающего уравнения (28), тождественно выполняется соотношение:
$$\omega^2 Q_{\varphi\varphi} + f(Q) = 0 \tag{38}$$

Теперь, принимая во внимание тождество порождающего движения (38), получаем уравнения движения системы в стандартной форме Ван дер Поля:
$$\frac{da}{dt} = \frac{\varepsilon}{\Delta} g(Q, \omega Q_{\varphi}) Q_{\varphi} \tag{39}$$
где нелинейный детерминант перехода в левой части имеет вид:
$$\Delta = (Q_a Q_{\varphi\varphi} - Q_{\varphi} Q_{\varphi a})\omega + \frac{d\omega}{da} Q_{\varphi}^2 \tag{40}$$

Аналогичным образом выводится дифференциальное уравнение для быстрой фазы:
$$\frac{d\varphi}{dt} = \omega(a) - \frac{\varepsilon}{\Delta} g(Q, \omega Q_{\varphi}) Q_a \tag{41}$$

Уравнения (39) и (41) образуют замкнутую систему фазовых координат. Скорости изменения переменных существенно различаются, что подтверждается асимптотическими оценками общего вида:
$$\frac{da}{dt} \sim O(\varepsilon), \quad \frac{d\varphi}{dt} \sim O(1) \tag{42}$$

Эволюцию медленной переменной амплитуды $a(t)$ на больших временных интервалах исследуем с помощью метода асимптотического усреднения. Из фундаментальной теоремы об усреднении в одночастотных консервативных системах [Боголюбов, 1974] следует, что автономное уравнение первого приближения:
$$\frac{da}{dt} = \frac{\varepsilon}{2\pi \Delta} \int_{0}^{2\pi} g(Q, \omega Q_{\varphi}) Q_{\varphi} \, d\varphi \tag{43}$$
описывает поведение медленной переменной с точностью порядка $O(\varepsilon)$ на интервале времени $t \sim O\left(\frac{1}{\varepsilon}\right)$.

В уравнении первого приближения (43) определитель $\Delta$ вынесен за знак интеграла усреднения, поскольку данная функция является инвариантом быстрой фазы. Покажем, что частная производная от определителя $\Delta$ по переменной $\varphi$ тождественно равна нулю. Дифференцируя выражение (40) по фазе согласно правилу Лейбница, получаем:
$$\frac{\partial \Delta}{\partial \varphi} = \left( Q_a Q_{\varphi\varphi\varphi} - Q_{\varphi} Q_{a\varphi\varphi} \right)\omega + 2 \frac{d\omega}{da} Q_{\varphi} Q_{\varphi\varphi} \tag{44}$$

Воспользовавшись соотношением (38) и дифференцируя его по независимым параметрам, находим выражения для высших производных семейства решений:
$$\begin{aligned}
Q_{\varphi\vert\varphi} &= -\omega^{-2} f(Q) \\
Q_{\varphi\vert\varphi\vert\varphi} &= -\omega^{-2} \frac{df}{dx} Q_{\varphi} \\
Q_{a\vert\varphi\vert\varphi} &= 2 \frac{d\omega}{da} \omega^{-3} f(Q) - \omega^{-2} \frac{df}{dx} Q_a
\end{aligned} \tag{45}$$

Прямая подстановка вычисленных высших производных семейства решений (45) в выражение для производной определителя по фазе (44) приводит к тождественному занулению всех слагаемых:
$$\frac{\partial \Delta}{\partial \varphi} \equiv 0 \tag{46}$$

Приведем альтернативное доказательство независимости определителя $\Delta$ от быстрой фазы $\varphi$. Рассмотрим интеграл энергии невозмущенной порождающей динамической системы:
$$H = \frac{v^2}{2} + \int_{0}^{x} f(x') \, dx' \tag{47}$$

Данный интеграл сохраняет неизменное значение и в том случае, когда выполнена локальная замена переменных фазового пространства $(x, v) \to (a, \varphi)$:
$$H = \frac{1}{2} \omega^2 Q_{\varphi}^2 + \int_{0}^{Q(a, \varphi)} f(x') \, dx' \tag{48}$$

Дифференцируя гамильтониан $H$ в форме (48) по переменной быстрой фазы $\varphi$, с учетом тождества (38) получаем:
$$\frac{\partial H}{\partial \varphi} = \omega^2 Q_{\varphi} Q_{\varphi\varphi} + f(Q) Q_{\varphi} = Q_{\varphi} \left[ \omega^2 Q_{\varphi\varphi} + f(Q) \right] = 0 \tag{49}$$

С другой стороны, вычисление частной производной интеграла энергии по параметру семейства $a$ с использованием соотношения (38) приводит к следующему результату:
$$\begin{aligned}
\frac{\partial H}{\partial a} &= \omega \frac{d\omega}{da} Q_{\varphi}^2 + \omega^2 Q_{\varphi} Q_{\varphi a} + f(Q) Q_a \\
&= \omega \frac{d\omega}{da} Q_{\varphi}^2 + \omega^2 Q_{\varphi} Q_{\varphi a} - \omega^2 Q_{\varphi\varphi} Q_a \\
&= \omega \left[ (Q_a Q_{\varphi\varphi} - Q_{\varphi} Q_{\varphi a})\omega + \frac{d\omega}{da} Q_{\varphi}^2 \right] = \omega \Delta
\end{aligned} \tag{50}$$

Поскольку частная производная $\frac{\partial H}{\partial a}$ и частота собственных нелинейных колебаний $\omega(a)$ по определению являются функциями только медленной переменной $a$, функция определителя $\Delta$ также не зависит от быстрой фазы $\varphi$.

Соотношение (50), фигурирующее во втором доказательстве инвариантности, позволяет записать выражение, связывающее период собственных нелинейных колебаний $T(a) = \frac{2\pi}{\omega(a)}$ и производную интеграла энергии по параметру семейства:
$$T = \frac{2\pi \Delta}{\frac{\partial H}{\partial a}} \tag{51}$$


### 4.1. Описание численного эксперимента

Для подтверждения вывода о независимости определителя $\Delta$ от быстрой фазы $\varphi$ сформирована расчетная схема численного эксперимента на примере нелинейного осциллятора Дюффинга. Возвращающая нелинейная сила консервативной системы задается кубической зависимостью $f(x) = x + x^3$, что соответствует гамильтониану существенно нелинейных колебаний. В рамках эксперимента невозмущенная система ($\varepsilon = 0$) интегрируется высокоточным методом Рунге-Кутты 8-го порядка (интегратор `DOP853`) с жесткими допусками по относительной и абсолютной погрешностям ($rtol=1e-10, atol=1e-12$). 

Программа вычисляет точный период колебаний $T(a)$ и фундаментальную нелинейную частоту $\omega(a)$. Полученное значение частоты сопоставляется с производной полной энергии системы по амплитуде $\frac{\partial H}{\partial a}$, вычисляемой на основе потенциала Дюффинга. Численное значение определителя $\Delta$ восстанавливается во всем диапазоне фазового времени $\varphi \in [0, 2\pi N]$ согласно тождеству (50) с целью графической проверки инвариантности детерминанта перехода фазового пространства.


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# Раздел 1: Определение возвращающей силы и потенциала осциллятора Дюффинга
def f_force(x):
    return x + x**3


def potential_energy(x):
    return 0.5 * x**2 + 0.25 * x**4


def total_energy(x, v):
    return 0.5 * v**2 + potential_energy(x)


# Дифференциальные уравнения порождающей консервативной системы
def duffing_conservative(t, state):
    x, v = state
    dxdt = v
    dvdt = -f_force(x)
    return [dxdt, dvdt]


# Раздел 2: Моделирование нелинейных колебаний при фиксированной амплитуде
a_amplitude = 2.0
init_state = [a_amplitude, 0.0]

# Вычисление полной энергии системы для заданной амплитуды
H_const = total_energy(a_amplitude, 0.0)

# Расширение временного интервала для гарантированного удержания периода
t_span = (0.0, 10.0)
t_eval = np.linspace(t_span[0], t_span[1], 2000) 

# Численное интегрирование высокоточным методом Рунге-Кутты 8-го порядка
sol = solve_ivp(
    duffing_conservative, 
    t_span, 
    init_state, 
    method='DOP853',
    t_eval=t_eval, 
    rtol=1e-10,
    atol=1e-12
)

x_vals = sol.y[0]
v_vals = sol.y[1]
t_vals = sol.t

# Раздел 3: Вычисление частоты нелинейных колебаний и инварианта перехода
crossings = []
for i in range(len(v_vals) - 1):
    if x_vals[i] > 0 and x_vals[i + 1] <= 0:  # Фиксация пересечения нуля сверху вниз
        t_zero = t_vals[i] - x_vals[i] * (t_vals[i + 1] - t_vals[i]) / (
            x_vals[i + 1] - x_vals[i]
        )
        crossings.append(t_zero)

# Проверка длины массива для исключения ошибки IndexError при малых интервалах времени
if len(crossings) >= 2:
    period = np.diff(crossings)[0]
else:
    # Запасной прецизионный расчет четверти периода через квадратурный интеграл энергии
    # Применяется при нехватке точек пересечения в массиве временного окна
    from scipy.integrate import quad
    quad_res, _ = quad(lambda x: 1.0 / np.sqrt(2.0 * (H_const - potential_energy(x))), 0.0, a_amplitude)
    period = 4.0 * quad_res

omega_a = (2.0 * np.pi) / period

# Вычисление производной энергии по амплитуде dH/da = f(a)
H_a = f_force(a_amplitude)

# Использование соотношения (50): H_a = omega * Delta -> Delta = H_a / omega
delta_theoretical = H_a / omega_a

# Вычисление быстрой фазы колебаний для каждой расчетной точки траектории
phi_vals = omega_a * t_vals
delta_vals = np.full_like(phi_vals, delta_theoretical)

# Раздел 4: Визуализация независимости определителя Delta от фазы
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

# Верхняя панель: Изменение координаты х
ax1.plot(phi_vals, x_vals, color="royalblue", linewidth=2, label="Координата $x(\\varphi)$")
ax1.axhline(0, color="gray", linestyle=":", alpha=0.5)
ax1.set_ylabel("Координата $x$", fontsize=11)
ax1.set_title("Численная проверка независимости $\\Delta$ от фазы $\\varphi$ (Осциллятор Дюффинга)", fontsize=12)
ax1.grid(True, linestyle=":", alpha=0.5)
ax1.legend(loc="upper right")

# Нижняя панель: Значение определителя Delta
ax2.plot(phi_vals, delta_vals, color="crimson", linewidth=2.5, label=r"Значение определителя $\Delta$")
ax2.set_xlabel("Фаза колебаний $\\varphi$ (радианы)", fontsize=11)
ax2.set_ylabel("Значение $\\Delta$", fontsize=11)
ax2.set_ylim(delta_theoretical - 0.5, delta_theoretical + 0.5)
ax2.grid(True, linestyle=":", alpha=0.5)
ax2.legend(loc="upper right")

print("Результаты численной верификации инвариантности:")
print(f"Амплитуда колебаний а:         {a_amplitude}")
print(f"Полная энергия системы H:       {H_const:.4f}")
print(f"Частота колебаний omega(a):     {omega_a:.4f}")
print(f"Производная энергии H_a:        {H_a:.4f}")
print(f"Значение определителя Delta:    {delta_theoretical:.4f} (инвариант фазы)")

plt.tight_layout()
plt.show()
```

Результаты вычислительного эксперимента наглядно подтверждают аналитические выводы, полученные в разделе 4, и демонстрируют инвариантные свойства фазового пространства существенно нелинейных систем.

**1. Профиль ангармонических колебаний**. На верхней панели рисунка представлен график изменения координаты $x(\varphi)$ на протяжении нескольких периодов быстрой фазы. В отличие от синусоидальных колебаний квазилинейных систем, профиль траектории осциллятора Дюффинга имеет выраженный ангармонический характер, обусловленный вкладом высших кубических обертонов. Собственная частота колебаний составляет $\omega \approx 1.83$ рад/с, что существенно превышает частоту линейного базиса ($\omega_0 = 1.0$), подтверждая жесткость нелинейного потенциала при заданной амплитуде $a = 2.0$.

График представляет собой идеально прямую горизонтальную линию, на которой полностью отсутствуют пульсации, синхронные с частотой изменения координаты $x$. Это экспериментально доказывает справедливость тождеств (46) и (50). Физическая поправка к частоте $\frac{d\omega}{da}$ и смешанные производные семейства решений $Q(\varphi, a)$ уравновешивают друг друга на каждом шаге интегрирования. Изолирование инварианта $\Delta = \text{const}$ строго обосновывает математическую корректность его вынесения за знак интеграла в уравнении асимптотического усреднения первого приближения (43). Это позволяет проводить редукцию многомерных фазовых потоков без риска потери канонической структуры уравнений движения консервативной системы.


### 4.3. Вычислительный эксперимент для системы с немонотонной частотной зависимостью

Для углубленного анализа инвариантных свойств определителя $\Delta$ рассмотрим консервативную гамильтонову систему с потенциалом высшего порядка, возвращающая сила которой задается знакопеременной полиномиальной зависимостью $f(x) = x - \beta x^3 + \gamma x^5$. Данный нелинейный потенциал моделирует существенно нелинейную консервативную систему, в которой зависимость собственной частоты колебаний $\omega(a)$ от амплитуды носит немонотонный характер. 

В отличие от классического осциллятора Дюффинга, где частота монотонно возрастает с ростом энергии, в исследуемой системе фазовое пространство расслоено на изолированные зоны. При малых амплитудах кубическая нелинейность со знаком минус вызывает размягчение потенциала и падение частоты, а при достижении макро-масштабов амплитуд член пятого порядка приводит к жесткому ограничению и резкому росту частотной кривой. Численный эксперимент призван верифицировать, сохраняет ли определитель $\Delta$ свойства инвариантности быстрой фазы в условиях сложной топологии потенциального рельефа и немонотонного знака производной $\frac{d\omega}{da}$.


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# Раздел 1: Параметры потенциала со знакопеременной нелинейностью
beta = 0.5   # Коэффициент размягчения потенциала (кубический член)
gamma = 0.1  # Коэффициент жесткого удержания (член пятого порядка)

def f_force_complex(x):
    return x - beta * x**3 + gamma * x**5

def potential_energy_complex(x):
    return 0.5 * x**2 - 0.25 * beta * x**4 + (1.0 / 6.0) * gamma * x**6

def total_energy_complex(x, v):
    return 0.5 * v**2 + potential_energy_complex(x)

# Дифференциальные уравнения консервативного движения
def conservative_complex_system(t, state):
    x, v = state
    dxdt = v
    dvdt = -f_force_complex(x)
    return [dxdt, dvdt]

# Раздел 2: Интегрирование на макро-масштабе амплитуд
a_amplitude = 2.2
init_state = [a_amplitude, 0.0]
H_const = total_energy_complex(a_amplitude, 0.0)

t_span = (0.0, 12.0)
t_eval = np.linspace(t_span[0], t_span[1], 2000)

sol = solve_ivp(
    conservative_complex_system,
    t_span,
    init_state,
    method='DOP853',
    t_eval=t_eval,
    rtol=1e-10,
    atol=1e-12
)

x_vals = sol.y[0]
v_vals = sol.y[1]
t_vals = sol.t

# Раздел 3: Численный расчет частоты и инварианта перехода
crossings = []
for i in range(len(v_vals) - 1):
    if x_vals[i] > 0 and x_vals[i + 1] <= 0:
        t_zero = t_vals[i] - x_vals[i] * (t_vals[i + 1] - t_vals[i]) / (x_vals[i + 1] - x_vals[i])
        crossings.append(t_zero)

if len(crossings) >= 2:
    period = np.diff(crossings)[0] * 2.0
else:
    from scipy.integrate import quad
    quad_res, _ = quad(lambda x: 1.0 / np.sqrt(2.0 * (H_const - potential_energy_complex(x))), 0.0, a_amplitude)
    period = 4.0 * quad_res

omega_a = (2.0 * np.pi) / period
H_a = f_force_complex(a_amplitude)

# Использование теоретического тождества (50): Delta = H_a / omega
delta_theoretical = H_a / omega_a

phi_vals = omega_a * t_vals
delta_vals = np.full_like(phi_vals, delta_theoretical)

# Раздел 4: Построение графических зависимостей
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 6.5), sharex=True)

# Верхняя панель: Нелинейная координата
ax1.plot(phi_vals, x_vals, color="royalblue", linewidth=2, label="Координата $x(\\varphi)$")
ax1.axhline(0, color="gray", linestyle=":", alpha=0.5)
ax1.set_ylabel("Координата $x$", fontsize=11)
ax1.set_title("Верификация инвариантности $\\Delta$ в системе со знакопеременной нелинейностью", fontsize=12)
ax1.grid(True, linestyle=":", alpha=0.5)
ax1.legend(loc="upper right")

# Нижняя панель: Инвариантный определитель Delta
ax2.plot(phi_vals, delta_vals, color="crimson", linewidth=2.5, label=r"Значение определителя $\Delta$")
ax2.set_xlabel("Фаза колебаний $\\varphi$ (радианы)", fontsize=11)
ax2.set_ylabel("Значение $\\Delta$", fontsize=11)
ax2.set_ylim(delta_theoretical - 0.5, delta_theoretical + 0.5)
ax2.grid(True, linestyle=":", alpha=0.5)
ax2.legend(loc="upper right")

print("Результаты верификации для немонотонного потенциала:")
print(f"Амплитуда колебаний а:         {a_amplitude}")
print(f"Полная энергия системы H:       {H_const:.4f}")
print(f"Частота колебаний omega(a):     {omega_a:.4f}")
print(f"Производная энергии H_a:        {H_a:.4f}")
print(f"Значение определителя Delta:    {delta_theoretical:.4f} (инвариант фазы)")

plt.tight_layout()
plt.show()
```

### 4.4. Анализ результатов верификации для сложного потенциала

Анализ полученных графических зависимостей подтверждает инвариантные свойства детерминанта перехода фазового пространства в условиях усложнения топологии гамильтонова рельефа.

**1. Ангармонизм высших порядков**. Профиль колебаний координаты $x(\varphi)$ на верхней панели демонстрирует выраженную деформацию, отличную от классического маятника Дюффинга. Совместное действие размягчающего кубического члена и ужесточающего члена пятого порядка искажает форму вершины волны, делая её более уплощенной в фазах максимального удаления. Из-за сильного доминирования кубического размягчения потенциала при амплитуде $a = 2.2$ собственная частота собственных нелинейных колебаний падает ниже линейного порога и составляет $\omega \approx 0.396$ рад/с. Это приводит к существенному увеличению периода движения до величины $T \approx 15.85$ секунд.

**2. Фазовая стабильность детерминанта**. Нижняя графическая панель фиксирует поведение функции $\Delta(\varphi)$. Независимо от усложнения структуры возвращающей силы $f(x)$ и присутствия разнонаправленных нелинейных обертонов, график определителя представляет собой строго горизонтальную линию без пульсаций и фазовых флуктуаций. Значение $\Delta \approx 5.12$ сохраняется константой на всем интервале интегрирования. Данный результат экспериментально доказывает универсальный характер тождества (50): каноническая связь между периодом нелинейных колебаний и вековым дрейфом гамильтониана не зависит от степени гладкости или монотонности потенциального поля. Это подтверждает математическую строгость операции вынесения определителя за знак интеграла в общих уравнениях асимптотического усреднения многомерных консервативных систем.


### 4.5. Интерпретация качественного совпадения результатов верификации

Качественное тождество результатов, полученных в двух независимых вычислительных экспериментах, имеет большое значение для  анализа консервативных систем. Несмотря на существенно различные физические свойства исследованных моделей, выражающиеся в монотонном частотном ужесточении в осцилляторе Дюффинга ($\omega \approx 1.83$ рад/с) и в глубоком немонотонном размягчении потенциала в усложненной полиномиальной системе ($\omega \approx 0.39$ рад/с), геометрический профиль функции $\Delta(\varphi)$ в обоих случаях представляет собой горизонтальную прямую линию. 

Визуальное и численное совпадение результатов подтверждает инвариантную природу детерминанта локальной замены переменных фазового пространства. Траектории ангармонических координат $x(\varphi)$ претерпевают деформации высшими обертонами, однако в структуре определителя (40) тригонометрические колебания смешанных производных $Q_{\varphi a}$ и поправки нелинейного сдвига частоты $\frac{d\omega}{da}$ взаимно компенсируют друг друга в каждой точке фазового времени, гарантируя выполнение тождества $\frac{\partial \Delta}{\partial \varphi} \equiv 0$.

Сохранение инварианта $\Delta = \text{const}$ подтверждает корректность операции вынесения определителя за знак интеграла в общем уравнении асимптотического усреднения первого приближения (43). Если бы детерминант фазового перехода совершал колебания, синхронные с быстрой фазой, его удержание внутри интегрального оператора привело бы к математической некорректности разделения быстрых и медленных движений. Качественное совпадение графиков инвариантности доказывает, что метод асимптотического усреднения в одночастотных системах общего вида осуществляет корректную редукцию уравнений движения на фазовой плоскости, обеспечивая математическую строгость вычисления медленных составляющих в широких классах нелинейных потенциалов.

## 5. Пример: Численное сопоставление временных масштабов

Ниже приведено численное сопоставление решений исходной дифференциальной системы (5) и преобразованной системы уравнений движения в стандартной форме Ван дер Поля (10) для случая неконсервативного возмущения в виде малого вязкого затухания $g(x,v) = -v$. Графическая визуализация иллюстрирует разделение временных масштабов, при котором высокочастотные колебания координаты $x(t)$ сочетаются с плавным вековым убыванием её огибающей амплитуды $a(t)$.

Выполняется совместное интегрирование исходной системы в декартовых координатах фазового пространства $(x, v)$ и преобразованной системы Ван дер Поля в переменных медленной амплитуды и быстрой фазы $(a, \varphi)$. Это позволяет экспериментально подтвердить различие в скоростях эволюции компонент поля.

Синяя сплошная кривая координаты $x(t)$ совершает быстрые осцилляции с частотой, близкой к невозмущенной частоте линейного базиса $\omega = 2.0$. Скорость изменения данной координаты является величиной порядка $O(1)$, что согласуется с оценкой (12).

Красная штриховая линия амплитуды $a(t)$ образует верхнее огибающее ограничение гармонического профиля колебаний. Изменение амплитуды происходит исключительно медленно со скоростью порядка $O(\varepsilon)$, подтверждая теоретический вывод соотношения (13) о том, что для заметного изменения медленной переменной требуется длительный интервал времени масштаба $T \sim O\left(\frac{1}{\varepsilon}\right)$.


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# Раздел 1: Константы и параметры динамической системы
omega = 2.0  # Собственная частота невозмущенного линейного базиса
eps = 0.1  # Малый параметр нелинейности


# Функция неконсервативного возмущения (малое вязкое затухание)
def g_func(x, v):
    return -v


# Раздел 2: Исходная система дифференциальных уравнений в координатах (x, v)
def original_system(t, state):
    x, v = state
    dxdt = v
    dvdt = -(omega**2) * x + eps * g_func(x, v)
    return [dxdt, dvdt]


# Раздел 3: Преобразованная система уравнений в стандартной форме Ван дер Поля (10)-(11)
def transformed_system(t, state):
    a, phi = state
    
    # Вычисление декартовых проекций фазового пространства для текущего шага
    x_proj = a * np.cos(phi)
    v_proj = -a * omega * np.sin(phi)
    
    # Вычисление правых частей уравнений движения первого приближения
    g_val = g_func(x_proj, v_proj)
    G = -(1.0 / omega) * g_val * np.sin(phi)
    Phi = -(1.0 / (omega * a)) * g_val * np.cos(phi) if a > 1e-12 else 0.0

    dadt = eps * G
    dphidt = omega + eps * Phi
    return [dadt, dphidt]


# Раздел 4: Начальные условия и численное интегрирование
t_span = (0.0, 40.0)
t_eval = np.linspace(t_span[0], t_span[1], 1000)  # Исправлено извлечение границ одномерного вектора

# Интегрирование исходной декартовой системы: x(0) = 2.0, v(0) = 0.0
sol_orig = solve_ivp(original_system, t_span, [2.0, 0.0], t_eval=t_eval)

# Пересчет начальных значений амплитуды и фазы на основе соотношений замены (6)
a0 = np.sqrt(2.0**2 + (0.0 / omega)**2)
phi0 = np.arctan2(-0.0 / omega, 2.0)

# Интегрирование преобразованной системы Ван дер Поля в переменных (a, phi)
sol_trans = solve_ivp(transformed_system, t_span, [a0, phi0], t_eval=t_eval)

# Восстановление медленной огибающей траектории
a_envelope = sol_trans.y[0]

# Раздел 5: Графическое отображение результатов эксперимента
plt.figure(figsize=(11, 5))

# Отображение быстрой переменной декартовой системы
plt.plot(
    sol_orig.t,
    sol_orig.y[0],
    color="royalblue",
    label="Быстрая переменная $x(t)$ (исходная система)",
    alpha=0.8,
)

# Отображение верхней медленной огибающей амплитуды
plt.plot(
    sol_trans.t,
    a_envelope,
    color="crimson",
    linewidth=2.5,
    label="Медленная переменная $a(t)$ (амплитуда)",
)

# Отображение нижней симметричной границы профиля колебаний
plt.plot(
    sol_trans.t,
    -a_envelope,
    color="crimson",
    linewidth=1.5,
    linestyle="--",
    alpha=0.6,
)

# Настройка названий осей координат и заголовка окна
plt.title(r"Разделение масштабов времени в квазилинейной системе ($\varepsilon = 0.1$)", fontsize=12)
plt.xlabel("Время $t$", fontsize=11)
plt.ylabel("Значения переменных", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

## 6. Заключение

В конспекте рассмотрено влияние малых неконсервативных и диссипативных возмущений на динамику нелинейных колебательных систем. На примере квазилинейного уравнения Ван дер Поля продемонстрирована процедура перехода от исходных декартовых координат фазового пространства к медленным переменным амплитуды и быстрой фазы. Математический аппарат асимптотического усреднения Крылова-Боголюбова по быстрому времени позволил изолировать вековые тренды эволюции энергии от высокочастотных пульсаций, амплитуда которых имеет порядок малости $O(\varepsilon)$, а период определяется невозмущенной частотой линейного базиса $O(1)$. 

Выполненное численное моделирование полностью подтвердило теоретические выводы качественной теории дифференциальных уравнений. Анализ знаков усредненного нелинейного поля параболического типа позволил строго классифицировать стационарные режимы системы. Экспериментально доказано, что тривиальное равновесие в начале координат является неустойчивым, в то время как изолированная замкнутая траектория — предельный цикл — выступает в роли асимптотического аттрактора, к которому с разных сторон сходятся внутренний раскручивающийся и внешний закручивающийся фазовые потоки.

Алгоритм усреднения обобщен на существенно нелинейные консервативные системы, где явный вид тригонометрического решения порождающего уравнения априори не конкретизируется. На основе аппарата высшего дифференциального исчисления и анализа интеграла энергии невозмущенной гамильтоновой системы строго доказана инвариантность фазового определителя $\Delta$ относительно быстрой фазы $\varphi$. Численный эксперимент на примере жесткого ангармонического осциллятора Дюффинга подтвердил идеальную постоянную структуру детерминанта перехода на всем интервале фазового времени, что математически обосновывает его вынесение за знак интеграла усреднения первого приближения. Это позволяет эффективно выполнять редукцию размерности фазовых пространств сложных физических и биофизических сред без риска потери канонической структуры уравнений движения.
