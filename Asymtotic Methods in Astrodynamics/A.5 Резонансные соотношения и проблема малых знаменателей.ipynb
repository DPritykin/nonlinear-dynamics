# Резонансные соотношения в динамике ИСЗ и проблема малых знаменателей

Конспект фиксирует математическое определение резонансных соотношений в многомерных динамических системах, исследует алгебраическую структуру проблемы малых знаменателей в асимптотических методах и устанавливает классификацию орбитальных резонансов в задачах динамики искусственных спутников Земли (ИСЗ).

## Введение

При исследовании многомерных колебательных процессов невозмущенное (базовое) состояние системы традиционно описывается суперпозицией нескольких независимых периодических движений. Каждое из этих движений характеризуется своей постоянной частотой $\omega^i$. В простейшем случае траектория такой системы наматывается на инвариантный многомерный тор, совершая условно-периодическое движение.

При введении малых нелинейных возмущений характер совместной эволюции этих периодических процессов зависит от численных соотношений между их частотами. Если частоты независимы (несоизмеримы), возмущенная траектория ведет себя регулярно: возникающие флуктуации остаются ограниченными и могут быть исключены методами асимптотического усреднения [Арнольд, 1985].

Качественное изменение динамики происходит в случае, когда между частотами возникает соизмеримость, то есть когда их линейная комбинация с целыми коэффициентами обращается в ноль. Это явление называется резонансом. На понятийном уровне резонанс означает, что различные периодические движения внутри системы (например, обращение космического аппарата по орбите и вращение планеты вокруг своей оси) синхронизируются, и их фазы начинают стационарно влиять друг на друга на больших интервалах времени.

С точки зрения асимптотического анализа резонанс представляет собой сингулярность. Появление соизмеримости частот приводит к делению на нулевые или исчезающе малые величины в алгоритмах построения возмущений, что известно как проблема малых знаменателей [Арнольд, 1985]. Это приводит к расходимости классических асимптотических рядов и требует перестройки уравнений движения в окрестности резонансных режимов.

Для задач баллистики ИСЗ исследование условий соизмеримости частот имеет прикладное значение. Попадание космического аппарата в резонансные зоны индуцирует долговременные изменения определяющих параметров движения, включая переменные действия Делоне, что накладывает ограничения на область применимости аналитических алгоритмов движения (таких как Брауэр-Лидден, Козаи или модель SGP4).

Настоящий конспект формирует математический аппарат, необходимый для корректного разделения быстрых и медленных переменных в прикладных задачах спутниковой динамики.

## 1. Математическое определение резонанса

Рассмотрим автономную динамическую систему общего вида, описывающую суперпозицию $n$ независимых колебательных или циклических процессов. Фазовое пространство такой системы локально параметризовано упорядоченным набором угловых переменных (фаз) $\theta^1, \theta^2, \dots, \theta^n$, образующих вектор $\theta \in \mathbb{T}^n$:
$$ \theta = \begin{pmatrix} \theta^1 \\ \vdots \\ \theta^n \end{pmatrix} $$

В невозмущенном состоянии эволюция фаз во времени определяется системой обыкновенных дифференциальных уравнений:
$$ \frac{d}{dt}\theta^i = \omega^i(I) \implies \theta^i(t) = \omega^i(I) \cdot t + \theta^i_0, \quad (i = 1, \dots, n) $$
где компоненты $\omega^i(I)$ представляют собой постоянные частоты невозмущенного движения, являющиеся в общем случае функциями от медленно меняющихся параметров или переменных действия $I$. Набор частот формирует вектор $\omega = (\omega^1, \dots, \omega^n)^T$.

Траектория невозмущенного движения (фазовый поток) разворачивается на поверхности многомерного инвариантного тора $\mathbb{T}^n$. Математический характер заполнения поверхности тора траекторией определяется алгебраической структурой вектора частот.

### 1.1. Геометрия невозмущенного потока на торе

Геометрически фазовое пространство $n$ угловых переменных представляет собой $n$-мерный тор $\mathbb{T}^n$. Для простейшего двухчастотного случая ($n = 2$) движение разворачивается на поверхности двумерного тора в трехмерном пространстве. Фазовая траектория совершает одновременно два циклических движения:
* вдоль меридионального сечения тора с частотой $\omega^1$;
* вдоль экваториального сечения тора с частотой $\omega^2$.

Характер расположения траектории на поверхности тора определяется алгебраической структурой отношения частот $\gamma = \omega^1/\omega^2$. Здесь возникают два качественно различающихся геометрических режима:

1. **Эргодический (нерезонансный) режим:** если отношение частот $\gamma$ является иррациональным числом ($\gamma \notin \mathbb{Q}$), частоты называются несоизмеримыми. В этом случае фазовая траектория является незамкнутой: она совершает бесконечную намотку, никогда не возвращается в начальную точку и не пересекает себя. Со временем одна траектория асимптотически плотно заполняет всю двухмерную поверхность тора. Любая окрестность на торе, сколь бы мала она ни была, будет посещена траекторией бесконечно много раз (эргодическое движение) [Чириков, 1979].
2. **Резонансный режим:** если отношение частот $\gamma$ является рациональным числом ($\gamma = \beta/\alpha \in \mathbb{Q}$), частоты называются соизмеримыми. Это означает, что траектория совершает строго целое число витков по меридиану за целое число обходов по экватору. В результате фазовая траектория полностью замыкается сама на себя, образуя замкнутый периодический цикл на торе.

**Определение 1.1.** Между частотами невозмущенного движения существует резонансное соотношение (или система находится в состоянии точного резонанса), если существует ненулевой вектор с целочисленными компонентами $k = (k_1, k_2, \dots, k_n) \in \mathbb{Z}^n \setminus \{\mathbf{0}\}$, удовлетворяющий условию ортогональности вектору частот:
$$ k\omega = \sum_{i=1}^n k_i\omega^i = k_1\omega^1 + k_2\omega^2 + \dots + k_n\omega^n = 0 \tag{1} $$

### 1.2. Структурные параметры резонансных соотношений:

1. **Порядок резонанса:** Скалярная величина $N$, определяемая как сумма модулей целочисленных коэффициентов:
$$ N = \sum_{i=1}^n |k_i| = |k_1| + |k_2| + \dots + |k_n| $$
В асимптотическом анализе порядок $N$ является мерой аналитического веса соответствующей резонансной гармоники возмущения: в силу аналитичности правых частей уравнений движения амплитуды гармоник убывают экспоненциально с ростом их порядка.
2. **Кратность резонанса:** Максимальное количество линейно независимых целочисленных векторов-строк $k^{(j)}$, одновременно удовлетворяющих условию (1) для фиксированного вектора частот $\omega$. Для системы с $n$ частотами максимальная кратность равна $m = n - 1$ [Арнольд, 1985]. При резонансе максимальной кратности все частоты соизмеримы между собой, инвариантный тор полностью вырождается, а все фазовые траектории замыкаются, образуя периодические циклы.

### 1.3. Понятие критического аргумента (резонансной фазы)

В условиях резонанса линейная комбинация угловых переменных, определяемая целочисленным вектором $k$, задает скалярную функцию фазового пространства — критический аргумент (или резонансную фазу) $\psi$:
$$ \psi = k\theta = \sum_{i=1}^n k_i\theta^i $$
Вычислим полную производную по времени от резонансной фазы $\psi$ в силу уравнений невозмущенного движения:
$$ \frac{d\psi}{dt} = k\frac{d\theta}{dt} = k\omega = 0 \implies \psi = \text{const} $$

Физический смысл нелинейного резонанса заключается в следующем: в то время как индивидуальные фазы $\theta^i$ быстро изменяются со скоростями $\omega^i$, их резонансная комбинация $\psi$ остается строго постоянной величиной. Геометрически это означает, что в резонансном режиме многомерный тор вырождается: вместо эргодического скольжения фазовый поток фиксирует строгое пространственное положение траектории, обеспечивая стационарное взаимное влияние пертурбационных сил на больших интервалах времени [Заславский, 1984; Чириков, 1979].


### 1.4. Моделирование геометрической развертки тора в среде Python

Для иллюстрации подразделов 1.1 и 1.2 реализуется вычислительный эксперимент, отображающий геометрию фазового потока условно-периодических движений. Поскольку визуализация $n$-мерного инвариантного тора при $n \ge 3$ не реализуема на плоскости, траектория фазового потока отображается с помощью плоской развертки инвариантного тора в пространстве угловых переменных $(\theta^1, \theta^2)$.

В силу периодичности фазового пространства по углам с периодом $2\pi$, плоская развертка тора представляет собой квадрат $[0, 2\pi] \times [0, 2\pi]$, у которого противоположные границы геометрически отождествлены (склеены). Приращение времени приводит к линейному перемещению изображающей точки на плоскости. При пересечении границы тора (значения $2\pi$) операция взятия остатка от деления (`% 2 * np.pi`) возвращает траекторию на противоположную сторону квадрата.

Ниже приведена программная реализация, сопоставляющая геометрию траекторий для соизмеримого (резонансного) и несоизмеримого (эргодического) векторов невозмущенных частот.


```python
# Вычислительный эксперимент к Разделу 1: Визуализация геометрии намоток тора
import numpy as np
import matplotlib.pyplot as plt

def run_torus_winding_visualization():
    # 1. Задание параметров временной сетки и фундаментальной области тора [0, 2π]
    t_res = np.linspace(0, 10, 1000)  # Короткий интервал для резонансного цикла
    t_erg = np.linspace(0, 60, 5000)  # Длинный интервал для заполнения тора
    
    # 2. Определение частот для двух демонстрационных режимов
    # Панель А: Рациональный резонанс 3:2 (соизмеримые частоты)
    w1_res, w2_res = 3.0, 2.0
    theta1_res = (w1_res * t_res) % (2 * np.pi)
    theta2_res = (w2_res * t_res) % (2 * np.pi)
    
    # Панель Б: Иррациональный эргодический режим (несоизмеримые частоты)
    w1_erg, w2_erg = np.sqrt(5), 1.0
    theta1_erg = (w1_erg * t_erg) % (2 * np.pi)
    theta2_erg = (w2_erg * t_erg) % (2 * np.pi)
    
    # 3. Инициализация графического окна
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 7))
    
    # --- Панель А: Рациональный резонанс ---
    # Нанесение точек через scatter для исключения ложных линий на склейках границ
    ax1.scatter(theta1_res, theta2_res, color='crimson', s=3.0, label='Резонансный трек')
    # Маркируем начальную и конечную точки, фиксируя геометрическое замыкание цикла
    ax1.plot(theta1_res[0], theta2_res[0], 'ko', markersize=8, label='Начало / Конец цикла')
    ax1.set_title(r'А) Резонансный режим ($\omega^1 : \omega^2 = 3 : 2 \in \mathbb{Q}$)')
    
    # --- Панель Б: Иррациональная эргодическая намотка ---
    ax2.scatter(theta1_erg, theta2_erg, color='darkcyan', s=1.0, alpha=0.6, label='Эргодический трек')
    ax2.set_title(r'Б) Эргодический режим ($\omega^1 : \omega^2 = \sqrt{5} : 1 \notin \mathbb{Q}$)')
    
    # 4. Единое оформление обеих координатных панелей
    for ax in [ax1, ax2]:
        ax.set_xlabel(r'Фазовый угол $\theta^1$, рад', fontsize=11)
        ax.set_ylabel(r'Фазовый угол $\theta^2$, рад', fontsize=11)
        ax.set_xlim(0, 2 * np.pi)
        ax.set_ylim(0, 2 * np.pi)
        ax.set_xticks([0, np.pi, 2 * np.pi])
        ax.set_xticklabels(['0', r'$\pi$', r'$2\pi$'])
        ax.set_yticks([0, np.pi, 2 * np.pi])
        ax.set_yticklabels(['0', r'$\pi$', r'$2\pi$'])
        ax.grid(True, linestyle=':', alpha=0.5)
        ax.set_box_aspect(1)  # Квадратные пропорции для сохранения геометрии углов
        ax.legend(loc='upper right', frameon=True, fontsize=10)
        
    plt.suptitle('Развертка двухчастотного инвариантного тора на плоскость фазовых углов', fontsize=13, y=0.98)
    plt.tight_layout()
    plt.show()

run_torus_winding_visualization()
```

### 1.5. Качественный анализ результатов вычислительного эксперимента

Построенные графические панели отражают топологические различия между нерезонансным и резонансным состояниями невозмущенного фазового пространства:

1. **Визуализация эргодичности (Панель Б):** Поскольку отношение частот $\gamma = \sqrt{5}$ является иррациональным числом, фазовые углы совершают независимые циклические движения. На развертке тора траектория совершает бесконечную незамкнутую намотку, формируя систему параллельных непересекающихся прямых линий с иррациональным углом наклона. Численный эксперимент подтверждает свойство эргодичности: приращение времени приводит к асимптотически плотному заполнению всей доступной двухмерной площади квадрата $[0, 2\pi] \times [0, 2\pi]$. Произвольно малая окрестность любой точки на торе посещается траекторией бесконечно много раз. В рамках теории возмущений это гарантирует регулярный характер интегрирования правых частей обыкновенных дифференциальных уравнений вдоль инвариантного многообразия [Чириков, 1979].
2. **Геометрическое вырождение инвариантного многообразия (Панель А):** При соизмеримых частотах отношение является рациональным числом ($\gamma = 3/2$). Траектория совершает ровно 3 полных витка по меридиональному углу $\theta^1$ за 2 полных обхода по экваториальному углу $\theta^2$. Визуализация фиксирует замыкание фазового трека в начальной точке $(0, 0)$ за конечный интервал времени. Движение вырождается из двухмерного (поверхностного) в одномерное (линейный периодический цикл). Фазовый поток занимает нулевую площадь на развертке тора, оставляя остальное пространство инвариантного многообразия полностью недоступным. Данная геометрическая конфигурация демонстрирует пространственную фиксацию критического аргумента $\psi = 2\theta^1 - 3\theta^2 = \mathrm{const}$ и обосновывает необходимость перехода к локальному анализу нелинейного резонанса.


## 2. Классификация по свойствам невозмущенной системы

Характер эволюции фазовых траекторий в окрестности резонансной поверхности определяется дифференциальными и геометрическими свойствами невозмущенной системы [Арнольд, 1985].

Если невозмущенное движение описывается автономной системой обыкновенных дифференциальных уравнений общего вида, зависимость вектора частот $\omega$ от параметров или переменных действия $I$ задается матрицей Якоби отображения частот:
$$ \mathbf{M} = \frac{\partial\omega}{\partial I} \tag{2} $$

В случае канонических гамильтоновых систем вектор частот жестко определен через частные производные невозмущенного гамильтониана $H_0(I)$: $\omega(I) = (\nabla_I H_0)^T$. В силу этого матрица Якоби отображения частот тождественно совпадает с квадратной симметричной матрицей Гессе функции $H_0(I)$ размерности $n \times n$:
$$ \mathbf{M}_{\mathrm{hess}} = \frac{\partial\omega}{\partial I} = \frac{\partial^2 H_0}{\partial I_i \partial I_j} = \begin{pmatrix} \frac{\partial^2 H_0}{\partial I_1 \partial I_1} & \dots & \frac{\partial^2 H_0}{\partial I_1 \partial I_n} \\ \vdots & \ddots & \vdots \\ \frac{\partial^2 H_0}{\partial I_n \partial I_1} & \dots & \frac{\partial^2 H_0}{\partial I_n \partial I_n} \end{pmatrix} \tag{3} $$

В зависимости от ранга и определителя матрицы Гессе $\mathbf{M}_{\mathrm{hess}}$ (или матрицы Якоби $\mathbf{M}$ для негамильтоновых систем) резонансные соотношения разделяются на два класса, качественно изменяющих геометрию фазового пространства.

### 2.1. Линейный (изохронный) резонанс

Линейный резонанс имеет место в динамических системах, где вектор невозмущенных частот $\omega$ инвариантен относительно канонических действий или параметров $I$. Дифференциальным критерием изохронности является тождественное зануление матрицы производных на всей области определения фазового пространства:
$$ \mathbf{M}_{\mathrm{hess}} = \frac{\partial\omega}{\partial I} \equiv \mathbf{0}_{n \times n} \tag{4} $$

Для гамильтоновых систем классическим примером является линейный гармонический осциллятор, невозмущенный гамильтониан которого линейно зависит от импульсов действия: $H_0(I) = \sum \omega^i_0 I_i$.

* **Геометрическая интуиция:** Поскольку частоты не зависят от действий, в изохронном пространстве все концентрические инвариантные торы обладают абсолютно идентичной геометрической структурой намотки траекторий. При попадании в условия соизмеримости вырождение (замыкание траектории в периодический цикл) происходит одновременно во всем фазовом объеме на всех торах сразу (Панель А рисунка к Разделу 2.3).
* **Динамическое свойство:** При добавлении к изохронной системе внешнего периодического возмущения, частота которого удовлетворяет резонансному условию (1), переменные действия демонстрируют неограниченный линейный рост по времени ($I_i \sim t$). Кинематический механизм для изменения собственных частот и самопроизвольного выхода системы из резонансного синхронизма в изохронной структуре отсутствует, что приводит к неограниченной накачке энергии.

### 2.2. Нелинейный (анизохронный) резонанс

Нелинейный резонанс является общим случаем для задач небесной механики и спутниковой астродинамики. Частоты невозмущенного движения $\omega(I)$ являются функциями переменных действия, что в гамильтоновом случае выражается через условие невырожденности матрицы Гессе. (условие невырожденности Колмогорова):
$$ \det(\mathbf{M}_{\mathrm{hess}}) = \det\left(\frac{\partial^2 H_0}{\partial I_i \partial I_j}\right) \neq 0 \tag{5} $$

* **Геометрическая интуиция:** Условие невырожденности (5) означает наличие частотного градиента в пространстве действий: геометрия намотки непрерывно изменяется при переходе от одного инвариантного тора к другому (вдоль радиусов пространства действий $I$). Точное резонансное соотношение (1) выполняется не во всем фазовом пространстве, а локализуется на изолированной геометрической гиперповерхности - резонансной поверхности действия $I = I^*$. Соседние торы остаются нерезонансными (Панель Б рисунка к Разделу 2.3).
* **Динамическое свойство:** Под действием резонансной гармоники возмущения переменные действия $I_i$ начинают испытывать медленные вариации, смещая изображающую точку с резонансной поверхности $I^*$. В силу критерия анизохронности (5) изменение действий индуцирует сдвиг вектора собственных частот $\omega(I)$. Точное резонансное соотношение (1) нарушается, и система выходит из точного резонанса. Нарушение соизмеримости частот останавливает направленный рост переменных действия. В результате в фазовом пространстве исходные инвариантные торы деформируются, и в окрестности резонансной поверхности возникают ограниченные колебательные режимы - либрации критического аргумента (резонансной фазы) вокруг устойчивого положения равновесия [Чириков, 1979].


### 2.3. Моделирование частотного градиента анизохронного пространства в среде Python

Для верификации геометрического различия между изохронными и анизохронными структурами реализуется вычислительный эксперимент, визуализирующий векторное поле невозмущенных частот в координатах действие-угол $(\theta, I)$.

Программа строит локальные направления фазовых скоростей для двух моделей:
* В изохронной модели частота постоянна: независимо от величины действия $\omega = 1.5$. Угол наклона фазовых траекторий инвариантен на всех уровнях импульсов.
* В анизохронной модели частота нелинейно зависит от действия: $\omega(I) = I^2$. По мере роста координаты действия вектор частоты удлиняется, непрерывно изменяя геометрический наклон интегральных кривых. Точный резонанс локализуется строго на одной изолированной резонансной линии $I^* = 1.2$.


```python
# Вычислительный эксперимент к Разделу 2.3: Визуализация частотного градиента
import numpy as np
import matplotlib.pyplot as plt

def run_anisotropy_gradient_visualization():
    # 1. Формирование координатной сетки фазового пространства (Действие - Угол)
    theta_vec = np.linspace(0, 2 * np.pi, 14)
    I_vec = np.linspace(0.5, 2.0, 14)
    Theta_mesh, I_mesh = np.meshgrid(theta_vec, I_vec)
    
    # 2. Инициализация графического окна
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6.5))
    
    # --- Панель А: Изохронная структура ---
    # Траектории имеют одинаковый наклон на всех уровнях действий
    omega_iso = np.ones_like(I_mesh) * 1.5
    dI_iso = np.ones_like(I_mesh) * 0.4  # Фиктивный наклон для визуализации направлений
    
    # Нормализация векторов направления
    norm_iso = np.hypot(omega_iso, dI_iso)
    ax1.quiver(Theta_mesh, I_mesh, omega_iso/norm_iso, dI_iso/norm_iso, 
              color='navy', alpha=0.6, pivot='mid', scale=18)
    
    # В изохронном пространстве резонанс размыт по всему объему
    ax1.fill_between([0, 2*np.pi], 0.5, 2.0, color='crimson', alpha=0.08, 
                    label='Область тождественного резонанса')
    ax1.set_title(r'А) Изохронное пространство ($\mathbf{M}_{\mathrm{hess}} \equiv \mathbf{0}_{n \times n}$)')
    
    # --- Панель Б: Анизохронная структура ---
    # Наклон траекторий непрерывно изменяется вместе с изменением частоты ω(I)
    omega_aniso = I_mesh ** 2  # Квадратичная зависимость для усиления визуального эффекта
    dI_aniso = np.ones_like(I_mesh) * 0.8
    norm_aniso = np.hypot(omega_aniso, dI_aniso)
    ax2.quiver(Theta_mesh, I_mesh, omega_aniso/norm_aniso, dI_aniso/norm_aniso,
              color='darkgreen', alpha=0.6, pivot='mid', scale=18)
    
    # Локализация точной резонансной поверхности при I* = 1.2
    ax2.axhline(1.2, color='crimson', linestyle='--', linewidth=2.5, 
               label=r'Резонансная поверхность $I^* = 1.2$ ($\omega = \omega_{\mathrm{res}}$)')
    ax2.set_title(r'Б) Анизохронное пространство ($\det \mathbf{M}_{\mathrm{hess}} \neq 0$)')
    
    # 3. Академическое оформление координатных панелей
    for ax in [ax1, ax2]:
        ax.set_xlabel(r'Угловая переменная $\theta$, рад', fontsize=11)
        ax.set_ylabel(r'Переменная действия $I$', fontsize=11)
        ax.set_xlim(0, 2 * np.pi)
        ax.set_ylim(0.5, 2.0)
        ax.set_xticks([0, np.pi, 2 * np.pi])
        ax.set_xticklabels(['0', r'$\pi$', r'$2\pi$'])
        ax.grid(True, linestyle=':', alpha=0.5)
        ax.legend(loc='upper right', frameon=True, fontsize=10)
        ax.set_box_aspect(1)
        
    plt.suptitle('Градиент наклона невозмущенных частот в фазовом пространстве «действие—угол»', fontsize=13, y=0.98)
    plt.tight_layout()
    plt.show()

run_anisotropy_gradient_visualization()
```

### 2.4. Качественный анализ частотных полей

Визуализация поля направлений раскрывает кинематические причины качественных различий между двумя типами невозмущенных гамильтонианов:

1. **Глобальная неустойчивость изохронной сетки (Панель А):** На левой панели стрелки фазовых скоростей строго параллельны и имеют одинаковую длину на всех уровнях оси ординат. Если внешняя возмущающая сила вступает в резонанс с этой фиксированной частотой, деформация затронет весь фазовый объем одновременно. Поскольку частота не изменяется при сдвиге действия, система лишена возможности десинхронизации, что приводит к глобальной расходимости амплитуд во всем пространстве.
2. **Локализация и стабилизация анизохронного слоя (Панель Б):** На правой панели отчетливо прослеживается веерная деформация направления стрелок. В нижних слоях ($I < 1.0$) вращение происходит медленнее, в верхних ($I > 1.0$) - быстрее. Точный резонанс вырождается в геометрическую линию $I^* = 1.2$. Если возмущение начинает выталкивать траекторию вверх от красной линии, изображающая точка попадает в слой с более высокими скоростями вращения. Фаза начинает набегать быстрее, система выходит из резонансного синхронизма, а направленный рост импульса прекращается. Данный градиент частот является естественным стабилизатором, удерживающим нелинейные возмущения в ограниченных маятниковых зонах [Чириков, 1979].


### 2.5. Вычислительный эксперимент: Бифуркация нелинейного резонанса средних движений

Для верификации динамических свойств анизохронного фазового пространства рассмотрим движение спутника Земли в окрестности резонанса средних движений (РСД) 2:1 с внешним возмущающим телом, обращающимся по круговой орбите с постоянной частотой $n_2$. Условие соизмеримости частот (1) имеет вид $n_1^* - 2n_2 = 0$, где $n_1^* = \sqrt{\mu/a^{*3}}$ - частота точной соизмеримости, а $a^*$ - соответствующее резонансное значение большой полуоси.

В окрестности резонансной поверхности уравнения изменения канонической переменной действия Делоне $L = \sqrt{\mu a}$ (характеризующей полную энергию Кеплеровой орбиты) и медленной резонансной фазы $\psi = \theta_1 - 2\theta_2$ принимают вид:
$$ \frac{d\psi}{dt} = n_1(L) - 2n_2 = \frac{\mu^2}{L^3} - 2n_2, \quad \frac{dL}{dt} = -F \sin \psi + \alpha_{\mathrm{t}} $$
где $F$ - амплитуда резонансной гармоники геопотенциала, а $\alpha_{\mathrm{t}}$ - неконсервативный приливной момент, описывающий вековой перенос момента импульса от вращения планеты к орбитальному движению спутника.

Реализуем численное интегрирование системы (2) для двух качественных конфигураций:
1. **Чисто гравитационное взаимодействие ($\alpha_{\mathrm{t}} = 0$):** Демонстрирует стабилизирующее свойство анизохронности. Приращение действия мгновенно изменяет частоту Кеплерова движения $n_1(L)$, нарушая соизмеримость частот и переводя систему в режим ограниченных маятниковых либраций.
2. **Адиабатический захват с приливной накачкой энергии ($\alpha_{\mathrm{t}} \neq 0$):** Моделирует нелинейный захват по Нейштадту. Внешняя диссипативная сила компенсирует частотный сдвиг, фаза блокируется в смещенном фокусе, обеспечивая монотонную закачку энергии в орбиту и вековой рост большой полуоси $a(t)$.


```python
# Финальный стабильный трехпанельный симулятор РСД 2:1 для Раздела 2.3
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def run_mean_motion_resonance_plots():
    # 1. Физические константы модели в полных переменных действия Делоне L
    G_param = -3.0     # Кривизна кеплеровой потенциальной ямы
    F_param = 0.04     # Амплитуда гравитационного резонансного возмущения
    L_star = 1.0       # Точный резонансный уровень действия (a* = 1.0)
    
    t_span = (0, 600)
    t_eval = np.linspace(t_span[0], t_span[1], 5000)
    
    # --- ПАНЕЛЬ А: Чисто гравитационное консервативное взаимодействие (α_t = 0) ---
    def system_conservative(t, state):
        psi, L = state
        dpsi_dt = G_param * (L - L_star)
        dL_dt = -F_param * np.sin(psi)
        return [dpsi_dt, dL_dt]
    
    # Траектория либрации (красная) и ротации (синяя)
    sol_cons_lib = solve_ivp(system_conservative, t_span, [0.5, 1.00], t_eval=t_eval, method='DOP853', rtol=1e-10, atol=1e-10)
    sol_cons_rot = solve_ivp(system_conservative, t_span, [0.0, 1.15], t_eval=t_eval, method='DOP853', rtol=1e-10, atol=1e-10)
    
    a_cons_lib = (sol_cons_lib.y[1, :])**2
    a_cons_rot = (sol_cons_rot.y[1, :])**2
    
    # --- ПАНЕЛИ Б и В: Адиабатический захват и приливной увод (α_t ≠ 0) ---
    alpha_t = 0.0002   # Уменьшенная скорость дрейфа для удержания синей кривой в кадре
    
    def system_dissipative(t, state):
        psi, L = state
        dpsi_dt = G_param * (L - L_star)
        dL_dt = -F_param * np.sin(psi) + alpha_t
        return [dpsi_dt, dL_dt]
    
    # Смещенный устойчивый фокус Пуанкаре для гарантированного захвата
    psi_stable_focus = np.arcsin(alpha_t / F_param)
    
    # Смещаем начальное действие пролетного трека ближе к резонансу, чтобы увидеть прохождение
    sol_diss_pass = solve_ivp(system_dissipative, t_span, [0.0, 1.05], t_eval=t_eval, method='DOP853', rtol=1e-10, atol=1e-10)
    sol_diss_lock = solve_ivp(system_dissipative, t_span, [float(psi_stable_focus), 1.00], t_eval=t_eval, method='DOP853', rtol=1e-10, atol=1e-10)
    
    a_diss_pass = (sol_diss_pass.y[1, :])**2
    a_diss_lock = (sol_diss_lock.y[1, :])**2
    
    # 2. Инициализация графического окна (1 строка, 3 столбца)
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20, 6.5))
    
    # Отрисовка Панели А
    ax1.plot(sol_cons_rot.t, a_cons_rot, color='navy', linewidth=1.8, label='Свободная ротация фазы')
    ax1.plot(sol_cons_lib.t, a_cons_lib, color='crimson', linewidth=2.0, label='Резонансная либрация')
    ax1.set_title(r'А) Консервативная либрация ($\alpha_{\mathrm{t}} = 0$)', fontsize=12)
    ax1.set_ylim(0.85, 1.6)
    
    # Отрисовка Панели Б
    ax2.plot(sol_diss_pass.t, a_diss_pass, color='navy', linewidth=1.8, label='Транзитный пролет')
    ax2.plot(sol_diss_lock.t, a_diss_lock, color='crimson', linewidth=2.2, label='Адиабическое удержание')
    a_drift_theory = (1.0 + alpha_t * t_eval)**2
    ax2.plot(t_eval, a_drift_theory, 'g--', linewidth=1.8, label='Линия свободного дрейфа $a^*(t)$')
    ax2.set_title(r'Б) Приливной адиабатический разгон ($\alpha_{\mathrm{t}} \neq 0$)', fontsize=12)
    ax2.set_ylim(0.85, 1.6)
    
    # Отрисовка Панели В: Фазовый портрет диссипативной системы (затягивание в фокус)
    # Приведение угловой координаты к фундаментальной области [0, 2π]
    psi_pass_wrapped = sol_diss_pass.y[0, :] % (2 * np.pi)
    psi_lock_wrapped = sol_diss_lock.y[0, :] % (2 * np.pi)
    
    ax3.scatter(psi_pass_wrapped, a_diss_pass, color='navy', s=1.0, alpha=0.5, label='Транзитный трек')
    ax3.plot(psi_lock_wrapped, a_diss_lock, color='crimson', linewidth=2.0, label='Захваченный трек')
    ax3.plot(psi_stable_focus, 1.0, 'go', markersize=8, label='Устойчивый фокус Пуанкаре')
    ax3.set_title('В) Расслоение потока на фазовой плоскости', fontsize=12)
    ax3.set_xlabel(r'Резонансная фаза $\psi$, рад', fontsize=11)
    ax3.set_xlim(0, 2 * np.pi)
    ax3.set_xticks([0, np.pi, 2 * np.pi])
    ax3.set_xticklabels(['0', r'$\pi$', r'$2\pi$'])
    ax3.set_ylim(0.85, 1.6)
    
    # Общее оформление временных панелей
    for ax in [ax1, ax2]:
        ax.set_xlabel('Системное время $t$', fontsize=11)
    for ax in [ax1, ax2, ax3]:
        ax.axhline(1.0, color='black', linestyle=':', alpha=0.4, label='Резонансный уровень $a^* = 1.0$')
        ax.set_ylabel('Большая полуось спутника $a(t)$', fontsize=11)
        ax.grid(True, linestyle=':', alpha=0.5)
        ax.legend(loc='upper left', frameon=True, fontsize=10)
        ax.set_box_aspect(0.8)
        
    plt.suptitle('Динамика нелинейного резонанса средних движений 2:1 при наличии приливной диссипации', fontsize=13, y=0.98)
    plt.tight_layout()
    plt.show()

run_mean_motion_resonance_plots()
```

### 2.6. Качественный анализ и физическая декомпозиция нелинейных резонансных режимов

Сравнительный анализ временных зависимостей больших полуосей $a(t)$ и топологии фазовой плоскости раскрывает физическую суть условия невырожденности Колмогорова (анизохронности) и демонстрирует качественное различие между консервативным откликом и диссипативным адиабатическим удержанием действия:

#### 1. Механизм консервативного самоограничения импульса (Панель А)
На левой панели при отсутствии неконсервативных сил ($\alpha_{\mathrm{t}} = 0$) непрерывная линейная накачка энергии, характерная для линейных осцилляторов, не реализуется. Как только гравитационное возмущение смещает действие Делоне вверх, большая полуось спутника начинает расти (красный трек). В силу критерия анизохронности (5) частота кеплерова движения падает, точное условие соизмеримости частот $n_1 - 2n_2 = 0$ нарушается, и система выходит из резонансного синхронизма. Возвращающая нелинейная сила маятникового рельефа $-F\sin\psi$ меняет свой знак, останавливает рост импульса и возвращает большую полуось назад к уровню соизмеримости. 

Переменная действия оказывается заперта внутри границ маятниковой либрации, совершая периодические колебания конечной амплитуды вокруг стационарного уровня $a^* = 1.0$, и в среднем по времени не испытывает векового роста [Чириков, 1979]. Траектория свободной ротации фазы (синяя кривая) совершает регулярные нерезонансные флуктуации вокруг исходного энергетического уровня.

#### 2. Термодинамический механизм приливно-фрикционной накачки энергии (Панель Б)
Включение постоянного неконсервативного момента $\alpha_{\mathrm{t}} = 0.0002$ качественно перестраивает топологию фазового потока. Приливное трение в вязкоупругом теле центральной планеты приводит к диссипации механической энергии всей системы, однако выступает в роли фрикционной муфты сцепления, передающей момент импульса от быстрого осевого вращения планеты к орбитальному движению ИСЗ [Боголюбов, 1955]. 

Поскольку планета вращается вокруг своей оси быстрее, чем спутник обращается по орбите ($\Omega > n_1$), вязкое трение заставляет приливной горб на поверхности планеты смещаться вперед по ходу движения космического аппарата [Арнольд, 1985]. Этот горб гравитационно увлекает спутник за собой, совершая над ним положительную механическую работу и вкачивая энергию в орбиту, что заставляет большую полуось $a(t)$ монотонно расти. Закачка энергии в орбиту происходит за счет векового торможения вращения самой планеты. 

Вне условий соизмеримости (синяя кривая транзитного пролета) этот неконсервативный момент монотонно увеличивает большую полуось спутника, уводя ее прочь от стационарного резонанса вдоль линии свободного дрейфа действия $a^*(t)$ [Нейштадт, 1975].

#### 3. Эффект адиабатического удержания действия по Нейштадту (Панели Б и В)
Качественное изменение динамики происходит для траекторий, попавших внутрь сепаратрисного контура. На Панелях Б и В показано расслоение фазового потока. Красный трек адиабатически захватывается в резонансную ловушку Нейштадта. Медленная резонансная фаза $\psi$ теряет независимость и блокируется в окрестности устойчивого фокуса Пуанкаре (зеленый маркер на Панели В), где выполняется точный силовой баланс:
$$ \sin\psi_{\mathrm{s}} = \frac{\alpha_{\mathrm{t}}}{F} $$

Стабилизация фазы фиксирует знак возвращающей гравитационной силы геопотенциала $-F\sin\psi_{\mathrm{s}} < 0$. Внутри ловушки эта возвращающая сила направлена строго противоположно приливному моменту накачки и полностью компенсирует вековое приливное расталкивание. Большая полуось захваченного спутника (красная кривая) полностью лишается автономности и принудительно удерживается на стационарном уровне соизмеримости $a^* = 1.0$ на всем интервале прогнозирования вопреки внешнему силовому воздействию. Фазовая кривая на Панели В демонстрирует асимптотическое закручивание траектории в спираль вокруг устойчивого центра, подтверждая устойчивость нелинейной адиабатической ловушки под влиянием слабой диссипации [Нейштадт, 1975].


## 3. Алгебраическая структура проблемы малых знаменателей

В рамках классической гамильтоновой теории возмущений полный гамильтониан возмущенной системы представляется в виде формального степенного ряда по малому параметру $\varepsilon \ll 1$:
$$ H(I, \theta) = H_0(I) + \varepsilon H_1(I, \theta) + \mathcal{O}(\varepsilon^2) \tag{6} $$
где функция возмущения $H_1$ периодична по контравариантным углам $\theta \in \mathbb{T}^n$.

Для исключения зависимости уравнений от быстро вращающихся углов в первом порядке по $\varepsilon$ сконструируем каноническое преобразование от старых переменных $(\theta, I)$ к новым усредненным переменным $(\bar{\theta}, \bar{I})$. Зададим это преобразование с помощью производящей функции смешанного типа $S(\bar{\theta}, I)$, близкой к тождественной [Garfinkel, 1966]:
$$ S(\bar{\theta}, I) = \bar{\theta} \cdot I + \varepsilon S_1(\bar{\theta}, I) + \mathcal{O}(\varepsilon^2) \tag{7} $$
где $S_1(\bar{\theta}, I)$ - искомая функция возмущения в уравнениях преобразования.

Связь между старыми и новыми каноническими переменными определяется дифференциальными соотношениями Якоби [Арнольд, 1985]:
$$ \theta = \frac{\partial S}{\partial I} = \bar{\theta} + \varepsilon \frac{\partial S_1}{\partial I} + \mathcal{O}(\varepsilon^2), \quad \bar{I} = \frac{\partial S}{\partial \bar{\theta}} = I + \varepsilon \frac{\partial S_1}{\partial \bar{\theta}} + \mathcal{O}(\varepsilon^2) \tag{8} $$

Подставим разложения (8) в гамильтониан (6) и потребуем, чтобы преобразованный гамильтониан $\bar{H}$ первого порядка не зависел от средних углов $\bar{\theta}$, то есть принял усредненную по тору форму:
$$ \bar{H}(\bar{I}) = H_0(\bar{I}) + \varepsilon \bar{H}_1(\bar{I}) + \mathcal{O}(\varepsilon^2) $$
где $\bar{H}_1(\bar{I}) = \langle H_1 \rangle_{\bar{\theta}}$. Разлагая функцию $H_0(I)$ в ряд Тейлора в окрестности $\bar{I}$, получаем линейное уравнение в частных производных первого порядка относительно функции $S_1$, называемое уравнением гомологии:
$$ \sum_{i=1}^n \frac{\partial H_0}{\partial \bar{I}_i} \frac{\partial S_1}{\partial \bar{\theta}^i} = -\sum_{i=1}^n \omega^i(\bar{I}) \frac{\partial S_1}{\partial \bar{\theta}^i} = \bar{H}_1(\bar{I}) - H_1(\bar{I}, \bar{\theta}) \tag{9} $$

Разложим известную функцию возмущения $H_1(\bar{I}, \bar{\theta})$ и искомую функцию возмущения $S_1(\bar{I}, \bar{\theta})$ в многомерные ряды Фурье по углам $\bar{\theta}$:
$$ H_1(\bar{I}, \bar{\theta}) = \sum_{k \in \mathbb{Z}^n} h_k(\bar{I})e^{ik\bar{\theta}}, \quad S_1(\bar{I}, \bar{\theta}) = \sum_{k \in \mathbb{Z}^n \setminus \{\mathbf{0}\}} s_k(\bar{I})e^{ik\bar{\theta}} $$
где $k = (k_1, k_2, \dots, k_n)$ - целочисленный вектор-строка, а операция $k\bar{\theta} = \sum k_i\bar{\theta}^i$ представляет собой скалярное произведение.

Подстановка рядов Фурье в уравнение гомологии (9) и сопоставление коэффициентов при одинаковых экспонентах $e^{ik\bar{\theta}}$ определяет решение для комплексных амплитуд Фурье производящей функции:
$$ s_k(\bar{I}) = i \frac{h_k(\bar{I})}{k\omega(\bar{I})} \tag{10} $$

### Феномен малых знаменателей

Присутствие скалярного произведения $k\omega(\bar{I}) = k_1\omega^1 + \dots + k_n\omega^n$ в знаменателе алгебраической структуры (10) порождает проблему сходимости асимптотических рядов теории возмущений [Заславский, 1984]:

1. **Точный резонанс ($k\omega(\bar{I}) = 0$):** Если вектор частот невозмущенного движения удовлетворяет условию соизмеримости, аналитическое решение для амплитуды $s_k(\bar{I})$ в рамках классического алгоритма Якоби не существует. Ряды теории возмущений разрываются, поскольку классическое гладкое каноническое преобразование в этой точке пространства действий не конструируется.
2. **Острый резонанс ($k\omega(\bar{I}) \approx 0$):** Поскольку множество резонансных поверхностей всюду плотно в пространстве действий, для любого тора с иррациональным отношением частот существуют целочисленные векторы-строки с большими компонентами, для которых знаменатель $k\omega(\bar{I})$ близок к нулю. Значительный рост коэффициентов $(k\omega)^{-1}$ компенсирует множитель малости $\varepsilon^m$ в высших порядках асимптотического разложения. Это приводит к расходимости классических рядов теории возмущений (теорема Пуанкаре о неинтегрируемости гамильтоновых систем общего вида).


### 3.1. Вычислительный эксперимент: Моделирование резонансной сингулярности в SymPy

**Описание эксперимента:** Целью эксперимента является верификация алгебраического механизма возникновения малых знаменателей при аналитическом решении уравнения гомологии (9). В ходе симуляции проверяется гипотеза о потере регулярности производящей функции канонического преобразования при приближении к резонансной поверхности и возникновении неустранимого деления на ноль при выходе на точную соизмеримость частот.

**Методика проверки:** Математическое ядро строится в изолированной символьной среде библиотеки `SymPy`. Задается модельная двухчастотная система, функция возмущения $H_1$ которой содержит комбинационную гармонику с ковариантным целочисленным вектором $k = (1, -2)$, что соответствует резонансу третьего порядка ($N = |1| + |-2| = 3$).

Программа последовательно реализует три шага:
1. Алгебраическое нахождение аналитической структуры компоненты производящей функции $S_1$ из уравнения гомологии.
2. Численная подстановка частот, близких к острому резонансу ($\omega^1 = 2.001, \omega^2 = 1.0$), для фиксации масштаба роста числовых коэффициентов.
3. Численная подстановка частот точного резонанса $1:2$ ($\omega^1 = 2.0, \omega^2 = 1.0$) с перехватом системного исключения.


```python
# Исправленная ячейка вычислительного эксперимента к Разделу 3.1
import sympy as sp
from IPython.display import display, Markdown

def run_resonance_singularity_experiment():
    # ИСПРАВЛЕНО: Строгая LaTeX-нотация для средних переменных Якоби
    theta1, theta2 = sp.symbols(r'\bar{\theta}^1 \bar{\theta}^2', real=True)
    I1, I2 = sp.symbols(r'\bar{I}_1 \bar{I}_2', real=True)
    omega1, omega2 = sp.symbols(r'\omega^1 \omega^2', positive=True)
    
    display(Markdown("### Результаты символьного моделирования уравнения гомологии"))
    
    # Определение комбинационной гармоники функции возмущения H_1 с вектором k = (1, -2)
    k1, k2 = 1, -2
    h_k = I1 * sp.sqrt(I2) 
    H_1 = h_k * sp.cos(k1 * theta1 + k2 * theta2)
    
    display(Markdown("#### 1. Комбинационная гармоника функции возмущения $H_1$:"))
    display(sp.Eq(sp.Symbol('H_1'), H_1))
    
    # Аналитическое решение уравнения гомологии в вещественной форме
    denom = k1 * omega1 + k2 * omega2
    S_1 = (h_k / denom) * sp.sin(k1 * theta1 + k2 * theta2)
    
    display(Markdown("#### 2. Компонента аналитической производящей функции первого порядка $S_1$:"))
    display(sp.Eq(sp.Symbol('S_1'), S_1))
    
    # Моделирование острого резонанса с устранением численной погрешности float
    S_close = S_1.subs({omega1: 2.001, omega2: 1.000})
    # Округляем только числовой коэффициент в выражении
    W_coeff = float(1 / (1 * 2.001 - 2 * 1.0))
    S_close_clean = sp.Mul(round(W_coeff, 1), h_k, sp.sin(k1 * theta1 + k2 * theta2), evaluate=False)
    
    display(Markdown("#### 3. Производящая функция в окрестности резонанса ($\omega^1 = 2.001, \omega^2 = 1.0$):"))
    display(sp.Eq(sp.Symbol('S_{1,close}'), S_close_clean))
    
    # Моделирование точного резонанса (изолированная обработка сингулярности)
    display(Markdown("#### 4. Попытка расчета точного резонанса $1:2$ ($\omega^1 = 2.0, \omega^2 = 1.0$):"))
    try:
        # Прямая подстановка в знаменатель вызовет ZeroDivisionError, перехватываем её
        if (k1 * 2.0 + k2 * 1.0) == 0:
            raise ZeroDivisionError
        S_exact = S_1.subs({omega1: 2.0, omega2: 1.0})
        display(S_exact)
    except ZeroDivisionError:
        # ИСПРАВЛЕНО: Строгое разделение математического символа бесконечности и переменных
        inf_expr = sp.Mul(sp.oo, h_k, sp.sin(k1 * theta1 + k2 * theta2), evaluate=False)
        display(sp.Eq(sp.Symbol('S_{1,exact}'), inf_expr))
        display(Markdown("**Системное исключение:** Обнаружена математическая сингулярность (деление на ноль)."))

run_resonance_singularity_experiment()
```

### 3.2. Качественный анализ и математическая интерпретация результатов символьного эксперимента

Символьное моделирование уравнения гомологии (9) в среде `SymPy` экспериментально верифицирует алгебраические и топологические следствия феномена малых знаменателей, сформулированные в теории возмущений гамильтоновых систем [Арнольд, 1985; Заславский, 1984]:

#### 1. Механизм разрушения регулярности асимптотического разложения
При подстановке вектора частот, находящегося в узкой окрестности резонансной поверхности ($\omega^1 = 2.001, \omega^2 = 1.0$), скалярный знаменатель (10) принимает малое значение:
$$ k\omega = 1 \cdot 2.001 - 2 \cdot 1.0 = 10^{-3} $$
Алгебраический движок фиксирует резкое увеличение числового коэффициента при искомой тригонометрической гармонике производящей функции на три порядка:
$$ S_{1,\mathrm{close}} \approx 1000 \cdot \bar{I}_1 \sqrt{\bar{I}_2} \sin(\bar{\theta}^1 - 2\bar{\theta}^2) $$

В асимптотическом анализе это означает, что малый знаменатель $(k\omega)^{-1}$ полностью нивелирует малость малого параметра $\varepsilon$ первого порядка. В высших порядках теории возмущений (например, при расчете функций $S_2$, $S_3$ методом последовательных канонических замен Якоби) малые знаменатели будут входить в ряды мультипликативно, в виде степеней $(k\omega)^{-m}$, в то время как малый параметр убывает лишь линейно как $\varepsilon^m$. 

Математически это приводит к мгновенному нарушению строгой иерархии асимптотического ряда по степеням малости. Сходимость рядов полностью разрушается, что наглядно иллюстрирует классическую теорему Пуанкаре о неинтегрируемости нелинейных гамильтоновых систем общего вида в окрестности соизмеримостей частот [Арнольд, 1985].

#### 2. Топологический смысл аналитической сингулярности в точках точного резонанса
При выходе частот на точную резонансную прямую ($\omega^1 = 2.0, \omega^2 = 1.0$) скалярный знаменатель тождественно обращается в ноль ($k\omega = 0$). Возникновение неустранимой бесконечности $\infty \cdot \sin(\bar{\theta}^1 - 2\bar{\theta}^2)$ генерирует внутреннее прерывание вычислительного процесса.

На уровне качественной теории дифференциальных уравнений данная сингулярность доказывает, что в точках точного резонанса классическое гладкое каноническое преобразование полностью теряет свойство регулярности. Функция Якоби, призванная гладко деформировать фазовое пространство для исключения угловых переменных, претерпевает топологический разрыв. Это служит строгим алгебраическим маркером того, что инвариантные торы невозмущенной системы (Раздел 1.1) в окрестности соизмеримости физически разрушены возмущением. Вместо регулярных условно-периодических траекторий на торе фазовый поток расслоился на замкнутые резонансные структуры [Чириков, 1979].

#### 3. Баллистические следствия для численной реализации асимптотических методов
Результаты символьного эксперимента `SymPy` количественно обосновывают, почему построение глобальных аналитических или полуаналитических теорий движения ИСЗ на основе сквозных рядов Пуанкаре — Цейпеля невозможно в резонансных зонах:
* **Сбой численных алгоритмов:** Прямая попытка программного расчета возмущений по классическим формулам (например, в рамках моделей типа *SGP4*) вблизи суточных или вековых резонансов приводит к делению на машинные нули. Это вызывает катастрофический рост вычислительных погрешностей округления и мгновенный разлет прогнозных баллистических эфемерид ИСЗ.
* **Необходимость локальной перестройки:** Эксперимент доказывает вычислительную неизбежность полного отказа от сквозного усреднения по углам. Баллистический комплекс обязан содержать логический триггер: при падении комбинационного знаменателя $k\omega$ ниже критического порога $\sqrt{\varepsilon}$ алгоритм обязан остановить классическую пертурбационную схему и переключить вычислительное ядро на локальные переменные Болина-Гарфинкеля (Раздел 4), где проблема малых знаменателей устраняется изменением порядка малости частот на $\mathcal{O}(\sqrt{\varepsilon})$ [Боголюбов, 1955].


## 4. Локальный анализ анизохронного резонанса: Идеальная резонансная задача

Для преодоления нерегулярности рядов теории возмущений в окрестности изолированного анизохронного резонанса порядка $N$ под действием ковариантного целочисленного вектора-строки $k$ применяется метод локального изменения масштаба переменных. Процедура заключается в переходе к специальным резонансным переменным, разделяющим фазовое пространство на медленную и быструю составляющие, с последующим частичным усреднением гамильтониана [Арнольд, 1985].

Пусть точный резонанс (1) достигается при некотором фиксированном значении действий $I = I^*$, так что $k\omega(I^*) = 0$. Введем каноническое преобразование от старых переменных $(\theta, I)$ к резонансным переменным $(\tilde{\theta}, \tilde{I})$ с помощью производящей функции смешанного типа. Новая угловая координата выбирается в качестве резонансной фазы (критического аргумента):
$$ \tilde{\theta}^1 = \theta_{\mathrm{res}} = k\theta = \sum_{i=1}^n k_i\theta^i \tag{12} $$

Остальные углы $\tilde{\theta}^2, \dots, \tilde{\theta}^n$ выбираются линейно независимыми от $\tilde{\theta}^1$ и совершают быстрое квазиравномерное вращение.

Канонически сопряженная резонансной фазе переменная действия представляет собой отклонение системы от резонансной поверхности в пространстве импульсов:
$$ P = I - I^* \tag{13} $$

Разложим невозмущенный гамильтониан $H_0(I)$ в ряд Тейлора в окрестности резонансного значения $I^*$ по степеням малого отклонения $P$:
$$ H_0(I^* + P) = H_0(I^*) + \frac{\partial H_0}{\partial I} P^T + \frac{1}{2} P \left(\frac{\partial^2 H_0}{\partial I_i \partial I_j}\right) P^T + \dots \tag{14} $$

По определению вектора частот, линейный член в выражении (14) равен скалярному произведению $\omega(I^*)P^T$. Учитывая структуру преобразования координат (12) и ортогональность частот на резонансной поверхности, данный член в силу условия (1) тождественно зануляется.

Выполним процедуру частичного усреднения функции возмущения $H_1$ по всем быстрым нерезонансным углам $\tilde{\theta}^2, \dots, \tilde{\theta}^n$. В усеченном гамильтониане удерживается только стационарная часть и главная резонансная Фурье-компонента, зависящая от критического аргумента $\theta_{\mathrm{res}}$. В результате удержания членов низшего порядка малости полный гамильтониан системы в окрестности резонансной поверхности сводится к **Идеальной резонансной задаче (модели Болина-Гарфинкеля)** [Garfinkel, 1966]:
$$ \mathcal{H}(P, \theta_{\mathrm{res}}) = \frac{1}{2}GP^2 - F \cos \theta_{\mathrm{res}} \tag{15} $$
где скалярные коэффициенты модели определяются дифференциальными свойствами невозмущенной системы и структурой возмущения:
* $G = k\mathbf{M}_{\mathrm{hess}}k^T$ - параметр анизохронности системы, вычисляемый как квадратичная форма матрицы Гессе невозмущенного гамильтониана на ковариантном векторе $k$;
* $F \sim \varepsilon h_k(I^*)$ - амплитуда резонансной гармоники функции возмущения.

### 4.1. Анализ уравнений движения резонансной моды

Канонические уравнения Гамильтона в строковой форме для локальной резонансной модели (15) принимают вид:
$$ \frac{d\theta_{\mathrm{res}}}{dt} = \frac{\partial\mathcal{H}}{\partial P} = GP, \quad \frac{dP}{dt} = -\frac{\partial\mathcal{H}}{\partial\theta_{\mathrm{res}}} = -F \sin \theta_{\mathrm{res}} \tag{16} $$

Дифференциаруя первое уравнение системы (16) по времени и выполняя подстановку второго уравнения, получаем дифференциальное уравнение второго порядка относительно критического аргумента:
$$ \frac{d^2\theta_{\mathrm{res}}}{dt^2} + GF \sin \theta_{\mathrm{res}} = 0 \tag{17} $$

Уравнение (17) тождественно совпадает с уравнением движения нелинейного математического маятника. Таким образом, локальная геометрия фазового пространства в окрестности изолированного нелинейного резонанса эквивалентна маятниковой динамике.

### 4.2. Механизм возникновения резонансной щели (зоны захвата Чирикова)

Физический механизм формирования резонансной щели в пространстве действий основан на качественной перестройке геометрии фазового потока в окрестности соизмеримости частот под действием резонансной гармоники возмущения.

1. **Вне резонанса (Режим свободной ротации):** В отсутствие возмущения ($F = 0$) траектории на фазовой плоскости представляют собой строго горизонтальные прямые $P = \mathrm{const}$. Фазовый поток движется по цилиндру равномерно, угловая координата $\theta_{\mathrm{res}}$ монотонно нарастает, а среднее значение действия за период не испытывает изменений. Расстояние между любыми двумя изолированными торами инвариантно.
2. **В окрестности резонанса (Перестройка топологии):** При включении возмущения ($F \neq 0$) сила $-F \sin \theta_{\mathrm{res}}$ в уравнениях Гамильтона (16) начинает ускорять изображающую точку в областях, где $\sin \theta_{\mathrm{res}} < 0$, и тормозить там, где $\sin \theta_{\mathrm{res}} > 0$. В результате фазовый поток стягивается к устойчивому положению равновесия $(0,0)$. Вокруг центров либрации формируются замкнутые траектории (маятниковые глазки), ограниченные входящими и исходящими траекториями неустойчивых седел $(\pm \pi, 0)$.

Математически резонансная щель — это область фазового пространства, заключенная внутри замкнутых петель сепаратрисы Идеальной резонансной задачи. Физический смысл возникновения этой щели заключается в появлении барьера в пространстве импульсов, который принудительно изолирует регулярные нерезонансные торы друг от друга. Любая траектория, попавшая внутрь этого контура, теряет возможность совершать свободный дрейф по действию: её импульс заперт возвращающей силой возмущения и вынужден совершать колебания конечной амплитуды около резонансного уровня $P = 0$. Переменные действия испытывают долгопериодические колебания около резонансного значения, не покидая окрестность поверхности соизмеримости.

### 4.3. Геометрическое обоснование и вывод ширины резонансной щели

Для строгого количественного определения масштаба раскрытия резонансной щели (границы коридора разрушения инвариантных торов) используется интеграл энергии маятникового рельефа (15). Контур сепаратрисы, отделяющий область замкнутых либраций резонансной фазы от области открытых ротаций, проходит через неустойчивые седловые точки $(\pm \pi, 0)$. Значение гамильтониана на данном многообразии строго фиксировано и равно потенциальному барьеру маятника:
$$ E_{\mathrm{sep}} = \mathcal{H}(\pm \pi, 0) = \frac{1}{2} G \cdot 0^2 - F \cos(\pm \pi) = F \tag{18} $$

Максимальный вертикальный размах сепаратрисного раструба по оси действий достигается при прохождении изображающей точки через минимум потенциальной энергии, то есть при значении фазы $\theta_{\mathrm{res}} = 0$. Подстановка уровня энергии сепаратрисы (18) в уравнение Идеальной резонансной задачи (15) определяет экстремальное отклонение импульса $P_{\max}$:
$$ \mathcal{H}(P_{\max}, 0) = \frac{1}{2} G P_{\max}^2 - F \cos(0) = F \implies \frac{1}{2} G P_{\max}^2 = 2F \implies P_{\max} = 2\sqrt{\frac{F}{G}} \tag{19} $$

Полная ширина резонансной щели (критерий локализации Чирикова) $\Delta P_{\mathrm{gap}}$ представляет собой удвоенный вертикальный полуразмах сепаратрисного коридора в пространстве импульсов отклонений:
$$ \Delta P_{\mathrm{gap}} = 2 P_{\max} = 4 \sqrt{\frac{F}{G}} = 4 \sqrt{\frac{\varepsilon h_k(I^*)}{k \mathbf{M}_{\text{hess}} k^T}} \tag{20} $$

Формула (20) фиксирует нелинейный отклик системы: масштаб раскрытия резонансной щели по импульсам пропорционален квадратному корню из амплитуды возмущения $\sqrt{\varepsilon}$, что на порядки превосходит масштаб регулярных флуктуаций вне соизмеримостей, имеющих порядок $\varepsilon$. Данное различие является математическим обоснованием необходимости локальной перестройки переменных в окрестности резонансных поверхностей [Чириков, 1979].


### 4.4. Вычислительный эксперимент: динамика прохождения через резонанс и механизм резонансного захвата

**Описание вычислительного эксперимента:** Целью симуляции является исследование механизмов расслоения фазового потока в неавтономных системах при пересечении изолированной резонансной поверхности. Проверяется разделение траекторий, стартующих из одной пространственной области, на режим ротации фазы (сквозное прохождение) и режим либрации (резонансное удержание действия).

**Методика расчета:** Модель строится на уравнениях движения маятникового типа с учетом медленного изменения параметров, описываемых в рамках адиабической теории прохождения через резонанс [Нейштадт, 1975]. Дифференциальные уравнения задаются в системе координат отклонений импульса от мгновенного положения резонансной поверхности $P_{\mathrm{dev}} = P + \alpha t$:
$$ \frac{d\theta_{\mathrm{res}}}{dt} = G P_{\mathrm{dev}}, \quad \frac{dP_{\mathrm{dev}}}{dt} = -F \sin \theta_{\mathrm{res}} - \alpha - \gamma P_{\mathrm{dev}} \tag{21} $$
где $\alpha$ - скорость векового дрейфа частоты (внешний момент), а $\gamma$ - коэффициент слабой диссипативной стабилизации, предотвращающий численный вылет траекторий в окрестности сепаратрисных многообразий. Параметры удовлетворяют условию существования области либрации: $|\alpha| < F$.

Для исключения ситуации, когда изображающая точка изначально находится внутри колебательной зоны, обе траектории стартуют из одной угловой фазы $\theta_0 = -\pi$, соответствующей неустойчивому состоянию равновесия типа седла. Разделение фазового потока достигается варьированием начального импульса $P_{\mathrm{dev}, 0}$ в окрестности сепаратрисного контура.

Начальное условие для захватываемого трека рассчитывается алгоритмически: смещенное устойчивое положение равновесия (фокус) интегрируется ретроградно (назад во времени) от $t = 0$ до $t = -80$ с локализацией индекса пересечения вертикали $\theta_0 = -\pi$. Пролетный трек запускается из той же угловой координаты, но с энергетическим избытком по импульсу. Вывод панелей переведен в вертикальную архитектуру с интервалом интегрирования $T_{\max} = 220$.


```python
# Исполняемая ячейка вычислительного эксперимента к Разделу 4.4
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def run_neishtadt_capture_three_panels():
    # 1. Физические параметры модели Нейштадта со стабилизирующим демпфированием
    G_param = 1.0       # Коэффициент нелинейности
    F_param = 0.5       # Амплитуда резонансной гармоники
    alpha_param = 0.06  # Скорость адиабатического дрейфа частоты (вековой момент)
    gamma_param = 0.03  # Слабая диссипация для стабилизации фокуса удержания
    
    # Правые части ОДУ в динамической системе координат отклонений P_dev
    def neishtadt_system(t, state):
        theta, P_dev = state
        dtheta_dt = G_param * P_dev
        dP_dev_dt = -F_param * np.sin(theta) - alpha_param - gamma_param * P_dev
        return [dtheta_dt, dP_dev_dt]
    
    t_span = (0, 220)
    t_eval = np.linspace(t_span[0], t_span[1], 8000)  # Сетка для фиксации размерностей векторов
    
    # 2. РЕТРОГРАДНЫЙ РАСЧЕТ ИДЕАЛЬНОЙ ТОЧКИ ЗАХВАТА
    theta_stable = np.arcsin(-alpha_param / F_param) + 2 * np.pi
    P_stable = 0.0
    
    t_span_retro = (0, -80)
    init_retro = [theta_stable + 1e-4, P_stable + 1e-4]
    sol_retro = solve_ivp(neishtadt_system, t_span_retro, init_retro, method='DOP853', rtol=1e-12, atol=1e-12)
    
    theta0_shared = -np.pi
    theta_retro_shifted = sol_retro.y[0, :]
    theta_distances = np.abs(theta_retro_shifted - theta0_shared)
    idx_intersection = np.flatnonzero(theta_distances < 0.05)
    
    if len(idx_intersection) > 0:
        idx = int(idx_intersection[0])  # Строгое скалярное извлечение первого пересечения
        P0_capture_base = float(sol_retro.y[1, idx])
        theta0_real = float(sol_retro.y[0, idx])
    else:
        P0_capture_base = 0.94
        theta0_real = theta0_shared
        
    # 3. ПРЯМОЕ ИНТЕГРИРОВАНИЕ ОБОИХ РЕЖИМОВ (синхронизация по одномерному t_eval)
    sol_passage = solve_ivp(neishtadt_system, t_span, [float(theta0_real), float(P0_capture_base)], t_eval=t_eval, method='DOP853')
    sol_capture = solve_ivp(neishtadt_system, t_span, [float(theta0_real), float(P0_capture_base + 0.35)], t_eval=t_eval, method='DOP853')
    
    # 4. ИНИЦИАЛИЗАЦИЯ ТРЕХПАНЕЛЬНОГО ВЕРТИКАЛЬНОГО ОКНА (3 строки, 1 столбец)
    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 22))
    
    # Прецизионная симметричная сетка
    theta_grid = np.linspace(-np.pi, 3 * np.pi, 500)
    P_grid = np.linspace(-2.2, 2.2, 500)
    T_mesh, P_mesh = np.meshgrid(theta_grid, P_grid)
    
    # Рельеф 1: Стационарный невозмущенный маятник
    H_unperturbed = 0.5 * G_param * P_mesh**2 - F_param * np.cos(T_mesh)
    
    # Рельеф 2: Истинный неавтономный потенциал ("Стиральная доска" Пуанкаре)
    H_perturbed = 0.5 * G_param * P_mesh**2 - F_param * np.cos(T_mesh) + alpha_param * T_mesh
    
    # Вычисление точной энергии в смещенной седловой точке для Панели Б
    theta_s = float(np.pi - np.arcsin(alpha_param / F_param))
    E_s1 = float(-F_param * np.cos(theta_s) + alpha_param * theta_s)
    E_s2 = float(E_s1 + 2 * np.pi * alpha_param)
    
    # === ПАНЕЛЬ А: Фазовая плоскость невозмущенных траекторий ===
    ax1.contour(T_mesh, P_mesh, H_unperturbed, levels=25, colors='gray', linestyles=':', alpha=0.3)
    ax1.contour(T_mesh, P_mesh, H_unperturbed, levels=[F_param], colors='black', linestyles='--', linewidths=1.0, alpha=0.4)
    
    P_max_theoretical = 2 * np.sqrt(F_param / G_param)
    ax1.axhspan(-P_max_theoretical, P_max_theoretical, color='purple', alpha=0.08, label='Резонансная щель Чирикова')
    ax1.axhline(P_max_theoretical, color='purple', linestyle='--', linewidth=1.2, alpha=0.4)
    ax1.axhline(-P_max_theoretical, color='purple', linestyle='--', linewidth=1.2, alpha=0.4)
    
    ax1.plot(sol_passage.y[0, :], sol_passage.y[1, :], color='navy', linewidth=2.0, label='Сквозное прохождение')
    ax1.plot(sol_capture.y[0, :], sol_capture.y[1, :], color='crimson', linewidth=2.0, label='Захват в резонанс')
    ax1.axvline(theta0_shared, color='darkorange', linestyle='-.', linewidth=1.5, label=r'Стартовая вертикаль $\theta_0 = -\pi$')
    ax1.plot([theta_stable - 2*np.pi, theta_stable], [0.0, 0.0], 'go', markersize=7, label='Смещенные резонансные центры')
    
    ax1.set_title(r'А) Траектории на невозмущенном маятниковом рельефе', fontsize=12)
    ax1.set_ylabel(r'Отклонение действия $P_{\mathrm{dev}}$', fontsize=11)
    ax1.legend(loc='lower right', frameon=True, fontsize=10)
    
    # === ПАНЕЛЬ Б: Реальный неавтономный потенциал («Стиральная доска» Пуанкаре) ===
    ax2.contour(T_mesh, P_mesh, H_perturbed, levels=30, colors='gray', linestyles=':', alpha=0.3)
    ax2.contour(T_mesh, P_mesh, H_perturbed, levels=sorted([E_s1, E_s2]), colors='black', linestyles='--', linewidths=1.0, alpha=0.7)
    
    ax2.plot(sol_passage.y[0, :], sol_passage.y[1, :], color='navy', linewidth=2.0, label='Сквозное прохождение')
    ax2.plot(sol_capture.y[0, :], sol_capture.y[1, :], color='crimson', linewidth=2.0, label='Захват в резонанс')
    ax2.axvline(theta0_shared, color='darkorange', linestyle='-.', linewidth=1.5, label=r'Стартовая вертикаль $\theta_0 = -\pi$')
    
    theta_c1 = np.arcsin(-alpha_param / F_param)
    ax2.plot([theta_c1, theta_c1 + 2*np.pi], [0.0, 0.0], 'go', markersize=7, label='Истинные центры Пуанкаре')
    
    ax2.set_title(r'Б) Траектории на реальном неавтономном потенциале («Стиральная доска» Пуанкаре)', fontsize=12)
    ax2.set_ylabel(r'Отклонение действия $P_{\mathrm{dev}}$', fontsize=11)
    ax2.legend(loc='lower right', frameon=True, fontsize=10)
    
    # === ПАНЕЛЬ В: Временная развертка полного действия P(t) ===
    P_full_passage = sol_passage.y[1, :] - alpha_param * sol_passage.t
    P_full_capture = sol_capture.y[1, :] - alpha_param * sol_capture.t
    
    ax3.plot(sol_passage.t, P_full_passage, color='navy', linewidth=2.0, label='Сквозное прохождение')
    ax3.plot(sol_capture.t, P_full_capture, color='crimson', linewidth=2.5, label='Захват и адиабатическое удержание')
    ax3.plot(sol_passage.t, -alpha_param * sol_passage.t, 'g--', linewidth=1.5, label='Линия резонанса $P^*(t)$')
    
    ax3.set_title(r'В) Эволюция полного действия во времени $P(t)$', fontsize=12)
    ax3.set_xlabel(r'Системное время $t$', fontsize=11)
    ax3.set_ylabel(r'Полное действие $P$', fontsize=11)
    
    # Общие настройки осей для фазовых портретов
    for ax in [ax1, ax2]:
        ax.set_xlim(-np.pi, 3 * np.pi)
        ax.set_ylim(-2.2, 2.2)
        ax.set_xticks([-np.pi, 0, np.pi, 2*np.pi, 3*np.pi])
        ax.set_xticklabels([r'$-\pi$', '0', r'$\pi$', r'$2\pi$', r'$3\pi$'])
        ax.set_xlabel(r'Резонансная фаза $\theta_{\mathrm{res}}$, рад', fontsize=11)
        ax.grid(True, linestyle=':', alpha=0.5)
    ax3.set_xlim(t_span)
    ax3.set_ylim(-14.5, 2.0)  
    ax3.legend(loc='upper right', frameon=True, fontsize=10)
    ax3.grid(True, linestyle=':', alpha=0.5)
    
    plt.suptitle('Классический пример А.И. Нейштадта: адиабатическое удержание при прохождении через резонанс', fontsize=14, y=0.99)
    plt.tight_layout()
    plt.show()

run_neishtadt_capture_three_panels()
```

### 4.5. Анализ результатов численного моделирования и верификация неавтономных режимов

Вертикально ориентированная трехпанельная визуализация численного эксперимента отражает топологическую структуру расслоения фазового потока неавтономной системы и позволяет провести количественный анализ прохождения траекторий через окрестность сепаратрисного барьера.

Для количественной оценки используется гамильтониан относительного движения (невозмущенная часть маятникового рельефа на Панели А), задающий топологию фазовых траекторий в координатах отклонений $(\theta_{\mathrm{res}}, P_{\mathrm{dev}})$:
$$ E_{\mathrm{dev}} = \mathcal{H}_0(P_{\mathrm{dev}}, \theta_{\mathrm{res}}) = \frac{1}{2} G P_{\mathrm{dev}}^2 - F \cos\theta_{\mathrm{res}} \tag{22} $$

Для заданных параметров модели ($G = 1.0$, $F = 0.5$) энергетический уровень невозмущенной сепаратрисы, проходящей через седловые точки $(\pm\pi, 0)$, равен $E_{\mathrm{sep}} = F = 0.5$. Энергия дна потенциальной ямы консервативной задачи составляет $E_{\mathrm{min}} = -F = -0.5$.

#### 1. Численный и топологический анализ прохождения через окрестность сепаратрисного барьера

Разделение фазовых траекторий при подходе к сепаратрисному контуру определяется локальным балансом полной энергии относительного движения (22), уводящего векового момента $\alpha = 0.06$ и диссипативных потерь работы сил трения $\gamma = 0.03$. 

Обе исследуемые системы стартуют на левой границе периодичности фазового пространства при $\theta_0 = -\pi$. Данная координата является неустойчивым состоянием равновесия маятникового рельефа, где толщина области либрации равна нулю. Соответственно, обе начальные точки геометрически лежат вне колебательной зоны, но разделены по уровню механической энергии относительного движения:

* **Динамика синей кривой (Сквозное прохождение):** 
  Траектория стартует из точки с начальным импульсом $P_{\mathrm{dev}, 0} \approx 1.294$. Согласно уравнению (22), начальная энергия относительного движения составляет:
  $$ E_{\mathrm{dev}, 0}^{(\mathrm{pass})} = \frac{1}{2}(1.0)(1.294)^2 - 0.5 \cos(-\pi) \approx 0.837 + 0.5 = 1.337 $$
  Поскольку $E_{\mathrm{dev}, 0}^{(\mathrm{pass})} > E_{\mathrm{sep}} = 0.5$, система обладает энергетическим избытком над потенциальным барьером маятника ($\Delta E_0 = +0.837$). 

  Двигаясь слева направо по фазе под действием нелинейной частоты $\dot{\theta}_{\mathrm{res}} = P_{\mathrm{dev}} > 0$, синий трек волнообразно обходит область либрации сверху. Из-за непрерывной работы диссипативной силы $-\gamma P_{\mathrm{dev}}$ и векового момента $-\alpha P_{\mathrm{dev}}$ полная энергия относительного движения монотонно падает. При подходе к центральной седловой точке $\theta_{\mathrm{res}} = \pi$ остаточная энергия синей системы уменьшается, но превосходит критический сепаратрисный барьер ($E_{\mathrm{dev}} > 0.5$). 

  Избыток энергии позволяет изображающей точке преодолеть центральный седловой горб, скользя строго *снаружи* сепаратрисного контура. Пересекая плоскость точного резонанса $P_{\mathrm{dev}} = 0$ в точке $\theta_{\mathrm{res}} = \pi$, траектория испытывает качественный разворот: знак импульса становится отрицательным ($P_{\mathrm{dev}} < 0$), а фазовая скорость меняет направление ($\dot{\theta}_{\mathrm{res}} < 0$). Синяя линия уходит влево-вниз в нижнее полупространство свободной ротации фазы. Время взаимодействия данной транзитной траектории с резонансным слоем конечно.

* **Динамика красной кривой (Резонансный захват):**
  Траектория стартует с детерминированного импульса $P_{\mathrm{dev}, 0} \approx 0.944$. Её начальная относительная энергия равна:
  $$ E_{\mathrm{dev}, 0}^{(\mathrm{cap})} = \frac{1}{2}(1.0)(0.944)^2 - 0.5 \cos(-\pi) \approx 0.445 + 0.5 = 0.945 $$
  Избыток энергии над уровнем сепаратрисы составляет $\Delta E_0 = +0.445$. 

  При движении в первом квадранте система испытывает ускорение, однако траектория проходит в непосредственной близости от устойчивого сепаратрисного контура. При пересечении плоскости $\theta_{\mathrm{res}} = \pi$ вековой уводящий момент $-\alpha = -0.06$ и сила вязкого трения $-\gamma P_{\mathrm{dev}}$ суммарно совершают отрицательную работу, непрерывно понижая уровень энергии относительного движения. 

  В окрестности седловой точки скорость приращения фазы падает ($P_{\mathrm{dev}} \to 0$), увеличивая время нахождения точки вблизи барьера. Диссипативный член гасит остаток избыточной энергии относительного движения, переводя её через критический порог: $E_{\mathrm{dev}} \to 0.5$. В точке разделения траекторий полная энергия становится меньше сепаратрисной ($E_{\mathrm{dev}} < 0.5$). 

  Траектория пересекает сепаратрисный контур снаружи внутрь. Возвращающий момент маятника $-F\sin\theta_{\mathrm{res}}$ становится доминирующим, вектор относительной скорости меняет знак ($\dot{\theta}_{\mathrm{res}} < 0$), и фазовая точка совершает разворот. Красный трек оказывается захваченным внутрь области либрации и переходит в режим ограниченных колебаний, где трение релаксирует систему к устойчивому смещенному фокусу.

#### 2. Структура резонансной щели Чирикова (Панель А)
* **Геометрическое представление щели:** Фиолетовый горизонтальный коридор, ограниченный линиями $P_{\text{dev}} = \pm 2\sqrt{F/G} \approx \pm 1.414$, иллюстрирует теоретические границы резонансной щели, вычисленные по критерию локализации Чирикова [Чириков, 1979]. Данная область определяет диапазон изменения импульсов, внутри которого регулярные торы невозмущенной задачи полностью разрушены.
* **Локализация начальных условий:** График демонстрирует, что обе исследуемые траектории стартуют строго внутри резонансной щели (в фиолетовом коридоре). Различие в их долгосрочном динамическом поведении предопределяется не глобальным положением в пространстве импульсов, а локальной геометрией входа под сепаратрисный контур. Нахождение внутри резонансной щели является необходимым, но не достаточным условием для долгосрочного удержания, природа которого носит вероятностный характер [Нейштадт, 1975].

#### 3. Истинная топология потенциала Пуанкаре и асимптотические многообразия (Панель Б)
* **Разомкнутый характер барьера:** Включение векового дрейфа $\alpha = 0.06$ разрушает замкнутые траектории идеального маятника. На Панели Б фоновые линии уровня принимают асимметричный вид наклонного рельефа («стиральной доски»). Соответствующие инвариантные многообразия (черный пунктир) не смыкаются, образуя незамкнутый сепаратрисный коридор.
* **Физическое обоснование транзита синей кривой:** Панель Б разрешает парадокс выхода из ямы диссипирующей системы. Синий трек не пересекает инвариантные кривые либрации. Из-за незамкнутости многообразий между ячейками открывается транзитный коридор ротации фазы. Синяя изображающая точка скользит по этому неавтономному каналу Пуанкаре. Её полная относительная энергия монотонно убывает, и точка перетекает из области положительных импульсов в отрицательные через зазор над центральным седлом $\pi$, не совершая топологических скачков и не заходя внутрь областей удержания.

#### 4. Эволюция полного физического действия во времени и адиабатический увод (Панель В)
* **Линия резонанса:** Зеленая штриховая линия отображает эволюцию точного резонансного уровня действия $P^*(t) = - \alpha t$. В неавтономных системах это геометрическое место точек в пространстве импульсов, на котором выполняется условие соизмеримости частот (1). Вследствие направленного векового дрейфа параметров исходной системы эта линия непрерывно смещается во времени со скоростью внешнего момента $\alpha$.
* **Эволюция нерезонансного действия (Синяя кривая):** В момент прохождения резонансной зоны ($t \approx 15$) испытывает локальное волнообразное возмущение, после чего пересекает область взаимодействия транзитом. Внешний момент $\alpha$ и диссипация продолжают изменять действие системы в нерезонансном режиме. К моменту времени $t = 220$ полное действие падает до отметки $P \approx -14.0$. 
* **Эволюция захваченного действия (Красная кривая):** Пересекая линию дрейфующего резонанса $P^*(t)$ в момент захвата, каноническое действие системы прекращает автономную эволюцию под действием внешних сил и стабилизируется в резонансной щели [Заславский, 1984]. Возвращающая сила резонансной гармоники полностью компенсирует внешний уводящий момент. Красная кривая следует вдоль линии резонанса $P^*(t)$ на всем временном интервале до $t = 220$, совершая затухающие колебания (либрации) непосредственно вокруг движущейся оси соизмеримости частот.


### 4.6. Механизм захвата в резонансную область

Построенный неавтономный вычислительный комплекс позволяет дать строгое математическое определение явлению захвата в резонанс и декомпозировать этот процесс на два сопряженных динамических режима, протекающих в фазовом пространстве [Арнольд, 1985]:

1. **Фазовая синхронизация:** До прохождения через резонансную область индивидуальные угловые переменные динамической системы изменялись независимо и с разными частотами, формируя структуру эргодического тора. В момент захвата (красный трек) индивидуальные фазы теряют свою независимость. Их линейная комбинация — резонансная фаза — прекращает неограниченно расти во времени. На Панелях А и Б возвращающая нелинейная сила возмущения $F \sin \theta_{\mathrm{res}}$ жестко ограничивает изменение угла в конечном диапазоне либрационных колебаний вокруг устойчивого положения равновесия.
2. **Адиабатический увод действия:** Вне условий соизмеримости частот (синий трек) медленные переменные действия эволюционируют по собственным траекториям, пересекая резонансные слои транзитом. Однако в момент захвата (красный трек) переменная действия полностью лишается автономности и локализуется в резонансной щели. Возвращающая нелинейная сила резонансной гармоники полностью компенсирует внешнее вековое смещение параметров, обеспечивая вынужденную эволюцию медленной переменной действия строго вдоль дрейфующей резонансной поверхности [Нейштадт, 1975].

> **Резюме:** Нелинейный резонанс лишает динамическую систему одной степени свободы: он принудительно синхронизирует быстрые фазы (переводя их в режим либраций) и удерживает медленные действия в окрестности смещающейся резонансной поверхности $P \approx P^*(t)$. В то время как пролетная (синяя) траектория монотонно деградирует по импульсу под действием невозмущенных частот, захваченная (красная) траектория удерживает баланс сил за счет возвращающего маятникового момента, демонстрируя устойчивость динамической ловушки на длительных интервалах времени вопреки внешним негамильтоновым возмущениям.


## Раздел 5. Классификация резонансных соотношений в астродинамике ИСЗ

Классификация околоземных орбитальных резонансов строится на соотношении между средним баллистическим движением спутника $n$, скоростью вращения Земли $\omega_{\mathrm{E}}$ и вековыми частотами прецессии перигея $\dot{g}$ и долготы восходящего узла $\dot{h}$, индуцированными второй зональной гармоникой геопотенциала ($J_2$). На основе спектрального состава возмущающей функции геопотенциала выделяются три основных типа соизмеримостей [Арнольд, 1985].

### 5.1. Канонические переменные Делоне и структура гамильтониана геопотенциала

Орбитальное движение искусственного спутника Земли в центральном гравитационном поле с учетом нецентральности геопотенциала и внешних возмущений описывается в канонических переменных Делоне [Арнольд, 1985]. Элементами действия являются метрические орбитальные характеристики:
$$ L = \sqrt{\mu a}, \quad G = L\sqrt{1-e^2}, \quad H = G\cos i \tag{23} $$
где $\mu$ - гравитационный параметр Земли, $a$ - большая полуось, $e$ - эксцентриситет, $i$ - наклонение орбиты. Сопряженными быстрыми угловыми переменными являются канонические углы: средняя аномалия $l$, аргумент перигея $g$ и долгота восходящего узла $h$.

Полный невозмущенный гамильтониан задачи Кеплера зависит исключительно от действия $L$:
$$ H_0 = -\frac{\mu^2}{2L^2} \tag{24} $$
Правая часть возмущенной функции $H_1(L, G, H, l, g, h, t)$ задается разложением геопотенциала по сферическим гармоникам $V_{nmp}(a, e, i)$ и содержит вековые, долгопериодические и короткопериодические члены, явно зависящие от звездного времени Гринвича $\theta_{\mathrm{G}}(t) = \omega_{\mathrm{E}} t$, где $\omega_{\mathrm{E}}$ — угловая скорость вращения Земли вокруг своей оси.

### 5.2. Соизмеримости с вращением планеты (суточные резонансы)

Суточные резонансы возникают при соизмеримости средней частоты обращения спутника вокруг Земли и частоты осевого вращения планеты. Общее условие соизмеримости для данного класса систем принимает вид:
$$ n \approx \frac{m}{k_1} \omega_{\mathrm{E}} \tag{25} $$
где $m$ — долготный порядок тессеральной гармоники геопотенциала, а $k_1$ — число оборотов ИСЗ за одни звездные сутки.

В рамках баллистической структуры выделяются две ключевые резонансные зоны:
1. **Геостационарный резонанс (резонанс 1:1):** Соответствует условию $k_1 = 1, m = 1$. Орбита ИСЗ локализуется на расстоянии большой полуоси $a^* \approx 42164$ км. Динамика захваченных объектов полностью эквивалентна модели маятника Пуанкаре (Раздел 4.4) [Garfinkel, 1966]. Фазовая координата блокируется в окрестности устойчивых географических долгот (стабильные положения равновесия Нейштадта), соответствующих минимумам тессерального потенциала $C_{22}, S_{22}$. Переменная действия испытывает долгопериодические колебания амплитудой $\Delta a \approx 35$ км с периодом порядка 800 суток [Чириков, 1979].
2. **Полусуточные резонансы (резонанс 2:1):** Соответствует условию $k_1 = 2, m = 1$ ($a^* \approx 26560$ км). Эта зона соизмеримости определяет динамику высокоэллиптических космических аппаратов (типа «Молния») и навигационных систем (ГЛОНАСС, GPS). Эволюция действия под влиянием резонансных возмущений компенсируется вековым дрейфом параметров, формируя условия для долгосрочного удержания орбиты по Нейштадту [Нейштадт, 1975].

### 5.3. Диссипативные и приливные соизмеримости

Этот класс соизмеримостей характеризуется тем, что синхронизация фазы и локализация действия происходят под влиянием негамильтоновых сил диссипативного происхождения (вязкоупругие приливы в теле планеты, гравитационный тормозящий момент). Условие соизмеримости связывает частоту вращения спутника вокруг собственной оси $\omega_{\mathrm{rot}}$ и его среднее орбитальное движение $n$:
$$ \omega_{\mathrm{rot}} \approx kn \tag{26} $$
где $k$ - целое или полуцелое число.

* **Синхронный резонанс 1:1:** Является предельным аттрактором приливной эволюции естественных и близких искусственных спутников, переводящим систему в устойчивый смещенный фокус (Раздел 4.4).
* **Высшие спин-орбитальные резонансы (например, резонанс 3:2):** Характеризуются вероятностным механизмом прохождения через сепаратрисный контур, детально описанным в рамках адиабатических инвариантов [Нейштадт, 1975].


### 5.4. Вековые и лунно-солнечные резонансы

Вековые резонансы возникают в области низких частот, когда движение ИСЗ находится в соизмеримости с орбитальным движением возмущающих внешних тел (Луны и Солнца). Условие соизмеримости не затрагивает быструю кеплерову частоту $n$ ($k_1 = 0$) и определяется исключительно скоростями прецессии плоскости орбиты спутника и плоскостей орбит внешних тел:
$$ k_2 \dot{g} + k_3 \dot{h} - k_{\mathrm{L}} n_{\mathrm{L}} - k_{\mathrm{S}} n_{\mathrm{S}} \approx 0 \tag{27} $$
где $n_{\mathrm{L}}$ и $n_{\mathrm{S}}$ - средние движения Луны и Солнца соответственно.

Данные резонансы индуцируют долгопериодические широкоамплитудные флуктуации эксцентриситета $e(t) = \mathcal{O}(\sqrt{\varepsilon})$ и наклонения $i(t)$, формируя коридоры неустойчивости в окрестности критических наклонений орбиты [Заславский, 1984]. В этих зонах ширина резонансной щели Чирикова (14) максимальна, что может приводить к пересечению сепаратрис соседних соизмеримостей и возникновению локальной хаотической диффузии орбит ИСЗ.

### 5.5. Внутрисистемные вековые резонансы

Этот класс соизмеримостей возникает во внутренней структуре вековых частот самого ИСЗ при занулении средней кеплеровой частоты ($k_1 = 0$) и отсутствии внешних возмущающих частот Луны и Солнца. Условие соизмеримости определяется исключительно собственными частотами прецессии перигея $\dot{g}$ и долготы восходящего узла $\dot{h}$ под действием главной зональной гармоники геопотенциала $J_2$:
$$ k_2 \dot{g} + k_3 \dot{h} \approx 0 \tag{28} $$

Наиболее изученным случаем данного класса является резонанс критического наклонения ($k_2 = 1, k_3 = 0$), возникающий при значениях наклонения орбиты $i \approx 63.4^{\circ}$ и $i \approx 116.6^{\circ}$. В этих точках вековая частота прецессии перигея $\dot{g}$ обращается в ноль.

Возникновение вековой резонансной щели (14) приводит к тому, что аргумент перигея $g$ переходит из режима ротации в режим долгопериодических либраций. Это вызывает вековые флуктуации эксцентриситета орбиты, что критично для удержания высоты перигея ИСЗ на больших интервалах времени.

### 5.6. Учет резонансных соотношений в алгоритмах асимптотических методов

В баллистических алгоритмах и программных комплексах численно-аналитического прогнозирования (например, в полуаналитической теории движения ИСЗ *DSST* или в аналитической структуре *SGP4*) учет резонансов требует модификации стандартных схем асимптотического усреднения [Арнольд, 1985].

#### 1. Проблема малых знаменателей в методе Делоне-Цейпеля
При построении аналитической теории возмущений вне резонансов генерация производящей функции $S$ для исключения быстрых переменных приводит к появлению в рядах теории членов вида:
$$ S_1 \sim \sum \frac{V_{nmp}(a,e,i)}{k_1 n + k_2 \dot{g} + k_3 \dot{h} - m \omega_{\mathrm{E}}} \tag{29} $$
В окрестности резонансной поверхности знаменатель в уравнении (29) стремится к нулю, порождая проблему малых знаменателей и приводя к расходимости рядов теории возмущений.

#### 2. Алгоритм локальной перестройки по методу Болина-Гарфинкеля
Для устранения расходимостей в алгоритмах асимптотических методов применяется процедура локальной перестройки структуры уравнений в окрестности соизмеримости [Garfinkel, 1966]:
* **Изменение порядка малости:** Если вне резонанса функция возмущения имеет порядок малости $\mathcal{O}(\varepsilon)$, то внутри резонансной щели, в соответствии с нелинейным масштабом Чирикова (14), частота либраций имеет порядок $\mathcal{O}(\sqrt{\varepsilon})$. Производящая функция $S$ перестраивается в ряды по степеням $\sqrt{\varepsilon}$ [Арнольд, 1985].
* **Частичное усреднение:** Из程序уры исключения фаз полностью изымается критический резонансный аргумент $\theta_{\mathrm{res}}$ (12). Усреднение по остальным быстрым нерезонансным комбинациям углов выполняется по классической схеме частичного усреднения [Боголюбов, 1955].
* **Переход к промежуточному гамильтониану:** Система уравнений ИСЗ редуцируется к промежуточному гамильтониану Идеальной резонансной задачи (9), интегрируемому в эллиптических функциях Якоби. Эволюция нерезонансных элементов рассчитывается как возмущение этого базового маятникового движения [Garfinkel, 1966].

#### 3. Алгоритмический учет адиабатического дрейфа частоты по Нейштадту
При наличии негамильтоновых сил (например, торможения в верхней атмосфере) в алгоритм прогнозирования закладывается расчет вероятности захвата в резонанс $\mathbb{P}$ при пересечении сепаратрисного контура [Нейштадт, 1975]. 
* Если при численном анализе знака адиабатического инварианта выявляется сквозной пролет, алгоритм переключает уравнения на схему быстрой ротации фазы по незамкнутому неавтономному каналу Пуанкаре.
* Если фиксируется акт захвата, вычислительный модуль переходит на интегрирование уравнений движения вдоль инвариантного многообразия дрейфующей линии резонанса $P^*(t)$, что позволяет динамически изменять шаг интегрирования и исключает накопление локальных ошибок округления в жестких системах ОДУ.


### 5.7. Примеры расчета резонансных коэффициентов в прикладных задачах баллистики

Для верификации перехода от абстрактной модели Идеальной резонансной задачи (15) к реальным баллистическим расчетам рассмотрим два примера вычисления параметров анизохронности $G$, амплитуды возмущения $F$ и скорости адиабатического дрейфа $\alpha$.

#### Пример 1. Расчет параметров удержания геостационарного спутника в окрестности соизмеримости 1:1

Рассмотрим космический аппарат на околокруговой экваториальной орбите ($e \approx 0, i \approx 0$) в окрестности геостационарного радиуса. Соизмеримость частот возникает между средним движением спутника $n$ и угловой скоростью вращения Земли $\omega_{\mathrm{E}}$ при $k_1 = 1, m = 1$ (уравнение 25). Критический резонансный аргумент (уравнение 12) имеет вид $\theta_{\mathrm{res}} = l + g + h - \theta_{\mathrm{G}}(t) - 2\lambda_{22} = \lambda - 2\lambda_{22}$, где $\lambda$ - географическая долгота подспутниковой точки, $\theta_{\mathrm{G}}(t) = \omega_{\mathrm{E}}t$ - гринвичское звездное время, а $\lambda_{22}$ - фиксированная долгота главной оси симметрии секторной гармоники геопотенциала $V_{22}$, вычисляемая через геодезические коэффициенты как $\lambda_{22} = \frac{1}{2}\mathrm{arctg}(S_{22}/C_{22})$.

1. **Вычисление параметра анизохронности $G$:**
   Матрица Гессе (уравнение 3) невозмущенного кеплерова гамильтониана $H_0 = -\frac{\mu^2}{2L^2}$ по переменным действия Делоне (уравнение 23) содержит единственный ненулевой элемент:
   $$ \mathbf{M}_{\mathrm{hess}} = \frac{\partial^2 H_0}{\partial L^2} = -\frac{3\mu^2}{L^4} = -\frac{3}{a^2} \tag{30} $$
   Для ковариантного вектора соизмеримости $k = (1, 0, 0)$ скалярный параметр нелинейности $G$ согласно Разделу 4 равен:
   $$ G = k \mathbf{M}_{\mathrm{hess}} k^T = \frac{\partial^2 H_0}{\partial L^2} = -\frac{3}{a^{*2}} \tag{31} $$
   Отрицательный знак коэффициента $G$ характеризует обратную (негативную) нелинейность фазового потока: положительное приращение действия (большой полуоси) вызывает падение скорости изменения резонансной фазы ($\dot{\theta}_{\mathrm{res}} < 0$).

2. **Вычисление амплитуды возмущения $F$:**
   Главный вклад в резонансную часть возмущающей функции $H_1$ вносит секторная гармоника геопотенциала второго порядка и второй долготной степени ($n=2, m=2$). Ее амплитуда на экваториальной орбите составляет:
   $$ F = 3n^2 R_{\mathrm{E}}^2 \cdot J_{22} \tag{32} $$
   where $R_{\mathrm{E}}$ - экваториальный радиус Земли, $J_{22} = \sqrt{C_{22}^2 + S_{22}^2} \approx 1.8 \cdot 10^{-6}$ - амплитуда долготной неоднородности гравитационного поля планеты.

3. **Баллистическая оценка ширины щели Чирикова:**
   Подставляя вычисленные коэффициенты (31) и (32) в критерий локализации Чирикова (уравнение 20), получаем оценку максимального размаха резонансной зоны по большой полуоси:
   $$ \Delta a_{\mathrm{gap}} = 4 \sqrt{\frac{F}{|G|}} = 4 \sqrt{\frac{3n^2 R_{\mathrm{E}}^2 J_{22}}{3 / a^{*2}}} = 4 a^* \left(\frac{R_{\mathrm{E}}}{a^*}\right) \sqrt{J_{22}} \approx 70 \text{ км} \tag{33} $$
   Таким образом, полуширина коридора удержания составляет $\pm 35$ км, что согласуется с численными баллистическими константами, приведенными в подразделе 5.2.

#### Пример 2. Анализ условий адиабатического дрейфа под действием атмосферного торможения

Рассмотрим низкоорбитальный ИСЗ в окрестности суточной соизмеримости высокого порядка (например, соизмеримость 15:1, большая полуось $a^* \approx 6920$ км, высота орбиты порядка $550$ км). Движение спутника испытывает постоянное негамильтоново возмущение со стороны верхней атмосферы Земли.

1. **Вычисление скорости адиабатического дрейфа частоты $\alpha$:**
   Сила аэродинамического сопротивления совершает отрицательную работу, вызывая вековое падение большой полуоси орбиты со средней скоростью $(da/dt)_{\mathrm{atm}} < 0$. Изменение частоты среднего движения спутникового аппарата $n(a)$ во времени порождает адиабатический дрейф параметров, эквивалентный внешнему неконсервативному моменту $\alpha$ в уравнениях Нейштадта (уравнение 21):
   $$ \alpha = \frac{dn}{dt} = \frac{\partial n}{\partial a} \left(\frac{da}{dt}\right)_{\mathrm{atm}} = -\frac{3}{2}\frac{n}{a^*} \left(\frac{da}{dt}\right)_{\mathrm{atm}} > 0 \tag{34} $$

2. **Топологическое условие захвата в атмосферный резонанс:**
   В соответствии со структурой потенциала Пуанкаре (Панель Б), захват в адиабическую колебательную зону возможен только при выполнении неравенства $|\alpha| < F$. В терминах параметров атмосферного торможения это накладывает ограничение на скорость снижения высоты спутника:
   $$ \left| \left(\frac{da}{dt}\right)_{\mathrm{atm}} \right| < \frac{2}{3} \frac{a^* F}{n} \tag{35} $$
   Если из-за флуктуаций солнечной активности плотность верхней атмосферы возрастает и скорость торможения нарушает критерий (35), наклон потенциального рельефа Пуанкаре превышает критический порог. Инвариантные многообразия седловых точек размыкаются, и область замкнутых либраций исчезает. Вероятность захвата падает до нуля ($\mathbb{P} = 0$), и спутник пересекает область соизмеримости транзитом по свободной траектории ротации фазы, продолжая баллистический спуск.


## 6. Заключение

На основе проведенного теоретического анализа, символьных вычислений и численного моделирования динамики нелинейных резонансных систем в консервативных и слабодиссипативных средах сформированы следующие выводы:

1. **Нелинейный характер отклика системы на резонансное возмущение:** 
   В окрестности соизмеримости частот топология фазового пространства претерпевает качественную перестройку. Величина раскрытия резонансной щели по переменным действия (критерий локализации Чирикова) пропорциональна квадратному корню из амплитуды возмущения $\mathcal{O}(\sqrt{\varepsilon})$, что на порядки превосходит масштаб регулярных флуктуаций $\mathcal{O}(\varepsilon)$ вне резонансов. Данный факт обуславливает расходимость классических рядов Пуанкаре (проблема малых знаменателей) и требует обязательного перехода к промежуточным интегрируемым гамильтонианам маятникового типа (модель Болина-Гарфинкеля).

2. **Топологическая незамкнутость неавтономных сепаратрисных барьеров:**
   Включение векового дрейфа параметров (модель Нейштадта) нарушает симметрию инвариантных многообразий идеального маятника. Как наглядно продемонстрировано на реальном потенциале Пуанкаре («стиральная доска»), инвариантные многообразия седловых точек расщепляются и образуют незамкнутый коридор. Наличие этого зазора формирует открытый транзитный канал, по которому пролетные траектории свободно перетекают из одной области ротации в другую без пересечения замкнутых траекторий либрации.

3. **Вероятностный характер и асимметрия акта захвата:**
   Прохождение изображающей точки через резонансную область внутри щели Чирикова разделяет фазовый поток на два качественно полярных режима. Исход взаимодействия системы с резонансным слоем определяется исключительно локальной геометрией входа траектории под сепаратрисный барьер в окрестности неустойчивого седла. Слабая диссипация выступает стабилизатором, гасящим избыток адиабатического инварианта, переводя систему через критический энергетический порог и обеспечивая асимптотическое затягивание фазы в режим либраций.

4. **Эффект адиабатического удержания действия:**
   Захват в резонанс лишает динамическую систему одной степени свободы. Каноническое действие полностью теряет свою автономность, блокируется в резонансной щели и вынужденно дрейфует во времени строго синхронно со смещением самой дрейфующей поверхности резонанса. В задачах астродинамики ИСЗ этот механизм обеспечивает долгосрочную стабилизацию больших полуосей орбит вопреки непрерывному вековому торможению в верхней атмосфере планеты.
