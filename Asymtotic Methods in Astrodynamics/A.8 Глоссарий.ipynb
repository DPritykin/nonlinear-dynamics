# Глоссарий и предметный указатель: Часть A

### А

*   **Адиабатический инвариант** - скалярная величина (в интегрируемых или близких к ним гамильтоновых системах с одной быстрой фазой — переменная действия $I$), испытывающая лишь малые флуктуации порядка $\mathcal{O}(\varepsilon)$ и сохраняющая свое значение на эволюционных интервалах времени векового масштаба $t \sim 1/\varepsilon$ при наличии медленного непрерывного дрейфа параметров или векового изменения внешних сил [Арнольд, 1985; Нейштадт, 1975].  
    * *Встречается в разделах:* **А.2 (Секция 8.4)**, **А.5 (Секция 2.5, 4.4)**, **А.7 (Введение, Секция 5.2)**.
    
*   **Автономизация гамильтоновой системы** - математическая процедура приведения неавтономной системы ОДУ к эквивалентному автономному (стационарному) виду путем искусственного расширения размерности фазового пространства до $2n + 2$. При этом физическое время объявляется новой $(n+1)$-й обобщенной координатой ($q_{n+1} = t$), а в качестве сопряженного ему канонического импульса вводится фиктивный импульс энергии ($p_{n+1} = -E$), знак которого компенсирует появление частной производной по времени в расширенной скобке Пуассона [Whittaker, 1937].  
    * *Встречается в разделах:* **А.3 (Секция 4.4, Раздел 5.2)**
    
*   **Алгебра Ли гамильтоновых функций** - бесконечномерное линейное векторное пространство гладких функций $C^\infty(\mathbb{R}^{2n})$, заданных на кокасательном расслоении конфигурационного многообразия (фазовом пространстве), наделенное структурой алгебры Ли посредством канонической скобки Пуассона, играющей роль операторного коммутатора и удовлетворяющей аксиомам антисимметричности, билинейности и тождества Якоби [Новиков, Тайманов, 2014; Козлов, 1995].  
    * *Встречается в разделах:* **А.3 (Раздел 3, Заключение)**.

*   **Анизохронность (Жесткая нелинейность)** - свойство нелинейной динамической системы, заключающееся в наличии частотного градиента в пространстве канонических импульсов (в баллистике ИСЗ — функциональной зависимости невозмущенных частот $\omega^i$ от вектора действий Делоне $I$). Дифференциальным критерием анизохронности выступает отличие от нуля определителя матрицы Гессе невозмущенного гамильтониана системы $\det(\partial^2 H_0 / \partial I_i \partial I_j) \neq 0$ (условие невырожденности Колмогорова) [Арнольд, 1989; Чириков, 1979].  
    * *Встречается в разделах:* **А.2 (Секция 8.5)**, **А.5 (Раздел 2.2, Секция 4.3)**, **А.7 (Секция 2.4, 5.1)**.

*   **Асимптотический метод** - строгое математическое приближенное руководство (метод), оперирующее построением формальных функциональных рядов по малому безразмерному параметру $\varepsilon \ll 1$ для нахождения приближенных аналитических решений неинтегрируемых систем дифференциальных уравнений [Боголюбов, Митропольский, 1955]. В отличие от численного интегрирования, асимптотический метод не ищет точное единичное решение, а выявляет качественную геометрию фазового пространства (инвариантные торы, резонансные ловушки, вековые тренды дрейфа), удерживая погрешность в рамках заданной степени точности $\mathcal{O}(\varepsilon^m)$ на больших интервалах времени [Арнольд, 1989].  
    * *Встречается в разделах:* **А.2 (Раздел 5, Секция 8.5)**, **А.3 (Введение, Раздел 1.4)**, **А.5 (Введение, Секция 1.2)**, **А.7 (Введение, Секция 1.1, 5.6)**.

### Б

*   **Базисные скобки Пуассона** - скобки Пуассона, вычисляемые непосредственно для новых координатных функций фазового пространства $Q^i(q,p)$ и $P_i(q,p)$ по исходным каноническим переменным. Выполнение базисных тождеств $\{Q^i, Q^j\} = 0$, $\{P_i, P_j\} = 0$, $\{P_i, Q^j\} = \delta_i^j$ (где $\delta_i^j$ - символ Кронекера) является необходимым и достаточным критерием каноничности аналитического преобразования, гарантирующим инвариантность пуассоновой структуры для любых произвольных функций [Лурье, 2019; Маркеев, 2024].  
    * *Встречается в разделах:* **А.2 (Раздел 6.1, Секция 10.5)**, **А.5 (Секция 3.1)**.

*   **Билинейность скобки Пуассона** - свойство пуассоновой алгебраической структуры, заключающееся в линейности оператора по каждому из своих функциональных аргументов: $\{c_1 F_1 + c_2 F_2, G\} = c_1\{F_1, G\} + c_2\{F_2, G\}$ для любых произвольных гладких функций $F_1, F_2, G$ и скалярных констант $c_1, c_2 \in \mathbb{R}$ [Лурье, 2019; Козлов, 1995].  
    * *Встречается в разделах:* **А.2 (Раздел 2.1)**.

*   **Бифуркация фазового потока** - качественное изменение топологической структуры траекторий динамической системы в фазовом пространстве (изменение типа положений равновесия, рождение или разрушение инвариантных многообразий) при непрерывном изменении пассивных параметров или медленном адиабическом дрейфе частот в окрестности нелинейного резонанса [Андронов и др., 1967; Арнольд, 1989].  
    * *Встречается в разделах:* **А.5 (Раздел 2.5, 4.4)**, **А.7 (Секция 4.3)**.

*   **Блочный фазовый вектор** - составной вертикальный вектор-столбец размерности $2n \times 1$, формируемый путем вертикального объединения предварительно транспонированного ковектора импульсов $p^T$ и вектора обобщенных координат $q$. Используется в качестве сквозного стандарта в матричных критериях симплектичности и алгоритмах численного интегрирования систем компьютерной алгебры [Whittaker, 1937].  
    * *Встречается в разделах:* **А.1 (Раздел 3.3)**, **А.2 (Раздел 3)**, **А.3 (Раздел 1.1)**.

### В

*   **Вектор-столбец (Контравариантные компоненты)** - матричная форма представления элементов исходного конечномерного вещественного линейного пространства $V$ (обобщенных координат $q^i$, обобщенных скоростей $\dot{q}^i$, вариаций координат $\delta q^i$). При изменении базиса компоненты данного столбца преобразуются с помощью матрицы, обратной матрице перехода (контравариантным образом), обеспечивая инвариантность самого геометрического объекта [Новиков, Тайманов, 2014].  
    * *Встречается в разделах:* **А.1 (Раздел 1.1, Раздел 2.1)**, **А.2 (Раздел 1)**.

*   **Векторное поле в фазовом пространстве** - гладкое отображение, сопоставляющее каждой точке $x \in \mathbb{R}^{2n}$ касательный вектор $X(x) \in T_x(\mathbb{R}^{2n})$, который однозначно определяет направление и скорость изменения динамического состояния системы. В пространстве гладких функций $C^\infty(\mathbb{R}^{2n})$ всякое гладкое векторное поле реализуется как линейный дифференциальный оператор первого порядка (направленная производная) [Арнольд, 1989].  
    * *Встречается в разделах:* **А.3 (Раздел 1.1, Раздел 2.1)**.

*   **Вековой (эволюционный) масштаб времени** - долговременный интервал времени масштаба $t \sim 1/\varepsilon$, на котором малые пертурбационные силы, интегрируясь вдоль траектории движения, вызывают макроскопическое изменение медленных переменных (действий) на величину порядка единицы $\mathcal{O}(1)$. На данном интервале классические разложения решений в ряды Тейлора по времени теряют аналитический смысл из-за появления расходящихся секулярных членов [Боголюбов, Митропольский, 1955; Арнольд, 1989].  
    * *Встречается в разделах:* **А.7 (Раздел 1.1, Секция 2.3.2)**.

*   **Возмущений, теория** - общий раздел математической физики и аналитической механики, исследующий качественное и количественное поведение динамических систем, близких к вполне интегрируемым (невозмущенным) задачам [Whittaker, 1937; Арнольд, 1989]. Геометрический смысл гамильтоновой теории возмущений заключается в поиске локального канонического диффеоморфизма фазового пространства (близкого к тождественному), который исключает явную зависимость деформированного потенциала от быстрых циклических фаз, аппроксимируя реальный сложный фазовый поток прямолинейными траекториями законов сохранения [Deprit, 1969; Hori, 1966].  
    * *Встречается в разделах:* **А.2 (Раздел 5, Секция 8.4)**, **А.3 (Введение, Раздел 3.2)**, **А.5 (Раздел 3)**, **А.7 (Введение, Раздел 1.1)**.

*   **Временное среднее (Среднее по траектории)** - предел интеграла от гладкой функции возмущения вдоль истинной интегральной кривой невозмущенного фазового потока при стремлении верхнего предела интегрирования (времени) к бесконечности. Зависит от выбора конкретной траектории и начальной угловой фазы системы [Арнольд, 1989; Чириков, 1979].  
    * *Встречается в разделах:* **А.7 (Раздел 1.4, Секция 1.4.1)**.

*   **Вырождение гамильтоновой системы** - математическая ситуация, при которой невозмущенный гамильтониан $H_0$ зависит от меньшего числа канонических действий, чем полное число степеней свободы системы $n$, вследствие чего часть невозмущенных частот тождественно обращается в ноль. Предельным случаем является задача Кеплера в переменных Делоне, где гамильтониан $H_0(L)$ вырожден по орбитальному $G$ и проекционному $H_{\mathrm{del}}$ действиям [Арнольд, 1985; Брумберг, 1980].  
    * *Встречается в разделах:* **А.2 (Раздел 9.3, Секция 9.4)**, **А.5 (Секция 5.5)**.

### Г

*   **Гамильтониан (Функция Гамильтона)** - скалярная функция фазовых переменных (координат, импульсов и, в общем случае, времени), задающая полную механическую энергию системы и полностью определяющая ее динамику на кокасательном расслоении. Является результатом двойственного преобразования Лежандра для функции Лагранжа по обобщенным скоростям [Лурье, 2019; Маркеев, 2024].  
    * *Встречается в разделах:* **А.1 (Раздел 3.2)**, **А.2 (Раздел 1, Секция 3.2)**, **А.3 (Секция 1.3, Раздел 3)**.

*   **Геодезические линии риманова пространства** - гладкие кривые на римановом многообразии, являющиеся экстремалями функционала длины (кривыми наименьшей кривизны) [Новиков, Тайманов, 2014]. В геометрической трактовке лагранжева формализма (принцип Мопертюи—Лагранжа) истинные траектории движения консервативной механической системы в отсутствие внешних сил являются геодезическими линиями в конфигурационном пространстве, риманова метрика которого задается кинетической энергией системы $T = \frac{1}{2} \sum g_{ij}(q)\dot{q}^i\dot{q}^j$ [Арнольд, 1989].  
    * *Встречается в разделах:* **А.1 (Раздел 2.2)**.

*   **Градиент скалярной функции** - вектор-столбец, составленный из частных производных скалярного поля по векторному аргументу фазового пространства. Согласно принятому геометрическому протоколу, оператор градиента связан с матрицей-строкой производных операцией транспонирования: $\operatorname{grad} F = (\partial F / \partial x)^T$ [Лурье, 2019].  
    * *Встречается в разделах:* **А.1 (Раздел 1.2.4)**, **А.2 (Раздел 2.2)**, **А.3 (Секция 2.1)**.

*   **Группа Ли преобразований фазового пространства** - гладкое топологическое многообразие, наделенное групповой структурой, при которой операции группового умножения элементов и взятия обратного элемента являются гладкими отображениями. В гамильтоновой теории возмущений реализуется как непрерывное однопараметрическое семейство диффеоморфизмов, индуцированное стационарным фазовым потоком генерирующего векторного поля [Новиков, Тайманов, 2014; Арнольд, 1989].  
    * *Встречается в разделах:* **А.3 (Раздел 1, Раздел 1.3)**.

### Д

*   **Действие-угол, переменные** - канонический набор сопряженных фазовых переменных $(I, \theta)$ для автономных вполне интегрируемых гамильтоновых систем [Арнольд, 1989]. Компоненты ковектора действий $I_i$ являются строгими первыми интегралами движения и геометрически определяют радиусы инвариантного многомерного тора Лиувилля, а компоненты вектора углов $\theta^i$ задают координаты на поверхности этого тора и изменяются во времени абсолютно равномерно во всем фазовом объеме с постоянными нелинейными частотами $\omega^i(I)$ [Арнольд, 1989].  
    * *Встречается в разделах:* **А.2 (Раздел 8, Секция 8.1–8.4)**, **А.5 (Раздел 1.1)**, **А.7 (Раздел 1.2, Раздел 2.4)**.

*   **Делоне, переменные** - классический канонический набор сопряженных переменных типа действие-угол в гамильтоновой небесной механике, осуществляющий разделение геометрических размеров кеплеровской орбиты, формы эллиптического трека и пространственной ориентации плоскости движения ИСЗ [Брумберг, 1980]. Роли импульсов выполняет ковектор действий $I = (L, G, H_{\mathrm{del}})$, жестко связанный с интегралами энергии и момента импульса, а роли координат — вектор угловых фаз $\theta = (l, g, h)^T$, составленный из средней аномалии, аргумента перицентра и долготы восходящего узла [Брумберг, 1980]. 
    * *Встречается в разделах:* **А.2 (Раздел 9, Секция 9.2)**, **А.3 (Секция 3.3)**, **А.5 (Секция 5.1)**.

*   **Диффеоморфизм** - взаимно однозначное (биективное) отображение одного гладкого многообразия на другое, при котором как само отображение, так и обратное к нему принадлежат классу гладкости $C^\infty$ [Новиков, Тайманов, 2014]. В теории возмущений фазовый поток автономного гамильтонова векторного поля на любом фиксированном шаге группового деформационного параметра $\varepsilon$ задает диффеоморфизм фазового пространства на себя [Арнольд, 1989].  
    * *Встречается в разделах:* **А.1 (Раздел 2.1)**, **А.3 (Раздел 1.2, Секция 2.2)**.

*   **Дифференциал векторного отображения** - главная линейная часть приращения векторной функции, вычисляемая в рамках принятого матричного протокола как прямое произведение функциональной матрицы Якоби размерности $k \times m$ на вертикальный вектор-столбец приращений аргументов размерности $m \times 1$: $d\mathbf{Y} = \frac{\partial \mathbf{Y}}{\partial \mathbf{A}} d\mathbf{A}$ [Лурье, 2019].  
    * *Встречается в разделах:* **А.1 (Раздел 1.2.3, Раздел 1.2.4)**.

*   **Дифференциал скалярной функции** - главная линейная часть приращения скалярной функции, вычисляемая как матричное произведение ковариантной строки частных производных на вертикальный контравариантный столбец приращений независимых аргументов: $dF = \frac{\partial F}{\partial \mathbf{A}} d\mathbf{A} = \sum_{i=1}^m \frac{\partial F}{\partial A^i} dA^i$ [Лурье, 2019].  
    * *Встречается в разделах:* **А.1 (Раздел 1.2.4)**, **А.2 (Раздел 3.2)**, **А.3 (Раздел 1.4)**.

### И

*   **Идеальная резонансная задача (Модель Болина-Гарфинкеля)** - промежуточный интегрируемый гамильтониан, описывающий локальное поведение анизохронной системы в окрестности изолированной резонансной поверхности [Garfinkel, 1966]. Получается методом частичного усреднения по нерезонансным фазам и удержания ведущих членов разложения Тейлора по отклонениям действий от резонансного слоя. Математическая структура модели полностью эквивалентна уравнениям движения нелинейного математического маятника: $\mathcal{H} = \frac{1}{2} G P^2 - F \cos \theta_{\mathrm{res}}$ [Чириков, 1979].  
    * *Встречается в разделах:* **А.5 (Раздел 4, Секция 4.3)**, **А.7 (Секция 4.3, 5.6)**.

*   **Изохронный (линейный) резонанс** - динамический режим соизмеримости частот в линейных системах, где невозмущенный вектор частот полностью инвариантен относительно канонических действий или параметров: $\mathbf{M}_{\mathrm{hess}} \equiv \mathbf{0}_{n \times n}$ [Арнольд, 1985]. В отличие от анизохронного случая, попадание в изохронный резонанс вызывает неограниченный линейный рост амплитуд действий по времени ($I \sim \varepsilon t$), поскольку в линейной фазовой структуре отсутствует кинематический механизм частотного сдвига и самопроизвольного выхода системы из синхронизма [Чириков, 1979].  
    * *Встречается в разделах:* **А.5 (Раздел 2.1, Секция 2.4)**, **А.7 (Секция 1.5.1)**.

*   **Интеграл движения (Первый интеграл)** - гладкая скалярная функция фазовых переменных $F(q,p,t)$, сохраняющая постоянное значение вдоль любой истинной траектории динамической системы. Необходимым и достаточным условием является тождественное зануление полной производной по времени в силу уравнений движения: $\frac{\partial F}{\partial t} + \{H, F\} \equiv 0$ [Лурье, 2019; Козлов, 1995].  
    * *Встречается в разделах:* **А.2 (Раздел 1, Раздел 2.1)**.

*   **Инфинитезимальный генератор** - бесконечно малый порождающий элемент гладкого однопараметрического семейства преобразований [Новиков, Тайманов, 2014]. В гамильтоновом исчислении Ли представляет собой линейный дифференциальный оператор первого порядка касательной алгебры Ли, задающий направленную производную вдоль интегральных кривых фазового потока в окрестности тождественного отображения [Арнольд, 1989; Deprit, 1969].  
    * *Встречается в разделах:* **А.3 (Раздел 1.1, Раздел 1.4)**.

### К

*   **КАМ-теория (Теория Колмогорова-Арнольда-Мозера)** - математическая теория возмущений гамильтоновых систем, исследующая условия сохранения условно-периодических движений на инвариантных торах при добавлении малых нелинейных возмущений. Теория доказывает, что для большинства иррациональных (нерезонансных) векторов частот, удовлетворяющих условиям диофантовых приближений и критерию анизохронности, инвариантные торы не разрушаются, а лишь деформируются возмущением, обеспечивая вековую устойчивость по Ляпунову [Арнольд, 1985; Арнольд, 2024].  
    * *Встречается в разделах:* **А.2 (Раздел 9.4)**, **А.5 (Секция 3.2)**, **А.7 (Введение, Секция 3.5, 5.1)**.

*   **Канонические уравнения Гамильтона** - симплектическая система $2n$ обыкновенных дифференциальных уравнений первого порядка, описывающая эволюцию механической системы в фазовом пространстве через частные производные гамильтониана по сопряженным парам переменных: $\frac{d}{dt}q = (\frac{\partial H}{\partial p})^T$, $\frac{d}{dt}p = -\frac{\partial H}{\partial q}$ [Лурье, 2019; Маркеев, 2024]. В инвариантной матричной форме записывается через действие матрицы косого произведения на вектор-столбец градиента: $\frac{dx}{dt} = \mathbf{J}^T \operatorname{grad} H$ [Whittaker, 1937].  
    * *Встречается в разделах:* **А.1 (Раздел 3.3)**, **А.2 (Раздел 1, Раздел 3)**, **А.3 (Раздел 4.1)**.

*   **Каноническое преобразование (Симплектоморфизм)** - гладкая замена фазовых переменных, сохраняющая инвариантность канонической кососимметрической структуры пространства и форму дифференциальных уравнений Гамильтона в новой системе координат. Необходимым и достаточным условием каноничности отображения служит сохранение базисных скобок Пуассона или выполнение матричного критерия симплектичности $\mathbf{M}^T \mathbf{J} \mathbf{M} = \mathbf{J}$ для матрицы Якоби преобразования $\mathbf{M}$ [Арнольд, 1989; Лурье, 2019].  
    * *Встречается в разделах:* **А.2 (Раздел 5, Раздел 6)**, **А.3 (Секция 2.2, 4.2)**.

*   **Касательное расслоение ($TN^n$)** - гладкое дифференцируемое многообразие размерности $2n$, сформированное объединением линейных касательных пространств обобщенных скоростей $T_q N^n$ по всем точкам конфигурационного многообразия $N^n$. Является естественным пространством состояний для лагранжева формализма механики [Новиков, Тайманов, 2014; Арнольд, 1989].  
    * *Встречается в разделах:* **А.1 (Раздел 2.1, Раздел 2.2)**.

*   **Кокасательное расслоение ($T^*N^n$)** - гладкое дифференцируемое многообразие размерности $2n$, слоями которого в каждой точке конфигурационного многообразия служат линейные пространства ковекторов обобщенных импульсов $T^*_q N^n$. Является естественным геометрическим фазовым пространством для гамильтонова формализма механики [Новиков, Тайманов, 2014; Арнольд, 1989].  
    * *Встречается в разделах:* **А.1 (Раздел 3.1)**, **А.2 (Раздел 3.2)**, **А.3 (Раздел 1.1)**.

*   **Ковариантная строка (Ковекторы / Линейные формы)** - матричная форма представления элементов сопряженного линейного пространства $V^*$, состоящего из всех линейных скалярных функций (форм), заданных на векторах исходного пространства $V$. Физическими примерами в механике служат ковекторы обобщенных импульсов $p_i$, обобщенных сил $Q_i$ и градиенты скалярных полей. Координаты нумеруются нижними тензорными индексами [Новиков, Тайманов, 2014; Лурье, 2019].  
    * *Встречается в разделах:* **А.1 (Раздел 1.1, Раздел 1.2)**, **А.2 (Раздел 1)**.

*   **Конфигурационное многообразие ($N^n$)** - гладкое вещественное многообразие размерности $n$, точки которого однозначно задают мгновенные пространственные положения механической системы. Введение структуры искривленного многообразия вместо евклидова пространства обусловлено наличием в системе геометрических (голономных) связей [Арнольд, 1989; Лурье, 2019].  
    * *Встречается в разделах:* **А.1 (Раздел 2.1, Раздел 2.2)**.

*   **Кососимметричность скобки Пуассона** - фундаментальное свойство пуассоновой структуры, заключающееся в знаковой инверсии оператора при перестановке его функциональных аргументов: $\{F, G\} = -\{G, F\}$ для любых произвольных гладких функций фазового пространства [Козлов, 1995; Маркеев, 2024].  
    * *Встречается в разделах:* **А.2 (Раздел 2.1)**.

*   **Критический аргумент (Резонансная фаза)** - скалярная функция фазового пространства, определяемая как целочисленная линейная комбинация быстрых угловых переменных ($\psi = k\theta = \sum k_i\theta^i$). Полная производная по времени от резонансной фазы в силу уравнений невозмущенного движения тождественно равна нулю ($\dot{\psi} = k\omega(I) = 0$), что обеспечивает стационарное направленное влияние возмущающих сил на вековых масштабах времени [Арнольд, 1985; Чириков, 1979].  
    * *Встречается в разделах:* **А.5 (Раздел 1.3, Раздел 4)**, **А.7 (Раздел 4, Секция 5.6)**.

### Л

*   **Лагранжа уравнения движения второго рода** - инвариантная система $n$ дифференциальных уравнений второго порядка, описывающая динамику голономной механической системы в конфигурационном пространстве через обобщенные координаты $q^i$ и обобщенные скорости $\dot{q}^i$: $\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right) - \frac{\partial L}{\partial q} = \mathbf{0}_{1 \times n}$ [Лурье, 2019; Маркеев, 2024]. Согласно принятому геометрическому протоколу, уравнения записываются в строковой форме, где правая часть тождественно равна нулевой матрице-строке $\mathbf{0}_{1 \times n}$, согласуя размерности ковариантных сил и импульсов [Whittaker, 1937].  
    * *Встречается в разделах:* **А.1 (Раздел 2.1, Раздел 2.3)**, **А.2 (Раздел 1)**.

*   **Лейбница правило для скобки Пуассона** - фундаментальное алгебраическое свойство пуассоновой структуры, выражающееся в дифференциальном законе распределения оператора скобки на произведение функций: $\{F \cdot G, K\} = F\{G, K\} + \{F, K\}G$ для любых произвольных гладких функций фазового пространства [Козлов, 1995; Маркеев, 2024]. Гарантирует, что действие скобки Пуассона удовлетворяет свойствам ковариантного дифференцирования.  
    * *Встречается в разделах:* **А.2 (Раздел 2.1, Секция 3.2)**.
    
*   **Либрация** - динамический режим нелинейного (анизохронного) резонанса, характеризующийся ограниченными замкнутыми колебаниями критического аргумента $\psi$ и сопряженного ему действия Делоне внутри области фазового пространства, ограниченной петлями сепаратрисы [Чириков, 1979]. В баллистике ИСЗ определяет устойчивое удержание орбитальных элементов в резонансных ловушках.  
    * *Встречается в разделах:* **А.2 (Секция 8.5)**, **А.5 (Раздел 2.2, Секция 4.2)**, **А.7 (Секция 4.3, 4.5)**.
    
*   **Ловушка Нейштадта (Адиабатический захват в резонанс)** - бифуркационный механизм в неавтономных или слабодиссипативных многочастотных системах, при котором медленная фазовая траектория пересекает расщепляющееся сепаратрисное многообразие, вследствие чего критический аргумент синхронизируется и переходит в режим долгопериодической либрации около устойчивого смещенного фокуса Пуанкаре [Нейштадт, 1975].  
    * *Встречается в разделах:* **А.5 (Секция 2.5, Раздел 4.4, Секция 4.6)**, **А.7 (Секция 4.3, 5.2)**.

### М

*   **Малые знаменатели** - аналитические сингулярности вида $(k\omega(I))^{-1}$, возникающие в комплексных амплитудах Фурье производящих функций при решении линейных дифференциальных уравнений гомологии в окрестности соизмеримостей частот. Приводят к мгновенному нарушению иерархии по степеням малости и расходимости классических асимптотических рядов Пуанкаре—Цейпеля [Арнольд, 1985; Заславский, 1984].  
    * *Встречается в разделах:* **А.2 (Раздел 9.4)**, **А.5 (Введение, Раздел 3)**, **А.7 (Введение, Секция 5.6)**.
    
*   **Матрица Якоби** - функциональная матрица частных производных векторного отображения по векторному аргументу, тождественная производной вектора по вектору. В рамках принятого матричного протокола имеет размерность $k \times m$ при отображении $\mathbb{R}^m \to \mathbb{R}^k$ и служит базисным коэффициентом дифференциала [Лурье, 2019].  
    * *Встречается в разделах:* **А.1 (Раздел 1.2.2, Раздел 1.2.4)**, **А.2 (Раздел 6.2)**, **А.3 (Секция 2.2)**.
    
*   **Метод усреднения** - асимптотическая математическая процедура проецирования полного нелинейного фазового потока многочастотной системы на инвариантное подмногообразие медленных движений. Основана на эргодической замене динамического интегрирования уравнений по времени регулярным геометрическим интегрированием по объёму (мере Лебега) инвариантного тора $\mathbb{T}^n$, что позволяет исключить короткопериодические флуктуации без потери точности в главном порядке по малому параметру [Боголюбов, Митропольский, 1955; Арнольд, 1989].  
    * *Встречается в разделах:* **А.7 (Введение, Раздел 1.2, Секция 1.2.3, 1.4)**.

### Н

*   **Ньютона уравнения движения (в конфигурационном пространстве)** - векторная форма дифференциальных уравнений второго порядка, описывающая динамику механической системы в вещественном пространстве $\mathbb{R}^n$ под действием статического потенциала $U(x)$ и быстрых осциллирующих сил $\varphi(x, \tau)$, имеющих нулевое пространственное среднее [Арнольд, 1989]. Служат отправной точкой для вывода эффективного потенциала и анализа динамической стабилизации макро-движений методом одночастотного усреднения по времени.

### О

*   **Однопараметрическая группа преобразований Ли** - непрерывное семейство диффеоморфизмов фазового пространства $\{g^\varepsilon\}_{\varepsilon \in \mathbb{R}}$, порожденное стационарным фазовым потоком автономного гамильтонова векторного поля и параметризованное скалярным шагом деформации (в теории возмущений — малым параметром $\varepsilon$). Семейство обладает структурой гладкого топологического многообразия и удовлетворяет групповым аксиомам тождественного отображения ($g^0 = \mathrm{Id}$), аддитивности аргумента ($g^{\varepsilon_1} \cdot g^{\varepsilon_2} = g^{\varepsilon_1 + \varepsilon_2}$) и наличия обратного элемента ($(g^\varepsilon)^{-1} = g^{-\varepsilon}$) [Новиков, Тайманов, 2014; Арнольд, 1989].  

    * *Встречается в разделах:* **А.3 (Раздел 1, Раздел 1.3)**.
*   **Оператор Ли ($L_W$)** - линейный дифференциальный оператор первого порядка в пуассоновой алгебре гамильтоновых функций, сопоставляющий любой произвольной гладкой функции состояния её каноническую скобку Пуассона с фиксированной производящей функцией (генератором преобразования) $W$, строго занимающей левую позицию: $L_W F = \{W, F\}$ [Deprit, 1969]. Оператор высшего порядка $L_W^k$ определяет $k$-кратное последовательное вычисление вложенных скобок Пуассона.  
    * *Встречается в разделах:* **А.3 (Раздел 4.1, Секция 5.1)**.

*   **Оскулирующие элементы орбиты** - набор шести независимых геометрических и кинематических параметров, однозначно задающих мгновенное положение и скорость ИСЗ в инерциальной системе координат в предположении, что с текущего момента времени $t$ все возмущающие силы мгновенно зануляются, а движение космического аппарата становится строго кеплеровским [Брумберг, 1980]. Под влиянием нецентральности геопотенциала и неконсервативных сил оскулирующие элементы непрерывно изменяются, приобретая характер медленных фазовых переменных [Эльясберг, 1965].  
    * *Встречается в разделах:* **А.2 (Раздел 9.1)**, **А.5 (Раздел 5)**, **А.7 (Секция 5.2)**.

### П

*   **Паутина Арнольда (Резонансная паутина)** - топологически связанная, всюду плотная многомерная сеть пересекающихся под различными углами резонансных поверхностей в пространстве действий систем с числом быстрых частот $n \geq 3$ [Арнольд, 1989]. В отличие от двухчастотных конфигураций, где регулярные торы полностью разделяют фазовые слои, в геометрии пятимерного энергетического многообразия трехчастотной пространственной задачи ИСЗ торы не способны изолировать объем. Резонансная паутина прошивает фазовое пространство насквозь, формируя сквозные многомерные каналы стохастичности, вдоль которых возможен макроскопический дрейф элементов орбиты (диффузия Арнольда) сколь угодно далеко от начального состояния [Арнольд, 1989; Чириков, 1979].  
    * *Встречается в разделах:* **А.7 (Секция 3.4.2, Раздел 3.5)**.

*   **Переменные Делоне** - классический канонический набор сопряженных переменных типа действие-угол в гамильтоновой небеской механике, осуществляющий разделение геометрических размеров кеплеровской орбиты, формы эллиптического трека и пространственной ориентации плоскости движения ИСЗ [Брумберг, 1980]. Теряет свойство регулярности (испытывает координатное вырождение) при стремлении эксцентриситета или наклонения к нулю, что требует остановки сквозных схем усреднения.  
    * *Встречается в разделах:* **А.2 (Раздел 9, Секция 9.2)**, **А.3 (Секция 3.3)**, **А.4 (Раздел введения)**, **А.5 (Секция 5.1)**.

*   **Переменные Пуанкаре (Модифицированные)** - регулярный наклонно-эксцентрический набор канонических фазовых переменных $(\lambda, \xi, q_P, \Lambda, \eta, p_P)^T$, полученный канонической деформацией элементов Делоне и полностью устраняющий неопределенность аргумента перицентра и долготы узла при околокруговых ($e \to 0$) и экваториальных наклонениях ($i \to 0^{\circ}$) орбит ИСЗ [Брумберг, 1980]. Трактуется как фундаментальный рабочий базис для построения регулярных аналитических спутниковых теорий возмущений, не имеющих координатных сингулярностей в окрестности рабочих орбит ИСЗ.  
    * *Встречается в разделах:* **А.2 (Раздел 9.5)**, **А.4 (Полный специализированный конспект по элементам Пуанкаре)**.

*   **Правило Лейбница для скобки Пуассона** - фундаментальное алгебраическое свойство пуассоновой структуры, выражающееся в дифференциальном законе распределения оператора скобки на произведение функций: $\{F \cdot G, K\} = F\{G, K\} + \{F, K\}G$ для любых произвольных гладких функций фазового пространства [Козлов, 1995; Маркеев, 2024]. Гарантирует, что действие скобки Пуассона удовлетворяет свойствам ковариантного дифференцирования.  
    * *Встречается в разделах:* **А.2 (Раздел 2.1, Секция 3.2)**.

*   **Производная Ли** - тензорный оператор, описывающий скорость изменения геометрического объекта (в аналитической механике — инвариантной кососимметрической 2-формы или дифференциальной функции состояния) вдоль интегральных кривых фавого потока, генерируемого заданным векторным полем $X$. Используется для матричного доказательства теоремы Лиувилля и построения рекуррентных соотношений теории возмущений [Новиков, Тайманов, 2014; Арнольд, 1989].  
    * *Встречается в разделах:* **А.3 (Секция 2.2)**.

*   **Производящая функция (Генератор канонического преобразования)** - скалярная функция фазовых переменных смешанного типа (в теории возмущений чаще всего второго типа — старых координат и новых импульсов: $S(q, P, t)$), дифференциальные соотношения Якоби от которой ($p_i = \partial S/\partial q^i$, $Q^i = \partial S/\partial P_i$) однозначно задают закон канонического преобразования координатных сеток фазового пространства [Лурье, 2019; Маркеев, 2024]. Конструируется в виде формального степенного ряда по малому параметру $\varepsilon$ для аналитической компенсации возмущающих членов гамильтониана.  
    * *Встречается в разделах:* **А.2 (Раздел 5, Секция 10.2)**, **А.5 (Раздел 3, Секция 4.4)**, **А.7 (Раздел 1.2.3, Раздел 2.2)**.

*   **Пространственное среднее (Среднее по многообразию)** - нормированный кратный интеграл от гладкой периодической функции возмущения по геометрическому объёму фундаментального домена $[0, 2\pi]^n$ многомерного инвариантного тора $\mathbb{T}^n$ по инвариантной мере Лебега. Представляет собой чистую геометрическую характеристику распределения пертурбационных полей, инвариантную относительно времени и не зависящую от начальных условий по фазам [Арнольд, 1989].  
    * *Встречается в разделах:* **А.7 (Раздел 1.4, Секция 1.4.2)**.

### Р

*   **Резонанс динамический** - явление синхронизации частот независимых периодических или циклических процессов внутри многомерной динамической системы, приводящее к качественному изменению топологии фазового потока. С точки зрения асимптотического анализа представляет собой сингулярность (проблему малых знаменателей), индуцирующую расходимость классических пертурбационных рядов по торам и требующую локальной перестройки уравнений движения [Арнольд, 1985; Заславский, 1984].  
    * *Встречается в разделах:* **А.2 (Раздел 9.4)**, **А.5 (Введение, Раздел 1)**, **А.7 (Введение, Раздел 3)**.
    
*   **Резонанса кратность** - максимальное количество линейно независимых целочисленных векторов-строк $k^{(j)}$, одновременно удовлетворяющих условию ортогональности фиксированному вектору невозмущенных частот системы. Для системы с $n$ частотами максимальная кратность равна $m = n - 1$, при которой инвариантный тор полностью вырождается в расслоение Зейферта, а все фазовые траектории замыкаются в периодические циклы [Арнольд, 1985].  
    * *Встречается в разделах:* **А.5 (Раздел 1.2)**, **А.7 (Секция 1.4.3)**.
    
*   **Резонанса порядок** - скалярная величина $N$, определяемая как сумма модулей компонент целочисленного вектора соизмеримости: $N = \sum |k_i|$. В асимптотическом анализе служит мерой аналитического веса соответствующей комбинационной гармоники возмущения, амплитуда которой для аналитических потенциалов убывает экспоненциально с ростом $N$ [Чириков, 1979; Арнольд, 1985].  
    * *Встречается в разделах:* **А.5 (Раздел 1.2)**, **А.7 (Секция 3.2, 5.1)**.
    
*   **Резонансная поверхность** - изолированная геометрическая гиперповерхность $\Sigma_k$ размерности $m - 1$ в пространстве медленных переменных (действий Делоне), на которой невозмущенный вектор частот удовлетворяет условию соизмеримости (ортогональности целочисленному вектору $k$): $\Sigma_k = \{I \in G \mid k \cdot \omega(I) = 0\}$ [Арнольд, 1989]. В анизохронных системах множество таких поверхностей всюду плотно в пространстве импульсов.  
    * *Встречается в разделах:* **А.5 (Раздел 2.2)**, **А.7 (Раздел 3.1, Секция 3.2, 5.2)**.
    
*   **Резонансная фаза (Критический аргумент)** - скалярная функция фазового пространства, представляющая собой ковариантную целочисленную линейную комбинацию быстрых угловых переменных: $\psi = k\theta = \sum k_i \theta^i$. В условиях точного резонанса полная производная по времени от критического аргумента в силу невозмущенных уравнений ОДУ обращается в ноль ($\dot{\psi} = 0$), фиксируя пространственное положение траектории и обеспечивая направленное вековое изменение действий [Арнольд, 1985; Чириков, 1979].  
    * *Встречается в разделах:* **А.5 (Раздел 1.3, Раздел 4)**, **А.7 (Раздел 4, Секция 5.6)**.

*   **Резонансная щель Чирикова** - изолированная область фазового пространства в окрестности соизмеримости частот, заключенная внутри замкнутых петель сепаратрисы Идеальной резонансной задачи, полная ширина которой по оси канонических действий определяется квадратичным корнем из амплитуды возмущения и параметром анизохронности системы: $\Delta P_{\mathrm{gap}} = 4 \sqrt{F/G}$ [Чириков, 1979]. Служит математической границей коридора локального разрушения инвариантных торов.  
    * *Встречается в разделах:* **А.5 (Раздел 4.2, Секция 4.3)**, **А.7 (Раздел 3.5)**.

*   **Резонансное соотношение** - алгебраическое условие соизмеримости частот невозмущенного движения, выражающееся в занулении линейной комбинации компонент вектора частот $\omega$ с элементами ненулевого целочисленного вектора решетки $k \in \mathbb{Z}^n \setminus \{0\}$: $k\omega = \sum_{i=1}^n k_i \omega^i = 0$ [Whittaker, 1937; Арнольд, 1985].  
    * *Встречается в разделах:* **А.2 (Раздел 9.4)**, **А.5 (Раздел 1.1)**, **А.7 (Раздел 3.1)**.
    
*   **Резонансные зоны в динамике ИСЗ (Классификация)** - спектральное разделение соизмеримостей в околоземной астродинамике, строящееся на соотношениях между кеплеровым средним движением спутника $n$, угловой скоростью осевого вращения Земли $\omega_{\mathrm{E}}$ и частотами прецессии плоскости орбиты ($\dot{g}, \dot{h}$) под действием возмущений [Брумберг, 1980]:
    1. *Суточные (тессеральные) резонансы* ($n \approx \frac{m}{k_1}\omega_{\mathrm{E}}$, геостационарные миссии 1:1, полусуточные высокоэллиптические траектории «Молния» и навигационные ИСЗ 2:1) [Garfinkel, 1966; Нейштадт, 1975];
    2. *Внутрисистемные вековые резонансы* (соизмеримости собственных низких частот ИСЗ, включая резонанс критического наклонения $i \approx 63.4^{\circ}$, где прецессия перигея зануляется: $\dot{g} \approx 0$) [Арнольд, 1985];
    3. *Внешние лунно-солнечные резонансы* (соизмеримости вековых частот спутника и частот движения возмущающих тел) [Заславский, 1984].  
    * *Встречается в разделах:* **А.5 (Раздел 5, Секция 5.2–5.5)**.

*   **Резонансные уравнения Болина** - каноническая система дифференциальных уравнений движения в окрестности соизмеримости частот, полученная посредством локального изменения масштаба медленных переменных в ряды по степеням $\sqrt{\varepsilon}$ [Bohlin, 1888]. Позволяет устранить аналитическую сингулярность малого знаменателя за счет изменения порядка малости частоты либраций до величины $\mathcal{O}(\sqrt{\varepsilon})$, редуцируя задачу к промежуточному гамильтониану маятникового типа [Garfinkel, 1966].  
    * *Встречается в разделах:* **А.5 (Секция 3.2, 4.4)**, **А.7 (Секция 1.5.3, 5.6)**.
    
*   **Ротация (Свободная фазовая ротация)** - динамический режим движения изображающей точки вне сепаратрисного контура соизмеримости частот, характеризующийся монотонным (квазиравномерным) изменением резонансного углового аргумента во времени и отсутствием долгопериодических колебаний или вековых уходов сопряженной переменной действия [Чириков, 1979]. В данном режиме интегральное влияние возмущающих сил на медленные переменные взаимно компенсируется на промежуточных интервалах времени.  
    * *Встречается в разделах:* **А.2 (Секция 8.5)**, **А.5 (Секция 4.2)**, **А.7 (Секция 4.3, 4.5)**.

### С

*   **Сепаратриса** - инвариантное многообразие фазового пространства (набор траекторий и особых точек), разделяющее области с качественно различным поведением фазовых потоков (в нелинейном резонансе — область колебательных либраций критического аргумента от областей его вращательной ротации). В маятниковых рельефах идеальной резонансной задачи состоит из неустойчивых седловых положений равновесия и асимптотических интегральных кривых, стремящихся к этим седлам при $t \to \pm\infty$ [Арнольд, 1989; Чириков, 1979].  
    * *Встречается в разделах:* **А.2 (Секция 8.5)**, **А.5 (Раздел 4.2, Секция 4.4)**, **А.7 (Секция 4.3)**.
    
*   **Симплектическая единица ($\mathbf{J}$)** - постоянная блочная кососимметрическая матрица размера $2n \times 2n$, определяющая кососимметрическое скалярное произведение (симплектическую форму) фазового пространства и осуществляющая косой поворот векторов градиента функции Гамильтона для вычисления векторов скоростей фазового потока: $\mathbf{J} = \begin{pmatrix} \mathbf{0} & \mathbf{I}_n \\ -\mathbf{I}_n & \mathbf{0} \end{pmatrix}$ [Whittaker, 1937; Арнольд, 1989].  
    * *Встречается в разделах:* **А.1 (Раздел 3.3)**, **А.2 (Раздел 2.2, Раздел 3)**, **А.3 (Секция 2.1)**.
    
*   **Скобка Пуассона каноническая** - кососимметрическое скалярное произведение канонических градиентов двух гладких функций состояния фазового пространства через симплектическую единицу $\mathbf{J}$, задающее пуассонову алгебраическую структуру и определяющее временную эволюцию динамических переменных: $\{F, G\} = \langle\nabla F, \mathbf{J}\nabla G\rangle$ [Whittaker, 1937; Арнольд, 1989].  
    * *Встречается в разделах:* **А.2 (Раздел 2, Секция 2.1)**, **А.3 (Раздел 3.1)**.
    
*   **Сопряженная (предельная) невырожденность Арнольда** - дифференциальный критерий анизохронности для предельно вырожденных систем небесной механики, требующий невырожденности матрицы вторых производных усредненной возмущающей функции первого порядка по медленным действиям Делоне при тождественном занулении гессиана невозмущенного гамильтониана задачи двух тел [Арнольд, 2024].  
    * *Встречается в разделах:* **А.2 (Раздел 9.4)**, **А.5 (Раздел 2.2, Секция 5.5)**.
    
*   **Спаривание (Свертка)** - вычисление значения ковектора (линейной формы) на векторе, математически эквивалентное стандартному матричному произведению строки координат ковектора на столбец координат вектора и инвариантное относительно выбора базиса линейного пространства: $B(A) = BA = \sum B_i A^i$ [Новиков, Тайманов, 2014].  
    * *Встречается в разделах:* **А.1 (Раздел 1.1, Раздел 1.1.1)**.

### Т

*   **Теорема Лиувилля (в геометрической форме)** - теорема, утверждающая, что фазовый поток, порожденный гамильтоновым векторным полем любой гладкой функции состояния, строго сохраняет каноническую структуру пространства (дифференциальную 2-форму), вследствие чего производная Ли от матрицы косого произведения вдоль потока тождественно равна нулю: $\mathcal{L}_{X_W} \mathbf{J} = 0$ [Whittaker, 1937; Арнольд, 1989].  
    * *Встречается в разделах:* **А.3 (Секция 2.2, Секция 5.1)**, **А.5 (Секция 1.5)**.
    
*   **Теорема переноса Депри** - операторная теорема теории возмущений Ли, утверждающая, что операция канонического преобразования координат коммутирует с оператором экспоненты Ли: $\bar{H} = \exp(\varepsilon L_W)H(x)$, позволяя находить трансформированные функции энергии в явном виде без подстановки неявных траекторий [Deprit, 1969].  
    * *Встречается в разделах:* **А.3 (Раздел 4.2, Секция 5.1)**.
    
*   **Тождество Якоби** - третья фундаментальная аксиома бесконечномерных алгебр Ли для вложенных канонических скобок Пуассона: $\{F, \{G, K\}\} + \{G, \{K, F\}\} + \{K, \{F, G\}\} \equiv 0$, служащая правилом редукции символьных выражений и гарантирующая алгебраическую замкнутость процедур аналитического усреднения [Козлов, 1995; Арнольд, 1989].  
    * *Встречается в разделах:* **А.2 (Раздел 2.1)**, **А.3 (Раздел 3.2, Секция 3.3)**.

### У

*   **Уравнение гомологии** - линейное дифференциальное уравнение в частных производных первого порядка относительно искомого генератора зануляющей поправки координат теории возмущений, разрешимость которого в классе гладких периодических функций гарантируется равенством нулю пространственного среднего от возмущающего потенциала [Арнольд, 1989].  
    * *Встречается в разделах:* **А.5 (Раздел 3, Секция 3.1)**, **А.7 (Раздел 2.2, Раздел 3)**.

### Ф

*   **Фазовая кривая (Траектория фазового потока)** - одномерное геометрическое многообразие (кривая) в $2n$-мерном фазовом пространстве динамической системы, параметризованное временем $t$ (или групповым деформационным параметром $\varepsilon$) и представляющее собой образ решения канонической задачи Коши при заданных начальных условиях [Арнольд, 1989].  
    * *Встречается в разделах:* **А.1 (Раздел 2.2, Секция 3.2)**, **А.2 (Секция 8.5)**, **А.5 (Раздел 1.1)**.

*   **Фазовый портрет** - глобальная геометрическая карта фазового пространства (или его двумерных сечений), отображающая полную совокупность всех возможных фазовых кривых, сепаратрисных многообразий, особых точек и областей регулярных движений исследуемой динамической системы [Андронов и др., 1967]. Позволяет проводить качественный топологический анализ устойчивости режимов движения без явного аналитического интегрирования ОДУ [Чириков, 1979].  
    * *Встречается в разделах:* **А.2 (Секция 8.5)**, **А.5 (Раздел 2.3, Секция 4.4)**.

*   **Фазовый поток** - однопараметрическое семейство диффеоморфизмов $g^t: \mathbb{R}^{2n} \to \mathbb{R}^{2n}$ фазового пространства на себя, сопоставляющее любой начальной точке её положение на интегральной кривой через промежуток времени $t$ [Арнольд, 1989]. В автономных гамильтоновых системах фазовый поток сохраняет инвариантной симплектическую меру пространства (фазовый объем) [Whittaker, 1937].  
    * *Встречается в разделах:* **А.1 (Раздел 2.3)**, **А.2 (Раздел 2.2, Раздел 5)**, **А.3 (Раздел 1.2, Секция 2.2)**.

### Э

*   **Элементы Кеплеровой орбиты** - канонический набор шести независимых геометрических параметров, полностью определяющих форму, размеры и пространственную ориентацию невозмущенной эллиптической траектории космического аппарата в центральном гравитационном поле планеты [Брумберг, 1980; Эльясберг, 1965]:
    *   $a$ - *большая полуось*, масштабный параметр, фиксирующий геометрический размер орбиты и определяющий полную механическую энергию невозмущенного кеплеровского движения;
    *   $e$ - *эксцентриситет*, параметр формы, определяющий степень отклонения эллиптической траектории от круговой геометрии;
    *   $i$ - *наклонение*, угол между полярной осью $Z$ инерциальной системы координат и вектором орбитального момента импульса ИСЗ;
    *   $\Omega$ - *долгота восходящего узла*, угловое расстояние от фиксированной оси абсцисс $X$ до линии пересечения плоскости орбиты с экваториальной плоскостью;
    *   $\omega$ - *аргумент перицентра*, угол в плоскости орбиты ИСЗ, отсчитываемый от направления на восходящий узел до точки минимального расстояния до центрального тела;
    *   $M$ - *средняя аномалия*, быстрая линейная фазовая координата, характеризующая текущее положение материальной точки на орбите относительно времени прохождения перицентра.  
    * *Встречается в разделах:* **А.2 (Раздел 9.1, Секция 9.2, 9.6)**, **А.5 (Раздел 5)**.
