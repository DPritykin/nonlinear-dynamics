# Вводный конспект: Нотация уравнений аналитической механики

Настоящий конспект фиксирует систему обозначений, базовых понятий и соглашений о знаках, используемых на протяжении всего цикла аналитических исследований.

---

## 1. Протокол алгебраических и аналитических обозначений

Для построения алгоритмов аналитической механики и их корректной реализации в системах компьютерной алгебры необходимо зафиксировать строгие правила матрично-векторного исчисления в конечномерных вещественных линейных пространствах.

### 1.1. Векторы и ковекторы в линейных пространствах
Пусть $V$ является вещественным линейным пространством размерности $m$. Элементы данного пространства называются векторами. 
Пусть $V^*$ является сопряженным к нему линейным пространством, которое состоит из всех линейных скалярных функций, заданных на пространстве $V$. Элементы сопряженного пространства называются ковекторами или линейными формами.

Для представления векторов и ковекторов в фиксированном базисе принимаются следующие матричные соглашения.

1. **Вектор-столбец**. По умолчанию любой вектор $A \in V$ рассматривается как вертикальный столбец своих координат размерности $m \times 1$:
   $$A = \begin{pmatrix} A^1 \\ \vdots \\ A^m \end{pmatrix} = (A^1, \dots, A^m)^T$$
   Верхние индексы применяются в качестве стандартных нумераторов координат вектора.

2. **Вектор-строка**. Любой ковектор $B \in V^*$ рассматривается как горизонтальная строка своих координат размерности $1 \times m$:
   $$B = \begin{pmatrix} B_1 & \dots & B_m \end{pmatrix}$$
   Нижние индексы применяются в качестве нумераторов координат ковектора.

Спаривание ковектора $B$ и вектора $A$, то есть вычисление значения линейной формы $B$ на векторе $A$, математически эквивалентно стандартному матричному произведению строки на столбец:
$$B(A) = B A = \begin{pmatrix} B_1 & \dots & B_m \end{pmatrix} \begin{pmatrix} A^1 \\ \vdots \\ A^m \end{pmatrix} = \sum_{i=1}^m B_i A^i$$
Результатом операции спаривания является скаляр.


### 1.1.1. Дифференциально-геометрический смысл: ковариантность и контравариантность

Принятые соглашения имеют строгий геометрический смысл и напрямую отражают понятия **контравариантности** и **ковариантности** при преобразовании базиса.

1. **Контравариантные компоненты (Векторы):**
   * *Суть:* Это элементы исходного линейного пространства $V$ (например, обобщенные координаты $q^i$, обобщенные скорости $\dot{q}^i$ или вариации координат $\delta q^i$).
   * *Тензорный индекс:* Пишется **сверху** ($A^i$).
   * *Матричная форма:* Представляются в виде **вектора-столбца** размерности $m \times 1$. При изменении координат компоненты этого столбца преобразуются с помощью матрицы, *обратной* матрице перехода (противоположным, или *контравариантным* образом), чтобы сам геометрический объект остался инвариантным.

2. **Ковариантные компоненты (Ковекторы / Линейные формы):**
   * *Суть:* Это элементы сопряженного пространства $V^*$, представляющие собой линейные функции на векторах. Физическими примерами в механике служат обобщенный импульс $p_i = \frac{\partial L}{\partial \dot{q}^i}$, градиент скалярного поля $\frac{\partial F}{\partial q^i}$ или обобщенная сила $Q_i$.
   * *Тензорный индекс:* Пишется **снизу** ($B_i$).
   * *Матричная форма:* Представляются в виде **вектора-строки** размерности $1 \times m$. При изменении координат их компоненты преобразуются с помощью исходной матрицы перехода (*согласованным*, или *ковариантным* образом).

#### Инвариантность операции спаривания (свертки)

Физические и динамические скалярные величины (такие как действие ковектора на вектор $B(A)$, кинетическая энергия $T = \frac{1}{2} p_i \dot{q}^i$ или виртуальная работа $\delta A = Q_i \delta q^i$) являются инвариантами и не должны зависеть от выбора системы координат. 

В тензорном анализе это обеспечивается автоматической сверткой верхнего (контравариантного) и нижнего (ковариантного) индексов. В рамках принятой нотации эта операция естественным образом эквивалентна стандартному произведению строки на столбец без необходимости искусственного применения операции транспонирования:

$$\text{Скаляр} = \sum_{i=1}^m B_i A^i = B A$$

$$\text{Виртуальная работа} = \sum_{i=1}^n Q_i \delta q^i = Q \delta q$$

Такое разделение позволяет системам компьютерной алгебры на уровне размерностей контролировать геометрическую корректность аналитических выражений: операция умножения столбца на столбец напрямую (без участия метрического тензора) интерпретируется алгоритмом как явная математическая ошибка.


### 1.2. Правила матричного дифференцирования и базовые операторы

Перед введением дифференциально-геометрических понятий зафиксируем правила вычисления производных для скалярных и векторных функций по векторным аргументам разной структуры. В рамках данной нотации операция дифференцирования скаляра по вертикальному объекту (вектору) всегда порождает горизонтальный объект (строку), а по горизонтальному (ковектору) - вертикальный (столбец).

#### 1.2.1. Дифференцирование по вектору и ковектору
Пусть $F$ является гладкой скалярной функцией.
1. **Дифференцирование по вектору-столбцу** $A = (A^1, \dots, A^m)^T$. Производная скаляра по вектору-столбцу определяется как матрица-строка частных производных размерности $1 \times m$:
   $$\frac{\partial F}{\partial A} = \begin{pmatrix} \frac{\partial F}{\partial A^1} & \dots & \frac{\partial F}{\partial A^m} \end{pmatrix}$$

2. **Дифференцирование по вектору-строке (ковектору)** $B = \begin{pmatrix} B_1 & \dots & B_m \end{pmatrix}$. Производная скаляра по вектору-строке определяется как вектор-столбец частных производных размерности $m \times 1$:
   $$\frac{\partial F}{\partial B} = \begin{pmatrix} \frac{\partial F}{\partial B_1} \\ \vdots \\ \frac{\partial F}{\partial B_m} \end{pmatrix} = \left( \frac{\partial F}{\partial B_1}, \dots, \frac{\partial F}{\partial B_m} \right)^T$$

#### 1.2.2. Дифференцирование вектора по вектору
Пусть задано гладкое отображение $\mathbf{Y}(A) = (Y^1(A), \dots, Y^k(A))^T$, преобразующее вектор-столбец $A \in \mathbb{R}^m$ в вектор-столбец $\mathbf{Y} \in \mathbb{R}^k$.
Производная вектора-столбца по вектору-столбцу определяется как операция последовательного построчного дифференцирования каждого скалярного компонента $Y^i$ по вектору $A$. Результатом является матрица размерности $k \times m$:
$$\frac{\partial \mathbf{Y}}{\partial A} = \begin{pmatrix} \frac{\partial Y^1}{\partial A} \\ \vdots \\ \frac{\partial Y^k}{\partial A} \end{pmatrix} = \begin{pmatrix} 
    \frac{\partial Y^1}{\partial A^1} & \dots & \frac{\partial Y^1}{\partial A^m} \\ 
    \vdots & \ddots & \vdots \\ 
    \frac{\partial Y^k}{\partial A^1} & \dots & \frac{\partial Y^k}{\partial A^m} 
    \end{pmatrix}$$
    
#### 1.2.3. Дифференцирование вектора-столбца по вектору-строке (ковектору)
Пусть задано гладкое отображение $\mathbf{Y}(B) = (Y^1(B), \dots, Y^k(B))^T$, преобразующее вектор-строку $B \in (\mathbb{R}^m)^*$ в вектор-столбец $\mathbf{Y} \in \mathbb{R}^k$.

Производная вектора-столбца по вектору-строке определяется как операция последовательного покомпонентного дифференцирования каждого скалярного элемента $Y^i$ по вектору-строке $B$. Поскольку согласно подразделу 1.2.1 производная скаляра по вектору-строке порождает вектор-столбец частных производных, итоговый объект формируется путем горизонтального (постолбцового) объединения полученных столбцов. Результатом операции является матрица размерности $k \times m$:

$$\frac{\partial \mathbf{Y}}{\partial B} = \begin{pmatrix} \frac{\partial Y^1}{\partial B_1} & \dots & \frac{\partial Y^1}{\partial B_m} \\ \vdots & \ddots & \vdots \\ \frac{\partial Y^k}{\partial B_1} & \dots & \frac{\partial Y^k}{\partial B_m} \end{pmatrix}$$

При такой фиксации правил сохраняется сквозная согласованность матричных размерностей. Дифференциал векторного отображения $\mathbf{Y}(B)$ принимает инвариантную форму:
$$d\mathbf{Y} = \frac{\partial \mathbf{Y}}{\partial B} (dB)^T$$
где матрица производных размерности $k \times m$ умножается на вертикальный столбец приращений ковектора $(dB)^T$ размерности $m \times 1$, возвращая столбец приращений $d\mathbf{Y}$ размерности $k \times 1$.


#### 1.2.4. Определения дифференциала, градиента и матрицы Якоби
На основе сформулированных правил введем фундаментальные аналитические операторы для скалярной функции $F(A)$ и векторного отображения $\mathbf{Y}(A)$ на пространстве $\mathbb{R}^m$. Примем вектор приращений аргументов в виде столбца $dA = (dA^1, \dots, dA^m)^T$.

*   **Дифференциал скалярной функции**. Дифференциал $dF$ представляет собой главную линейную часть приращения функции и вычисляется как матричное произведение строки производных на столбец приращений аргументов, возвращая скаляр:
    $$dF = \frac{\partial F}{\partial A} dA = \sum_{i=1}^m \frac{\partial F}{\partial A^i} dA^i$$
*   **Дифференциал векторного отображения**. Дифференциал $d\mathbf{Y}$ определяет вектор-столбец приращений размерности $k \times 1$ и вычисляется путем умножения матрицы производных на столбец приращений аргументов:
    $$d\mathbf{Y} = \frac{\partial \mathbf{Y}}{\partial A} dA$$
*   **Градиент скалярной функции**. Оператор градиента $\text{grad}\,F$ (в альтернативной нотации $\nabla F$) определяет вектор-столбец, который составлен из частных производных и связан с матрицей-строкой производных операцией транспонирования:
    $$\text{grad}\,F = \left( \frac{\partial F}{\partial A} \right)^T = \begin{pmatrix} \frac{\partial F}{\partial A^1} \\ \vdots \\ \frac{\partial F}{\partial A^m} \end{pmatrix}$$
*   **Матрица Якоби**. Матрицей Якоби векторного отображения $\mathbf{Y}(A)$ называется функциональная матрица $\mathbf{M}$ размерности $k \times m$, тождественная производной вектора по вектору и явно состоящая из частных производных каждого компонента:
    $$\mathbf{M} = \frac{\partial \mathbf{Y}}{\partial A} = \begin{pmatrix} 
    \frac{\partial Y^1}{\partial A^1} & \dots & \frac{\partial Y^1}{\partial A^m} \\ 
    \vdots & \ddots & \vdots \\ 
    \frac{\partial Y^k}{\partial A^1} & \dots & \frac{\partial Y^k}{\partial A^m} 
    \end{pmatrix}$$

При такой фиксации правил случай $k = 1$ полностью согласован. Матрица Якоби скалярной функции $\mathbf{M}$ в точности совпадает с её матрицей-строкой производных $\frac{\partial F}{\partial A}$, которая служит базисным коэффициентом дифференциала $dF$, а vector-столбец градиента всегда равен транспонированной матрице Якоби этой функции: $\text{grad}\,F = \mathbf{M}^T$.


## 2. Лагранжев формализм и геометрия конфигурационного пространства

### 2.1. Конфигурационное многообразие и касательное расслоение
Рассмотрим консервативную механическую систему, обладающую $n$ степенями свободы. Совокупность всех возможных пространственных положений, которые данная система может занимать мгновенно, образует гладкое $n$-мерное вещественное многообразие $N^n$. Это многообразие принято называть конфигурационным многообразием системы. В локальной координатной карте точка данного многообразия однозначно задается набором обобщенных координат, которые формируют вектор-столбец $q = (q^1, \dots, q^n)^T$.

Отказ от евклидова пространства в пользу дифференцируемого многообразия обусловлен наличием геометрических связей в системе. Например, сферический маятник или твердое тело с неподвижной точкой совершают движение на искривленных поверхностях, топология которых не эквивалентна плоскому пространству $\mathbb{R}^n$.

Для полного описания динамического состояния механической системы необходимо задать не только ее положение, но и обобщенные скорости $\dot{q} = (\dot{q}^1, \dots, \dot{q}^n)^T$. Множество векторов скоростей $\dot{q}$ в фиксированной точке $q$ образует линейное касательное пространство $T_q N^n$. С геометрической точки зрения элементы касательного пространства являются векторами скоростей кривых, проходящих через точку $q$.

Объединение касательных пространств, взятых по всем точкам конфигурационного многообразия, порождает новое гладкое многообразие размерности $2n$:
$$TN^n = \bigcup_{q \in N^n} T_q N^n$$
Данная геометрическая структура называется касательным расслоением многообразия $N^n$. Локальные координаты на многообразии касательного расслоения задаются упорядоченной парой векторов $(q, \dot{q})$, где координаты $q^i$ определяют точку на базе расслоения, а скорости $\dot{q}^i$ задают вектор в касательном пространстве к этой точке.

### 2.2. Функция Лагранжа и ее геометрический смысл
Динамика системы на касательном расслоении однозначно определяется гладкой скалярной функцией Лагранжа (лагранжианом), которая зависит от координат, скоростей и, в общем случае, от времени:
$$L(q, \dot{q}, t): TN^n \times \mathbb{R} \to \mathbb{R}$$

Геометрический смысл лагранжиана состоит в том, что его кинетическая часть задает на конфигурационном многообразии риманову метрику. Квадратичная форма кинетической энергии $T = \frac{1}{2} \sum_{i,j=1}^n g_{ij}(q)\dot{q}^i\dot{q}^j$ в точности определяет метрический тензор $g_{ij}(q)$ многообразия $N^n$. В рамках этой интуиции свободное движение механической системы (в отсутствие внешних сил) происходит по геодезическим линиям данного риманова пространства, то есть по путям наименьшей кривизны. Введение потенциальной энергии $V(q)$ интерпретируется как динамическое возмущение этой чистой геометрической метрики, искривляющее геодезические траектории.


### 2.3. Векторно-матричная форма уравнений Эйлера-Лагранжа

Траектории движения консервативной механической системы на касательном расслоении конфигурационного многообразия задаются системой $n$ обыкновенных дифференциальных уравнений второго порядка, которые называются уравнениями Эйлера-Лагранжа:
$$\frac{d}{dt}\left( \frac{\partial L}{\partial \dot{q}^i} \right) - \frac{\partial L}{\partial q^i} = 0, \quad (i = 1, \dots, n)$$

Для построения векторно-матричной формы этих уравнений примем во внимание правила исчисления, установленные в подразделе 1.2.1. Поскольку обобщенные координаты $q$ и обобщенные скорости $\dot{q}$ определены как векторы-столбцы, дифференцирование скалярной функции Лагранжа по этим вертикальным объектам автоматически порождает матрицы-строки частных производных размерности $1 \times n$:
$$\frac{\partial L}{\partial q} = \begin{pmatrix} \frac{\partial L}{\partial q^1} & \dots & \frac{\partial L}{\partial q^n} \end{pmatrix}, \quad \frac{\partial L}{\partial \dot{q}} = \begin{pmatrix} \frac{\partial L}{\partial \dot{q}^1} & \dots & \frac{\partial L}{\partial \dot{q}^n} \end{pmatrix}$$

Оператор полной производной по времени $\frac{d}{dt}$ применяется к матрице-строке $\frac{\partial L}{\partial \dot{q}}$ поэлементно, сохраняя её горизонтальную структуру. В результате инвариантная векторно-матричная запись уравнений Эйлера-Лагранжа в строковой форме принимает вид:
$$\frac{d}{dt}\left( \frac{\partial L}{\partial \dot{q}} \right) - \frac{\partial L}{\partial q} = \mathbf{0}_{1 \times n}$$

где $\mathbf{0}_{1 \times n}$ является нулевой матрицей-строкой размерности $1 \times n$. Данная горизонтальная форма записи полностью самосогласована с принятым алгебраическим протоколом, отражает ковариантную природу уравнений движения и напрямую транслируется в алгоритмы символьных вычислений систем компьютерной алгебры.

*Примечание*. В случае неконсервативных систем правая часть уравнений движения перестает быть нулевой. Для описания таких систем вводится вектор обобщенных сил. Пусть на систему действуют $N$ внешних активных сил $\mathbf{F}_\alpha$, приложенных к точкам с радиус-векторами $\mathbf{r}_\alpha(q, t)$. Обобщенная сила $Q_i$, соответствующая координате $q^i$, определяется через виртуальную работу сил на вариациях координат:
$$\delta A = \sum_{\alpha=1}^N \mathbf{F}_\alpha \cdot \delta \mathbf{r}_\alpha = \sum_{\alpha=1}^N \mathbf{F}_\alpha \cdot \sum_{i=1}^n \frac{\partial \mathbf{r}_\alpha}{\partial q^i} \delta q^i = \sum_{i=1}^n Q_i \delta q^i$$
Откуда компоненты силы имеют вид $Q_i = \sum_{\alpha=1}^N \mathbf{F}_\alpha \cdot \frac{\partial \mathbf{r}_\alpha}{\partial q^i}$. В соответствии с ковариантной структурой уравнений Лагранжа, обобщенные силы формируют матрицу-строку размерности $1 \times n$:
$$Q = \begin{pmatrix} Q_1 & \dots & Q_n \end{pmatrix}$$
Матричное уравнение движения неконсервативной системы в строковой записи принимает вид:
$$\frac{d}{dt}\left( \frac{\partial L}{\partial \dot{q}} \right) - \frac{\partial L}{\partial q} = Q$$


## 3. Гамильтонов формализм и переход к фазовому пространству

### 3.1. Кокасательное расслоение и преобразование Лежандра
В лагранжевом описании динамическое состояние механической системы характеризовалось упорядоченной парой векторов $(q, \dot{q})$, принадлежащих касательному расслоению $TN^n$. Гамильтонов подход базируется на переходе от обобщенных скоростей к сопряженным обобщенным импульсам, которые обладают принципиально иной геометрической природой.

С геометрической точки зрения обобщенные импульсы являются линейными функциями на касательном пространстве, то есть ковекторами. В соответствии с протоколом подраздела 1.1, ковекторы в фиксированном базисе представляются в виде матриц-строк. Вычисление вектора обобщенных импульсов реализуется через дифференцирование скалярной функции Лагранжа по вектору-столбцу скоростей $\dot{q}$. Эта операция согласно правилу 1.2.1 автоматически формирует матрицу-строку размерности $1 \times n$:
$$p = \frac{\partial L}{\partial \dot{q}} = \begin{pmatrix} \frac{\partial L}{\partial \dot{q}^1} & \dots & \frac{\partial L}{\partial \dot{q}^n} \end{pmatrix}$$

Совокупность всех импульсов (строк) в фиксированной точке $q$ образует линейное кокасательное пространство $T^*_q N^n$. Полное пространство состояний гамильтоновой механики, называемое фазовым пространством системы, представляет собой $2n$-мерное кокасательное расслоение:
$$T^*N^n = \bigcup_{q \in N^n} T^*_q N^n$$
Базой данного расслоения выступает конфигурационное многообразие $N^n$, а слоями в каждой точке служат кокасательные пространства $T^*_q N^n$.

Переход от координат касательного расслоения $(q, \dot{q})$ к координатам кокасательного расслоения $(q, p)$ осуществляется посредством преобразования Лежандра. Для математической возможности такого перехода предполагается, что функция Лагранжа является строго выпуклой по обобщенным скоростям, в силу чего функциональная матрица вторых производных (гессиан лагранжиана по скоростям) является невырожденной:
$$\det \left( \frac{\partial^2 L}{\partial \dot{q}^i \partial \dot{q}^j} \right) \neq 0$$

В силу правил подраздела 1.2.2 функциональная матрица вторых производных $\mathbf{M}_{hess}$ является матрицей Якоби векторного отображения строк $p(\dot{q})$. Так как $p$ является строкой, её дифференцирование по вектору-столбцу скоростей $\dot{q}$ порождает квадратную матрицу размерности $n \times n$:
$$\mathbf{M}_{hess} = \frac{\partial p}{\partial \dot{q}} = \begin{pmatrix} 
\frac{\partial p_1}{\partial \dot{q}^1} & \dots & \frac{\partial p_1}{\partial \dot{q}^n} \\ 
\vdots & \ddots & \vdots \\ 
\frac{\partial p_n}{\partial \dot{q}^1} & \dots & \frac{\partial p_n}{\partial \dot{q}^n} 
\end{pmatrix} = \begin{pmatrix} 
\frac{\partial^2 L}{\partial \dot{q}^1 \partial \dot{q}^1} & \dots & \frac{\partial^2 L}{\partial \dot{q}^1 \partial \dot{q}^n} \\ 
\vdots & \ddots & \vdots \\ 
\frac{\partial^2 L}{\partial \dot{q}^n \partial \dot{q}^1} & \dots & \frac{\partial^2 L}{\partial \dot{q}^n \partial \dot{q}^n} 
\end{pmatrix}$$

Выполнение условия невырожденности матрицы $\mathbf{M}_{hess}$ позволяет локально разрешить систему уравнений для импульсов относительно обобщенных скоростей и выразить их в явном виде как функции координат, импульсов и времени: $\dot{q} = \dot{q}(q, p, t)$.


### 3.2. Функция Гамильтона и ее аналитические свойства

Пусть условия преобразования Лежандра выполнены, и обобщенные скорости выражены в виде явных функций фазовых переменных: $\dot{q} = \dot{q}(q, p, t)$. Функция Гамильтона (гамильтониан) $H(q, p, t)$ определяется как двойственное преобразование лагранжиана $L$.

С учетом принятых соглашений, где вектор обобщенных импульсов $p$ является матрицей-строкой размерности $1 \times n$, а вектор обобщенных скоростей $\dot{q}$ является вектором-столбцом размерности $n \times 1$, инвариантная матричная запись функции Гамильтона принимает вид:
$$H(q, p, t) = p \dot{q} - L(q, \dot{q}(q, p, t), t)$$

Выражение $p \dot{q}$ представляет собой стандартное матричное произведение строки на столбец, результатом которого является скалярная величина.

Дифференциальные свойства гамильтониана устанавливаются путем вычисления его полного дифференциала на кокасательном расслоении. Применяя к выражению функции Гамильтона правило Лейбница для дифференциала произведения, получаем скалярное выражение:
$$dH = (dp) \dot{q} + p (d\dot{q}) - \frac{\partial L}{\partial q} dq - \frac{\partial L}{\partial \dot{q}} d\dot{q} - \frac{\partial L}{\partial t} dt$$

По определению обобщенного импульса из подраздела 3.1 справедливо тождество $p = \frac{\partial L}{\partial \dot{q}}$. В силу данного тождества члены $p (d\dot{q})$ и $\frac{\partial L}{\partial \dot{q}} d\dot{q}$ взаимно уничтожаются. Подстановка уравнений Эйлера-Лагранжа в строковой форме $\frac{\partial L}{\partial q} = \frac{d}{dt}p$ приводит выражение для дифференциала гамильтониана к виду:
$$dH = (dp) \dot{q} - \left(\frac{d}{dt}p\right) dq - \frac{\partial L}{\partial t} dt$$

С другой стороны, полный дифференциал функции Гамильтона как скалярной функции независимых переменных расписывается через частные производные. Чтобы каждый член выражения строго оставался скаляром (произведением строки на столбец), порядок умножения обязан учитывать матричную природу аргументов согласно подразделу 1.2.1:
$$dH = \frac{\partial H}{\partial q} dq + dp \left(\frac{\partial H}{\partial p}\right) + \frac{\partial H}{\partial t} dt$$

Поскольку координаты $q$ определены как столбец, приращение $dq$ является столбцом, а производная $\frac{\partial H}{\partial q}$ является строкой, что дает скалярное произведение $\frac{\partial H}{\partial q} dq$. Поскольку импульсы $p$ определены как строка, приращение $dp$ является строкой, а производная $\frac{\partial H}{\partial p}$ согласно правилу 1.2.1 является столбцом, что требует их умножения в порядке $dp \left(\frac{\partial H}{\partial p}\right)$ для получения скаляра.

В силу коммутативности скалярного произведения матриц-строк и матриц-столбцов, первый член в предыдущем уравнении для $dH$ переписывается в аналогичном порядке: $(dp) \dot{q} = dp (\dot{q})$. 

Сопоставление коэффициентов при одинаковых дифференциалах независимых приращений $dq$ и $dp$ в обоих представлениях для $dH$ порождает систему тождеств:
$$\frac{\partial H}{\partial q} = -\frac{d}{dt}p, \quad \frac{\partial H}{\partial p} = \dot{q}, \quad \frac{\partial H}{\partial t} = -\frac{\partial L}{\partial t}$$


### 3.3. Векторно-матричная форма канонических уравнений Гамильтона

Полученные в подразделе 3.2 тождества позволяют выразить производные канонических переменных по времени через частные производные функции Гамильтона. Перепишем систему уравнений в виде:

$$\frac{d}{dt}q = \frac{\partial H}{\partial p}, \quad \frac{d}{dt}p = -\frac{\partial H}{\partial q}$$

Согласно зафиксированному протоколу, обобщенные координаты $q$ и их производные по времени являются векторами-столбцами размерности $n \times 1$. Обобщенные импульсы $p$ и их производные по времени являются матрицами-строками размерности $1 \times n$.

Для построения единого фазового пространства объединим канонические переменные в составной вектор фазовых координат $x$. Чтобы поставить блок импульсов на первое место и сохранить строгую вертикальную структуру векторов состояний, используемую в алгоритмах численного интегрирования и компьютерной алгебры, вектор $x$ формируется как вектор-столбец размерности $2n \times 1$, в котором строка импульсов предварительно подвергается операции транспонирования:

$$x = \begin{pmatrix} p^T \\ q \end{pmatrix} = (p_1, \dots, p_n, q^1, \dots, q^n)^T$$

Производная фазового вектора по времени имеет аналогичную вертикальную структуру:

$$\frac{dx}{dt} = \begin{pmatrix} \frac{d}{dt}p^T \\ \frac{d}{dt}q \end{pmatrix}$$

Вычислим полную производную функции Гамильтона по фазовому вектору $x$. Поскольку гамильтониан является скалярной функцией, заданной на векторе-столбце $x$, то согласно правилу подраздела 1.2.1 его производная по $x$ образует матрицу-строку частных производных размерности $1 \times 2n$:

$$\frac{\partial H}{\partial x} = \begin{pmatrix} \left(\frac{\partial H}{\partial p}\right)^T & \frac{\partial H}{\partial q} \end{pmatrix}$$

Транспонирование строки производных согласно подразделу 1.2.4 порождает вектор-столбец градиента $\operatorname{grad} H$ размерности $2n \times 1$. С учетом правил дифференцирования скаляра по разноструктурным аргументам из подраздела 1.2.1, его блочный вид определяется выражением:

$$\operatorname{grad} H = \left(\frac{\partial H}{\partial x}\right)^T = \begin{pmatrix} \frac{\partial H}{\partial p} \\ \left(\frac{\partial H}{\partial q}\right)^T \end{pmatrix}$$

Для объединения уравнений движения в единую систему введем каноническую кососимметрическую симплектическую матрицу $\mathbf{J}$ размерности $2n \times 2n$, блочный вид которой обеспечивает инвариантность фазового потока при выбранном порядке переменных (сначала импульсы, затем координаты):

$$\mathbf{J} = \begin{pmatrix} \mathbf{0} & \mathbf{I}_n \\ -\mathbf{I}_n & \mathbf{0} \end{pmatrix}$$

где $\mathbf{I}_n$ является единичной матрицей порядка $n$, а $\mathbf{0}$ - нулевой матрицей размерности $n \times n$. 

Перемножим матрицу $\mathbf{J}^T$ и вектор-столбец градиента $\operatorname{grad} H$:

$$\mathbf{J}^T \operatorname{grad} H = \begin{pmatrix} \mathbf{0} & -\mathbf{I}_n \\ \mathbf{I}_n & \mathbf{0} \end{pmatrix} \begin{pmatrix} \frac{\partial H}{\partial p} \\ \left(\frac{\partial H}{\partial q}\right)^T \end{pmatrix} = \begin{pmatrix} -\left(\frac{\partial H}{\partial q}\right)^T \\ \frac{\partial H}{\partial p} \end{pmatrix}$$

Сопоставление полученного выражения с компонентами вектора фазовой скорости показывает их полное тождество, так как из канонических уравнений движения следует равенство $\frac{dx}{dt} = \begin{pmatrix} \frac{d}{dt}p^T \\ \frac{d}{dt}q \end{pmatrix} = \begin{pmatrix} -\left(\frac{\partial H}{\partial q}\right)^T \\ \frac{\partial H}{\partial p} \end{pmatrix}$. В результате инвариантная векторно-матричная запись канонических уравнений Гамильтона принимает вид:

$$\frac{dx}{dt} = \mathbf{J}^T \operatorname{grad} H = \mathbf{J}^T \left(\frac{\partial H}{\partial x}\right)^T$$
