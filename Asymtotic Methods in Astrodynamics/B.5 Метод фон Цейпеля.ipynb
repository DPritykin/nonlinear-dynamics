# Метод фон Цейпеля

Метод фон Цейпеля - асимптотический метод теории возмущений, предназначенный для интегрирования канонических систем уравнений Гамильтона [von Zeipel, 1916]. Его цель заключается в последовательном исключении циклических (быстрых) угловых переменных из функции Гамильтона с помощью канонических преобразований. Данные преобразования задаются производящей функцией смешанного типа, зависящей от старых координат и новых импульсов.

Базисом для построения асимптотических рядов фон Цейпеля служит фазовое пространство в канонических переменных «действие — угол» $(I, \theta)$, в котором невозмущенный гамильтониан зависит исключительно от импульсов действия. Ниже рассмотрены математическая структура и алгоритм метода на примере одномерного ангармонического осциллятора.

---

## 1. Постановка задачи

В качестве исходной модели исследуется гамильтониан одномерного линейного осциллятора с малой кубической ангармонической поправкой к потенциальной энергии [von Zeipel, 1916]:

$$H(q, p) = \frac{p^2 + q^2}{2} + \varepsilon q^3$$

где $\varepsilon$ — малый параметр возмущения ($\varepsilon \ll 1$).

Переход к каноническим переменным «действие-угол» $(I, \theta)$ для невозмущенной части системы осуществляется с помощью канонического преобразования:

$$q = \sqrt{2I^1} \cos \theta_1, \quad p = -\sqrt{2I^1} \sin \theta_1$$

После подстановки данных выражений в исходный гамильтониан система принимает вид:

$$H(I, \theta) = I^1 + \varepsilon (2I^1)^{3/2} \cos^3 \theta_1$$

На основании тригонометрического тождества $\cos^3 \theta_1 = \frac{3}{4}\cos \theta_1 + \frac{1}{4}\cos(3\theta_1)$ возмущенная часть гамильтониана представляется в виде суммы спектральных гармоник:

$$H(I, \theta) = I^1 + \varepsilon (2I^1)^{3/2} \left( \frac{3}{4}\cos \theta_1 + \frac{1}{4}\cos(3\theta_1) \right)$$

Переменная $\theta_1$ является быстрой фазой, совершающей полный оборот за период колебания системы. Целью алгоритма является построение канонического преобразования к новым переменным $(\bar{I}, \bar{\theta})$, в которых преобразованный гамильтониан не зависит от фазы $\bar{\theta}$. Данный подход позволяет определить новое действие $\bar{I}^1$, служащее интегралом движения усредненной системы в первом приближения по малому параметру $\varepsilon$.

### Физический смысл переменных «действие-угол»

Колебательные движения описываются через каноническую координату $q$ и канонический импульс $p$. На фазовой плоскости траектория невозмущенного гармонического осциллятора представляет собой окружность, задаваемую законом сохранения энергии:

$$\frac{p_1^2 + (q^1)^2}{2} = E$$

Переход к переменным «действие-угол» - каноническое преобразование, которое нормализует геометрию фазового пространства, связывая декартовы координаты с полярными на фазовой плоскости через радиус $R = \sqrt{2I^1}$:

*   **Переменная угла** $\theta_1$ определяет фазу колебаний и текущее угловое положение системы на фазовой траектории в диапазоне от $0$ до $2\pi$;
*   **Переменная действия** $I^1$ представляет собой канонический импульс, определяющий величину площади фазового портрета, ограниченного замкнутй траекторией [Уиттекер, 1937]:
    $$I^1 = \frac{1}{2\pi} \oint p \, dq = \frac{E}{\omega_0}$$

Для рассматриваемой системы собственная частота линейных колебаний $\omega_0 = 1$, откуда $I^1 = E$.

В невозмущенной автономной системе уравнения Гамильтона принимают интегрируемый линейный вид [Уиттекер, 1937]:

$$\frac{dI^1}{dt} = -\frac{\partial H_0}{\partial \theta_1} = 0 \implies I^1 = \text{const}$$
$$\frac{d\theta_1}{dt} = \frac{\partial H_0}{\partial I^1} = 1 \implies \theta_1 = t + \theta_{1,0}$$

Асимптотический алгоритм предназначен для построения в возмущенной нелинейной системе нового действия $\bar{I}^1$, которое является строгим интегралом движения ($\frac{d\bar{I}^1}{dt} = 0$), в то время как новый угол $\bar{\theta}_1$ изменяется равномерно с модифицированной нелинейной частотой: $\frac{d\bar{\theta}_1}{dt} = \bar{\omega}(\bar{I}^1)$.

## 2. Структура канонического преобразования фон Цейпеля

Для перехода от старых переменных к новым усредненным переменным строится каноническое преобразование второго типа. Данное преобразование задается производящей функцией смешанного типа $S(\bar{I}, \theta, \varepsilon)$, зависящей от вектора-строки новых импульсов действия $\bar{I}$ и вектора-столбца старых угловых координат $\theta$:

$$S(\bar{I}, \theta, \varepsilon) = \bar{I}\theta + \varepsilon S_1(\bar{I}, \theta) + \varepsilon^2 S_2(\bar{I}, \theta) + O(\varepsilon^3)$$

где $\bar{I}\theta = \sum_{i=1}^n \bar{I}_i \theta^i$ - скалярное произведение, порождающее тождественный переход, а функции $S_k(\bar{I}, \theta)$ являются $2\pi$-периодическими по угловой переменной и описывают короткопериодические возмущения системы.

В соответствии с правилами матричного дифференцирования, связь между исходными и новыми фазовыми переменными определяется дифференциальными соотношениями [Голдстейн, Пул, Сафко, 2012]:

$$I = \frac{\partial S}{\partial \theta} = \bar{I} + \varepsilon \frac{\partial S_1}{\partial \theta} + \varepsilon^2 \frac{\partial S_2}{\partial \theta} + O(\varepsilon^3)$$
$$\bar{\theta} = \frac{\partial S}{\partial \bar{I}} = \theta + \varepsilon \frac{\partial S_1}{\partial \bar{I}} + \varepsilon^2 \frac{\partial S_2}{\partial \bar{I}} + O(\varepsilon^3)$$

Операция $\frac{\partial S}{\partial \theta}$ (дифференцирование по столбцу) автоматически формирует матрицу-строку старых импульсов $I$, а операция $\frac{\partial S}{\partial \bar{I}}$ (дифференцирование по строке) формирует вектор-столбец новых координат $\bar{\theta}$ [Новиков, Тайманов, 2014].

Новый преобразованный гамильтониан $\bar{H}(\bar{I}, \varepsilon)$ по условию метода фон Цейпеля должен быть инвариантен относительно быстрой фазы $\theta$ и зависеть исключительно от новых импульсов действия. Разложение функции Гамильтона в асимптотический ряд по степеням малого параметра имеет вид:

$$\bar{H}(\bar{I}, \varepsilon) = \bar{H}_0(\bar{I}) + \varepsilon \bar{H}_1(\bar{I}) + \varepsilon^2 \bar{H}_2(\bar{I}) + O(\varepsilon^3)$$

Связь между старым и преобразованным гамильтонианами определяется тождеством Гамильтона-Якоби для автономных систем [Арнольд, 2024]:

$$H\left(\frac{\partial S}{\partial \theta}, \theta\right) = \bar{H}\left(\bar{I}, \varepsilon\right)$$


### Вспомним, что такое производящая функция

Преобразование переменных $(q, p) \to (Q, P)$ называется каноническим, если оно сохраняет инвариантность формы записи канонических уравнений Гамильтона для любой функции Гамильтона [Маркеев, 2024]. Связь между старыми и новыми переменными задается через частные производные скалярной производящей функции второго типа $S(P, q)$, аргументами которой служат новые импульсы $P$ и старые координаты $q$ [Голдстейн, Пул, Сафко, 2012]:

$$p = \frac{\partial S}{\partial q}, \quad Q = \frac{\partial S}{\partial P}$$

#### Пример 1: Тождественное каноническое преобразование
Рассмотрим производящую функцию вида:
$$S(P, q) = P q = \sum_{i=1}^n P_i q^i$$

Применение уравнений связи дает:
*   Старый импульс: $p = \frac{\partial (P q)}{\partial q} = P$
*   Новая координата: $Q = \frac{\partial (P q)}{\partial P} = q$

Функция $S = P q$ задает тождественный переход, при котором фазовые переменные не изменяются. При переходе к переменным «действие-угол» роль старых координат $q$ занимает вектор $\theta$, а роль импульсов $p$ и $P$ - элементы $I$ и $\bar{I}$ соответственно.

#### Пример 2: Канонический сдвиг координат
Рассмотрим случай, когда центр колебаний системы смещен на постоянную величину, определяемую малым параметром $\varepsilon$. Зададим производящую функцию, добавив к тождественному генератору линейное возмущение:

$$S(P, q) = P q + \varepsilon P c$$

где $c$ — постоянный вектор сдвига. Уравнения связи принимают вид:
*   Старый импульс: $p = \frac{\partial S}{\partial q} = P$ (импульс сохраняет свое значение);
*   Новая координата: $Q = \frac{\partial S}{\partial P} = q + \varepsilon c$

Из последнего выражения следует алгебраическое соотношение переноса начала координат: $q = Q - \varepsilon c$.

В методе фон Цейпеля реализуется аналогичный подход. Производящая функция представляется в виде асимптотического ряда по степеням малого параметра: $S = \bar{I}\theta + \varepsilon S_1 + \varepsilon^2 S_2 + \dots$ Периодические поправки $S_k(\bar{I}, \theta)$ подбираются таким образом, чтобы их частные производные порождали компенсирующие возмущения, которые нивелируют влияние нелинейного поля и приводят трансформированные уравнения движения к интегрируемому виду [von Zeipel, 1916].



## 3. Нахождение уравнений первого приближения

После подстановки дифференциальных соотношений связи в исходную функцию Гамильтона выполняется разложение невозмущенной части $H_0(I)$ в ряд Тейлора в окрестности нового импульса действия $\bar{I}$. Ограничиваясь членами первого порядка малости относительно $\varepsilon$, полученное выражение приравнивается к преобразованному гамильтониану [von Zeipel, 1916]:

$$\bar{H} = \bar{H}_0 + \varepsilon \bar{H}_1 + O(\varepsilon^2)$$

Подстановка выражения для старого действия $I$ в невозмущенную часть $H_0(I) = I$ порождает линейный по малому параметру член. В возмущенной части аргумент в первом приближении заменяется на $\bar{I}$, поскольку сама поправка первого порядка $H_1$ уже содержит множитель $\varepsilon$:

$$\left(\bar{I} + \varepsilon \frac{\partial S_1}{\partial \theta}\right) + \varepsilon (2\bar{I})^{3/2} \left( \frac{3}{4}\cos \theta + \frac{1}{4}\cos(3\theta) \right) + O(\varepsilon^2) = \bar{H}_0(\bar{I}) + \varepsilon \bar{H}_1(\bar{I})$$

Расщепление полученного уравнения по порядкам малости параметра $\varepsilon$ определяет систему зацепляющихся уравнений теории возмущений [Пуанкаре, 1971]:

*   **Порядок $O(1)$** определяет гамильтониан невозмущенной системы в новых переменных:
    $$\bar{H}_0(\bar{I}) = \bar{I}$$

*   **Порядок $O(\varepsilon)$** формирует гомологическое уравнение первого порядка, связывающее неизвестную поправку к гамильтониану $\bar{H}_1$ и частную производную функции $S_1$ по быстрой угловой переменной $\theta$:
    $$\frac{\partial S_1}{\partial \theta} + (2\bar{I})^{3/2} \left( \frac{3}{4}\cos \theta + \frac{1}{4}\cos(3\theta) \right) = \bar{H}_1(\bar{I})$$

## 4. Разделение на среднюю и осциллирующую компоненты

Разделение гомологического уравнения первого порядка на вековую и короткопериодическую составляющие осуществляется с помощью линейного оператора усреднения по быстрой фазе на периоде $[0, 2\pi]$ [von Zeipel, 1916]:

$$\langle \dots \rangle_{\theta} = \frac{1}{2\pi} \int_{0}^{2\pi} \dots d\theta_1$$

*   **Усредненная часть** определяет преобразованный гамильтониан первого порядка $\bar{H}_1$ и содержит члены, инвариантные относительно быстрой фазы $\theta_1$.
*   **Осциллирующая часть** определяет периодическую поправку к производящей функции $S_1$.

Поскольку среднее значение тригонометрических функций на периоде равно нулю ($\langle \cos \theta_1 \rangle_{\theta} = 0$, $\langle \cos(3\theta_1) \rangle_{\theta} = 0$), в левой части гомологического уравнения постоянные члены отсутствуют. Отсюда поправка первого порядка к усредненному гамильтониану равна нулю [Пуанкаре, 1971]:

$$\bar{H}_1(\bar{I}^1) = 0$$

Для оставшейся осциллирующей части формируется дифференциальное уравнение относительно функции $S_1$:

$$\frac{\partial S_1}{\partial \theta_1} = -(2\bar{I}^1)^{3/2} \left( \frac{3}{4}\cos \theta_1 + \frac{1}{4}\cos(3\theta_1) \right)$$

Интегрирование данного уравнения по быстрой фазе при условии нулевого среднего значения $\langle S_1 \rangle_{\theta} = 0$ определяет явный вид поправки к производящей функции первого порядка [von Zeipel, 1916]:

$$S_1(\bar{I}^1, \theta_1) = -(2\bar{I}^1)^{3/2} \left( \frac{3}{4}\sin \theta_1 + \frac{1}{12}\sin(3\theta_1) \right)$$

### Аналитический результат первого приближения

1.  Преобразованный гамильтониан системы с точностью до членов второго порядка малости совпадает с невозмущенным:
    $$\bar{H}(\bar{I}^1) = \bar{I}^1 + O(\varepsilon^2)$$
    Кубическая ангармоническая нелинейность потенциала в первом порядке теории возмущений не вызывает сдвига частоты колебаний осциллятора ($\bar{\omega} = \frac{\partial \bar{H}}{\partial \bar{I}^1} = 1$). Нелинейное изменение частоты проявляется в высших приближениях, начиная со второго порядка малости.

2.  Связь между исходным и новым действием определяется дифференциальным соотношением:
    $$I^1 = \bar{I}^1 + \varepsilon \frac{\partial S_1}{\partial \theta_1} = \bar{I}^1 - \varepsilon (2\bar{I}^1)^{3/2} \left( \frac{3}{4}\cos \theta_1 + \frac{1}{4}\cos(3\theta_1) \right) + O(\varepsilon^2)$$

Поскольку преобразованный гамильтониан не зависит от быстрой фазы, новое действие является интегралом движения ($\frac{d\bar{I}^1}{dt} = 0$). Исходное действие $I^1(t)$ совершает периодические колебания с амплитудой порядка $\varepsilon$ вокруг среднего значения $\bar{I}^1$.

## 5. Вычислительный эксперимент: верификация первого приближения фон Цейпеля

Для количественной проверки результатов первого приближения теории возмущений ниже реализован алгоритм численного моделирования нелинейных колебаний ангармонического осциллятора.

### 5.1. Цели эксперимента
Численный эксперимент спроектирован с целью проверки следующих аналитических следствий метода фон Цейпеля:
1. **Инвариантность преобразованного действия**. Проверяется положение о том, что после исключения быстрой фазы $\theta_1$ новое действие $\bar{I}^1$ становится строгим интегралом движения усредненной системы, сохраняя постоянное значение на всем интервале времени.
2. **Амплитуда короткопериодических пульсаций**. Проверяется точность выведенного уравнения связи первого порядка. Численный расчет оскулирующего действия $I^1(t)$ сопоставляется с аналитическим треком, вычисленным через частную производную функции поправок $\frac{\partial S_1}{\partial \theta_1}$, для подтверждения того, что истинная траектория совершает гладкие периодические колебания амплитуды $O(\varepsilon)$ вокруг инварианта $\bar{I}^1$.
3. **Отсутствие сдвига частоты в первом порядке**. Моделирование призвано показать, что в первом приближении по малому параметру кубическая нелинейность не вносит вековых фазовых отклонений, и средняя частота системы остается равной единице.

Интегрирование точной системы канонических уравнений в декартовых координатах $(q, p)$, полученных из гамильтониана $H = \frac{p^2+q^2}{2} + \varepsilon q^3$, выполняется высокоточным методом Рунге-Кутты 8-го порядка (интегратор `DOP853`). Начальные условия $(q_0, p_0)$ рассчитываются в точке $\theta_{1,0} = 0$ с учетом канонической поправки первого порядка. Мгновенные значения полярных переменных восстанавливаются на каждом шаге по формулам: $I^1(t) = \frac{p_1^2 + (q^1)^2}{2}$ и $\theta_1(t) = \text{atan2}(-p_1, q^1)$.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# =====================================================================
# 1. ПАРАМЕТРЫ СИСТЕМЫ И ИНИЦИАЛИЗАЦИЯ
# =====================================================================
eps = 0.05    # Малый параметр возмущения (кубическая ангармоничность)
bar_I1 = 1.0  # Новое инвариантное действие первого приближения

# =====================================================================
# 2. КАНОНИЧЕСКИЕ УРАВНЕНИЯ ДВИЖЕНИЯ ИСХОДНОЙ СИСТЕМЫ
# =====================================================================
# dq1/dt =  dH/dp1 = p1
# dp1/dt = -dH/dq1 = -q1 - 3 * eps * (q1)^2
def hamiltonian_system(t, state):
    q1, p1 = state
    dq1dt = p1
    dp1dt = -q1 - 3 * eps * q1**2
    return [dq1dt, dp1dt]

# =====================================================================
# 3. РАСЧЕТ НАЧАЛЬНЫХ УСЛОВИЙ (q0, p0) ПО МЕТОДУ ФОН ЦЕЙПЕЛЯ
# =====================================================================
# Исходное действие I1(0) выражается через новое инвариантное действие bar_I1 при t=0 (theta1=0)
theta1_0 = 0.0
I1_0 = bar_I1 - eps * (2 * bar_I1)**1.5 * (0.75 * np.cos(theta1_0) + 0.25 * np.cos(3 * theta1_0))

# Переход от канонических переменных «действие — угол» к декартовым фазовым координатам
q1_0 = np.sqrt(2 * I1_0) * np.cos(theta1_0)
p1_0 = -np.sqrt(2 * I1_0) * np.sin(theta1_0)

# =====================================================================
# 4. ЧИСЛЕННОЕ ИНТЕГРИРОВАНИЕ (DOP853)
# =====================================================================
t_span = (0.0, 30.0)
t_eval = np.linspace(0.0, 30.0, 3000)

# Интегрирование гамильтоновой системы методом DOP853 с жесткими допусками
sol = solve_ivp(hamiltonian_system, t_span, [q1_0, p1_0], t_eval=t_eval, 
                method='DOP853', rtol=1e-13, atol=1e-16)
q1_vals = sol.y[0]
p1_vals = sol.y[1]

# =====================================================================
# 5. ВЫЧИСЛЕНИЕ МГНОВЕННЫХ ЗНАЧЕНИЙ В ПЕРЕМЕННЫХ «ДЕЙСТВИЕ — УГОЛ»
# =====================================================================
# Мгновенное (оскулирующее) исходное действие I1(t)
I1_numerical = 0.5 * (p1_vals**2 + q1_vals**2)
# Истинная фаза колебаний theta1(t)
theta1_numerical = np.arctan2(-p1_vals, q1_vals)

# Аналитическое представление исходного действия первого порядка (уравнение связи (5))
I1_analytical = bar_I1 - eps * (2 * bar_I1)**1.5 * (0.75 * np.cos(theta1_numerical) + 0.25 * np.cos(3 * theta1_numerical))

# =====================================================================
# 6. ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ МЕТОДА ФОН ЦЕЙПЕЛЯ
# =====================================================================
plt.figure(figsize=(11, 5))

# Истинное осциллирующее действие из высокоточного численного расчета
plt.plot(sol.t, I1_numerical, color="royalblue", linewidth=2,
         label=r"Исходное действие $I^1(t)$ (численный расчет DOP853)")

# Аналитическое приближение (поправка первого порядка метода фон Цейпеля)
plt.plot(sol.t, I1_analytical, color="black", linestyle=":", linewidth=2,
         label=r"Аналитическое приближение $I^1(\theta_1) \approx \bar{I}^1 - \dots$")

# Интеграл движения усредненной системы
plt.axhline(bar_I1, color="crimson", linewidth=2.5,
            label=r"Преобразованное действие $\bar{I}^1$ (инвариант первого порядка)")

# Оформление графика согласно академическим стандартам
plt.title(r"Исключение быстрой фазы по методу фон Цейпеля (кубический ангармонизм, $\varepsilon = 0.05$)", fontsize=12)
plt.xlabel("Время $t$, с", fontsize=11)
plt.ylabel("Значение действия $I^1$", fontsize=11)
plt.ylim(bar_I1 - 0.3, bar_I1 + 0.3)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right", fontsize=10)
plt.tight_layout()
plt.show()
```

### 5.2. Анализ результатов численного моделирования первого приближения

Вычислительный эксперимент подтверждает теоретические выводы первого приближения метода фон Цейпеля, наглядно иллюстрируя разделение динамики канонической системы на инвариантную и осциллирующую составляющие:

1.  **Поведение преобразованной переменной**. Преобразованное действие $\bar{I}^1$ (красная горизонтальная линия) строго сохраняет постоянное значение во времени ($\frac{d\bar{I}^1}{dt} = 0$). Это экспериментально подтверждает инвариантность усредненного гамильтониана первого порядка относительно быстрой угловой переменной $\theta_1$. Данная величина выступает в роли точного интеграла движения усредненной системы на исследуемом интервале времени.
2.  **Периодические пульсации оскулирующего элемента**. Мгновенное исходное действие $I^1(t)$ в нелинейной системе (синяя сплошная линия) не сохраняется во времени. Вследствие ангармонического кубического возмущения потенциальной энергии величина $I^1(t)$ совершает короткопериодические колебания с амплитудой порядка $\varepsilon = 0.05$ вокруг среднего уровня.
3.  **Точность аналитического уравнения связи**. Аналитический трек исходного действия (черная пунктирная линия), рассчитанный по уравнению связи через частную производную функции поправок $\frac{\partial S_1}{\partial \theta_1}$, совпадает с траекторией высокоточного численного интегрирования. Периодическая поправка первого порядка полностью описывает высокочастотные колебания системы, обеспечивая асимптотическую погрешность порядка $O(\varepsilon^2)$ на ограниченных интервалах времени.

## 6. Второе приближение метода фон Цейпеля: ангармонический сдвиг частоты

Для учета нелинейных эффектов второго порядка малости асимптотические разложения производящей функции $S(\bar{I}, \theta, \varepsilon)$ и преобразованного гамильтониана $\bar{H}(\bar{I}, \varepsilon)$ расширяются до членов с точностью до $O(\varepsilon^2)$ включительно [von Zeipel, 1916]:

$$S(\bar{I}, \theta) = \bar{I}\theta + \varepsilon S_1(\bar{I}, \theta) + \varepsilon^2 S_2(\bar{I}, \theta) + O(\varepsilon^3)$$
$$\bar{H}(\bar{I}) = \bar{H}_0(\bar{I}) + \varepsilon \bar{H}_1(\bar{I}) + \varepsilon^2 \bar{H}_2(\bar{I}) + O(\varepsilon^3)$$

Каноническая связь между исходным и преобразованным действиями принимает вид:

$$I^1 = \bar{I}^1 + \varepsilon \frac{\partial S_1}{\partial \theta_1} + \varepsilon^2 \frac{\partial S_2}{\partial \theta_1} + O(\varepsilon^3)$$

После подстановки данного разложения в исходную функцию Гамильтона выполняется разложение составляющих в ряд Тейлора в окрестности нового действия $\bar{I}^1$. Группировка членов при второй степени малого параметра $\varepsilon^2$ приводит к гомологическому уравнению второго порядка [Пуанкаре, 1971]:

$$\frac{\partial S_2}{\partial \theta_1} \frac{\partial H_0}{\partial \bar{I}^1} + \frac{1}{2}\frac{\partial^2 H_0}{\partial (\bar{I}^1)^2}\left(\frac{\partial S_1}{\partial \theta_1}\right)^2 + \frac{\partial H_1(\bar{I}, \theta)}{\partial \bar{I}^1} \frac{\partial S_1}{\partial \theta_1} = \bar{H}_2(\bar{I}^1)$$

С учетом специфики невозмущенной части исследуемого ангармонического осциллятора ($H_0(I) = I^1$) частные производные принимают значения:

$$\frac{\partial H_0}{\partial \bar{I}^1} = 1, \quad \frac{\partial^2 H_0}{\partial (\bar{I}^1)^2} = 0$$

В силу данных соотношений гомологическое уравнение второго порядка упрощается и принимает вид:

$$\frac{\partial S_2}{\partial \theta_1} + \frac{\partial H_1(\bar{I}, \theta)}{\partial \bar{I}^1} \frac{\partial S_1}{\partial \theta_1} = \bar{H}_2(\bar{I}^1)$$

### 6.1. Вычисление вековой части гамильтониана второго порядка
На основании выражений первого приближения (Раздел 4) частные производные функций $H_1$ и $S_1$ равны:

$$\frac{\partial H_1}{\partial \bar{I}^1} = 3(2\bar{I}^1)^{1/2} \left( \frac{3}{4}\cos \theta_1 + \frac{1}{4}\cos(3\theta_1) \right)$$
$$\frac{\partial S_1}{\partial \theta_1} = -(2\bar{I}^1)^{3/2} \left( \frac{3}{4}\cos \theta_1 + \frac{1}{4}\cos(3\theta_1) \right)$$

После перемножения данных выражений определяется перекрестный нелинейный вклад во втором порядке малости:

$$\frac{\partial H_1}{\partial \bar{I}^1} \frac{\partial S_1}{\partial \theta_1} = -3(2\bar{I}^1)^2 \left( \frac{3}{4}\cos \theta_1 + \frac{1}{4}\cos(3\theta_1) \right)^2$$

Усреднение полученного выражения по быстрой фазе $\theta_1$ выполняется на интервале $[0, 2\pi]$. Учитывая свойства ортогональности тригонометрических функций и значения средних для их квадратов ($\langle \cos^2 \theta_1 \rangle_{\theta} = \frac{1}{2}$, $\langle \cos^2(3\theta_1) \rangle_{\theta} = \frac{1}{2}$, $\langle \cos \theta_1 \cos(3\theta_1) \rangle_{\theta} = 0$), среднее значение принимает вид [Пуанкаре, 1971]:

$$\left\langle \frac{\partial H_1}{\partial \bar{I}^1} \frac{\partial S_1}{\partial \theta_1} \right\rangle_{\theta} = -12(\bar{I}^1)^2 \left( \frac{9}{16}\cdot\frac{1}{2} + \frac{1}{16}\cdot\frac{1}{2} \right) = -12(\bar{I}^1)^2 \left( \frac{10}{32} \right) = -\frac{15}{4}(\bar{I}^1)^2$$

Поскольку среднее значение производной по фазе от периодической функции второго порядка равно нулю ($\langle \frac{\partial S_2}{\partial \theta_1} \rangle_{\theta} = 0$), из гомологического уравнения окончательно определяется вековая часть функции Гамильтона второго приближения [von Zeipel, 1916]:

$$\bar{H}_2(\bar{I}^1) = -\frac{15}{4}(\bar{I}^1)^2$$

### 6.2. Преобразованный гамильтониан и ангармонический сдвиг частоты
Преобразованный гамильтониан системы, определенный с погрешностью порядка $O(\varepsilon^3)$, принимает вид:

$$\bar{H}(\bar{I}^1) = \bar{I}^1 - \varepsilon^2 \frac{15}{4}(\bar{I}^1)^2$$

На основании канонических уравнений Гамильтона определяется модифицированная частота нелинейных колебаний осциллятора в преобразованном фазовом пространстве [Пуанкаре, 1971]:

$$\bar{\omega}(\bar{I}^1) = \frac{\partial \bar{H}}{\partial \bar{I}^1} = 1 - \varepsilon^2 \frac{15}{2}\bar{I}^1$$

Аналитическое исследование показывает, что кубическая ангармоническая нелинейность потенциала во втором порядке теории возмущений обуславливает вековое снижение частоты колебаний осциллятора. Величина ангармонического сдвига частоты прямо пропорциональна квадрату параметра возмущения $\varepsilon^2$ и значению новой переменной действия $\bar{I}^1$.


### 6.3. Вычислительный эксперимент: верификация нелинейного сдвига частоты

Для количественной проверки аналитических результатов, полученных во втором приближении метода фон Цейпеля, ниже реализован алгоритм моделирования канонической системы на длительном интервале времени.

#### 6.3.1. Цели эксперимента
Вычислительный эксперимент спроектирован с целью верификации следующих следствий второго порядка теории возмущений:
1. **Накопление фазового расхождения**. Проверяется положение о том, что кубическая ангармоничность во втором порядке малости по $\varepsilon$ приводит к замедлению колебаний. Моделирование призвано проиллюстрировать вековой уход по фазе реальной траектории ангармонического осциллятора относительно эталонного линейного базиса.
2. **Точность аналитического сдвига частоты**. Программа осуществляет численный расчет истинной нелинейной частоты по точкам пересечения нуля и сопоставляет её с теоретическим значением $\bar{\omega} = 1 - \varepsilon^2 \frac{15}{2}\bar{I}^1$, выведенным из векового гамильтониана второго порядка.
3. **Оценка погрешности второго порядка**. На длительном временном интервале $t \sim O(1/\varepsilon^2)$ проверяется стабильность асимптотического решения.

Интегрирование точной системы канонических уравнений выполняется на интервале $t \in [0, 100]$ с помощью интегратора `DOP853` со строгими допусками сходимости. Начальные условия $(q^1_0, p_{10})$ рассчитываются с учетом канонических поправок первого порядка. Визуализация фазового расхождения осуществляется в конечной части временного окна ($t \in [70, 100]$), где эффект отставания по фазе проявляется наиболее наглядно.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# =====================================================================
# 1. ПАРАМЕТРЫ СИСТЕМЫ И ИНИЦИАЛИЗАЦИЯ
# =====================================================================
# Малый параметр принимается равным 0.025 для явной визуализации фазового расхождения
eps = 0.025
bar_I1 = 1.0

# Канонические уравнения движения нелинейного осциллятора
def hamiltonian_system(t, state):
    q1, p1 = state
    return [p1, -q1 - 3 * eps * q1**2]

# Начальные условия при theta1=0 (с учетом канонических поправок первого порядка)
theta1_0 = 0.0
I1_0 = bar_I1 - eps * (2 * bar_I1) ** 1.5 * (0.75 * np.cos(theta1_0) + 0.25 * np.cos(3 * theta1_0))

q1_0 = np.sqrt(2 * I1_0) * np.cos(theta1_0)
p1_0 = -np.sqrt(2 * I1_0) * np.sin(theta1_0)

# Интегрирование на длительном интервале времени для определения фазового расхождения
t_span = (0.0, 100.0)
t_eval = np.linspace(t_span[0], t_span[1], 2000)

sol = solve_ivp(
    hamiltonian_system, 
    t_span, 
    [q1_0, p1_0], 
    method='DOP853', # Высокоточный метод интегрирования гамильтоновых систем
    t_eval=t_eval, 
    rtol=1e-11, 
    atol=1e-13
)
q1_vals = sol.y[0]
p1_vals = sol.y[1]
t_vals = sol.t

# =====================================================================
# 2. ОПРЕДЕЛЕНИЕ ИСТИННОЙ ЧИСЛЕННОЙ ЧАСТОТЫ ПО ПЕРЕСЕЧЕНИЯМ НУЛЯ
# =====================================================================
crossings = []
for i in range(len(p1_vals) - 1):
    if q1_vals[i] < 0 and q1_vals[i + 1] >= 0:
        # Линейная интерполяция корня для достижения субшаговой точности
        t_zero = t_vals[i] - q1_vals[i] * (t_vals[i + 1] - t_vals[i]) / (q1_vals[i + 1] - q1_vals[i])
        crossings.append(t_zero)
        
numerical_period = np.mean(np.diff(crossings))
omega_numerical = (2 * np.pi) / numerical_period

# =====================================================================
# 3. ТЕОРЕТИЧЕСКИЙ СДВИГ ЧАСТОТЫ ВТОРОГО ПОРЯДКА МЕТОДА ФОН ЦЕЙПЕЛЯ
# =====================================================================
omega_theoretical = 1.0 - (eps**2) * (15.0 / 2.0) * bar_I1
omega_unperturbed = 1.0

# =====================================================================
# 4. ГРАФИЧЕСКИЙ АНАЛИЗ ФАЗОВОГО РАСХОЖДЕНИЯ
# =====================================================================
plt.figure(figsize=(11, 5))

# Линейный невозмущенный осциллятор (собственная частота равна единице)
q1_unperturbed = np.sqrt(2 * bar_I1) * np.cos(t_vals)

# Визуализация осуществляется в конечной части временного интервала с наибольшим фазовым расхождением
mask = (t_vals >= 70) & (t_vals <= 100)
plt.plot(
    t_vals[mask],
    q1_unperturbed[mask],
    color="gray",
    linestyle="--",
    alpha=0.6,
    label="Гармонический осциллятор ($\omega_0 = 1$)",
)
plt.plot(
    t_vals[mask],
    q1_vals[mask],
    color="darkorange",
    linewidth=2,
    label=fr"Ангармонический осциллятор ($\varepsilon = {eps}$)",
)

plt.title(fr"Накопление фазового расхождения вследствие нелинейного сдвига частоты $O(\varepsilon^2)$", fontsize=12)
plt.xlabel("Время $t$, с", fontsize=11)
plt.ylabel("Координата $q^1$", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right")

# Вывод результатов сопоставления частот в академическом стиле
print("Результаты сопоставления частот нелинейных колебаний:")
print("-" * 65)
print(f"Невозмущенная частота (нулевой порядок):   {omega_unperturbed:.4f}")
print(f"Теоретическая частота (метод фон Цейпеля): {omega_theoretical:.4f}")
print(f"Истинная численная частота из симуляции:   {omega_numerical:.4f}")
print(f"Абсолютная погрешность асимптотической теории: {abs(omega_theoretical - omega_numerical):.5f}")
print("-" * 65)
plt.tight_layout()
plt.show()
```

#### 6.3.2. Анализ результатов моделирования второго порядка

Визуализация результатов численного моделирования в конечной части временного интервала $t \in [70, 100]$ экспериментально подтверждает теоретические выводы второго приближения метода фон Цейпеля:

1.  **Накопление фазового расхождения**. Под влиянием кубической ангармонической поправки к потенциальной энергии траектория реального осциллятора (оранжевая сплошная линия) систематически отстает по фазе от невозмущенных линейных колебаний (серая пунктирная линия). Данный вековой уход фазы вызван ангармонизмом и наглядно иллюстрирует эффект нелинейного замедления движения с ростом энергии.
2.  **Точность аналитического выражения частоты**. Сопоставление частот показывает совпадение теоретического значения $\bar{\omega} = 0.9953$ с истинной численной частотой симуляции, определенной по точкам пересечения нуля с субшаговой точностью. Абсолютная погрешность асимптотической теории второго порядка на длительном интервале интегрирования составляет $3 \cdot 10^{-5}$, что строго соответствует порядку малости отброшенных членов третьего приближения $O(\varepsilon^3)$.


## 7. Связь метода фон Цейпеля с усреднением Брауэра

В гамильтоновой небесной механике орбитальное движение космических тел описывается системами, близкими к интегрируемым. Задача аналитической теории возмущений в таких системах заключается в разделении фазового потока на быструю составляющую (орбитальное вращение тела за один виток) и медленную вековую составляющую (эволюция формы и пространственной ориентации орбиты под влиянием возмущающих факторов, таких как полярное сжатие центральной планеты) [Брумберг, 1980].

### 7.1. Исторический контекст
*   **Метод фон Цейпеля (1916 г.)**: Х. фон Цейпель адаптировал разработанный Ш. Делоне метод канонических переменных «действие — угол» для исследования вековых и долгопериодических возмущений в динамике малых планет и астероидов [von Zeipel, 1916]. В рамках данного подхода был применен аппарат канонических преобразований смешанного типа, позволяющих исключить среднюю аномалию (быструю угловую координату) из функции Гамильтона.
*   **Усреднение Брауэра (1959 г.)**: Д. Брауэр применил математический аппарат фон Цейпеля к задаче анализа движения искусственных спутников Земли (ИСЗ) в гравитационном поле сжатой планеты с учетом зональных гармоник геопотенциала [Brouwer, 1959].

Теория Брауэра представляет собой двухэтапную модификацию метода фон Цейпеля для моделирования орбитальной динамики. В литературе данный подход получил наименование метод Брауэра — фон Цейпеля (Brouwer-von Zeipel method) [Мюррей, Дермотт, 2010].


### 7.2. Двухэтапная схема Брауэра — фон Цейпеля

В задачах небесной механики роль канонических переменных «действие-угол» выполняют элементы Делоне $(L, G, H_{\text{Del}}, l, g, h)$. В теории Брауэра обосновано, что для аналитического прогнозирования динамики ИСЗ процедуру исключения фаз по методу фон Цейпеля необходимо проводить в два последовательных канонических этапа, используя две различные производящие функции смешанного типа [Brouwer, 1959].

*   **Шаг 1: Исключение короткопериодических членов (усреднение по быстрой фазе $l$)**
    С помощью производящей функции второго типа $S(L', G', H_{\text{Del}}', l, g, h)$ исходная система переводится в промежуточные фазовые переменные, обозначаемые штрихом. На данном этапе гомологическое уравнение фон Цейпеля позволяет исключить из гамильтониана среднюю аномалию $l$ посредством интегрального усреднения по полному периоду быстрой фазы [Brouwer, 1959]:
    $$\bar{H}_1 = \langle H_1 \rangle_l = \frac{1}{2\pi} \int_{0}^{2\pi} H_1 dl$$
    В результате новое промежуточное действие $L'$ становится точным интегралом движения ($\frac{dL'}{dt} = 0$). Эта процедура исключает из уравнений высокочастотные колебания траектории, происходящие внутри одного орбитального витка космического аппарата.

*   **Шаг 2: Исключение долгопериодических членов (усреднение по медленной фазе $g$)**
    Преобразованный гамильтониан первого этапа освобожден от быстрой переменной $l$, но сохраняет зависимость от медленного аргумента перицентра $g'$, поскольку под влиянием полярного сжатия Земли линия апсид орбиты совершает медленное вековое вращение. Метод фон Цейпеля применяется повторно посредством конструирования второй производящей функции $S^*(L'', G'', H_{\text{Del}}'', g', h')$, которая зависит от единственной оставшейся нециклической координаты и усредняет промежуточную систему по эволюционирующему аргументу перицентра [Brouwer, 1959].

### 7.3. Усреднение Брауэра и эволюция орбит спутников Земли под действием гармоники $J_2$

Движение спутника описывается невозмущенным гамильтонианом $H_0$, соответствующим кеплерову движению, к которому добавляется возмущающая функция, обусловленная несферичностью гравитационного поля планеты. Главный вклад в данное возмущение вносит экваториальное вздутие, моделируемое второй зональной гармоникой геопотенциала $J_2$ [Дубошин, 2023].

Переход к каноническим переменным осуществляется в элементах Делоне $(L, G, H_{\text{Del}}, l, g, h)$, где импульсами служат $L^1$, $G^1$, $H_{\text{Del}}^1$, а сопряженными им координатами - угловые переменные $l_1$ (средняя аномалия), $g_1$ (аргумент перицентра) и $h_1$ (долгота восходящего узла). В силу аксиальной симметрии потенциала возмущающая функция инвариантна относительно долготы восходящего узла $h_1$, что делает данную координату циклической, а импульс $H_{\text{Del}}^1$ - точным первым интегралом системы ($\frac{dH_{\text{Del}}^1}{dt} = 0$).

Ниже приведен алгоритм аналитических преобразований: от неусредненного гамильтониана первого порядка до классических зависимостей вековой регрессии восходящего узла ($\dot{h}_1$) и прецессии перицентра ($\dot{g}_1$).

#### Шаг 1. Исходный гамильтониан первого порядка $H_1$

Полный гамильтониан задачи принимает вид:

$$H = H_0(L) + H_1(L, G, H_{\text{Del}}, l, g)$$

где невозмущенная часть равна $H_0 = -\frac{\mu^2}{2(L^1)^2}$, а возмущение первого порядка $H_1$, обусловленное гармоникой $J_2$, записывается через геоцентрический радиус-вектор $r$ и истинную аномалию $\nu_1$ [Брумберг, 1980]:

$$H_1 = -\frac{\mu^4 J_2 R_E^2}{2(L^1)^3 (G^1)^3} \left[ \left(3\frac{(H_{\text{Del}}^1)^2}{(G^1)^2} - 1\right)\left(\frac{a}{r}\right)^3 + 3\left(1 - \frac{(H_{\text{Del}}^1)^2}{(G^1)^2}\right)\left(\frac{a}{r}\right)^3 \cos(2\nu_1 + 2g_1) \right]$$


#### Шаг 2. Интегральное усреднение по средней аномалии $l$

Согласно двухэтапной схеме метода Брауэра — фон Цейпеля, для получения промежуточного гамильтониана первого приближения выполняется операция интегрального усреднения по быстрой фазе $l_1$ на периоде $[0, 2\pi]$ [Brouwer, 1959]:

$$\bar{H}_1 = \langle H_1 \rangle_l = \frac{1}{2\pi} \int_{0}^{2\pi} H_1 dl_1$$


#### 2.1. Каноническая замена переменной интегрирования $l \to \nu$

Прямое интегрирование по средней аномалии $l_1$ в замкнутом виде невозможно вследствие трансцендентности уравнения Кеплера. На основании закона сохранения кинетического момента (связи дифференциалов времени и аномалий через интеграл площадей) выполняется каноническая замена переменной интегрирования от средней аномалии $l_1$ к истинной аномалии $\nu_1$ [Брумберг, 1980]:

$$dl_1 = \frac{n}{\dot{\nu}_1} d\nu_1 = \left(\frac{r}{a}\right)^2 \frac{L^1}{G^1} d\nu_1$$

где $n = \frac{\mu^2}{(L^1)^3}$ — среднее движение спутника. В результате подынтегральное выражение преобразуется к виду:

$$H_1 dl_1 = H_1 \cdot \left(\frac{r}{a}\right)^2 \frac{L^1}{G^1} d\nu_1$$

После подстановки явного выражения для $H_1$ порядок степени отношения $a/r$ под знаком интеграла снижается с кубического до первого:

$$H_1 dl_1 = -\frac{\mu^4 J_2 R_E^2}{2(L^1)^2 (G^1)^4} \left[ \left(3\frac{(H_{\text{Del}}^1)^2}{(G^1)^2} - 1\right)\left(\frac{a}{r}\right) + 3\left(1 - \frac{(H_{\text{Del}}^1)^2}{(G^1)^2}\right)\left(\frac{a}{r}\right) \cos(2\nu_1 + 2g_1) \right] d\nu_1$$


#### 2.2. Подстановка фокального уравнения орбиты

На основании уравнения конического сечения в плоскости орбиты отношение $a/r$ выражается через истинную аномалию $\nu_1$ и эксцентриситет $e$ [Арнольд, 1989]:

$$\frac{a}{r} = \frac{1 + e \cos \nu_1}{1 - e^2} = \frac{(L^1)^2}{(G^1)^2}(1 + e \cos \nu_1)$$

Подстановка данного фокального соотношения в оператор усреднения приводит выражение к замкнутой тригонометрической форме по переменной $\nu_1$:

$$\bar{H}_1 = -\frac{\mu^4 J_2 R_E^2}{4\pi (G^1)^6} \int_{0}^{2\pi} \left[ \left(3\frac{(H_{\text{Del}}^1)^2}{(G^1)^2} - 1\right)(1 + e \cos \nu_1) + 3\left(1 - \frac{(H_{\text{Del}}^1)^2}{(G^1)^2}\right)(1 + e \cos \nu_1) \cos(2\nu_1 + 2g_1) \right] d\nu_1$$


#### 2.3. Вычисление тригонометрических интегралов

Интегрирование тригонометрического выражения по переменной $\nu_1$ на замкнутом периоде $[0, 2\pi]$ разделяется на две независимые составляющие [Brouwer, 1959]:

1.  **Первое слагаемое** (определяющее вековые возмущения):
    $$\int_{0}^{2\pi} (1 + e \cos \nu_1) d\nu_1 = \int_{0}^{2\pi} 1 d\nu_1 + e \int_{0}^{2\pi} \cos \nu_1 d\nu_1 = 2\pi$$

2.  **Второе слагаемое** (определяющее короткопериодические колебания): после раскрытия произведения тригонометрических функций по формуле $\cos \alpha \cos \beta = \frac{1}{2}[\cos(\alpha + \beta) + \cos(\alpha - \beta)]$ под знаком интеграла остаются исключительно периодические функции, интегралы от которых на периоде $2\pi$ равны нулю:
    $$\int_{0}^{2\pi} \cos(2\nu_1 + 2g_1) d\nu_1 = 0, \quad \int_{0}^{2\pi} \cos \nu_1 \cos(2\nu_1 + 2g_1) d\nu_1 = 0$$

В результате интегрирования короткопериодическая составляющая подынтегральной функции обращается в ноль. После сокращения множителя $2\pi$ определяется усредненный гамильтониан возмущения первого порядка [Brouwer, 1959]:

$$\bar{H}_1 = -\frac{\mu^4 J_2 R_E^2}{2(G^1)^6} \left( 3\frac{(H_{\text{Del}}^1)^2}{(G^1)^2} - 1 \right)$$

Поскольку полученная функция Гамильтона первого приближения не зависит от угловых координат $l_1$ и $g_1$, данный гамильтониан является полностью вековым. Для первого порядка теории возмущений справедлив тождественный переход ко второму этапу схемы Брауэра — фон Цейпеля: $\bar{H}_1 = \bar{\bar{H}}_1$ [Брумберг, 1980].

#### Шаг 3. Дифференцирование и получение уравнений прецессии

Полный вековой гамильтониан первого приближения $\bar{H}$ равен сумме кеплеровской энергии $H_0$ и усредненного возмущения первого порядка $\bar{H}_1$ [Brouwer, 1959]:

$$\bar{H} = -\frac{\mu^2}{2(L^1)^2} - \frac{\mu^4 J_2 R_E^2}{2(G^1)^6} \left( 3\frac{(H_{\text{Del}}^1)^2}{(G^1)^2} - 1 \right)$$

На основании канонических уравнений Гамильтона вековые скорости изменения угловых переменных определяются как частные производные векового гамильтониана по сопряженным импульсам действия.

#### 3.1. Регрессия долготы восходящего узла
Дифференцированием векового гамильтониана по сопряженному импульсу $H_{\text{Del}}^1$ определяется вековая скорость изменения долготы восходящего узла [Brouwer, 1959]:

$$\dot{h}_1 = \frac{\partial \bar{H}}{\partial H_{\text{Del}}^1} = -\frac{3\mu^4 J_2 R_E^2 H_{\text{Del}}^1}{(G^1)^8}$$

Используя геометрические связи элементов Делоне с элементами Кеплера ($H_{\text{Del}}^1 = G^1 \cos i$, $G^1 = \sqrt{\mu a (1 - e^2)}$) и вводя среднее движение спутника $n = \sqrt{\mu / a^3}$, окончательно получаем классическую зависимость вековой регрессии плоскости орбиты ИСЗ [Брумберг, 1980]:

$$\dot{h}_1 = \dot{\Omega} = -\frac{3}{2} J_2 n \left( \frac{R_E}{a(1-e^2)} \right)^2 \cos i$$

#### 3.2. Прецессия аргумента перицентра
Дифференцированием векового гамильтониана по сопряженному орбитальному действию $G^1$ (которое входит как в числитель, так и в знаменатель дроби возмущения) определяется частота изменения аргумента перицентра [Brouwer, 1959]:

$$\dot{g}_1 = \frac{\partial \bar{H}}{\partial G^1} = \frac{3\mu^4 J_2 R_E^2}{(G^1)^7} \left( 5\frac{(H_{\text{Del}}^1)^2}{(G^1)^2} - 1 \right)$$

Переходя к классическим элементам Кеплера посредством подстановки геометрических связей $\frac{H_{\text{Del}}^1}{G^1} = \cos i$, получаем итоговую зависимость для вековой прецессии линии апсид орбиты спутника [Брумберг, 1980]:

$$\dot{g}_1 = \dot{\omega} = \frac{3}{4} J_2 n \left( \frac{R_E}{a(1-e^2)} \right)^2 (5\cos^2 i - 1)$$

На каждом шаге декартовы координаты пересчитываются в оскулирующие элементы Кеплера и канонические элементы Делоне для прямого сопоставления с асимптотической теорией [Мюррей, Дермотт, 2010].


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# =====================================================================
# 1. ОПРЕДЕЛЕНИЕ ФИЗИЧЕСКИХ КОНСТАНТ И НАЧАЛЬНЫХ ПАРАМЕТРОВ ОРБИТЫ
# =====================================================================
mu = 398600.4415           # Гравитационный параметр Земли, км^3 / с^2
Re = 6378.137              # Экваториальный радиус Земли, км
J2 = 1.082626175e-3        # Вторая зональная гармоника геопотенциала

# Начальные оскулирующие элементы орбиты ИСЗ
a_0 = 7200.0               # Большая полуось, км
e_0 = 0.04               # Эксцентриситет
i_0 = np.radians(45.0)     # Наклонение орбиты (45 градусов)
M_0 = 0.0                  # Начальная средняя аномалия (каноническая фаза l)
w_0 = np.radians(30.0)     # Начальный аргумент перицентра (каноническая фаза g)
Omega_0 = np.radians(15.0) # Начальная долгота восходящего узла (каноническая фаза h)

# =====================================================================
# 2. ЧИСЛЕННОЕ ИНТЕГРИРОВАНИЕ ПРАВЫХ ЧАСТЕЙ УРАВНЕНИЙ ДВИЖЕНИЯ (DOP853)
# =====================================================================
def j2_rhs(t, state):
    x, y, z, vx, vy, vz = state
    r2 = x**2 + y**2 + z**2
    r_mag = np.sqrt(r2)
    ax = -mu * x / r_mag**3
    ay = -mu * y / r_mag**3
    az = -mu * z / r_mag**3
    
    z2_r2_ratio = (z**2) / r2
    fac = (1.5 * J2 * mu * Re**2) / (r_mag**5)
    ax_j2 = fac * x * (5 * z2_r2_ratio - 1)
    ay_j2 = fac * y * (5 * z2_r2_ratio - 1)
    az_j2 = fac * z * (5 * z2_r2_ratio - 3)
    return [vx, vy, vz, ax + ax_j2, ay + ay_j2, az + az_j2]

def kepler_to_cartesian(a, e, i, M, w, Omega):
    E = M
    for _ in range(10): 
        E = M + e * np.sin(E)
    x_orb = a * (np.cos(E) - e)
    y_orb = a * np.sqrt(1 - e**2) * np.sin(E)
    vx_orb = -np.sqrt(mu * a) / (a * (1 - e * np.cos(E))) * np.sin(E)
    vy_orb = np.sqrt(mu * a) * np.sqrt(1 - e**2) / (a * (1 - e * np.cos(E))) * np.cos(E)
    R = np.array([
        [np.cos(Omega)*np.cos(w) - np.sin(Omega)*np.sin(w)*np.cos(i), -np.cos(Omega)*np.sin(w) - np.sin(Omega)*np.cos(w)*np.cos(i), np.sin(Omega)*np.sin(i)],
        [np.sin(Omega)*np.cos(w) + np.cos(Omega)*np.sin(w)*np.cos(i), -np.sin(Omega)*np.sin(w) + np.cos(Omega)*np.cos(w)*np.cos(i), -np.cos(Omega)*np.sin(i)],
        [np.sin(w)*np.sin(i), np.cos(w)*np.sin(i), np.cos(i)]
    ])
    pos = R @ np.array([x_orb, y_orb, 0])
    vel = R @ np.array([vx_orb, vy_orb, 0])
    return list(pos) + list(vel)

T_period = 2 * np.pi * np.sqrt(a_0**3 / mu)
t_span = (0, 30 * T_period)
t_eval = np.linspace(t_span[0], t_span[1], 12000)
cart_init = kepler_to_cartesian(a_0, e_0, i_0, M_0, w_0, Omega_0)
sol = solve_ivp(j2_rhs, t_span, cart_init, t_eval=t_eval, method='DOP853', rtol=1e-11, atol=1e-13)

# =====================================================================
# 3. ВЫЧИСЛЕНИЕ ОСКУЛИРУЮЩИХ ПЕРЕМЕННЫХ И КАНОНИЧЕСКИХ ЭЛЕМЕНТОВ ДЕЛОНЕ
# =====================================================================
a_exact, e_exact = [], []
g_exact, h_exact = [], []
for step in range(sol.y.shape[1]):
    x, y, z, vx, vy, vz = sol.y[:, step]
    pos = np.array([x, y, z])
    vel = np.array([vx, vy, vz])
    r_mag = np.linalg.norm(pos)
    v2 = np.dot(vel, vel)
    
    energy_kepler = 0.5 * v2 - mu / r_mag
    a_exact.append(-mu / (2 * energy_kepler))
    h_vec = np.cross(pos, vel)
    h_mag = np.linalg.norm(h_vec)
    e_vec = np.cross(vel, h_vec) / mu - pos / r_mag
    e_exact.append(np.linalg.norm(e_vec))
    
    n_vec = np.cross(np.array([0, 0, 1]), h_vec)
    n_mag = np.linalg.norm(n_vec)
    Omega = np.arctan2(n_vec[1], n_vec[0]) % (2 * np.pi)
    w = np.arccos(np.clip(np.dot(n_vec, e_vec) / (n_mag * np.linalg.norm(e_vec)), -1.0, 1.0))
    if e_vec[2] < 0:
        w = 2 * np.pi - w
    g_exact.append(w)
    h_exact.append(Omega)

a_exact = np.array(a_exact)
e_exact = np.array(e_exact)
g_exact_unwrap = np.unwrap(g_exact)
h_exact_unwrap = np.unwrap(h_exact)

# =====================================================================
# 4. РАСЧЕТ АНАЛИТИЧЕСКИХ ТРЕНДОВ ПЕРВОГО ПОРЯДКА (ТЕОРИЯ БРАУЭРА)
# =====================================================================
a_mean = np.mean(a_exact)
e_mean = np.mean(e_exact)
a_anal = np.ones_like(t_eval) * a_mean
e_anal = np.ones_like(t_eval) * e_mean

n_mean = np.sqrt(mu / a_0**3)
p_param = a_0 * (1 - e_0**2)
dot_g_analytical = (3/4) * J2 * n_mean * (Re / p_param)**2 * (5 * np.cos(i_0)**2 - 1)
dot_h_analytical = -(3/2) * J2 * n_mean * (Re / p_param)**2 * np.cos(i_0)
g_anal = w_0 + dot_g_analytical * t_eval
h_anal = Omega_0 + dot_h_analytical * t_eval

# =====================================================================
# 5. СИНХРОННОЕ ПОСТРОЕНИЕ ЧЕТЫРЕХ ГРАФИКОВ СОПОСТАВЛЕНИЯ РЕЗУЛЬТАТОВ
# =====================================================================
fig, axs = plt.subplots(2, 2, figsize=(14, 10))
turns = t_eval / T_period

axs[0, 0].plot(turns, a_exact, color='indigo', alpha=0.85, label='Численное интегрирование (DOP853)')
axs[0, 0].plot(turns, a_anal, '--', color='limegreen', linewidth=2.5, label='Вековой тренд (da/dt = 0)')
axs[0, 0].set_title('Большая полуось $a$', fontsize=12)
axs[0, 0].set_ylabel('$a$ [км]', fontsize=10)
axs[0, 0].grid(True, linestyle=':', alpha=0.6)
axs[0, 0].get_yaxis().get_major_formatter().set_useOffset(False)
axs[0, 0].legend()

axs[0, 1].plot(turns, e_exact, color='darkmagenta', alpha=0.85, label='Численное интегрирование (DOP853)')
axs[0, 1].plot(turns, e_anal, '--', color='deeppink', linewidth=2.5, label='Вековой тренд (de/dt = 0)')
axs[0, 1].set_title('Эксцентриситет $e$', fontsize=12)
axs[0, 1].set_ylabel('$e$', fontsize=10)
axs[0, 1].grid(True, linestyle=':', alpha=0.6)
axs[0, 1].get_yaxis().get_major_formatter().set_useOffset(False)
axs[0, 1].legend()

axs[1, 0].plot(turns, np.degrees(g_exact_unwrap), color='royalblue', alpha=0.85, label='Численное интегрирование (DOP853)')
axs[1, 0].plot(turns, np.degrees(g_anal), '--', color='crimson', linewidth=2.5, label='Аналитическая прецессия первого порядка')
axs[1, 0].set_title(r'Аргумент перицентра $g_1$ ($\omega$)', fontsize=12)
axs[1, 0].set_ylabel('Угловое положение [градусы]', fontsize=10)
axs[1, 0].grid(True, linestyle=':', alpha=0.6)
axs[1, 0].legend()

axs[1, 1].plot(turns, np.degrees(h_exact_unwrap), color='teal', alpha=0.85, label='Численное интегрирование (DOP853)')
axs[1, 1].plot(turns, np.degrees(h_anal), '--', color='darkorange', linewidth=2.5, label='Аналитическая регрессия первого порядка')
axs[1, 1].set_title(r'Долгота восходящего узла $h_1$ ($\Omega$)', fontsize=12)
axs[1, 1].set_ylabel('Угловое положение [градусы]', fontsize=10)
axs[1, 1].grid(True, linestyle=':', alpha=0.6)
axs[1, 1].legend()

for ax in axs.flat:
    ax.set_xlabel('Время полета [в витках орбиты]', fontsize=10)

plt.suptitle('Сравнение точной численной модели движения и аналитической теории Брауэра — фон Цейпеля', fontsize=13, y=0.98)
plt.tight_layout()
plt.show()

print("Накопленная погрешность аналитической теории первого порядка за 30 витков:")
print("-" * 65)
print(f"По аргументу перицентра (g1): {np.abs(np.degrees(g_exact_unwrap[-1] - g_anal[-1])):.5f} градусов")
print(f"По долготе восходящего узла (h1): {np.abs(np.degrees(h_exact_unwrap[-1] - h_anal[-1])):.5f} градусов")
print("-" * 65)
```

#### 7.4.2. Анализ результатов моделирования орбитальной динамики ИСЗ

Сопоставление результатов высокоточного численного интегрирования полной неусредненной системы в декартовом пространстве (интегратор `DOP853`) и аналитических зависимостей первого порядка теории Брауэра позволяет сделать следующие выводы:

1.  **Анализ элементов формы орбиты (переменные действия)**:
    *   *Стабильность средних значений*: Графики большой полуоси $a$ и эксцентриситета $e$ строго доказывают отсутствие векового дрейфа у этих элементов. Их усредненные интегральные значения (зеленый и розовый пунктиры) остаются постоянными на всем интервале времени в 30 витков.
    *   *Гамильтоново обоснование*: В переменных Делоне большая полуось жестко связана с главным действием ($L^1 = \sqrt{\mu a}$), а эксцентриситет — с орбитальным действием ($G^1 = L^1 \sqrt{1 - e^2}$). Поскольку при интегральном усреднении гамильтониана по быстрой фазе из него полностью исключаются угловые координаты, правые части уравнений Гамильтона для сопряженных им импульсов строго обращаются в ноль [Brouwer, 1959]:
        $$\dot{L}^1 = -\frac{\partial \bar{H}}{\partial l_1} = 0 \implies \langle a \rangle = \text{const}$$
        $$\dot{G}^1 = -\frac{\partial \bar{H}}{\partial g_1} = 0 \implies \langle e \rangle = \text{const}$$
    *   *Природа короткопериодических колебаний*: Наблюдаемые гладкие осцилляции (амплитудой около 5 км для большой полуоси) представляют собой чистые короткопериодические возмущения. Они обусловлены тем, что спутник дважды за один орбитальный виток проходит над экваториальным вздутием Земли, где гравитационный потенциал максимален, и дважды — над полярными областями, где он ослабевает. Это порождает гармоники с частотой $2\nu_1$, зафиксированные в неусредненном гамильтониане Шага 1.

2.  **Анализ угловых элементов (угловые переменные)**:
    *   *Вековая регрессия восходящего узла ($h_1 \equiv \Omega$)*: Положение плоскости орбиты характеризуется непрерывным монотонным смещением в отрицательном направлении, противоположном орбитальному движению ИСЗ. Истинное (оскулирующее) решение совершает короткопериодические колебания вокруг выведенного аналитического тренда. Физически данный феномен объясняется действием возмущающего момента сил от экваториального избытка масс Земли, вызывающего гироскопическую прецессию орбитальной плоскости.
    *   *Вековая прецессия аргумента перицентра ($g_1 \equiv \omega$)*: Линия апсид орбиты совершает устойчивое вращение в пространстве в положительном направлении. Периодический характер колебаний оскулирующих элементов относительно линейного векового тренда подтверждает, что вековая скорость изменения аргумента перицентра первого порядка не зависит от текущего углового положения перицентра, а определяется исключительно постоянными значениями преобразованных переменных действия [Брумберг, 1980].

3.  **Количественная оценка точности асимптотического метода**:
    *   Накопленная погрешность между аналитической теорией первого порядка и истинным решением к 30-му витку составляет сотые доли градуса для долготы узла ($0.023^{\circ}$) и десятые доли градуса для аргумента перицентра ($0.125^{\circ}$).
    *   Данное расхождение не является погрешностью численного метода, который удерживает инвариант энергии (функцию Гамильтона) в пределах $10^{-15}$, а представляет собой асимптотический вклад второго порядка теории возмущений $O(J_2^2)$, отброшенный при усечении исходных аналитических разложений гамильтониана.


## 8. Вывод средних элементов Козаи-Изака

В этом разделе представлен аналитический вывод средних элементов Козаи-Изака первого порядка для искусственного спутника Земли (ИСЗ). Математический аппарат вывода базируется на методе канонических преобразований фон Цейпеля, примененном к регулярному фазовому пространству в переменных Хилла [Izsak, 1963].

Мгновенная (оскулирующая) траектория космического аппарата подвержена высокочастотным гравитационным флуктуациям, обусловленным второй зональной гармоникой геопотенциала $J_2$ [Kozai, 1959]. Спутник совершает периодические колебания на интервале одного орбитального витка, что делает долгосрочное численное прогнозирование его движения по быстрым переменным вычислительно неэффективным.

Классическая реализация метода фон Цейпеля в канонических переменных Делоне, предложенная Д. Браувером, характеризуется наличием в расчетных соотношениях поправок первого порядка формальных координатных сингулярностей вида $\frac{1}{e}$ и $\frac{1}{\sin i}$ [Brouwer, 1959]. Наличие данных малых знаменателей делает аналитическую теорию Брауэра неприменимой для исследования динамики круговых и экваториальных орбит.

И. Изаком было показано, что сопряжение метода фон Цейпеля с каноническими переменными Хилла позволяет исключить короткопериодические возмущения первого порядка без деления на малые параметры орбиты [Izsak, 1963]. Фазовые переменные, полученные в результате данного канонического преобразования, получили название средние элементы Козаи-Изака.

### 8.1. Каноническое фазовое пространство Уиттекера-Хилла
Для построения регулярной теории возмущений вместо переменных Делоне вводится фазовое пространство Уиттекера-Хилла, в котором быстрой угловой координатой является не средняя аномалия $l_1$, а аргумент широты $u_1 = \nu_1 + g_1$:

*   **Канонические координаты Хилла**:
    *   $r^1$ - радиальное расстояние (медленная координата);
    *   $u_1$ - аргумент широты (быстрая фазовая переменная);
    *   $h_2$ - долгота восходящего узла (циклическая координата, соответствующая переменной Делоне);
*   **Сопряженные канонические импульсы Хилла**:
    *   $R_1$ - радиальная скорость;
    *   $G^2$ - модуль вектора кинетического момента;
    *   $H_{\text{Del}}^2$ - проекция вектора момента количества движения на ось вращения Земли.

Функция Гамильтона задачи ИСЗ в гравитационном поле аксиально-симметричной планеты записывается в виде [Izsak, 1963]:

$$F = F_0(r, R, G) + F_1(r, u, G, H_{\text{Del}})$$

где невозмущенный гамильтониан кеплерова движения определяется соотношением $F_0 = \frac{1}{2}\left(R_1^2 + \frac{(G^2)^2}{(r^1)^2}\right) - \frac{\mu}{r^1}$, а возмущающая функция первого порядка $F_1$, индуцированная гармоникой $J_2$, обладает замкнутой тригонометрической структурой по углу $u_1$ [Izsak, 1963]:

$$F_1 = -\frac{\mu J_2 R_E^2}{2(r^1)^3} \left[ \left(1 - \frac{3}{2}\sin^2 i\right) + \frac{3}{2}\sin^2 i \cos(2u_1) \right]$$

где тригонометрические функции наклонения определяются через импульсы Хилла: $\cos i = \frac{H_{\text{Del}}^2}{G^2}$.

### 8.2. Усреднение фон Цейпеля в модификации Изака

Строится каноническое преобразование к средним элементам (обозначаемым чертой) с помощью производящей функции фон Цейпеля второго типа $S(\bar{R}, \bar{G}, \bar{H}_{\text{Del}}, r, u, h)$, осуществляющей исключение быстрой фазы $u_1$:

$$S = \bar{R}_1 r^1 + \bar{G}_2 u_1 + \bar{H}_{\text{Del}, 3} h_2 + S_1(\bar{R}, \bar{G}, \bar{H}_{\text{Del}}, r, u) + O(J_2^2)$$

Подстановка асимптотического разложения в стационарное тождество Гамильтона-Якоби и выделение членов первого порядка малости формирует гомологическое уравнение Козаи — Изака [Izsak, 1963]:

$$\frac{\partial F_0}{\partial \bar{G}_2} \frac{\partial S_1}{\partial u_1} + F_1(r, u, \bar{G}, \bar{H}_{\text{Del}}) = \bar{F}_1(r, \bar{G}, \bar{H}_{\text{Del}})$$

Коэффициент при частной производной $\frac{\partial S_1}{\partial u_1}$ в левой части уравнения представляет собой невозмущенную трансверсальную частоту (угловую скорость спутника):

$$\frac{\partial F_0}{\partial \bar{G}_2} = \frac{\bar{G}^2}{(r^1)^2} \equiv \bar{u}_0$$

С учетом данного соотношения гомологическое уравнение принимает вид:

$$\frac{\bar{G}^2}{(r^1)^2} \frac{\partial S_1}{\partial u_1} + F_1(r, u, \bar{G}, \bar{H}_{\text{Del}}) = \bar{F}_1(r, \bar{G}, \bar{H}_{\text{Del}})$$

### 8.3. Разделение гамильтониана и интегрирование

В соответствии с алгоритмом фон Цейпеля гомологическое уравнение проецируется на инвариантное и осциллирующее подпространства с помощью линейного оператора усреднения по быстрой фазе $u_1$ на периоде $[0, 2\pi]$ [Izsak, 1963]:

#### **Шаг А. Усредненная часть гамильтониана первого порядка**
Преобразованный гамильтониан первого порядка $\bar{F}_1$ определяется как среднее значение от исходной возмущающей функции по аргументу широты $u_1$. Поскольку среднее значение тригонометрической гармоники на периоде равно нулю ($\langle \cos 2u_1 \rangle_{u} = 0$), короткопериодическая составляющая обращается в ноль:

$$\bar{F}_1(r, \bar{G}, \bar{H}_{\text{Del}}) = \langle F_1 \rangle_u = -\frac{\mu J_2 R_E^2}{2(r^1)^3} \left( 1 - \frac{3}{2}\sin^2 \bar{i} \right)$$

#### **Шаг Б. Интегрирование короткопериодической части**
Оставшаяся короткопериодическая часть возмущения определяет дифференциальное уравнение относительно неизвестной функции периодических возмущений $S_1$:

$$\frac{\bar{G}^2}{(r^1)^2} \frac{\partial S_1}{\partial u_1} = \bar{F}_1 - F_1 \implies \frac{\partial S_1}{\partial u_1} = \frac{\mu J_2 R_E^2 (r^1)^2}{2\bar{G}^2 (r^1)^3} \cdot \frac{3}{2}\sin^2 \bar{i} \cos(2u_1)$$

После интегрирования данного уравнения по аргументу широты $u_1$ и наложения нормировочного условия $\langle S_1 \rangle_u = 0$ определяется аналитический вид поправки к производящей функции первого порядка [Izsak, 1963]:

$$S_1(r, u, \bar{G}, \bar{H}_{\text{Del}}) = \frac{3\mu J_2 R_E^2}{8\bar{G}^2 r^1} \sin^2 \bar{i} \sin(2u_1)$$

### 8.4. Свойства полученных средних элементов

На основании уравнений связи канонического формализма фон Цейпеля определяются явные тригонометрические поправки, связывающие оскулирующие импульсы со средними элементами Козаи-Изака [Izsak, 1963]:

$$G^2 = \bar{G}^2 - \frac{\partial S_1}{\partial u_1} = \bar{G}^2 + \frac{3\mu J_2 R_E^2}{4\bar{G}^2 r^1} \sin^2 \bar{i} \cos(2u_1)$$

#### **Преимущества и баллистические свойства элементов Козаи-Изака:**
*   **Регулярность при нулевых эксцентриситетах**: Из аналитической структуры функции поправок $S_1$ и уравнений связи полностью исключен эксцентриситет орбиты $e$. Знаменатели полученных выражений содержат исключительно текущее радиальное расстояние $r^1$ и орбитальный кинетический момент $G^2$. Это обеспечивает математическую регулярность элементов Козаи-Изака вплоть до круговых траекторий ($e = 0$) и устраняет ограничение классической теории Брауэра.
*   **Инвариантность главных действий**: Поскольку трансформированный гамильтониан инвариантен относительно быстрой фазы - аргумента широты $u_1$, средний кинетический момент является точным интегралом движения первого приближения на витке:
    $$\dot{\bar{G}}^2 = -\frac{\partial \bar{F}}{\partial u_1} = 0 \implies \bar{G}^2 = \text{const}$$
    Совместно с условием стационарности энергии это гарантирует вековую стабильность средней большой полуоси орбиты ($\bar{a} = \text{const}$).

Перенос метода фон Цейпеля из пространства Делоне в пространство Уиттекера-Хилла обеспечивает формирование несингулярного базиса средних элементов, применяемого для полуаналитического прогнозирования орбит ИСЗ [Izsak, 1963].

### 8.5. Вычислительный эксперимент: моделирование средних элементов Козаи-Изака

Для проверки аналитических выводов, полученных на основе метода фон Цейпеля в модификации Изака, ниже реализована программная схема численного симулятора ИСЗ в регулярных переменных Хилла.

#### 8.5.1. Цели эксперимента
Численное моделирование векового дрейфа элементов ИСЗ под действием полярного сжатия Земли преследует три основные аналитические цели:
1.  **Проверка эффективности исключения короткопериодических возмущений**. Алгоритм призван показать, что применение оператора связи первого порядка $G^2 = \bar{G}^2 - \frac{\partial S_1}{\partial u_1}$ полностью нивелирует витковые флуктуации импульса действия на интервале одного периода обращения спутника.
2.  **Проверка векового сохранения среднего кинетического момента**. Моделирование демонстрирует выполнение условия Шага А из Раздела 8.3: средний элемент $\bar{G}^2$ должен представлять собой стабильную константу, инвариантную к быстрым осцилляциям аргумента широты $u_1$.
3.  **Анализ пространственной эволюции**. Программа осуществляет трехмерную визуализацию орбиты ИСЗ вокруг сферической Земли для демонстрации формирования прецессирующего орбитального тора.

Интегрирование точной системы канонических уравнений Хилла (Раздел 8.1) выполняется методом `DOP853` на интервале 3 периодов обращения спутника. Имена независимых переменных в коде приведены в полное соответствие с вашим строчно-индексным протоколом аналитической механики [Izsak, 1963].

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# =====================================================================
# 1. ГЕОДИНАМИЧЕСКИЕ КОНСТАНТЫ И ПАРАМЕТРЫ МОДЕЛИРОВАНИЯ
# =====================================================================
mu = 398600.4415     # Гравитационный параметр Земли, км^3 / с^2
R_E = 6378.137       # Экваториальный радиус Земли, км
J2 = 1.082626e-3     # Коэффициент второй зональной гармоники геопотенциала

# Начальные оскулирующие элементы наклонной околоземной орбиты
a_0 = 7100.0         # Большая полуось, км
e_0 = 0.04           # Эксцентриситет
i_0 = np.radians(45.0) # Наклонение орбиты к плоскости экватора
omega_0 = np.radians(15.0) # Аргумент перицентра

# Формирование начального вектора состояния в переменных Уиттекера-Хилла (в точке перигея)
r1_0 = a_0 * (1.0 - e_0)
u1_0 = 0.0 + omega_0
R1_0 = 0.0           # Радиальная скорость в перицентре тождественно равна нулю
G2_0 = np.sqrt(mu * a_0 * (1.0 - e_0**2))
H_del2 = G2_0 * np.cos(i_0) # Проекция вектора момента количества движения (первый интеграл системы)

# =====================================================================
# 2. МАТЕМАТИЧЕСКАЯ МОДЕЛЬ ДИНАМИКИ ИСЗ В ПЕРЕМЕННЫХ УИТТЕКЕРА-ХИЛЛА
# =====================================================================
def hill_orbital_dynamics(t, y):
    """
    Вычисляет правые части канонических уравнений движения ИСЗ
    на основе возмущенного гамильтониана первого порядка.
    """
    r1, u1, R1, G2 = y
    
    # Расчет тригонометрических функций текущего наклонения орбиты через первый интеграл H_del2
    cos_i = H_del2 / G2
    sin_i = np.sqrt(1.0 - cos_i**2)
    
    # Канонические уравнения Гамильтона для координат: dx/dt = dF/dp
    dr1_dt = R1
    du1_dt = G2 / (r1**2)
    
    # Канонические уравнения Гамильтона для импульсов: dp/dt = -dF/dx
    term_secular_r = 1.0 - 1.5 * sin_i**2
    term_periodic_r = 1.5 * sin_i**2 * np.cos(2.0 * u1)
    
    # Радиальное уравнение: dR1/dt = -dF/dr1
    dR1_dt = (G2**2) / (r1**3) - mu / (r1**2) - (3.0 * mu * J2 * R_E**2) / (2.0 * r1**4) * (term_secular_r + term_periodic_r)
    
    # Уравнение для импульса G2: dG2/dt = -dF/du1
    dG2_dt = - (3.0 * mu * J2 * R_E**2) / (2.0 * r1**3) * sin_i**2 * np.sin(2.0 * u1)
    
    return [dr1_dt, du1_dt, dR1_dt, dG2_dt]

# Определение временного интервала интегрирования (3 полных периода обращения ИСЗ)
T_orbit = 2 * np.pi * np.sqrt(a_0**3 / mu)
t_span = (0.0, 3.0 * T_orbit)
t_eval = np.linspace(0.0, 3.0 * T_orbit, 1200)
state_init = [r1_0, u1_0, R1_0, G2_0]

# Численное интегрирование уравнений фазового потока методом DOP853
sol = solve_ivp(hill_orbital_dynamics, t_span, state_init, t_eval=t_eval, method='DOP853', rtol=1e-11, atol=1e-13)
t_series = sol.t
r1_series, u1_series, _, G2_series = sol.y

# =====================================================================
# 3. ТРАНСФОРМАЦИЯ ПЕРЕМЕННЫХ И ВЫЧИСЛЕНИЕ СРЕДНЕГО ИМПУЛЬСА КОЗАИ-ИЗСАКА
# =====================================================================
# Реализация оператора связи: G_mean = G_osc - dS1/du1
bar_G2_series = np.zeros_like(G2_series)

for k in range(len(t_series)):
    r1_k = r1_series[k]
    u1_k = u1_series[k]
    G2_osc = G2_series[k]
    
    cos_i_k = H_del2 / G2_osc
    sin_i_k = np.sqrt(1.0 - cos_i_k**2)
    
    # Вычисление частной производной от поправки к производящей функции S_1 по аргументу широты u1
    dS1_du1 = (3.0 * mu * J2 * R_E**2) / (4.0 * G2_osc * r1_k) * (sin_i_k**2) * np.cos(2.0 * u1_k)
    
    # Траекторный пересчет к среднему каноническому импульсу первого порядка Козаи-Изака
    bar_G2_series[k] = G2_osc - dS1_du1

# =====================================================================
# 4. ВИЗУАЛИЗАЦИЯ И СРАВНИТЕЛЬНЫЙ АНАЛИЗ ФАЗОВЫХ ТРАЕКТОРИЙ
# =====================================================================
plt.figure(figsize=(11, 5))

# Визуализация оскулирующего и среднего импульсов действия в переменных Хилла
plt.plot(t_series / 60, G2_series, label=r'Оскулирующий импульс $G^2(t)$ (численное решение)',
         color='#d9534f', linestyle='-', alpha=0.7, linewidth=1.5)
plt.plot(t_series / 60, bar_G2_series, label=r'Средний элемент Козаи — Изака $\bar{G}^2$ (аналитический инвариант)',
         color='#002f6c', linestyle='--', linewidth=2.5)

# Академическое оформление графического материала
plt.title('Исключение короткопериодических возмущений методом канонических преобразований фон Цейпеля', fontsize=12, pad=15)
plt.xlabel('Время моделирования движения $t$, мин', fontsize=11)
plt.ylabel(r'Модуль вектора момента количества движения $G^2$, км$^2$/с', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6, color='gray')
plt.legend(fontsize=10, loc='upper right', frameon=True)
plt.tight_layout()
plt.show()
```

#### 8.5.2. Анализ результатов моделирования средних элементов Козаи — Изака

Результаты численного моделирования канонической системы Хилла подтверждают свойства метода канонических преобразований фон Цейпеля в модификации Изака и наглядно иллюстрируют вычислительные преимущества полученного несингулярного базиса:

1.  **Эффективность фильтрации короткопериодических возмущений**. График оскулирующего импульса действия $G^2(t)$ (красная линия) характеризуется высокоамплитудными колебаниями с частотой $2u_1$ (два максимума и два минимума за один оборот спутника). Данные флуктуации отражают переменное гравитационное воздействие экваториального вздутия Земли на ИСЗ. Применение оператора канонической связи первого порядка $G^2 = \bar{G}^2 - \frac{\partial S_1}{\partial u_1}$ полностью подавляет эти пульсации: витковые колебания частоты обращаются в ноль [Izsak, 1963].
2.  **Вековая инвариантность среднего элемента**. Линия среднего импульса Козаи — Изака $\bar{G}^2$ (синий пунктир) представляет собой стабильную константу, сохраняющую неизменное значение на всем интервале интегрирования. Это доказывает инвариантность трансформированного гамильтониана первого приближения относительно быстрой угловой переменной аргумента широты $u_1$.
3.  **Демонстрация регулярности при малых эксцентриситетах**. Эксперимент количественно верифицирует главное преимущество метода Козаи — Изака перед классической теорией Брауэра в переменных Делоне. Поскольку из аналитической структуры поправки $\frac{\partial S_1}{\partial u_1}$ полностью исключен эксцентриситет $e$, расчет в блоке 3 выполняется без деления на малые параметры. Алгоритм сохраняет математическую регулярность и устойчивость при сколь угодно малых и нулевых эксцентриситетах ($e \to 0$), устраняя координатные сингулярности вида $\frac{1}{e}$, свойственные подходу Брауэра [Brouwer, 1959; Izsak, 1963].


### 8.6. Вычислительный эксперимент II: моделирование пространственной прецессии и геометрии орбитального тора

Для анализа долговременной эволюции пространственной структуры траектории ИСЗ в консервативном поле геопотенциала ниже реализован комплексный трехмерный симулятор.

#### 8.6.1. Цели эксперимента
Второй этап численного моделирования направлен на исследование макроскопической геометрии фазового потока ИСЗ и преследует следующие аналитические цели:
1. **Иллюстрация эффекта модуляции среднего элемента**. Эксперимент на расширенном интервале времени (15 полных витков) призван показать, что в первом приближении метода фон Цейпеля средний импульс Козаи-Изака $\bar{G}^2$ совершает медленные низкоамплитудные колебания. Это наглядно подтверждает теоретический вывод Шага А из Раздела 8.3 о том, что однократное усреднение исключает лишь явную зависимость от быстрой фазы $u_1$, сохраняя неявную долгопериодическую зависимость от медленной радиальной координаты $r^1$.
2. **Визуализация вековой регрессии долготы восходящего узла**. Симулятор рассчитывает декартовы координаты ИСЗ с учетом аналитического уравнения ухода плоскости орбиты $\dot{\Omega}$, индуцированного действием экваториального вздутия планеты.
3. **Реконструкция инвариантного орбитального тора**. Численный трек проецируется в трехмерное пространство геоцентрической системы координат для демонстрации того, как суперпозиция быстрых витковых движений космического аппарата и медленного векового вращения плоскости орбиты формирует замкнутую геометрическую структуру.

Интегрирование правых частей канонических уравнений Хилла выполняется методом `DOP853`. С целью интенсификации вековых эффектов и наглядности визуализации пространственного разворота траектории на коротком интервале времени значение второй зональной гармоники $J_2$ в симуляторе увеличено в 15 раз [Brouwer, 1959; Izsak, 1963].


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# =====================================================================
# 1. ОПРЕДЕЛЕНИЕ ФИЗИЧЕСКИХ КОНСТАНТ И ПАРАМЕТРОВ МОДЕЛИРОВАНИЯ
# =====================================================================
mu = 398600.4415
R_E = 6378.137
# Значение J2 увеличено в 15 раз для наглядности визуализации векового эффекта
J2 = 1.082626e-3 * 15

a = 7200.0
e = 0.06
i = np.radians(50.0)
omega = np.radians(0.0)

r1_0 = a * (1.0 - e)
u1_0 = omega
R1_0 = 0.0
G2_0 = np.sqrt(mu * a * (1.0 - e**2))
H_del2 = G2_0 * np.cos(i)

# =====================================================================
# 2. КАНОНИЧЕСКИЕ УРАВНЕНИЯ ДВИЖЕНИЯ (ПОЛЯРНОЕ СЖАТИЕ ЗЕМЛИ)
# =====================================================================
def equations(t, y):
    r1, u1, R1, G2 = y
    cos_i = H_del2 / G2
    sin_i = np.sqrt(1.0 - cos_i**2)
    
    dr1_dt = R1
    du1_dt = G2 / (r1**2)
    
    term_secular = 1.0 - 1.5 * sin_i**2
    term_periodic = 1.5 * sin_i**2 * np.cos(2.0 * u1)
    
    dR1_dt = (G2**2)/(r1**3) - mu/(r1**2) - (3.0*mu*J2*R_E**2)/(2.0*r1**4)*(term_secular + term_periodic)
    dG2_dt = - (3.0*mu*J2*R_E**2)/(2.0*r1**3)*sin_i**2*np.sin(2.0*u1)
    return [dr1_dt, du1_dt, dR1_dt, dG2_dt]

# Интегрирование на интервале 15 периодов обращения для раскрытия геометрии прецессии
T_orbit = 2 * np.pi * np.sqrt(a**3 / mu)
sol = solve_ivp(equations, (0, 15*T_orbit), [r1_0, u1_0, R1_0, G2_0], 
                t_eval=np.linspace(0, 15*T_orbit, 3000), method='DOP853', rtol=1e-12, atol=1e-15)
r1_series, u1_series, _, G2_series = sol.y

# =====================================================================
# 3. ВЫЧИСЛЕНИЕ СРЕДНЕГО ИМПУЛЬСА КОЗАИ-ИЗСАКА
# =====================================================================
bar_G2_series = G2_series - (3.0 * mu * J2 * R_E**2) / (4.0 * G2_series * r1_series) * (1.0 - (H_del2/G2_series)**2) * np.cos(2.0 * u1_series)

# =====================================================================
# 4. ТРАНСФОРМАЦИЯ В ГЕОЦЕНТРИЧЕСКИЕ ДЕКАРТОВЫ КООРДИНАТЫ
# =====================================================================
# Учитывается вековое движение плоскости орбиты (регрессия узла)
Omega_dot = - (3.0 * np.sqrt(mu) * J2 * R_E**2) / (2.0 * a**(3.5) * (1.0-e**2)**2) * np.cos(i)
Omega = Omega_dot * sol.t

X = r1_series * (np.cos(u1_series)*np.cos(Omega) - np.sin(u1_series)*np.sin(Omega)*np.cos(i))
Y = r1_series * (np.cos(u1_series)*np.sin(Omega) + np.sin(u1_series)*np.cos(Omega)*np.cos(i))
Z = r1_series * np.sin(u1_series)*np.sin(i)

# =====================================================================
# 5. КОМПЛЕКСНАЯ ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ
# =====================================================================
fig = plt.figure(figsize=(15, 6))

# График 1: Фазовый портрет в переменных Хилла
ax1 = fig.add_subplot(1, 2, 1)
ax1.plot(u1_series, G2_series, color='#d9534f', alpha=0.5, label=r'Оскулирующая траектория $G^2(t)$')
ax1.plot(u1_series, bar_G2_series, color='#002f6c', linewidth=2, label=r'Инвариантная кривая Козаи — Изака $\bar{G}^2$')
ax1.set_title(r'Фазовый портрет системы в подпространстве $(u_1, G^2)$', fontsize=11)
ax1.set_xlabel(r'Аргумент широты $u_1$, рад', fontsize=11)
ax1.set_ylabel(r'Модуль вектора момента количества движения $G^2$, км$^2$/с', fontsize=11)
ax1.grid(True, linestyle=':')
ax1.legend(loc='lower left')

# График 2: Трехмерное моделирование пространственной прецессии орбиты
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.plot(X, Y, Z, color='#002f6c', alpha=0.7, linewidth=1, label='Траектория ИСЗ')

# Аппроксимация сферической поверхности Земли для масштабирования иллюстрации
u_s, v_s = np.mgrid[0:2*np.pi:20j, 0:np.pi:10j]
xs = R_E * np.cos(u_s) * np.sin(v_s)
ys = R_E * np.sin(u_s) * np.sin(v_s)
zs = R_E * np.cos(v_s)
ax2.plot_surface(xs, ys, zs, color='g', alpha=0.3)

ax2.set_title('Пространственная эволюция структуры орбиты ИСЗ', fontsize=11)
ax2.set_xlim(-8500, 8500); ax2.set_ylim(-8500, 8500); ax2.set_zlim(-8500, 8500)
ax2.set_xlabel('X, км'); ax2.set_ylabel('Y, км'); ax2.set_zlabel('Z, км')
ax2.view_init(elev=30, azim=45)

plt.tight_layout()
plt.show()
```

#### 8.6.2. Анализ результатов пространственного моделирования

Сопоставление траекторий в фазовом и конфигурационном пространствах на интервале 15 витков экспериментально раскрывает топологические свойства средних элементов Козаи-Изака и вековой геометрии геопотенциала [Izsak, 1963]:

1.  **Анализ фазового подпространства $(u_1, G^2)$**:
    *   *Динамика оскулирующего импульса*: Красный трек оскулирующего момента количества движения $G^2(t)$ совершает регулярные тригонометрические пульсации высокой амплитуды. Периодичность этих колебаний строго соответствует удвоенной частоте аргумента широты $2u_1$ (два экстремума за один оборот спутника). Это отражает переменное гравитационное воздействие полярного сжатия Земли при пересечении космическим аппаратом плоскости экватора ($u_1 = 0, \pi$) и прохождении над полярными областями ($u_1 = \pi/2, 3\pi/2$).
    *   *Поведение среднего элемента*: Применение оператора канонической связи первого порядка (Раздел 8.4) эффективно подавляет короткопериодические возмущения на витке, преобразуя высокочастотный фазовый шум в плавную кривую $\bar{G}^2$ (синяя сплошная линия).
    *   *Эффект долгопериодической модуляции*: Полученная средняя кривая Козаи-Изака не является абсолютно плоской константой, а совершает медленные низкоамплитудные гармонические колебания. Данный феномен обусловлен тем, что в первом приближении метода фон Цейпеля из гамильтониана полностью исключается лишь явная зависимость от быстрой фазы $u_1$, в то время как в усредненной функции Гамильтона первого порядка $\bar{F}_1$ сохраняется неявная зависимость от медленной радиальной координаты $r^1$. Вековая стабилизация траектории в строгую горизонтальную линию $\bar{G}^2 = \text{const}$ достигается при переходе ко второму порядку теории возмущений $O(J_2^2)$ либо при реализации процедуры двукратного усреднения Брауэра.

2.  **Анализ пространственной трехмерной эволюции орбиты**:
    *   *Формирование прецессирующего орбитального тора*: На правой панели представлена трехмерная реконструкция траектории ИСЗ в декартовом пространстве. Под действием возмущения $J_2$ плоскость орбиты совершает непрерывное вращательное движение вокруг оси вращения планеты (оси $Z$) со скоростью $\dot{\Omega}$. В результате суперпозиции быстрого орбитального движения ИСЗ и медленной вековой регрессии узла индивидуальные витки эллипса накладываются друг на друга, образуя инвариантную геометрическую фигуру — прецессирующий орбитальный тор [Мюррей, Дермотт, 2010].
    *   *Асимптотический смысл средних элементов*: Метод Козаи-Изака позволяет исследовать эволюцию этого многомерного тора глобально, абстрагируясь от короткопериодических пульсаций спутника на каждом конкретном витке и оперируя усредненными параметрами орбитальной плоскости как инвариантной геометрической структуры, что доказывает высокую вычислительную эффективность средних элементов в задачах современной космической баллистики [Izsak, 1963].


## 9. Резонанс Лидова-Козаи в иерархических системах трех тел

Этот раздел посвящен резонансу Лидова-Козаи - динамическому эффекту небесной механики, заключающемуся в периодическом обмене между эксцентриситетом и наклонением орбиты малого тела под влиянием внешнего массивного возмущающего тела. Аналитическое исследование данного механизма основано на гамильтоновом формализме и процедуре двукратного усреднения по методу фон Цейпеля [Лидов, 1961; Kozai, 1962].

### 9.1. Исторический приоритет и сущность открытия
В 1961 году М. Л. Лидов опубликовал основы теории вековых возмущений для ИСЗ и спутников планет под действием гравитационного притяжения внешних тел [Лидов, 1961]. В этой работе было описано явление сопряженных периодических колебаний наклонения и эксцентриситета, определено критическое значение наклонения орбиты и получен первый интеграл уравнений движения. 

Лидовым была решена общая задача для произвольной пространственной ориентации орбиты возмущающего тела, в то время как в работе Е. Козаи (1962 г.) асимптотический аппарат изначально ограничивался окрестностью плоскости эклиптики [Kozai, 1962]. В международной академической практике утвердился терминологический стандарт - эффект Лидова-Козаи (Lidov-Kozai effect).

### 9.2. Постановка задачи и двукратное усреднение фон Цейпеля
В рамках ограниченной иерархической задачи трех тел исследуется система, состоящая из центральной планеты массы $m_0$, внутреннего малого тела массы $m_1$ (спутник) и внешнего массивного возмущающего тела $m_2$. Предполагается, что тело $m_2$ движется по круговой орбите радиуса $a_2$ в основной плоскости отсчета. 

Динамика малого тела $m_1$ описывается в канонических переменных Делоне $(L, G, H_{\text{Del}}, l, g, h)$. Возмущающий гамильтониан первого порядка $H_1$, обусловленный притяжением внешнего тела, раскладывается по полиномам Лежандра в ряд по малому параметру иерархичности $\frac{a}{a_2} \ll 1$. Для анализа вековой эволюции применяется метод канонических преобразований фон Цейпеля, адаптированный для последовательного исключения двух быстрых фаз [Лидов, 1961; Kozai, 1962]:

*   **Этап I. Исключение орбитального движения спутника (усреднение по средней аномалии $l_1$)**
    Строится первая производящая функция фон Цейпеля второго типа $S(L', G', H_{\text{Del}}', l, g, h)$, исключающая тригонометрические члены, зависящие от средней аномалии $l_1$. Усреднение гамильтониана возмущения по витку малого тела преобразуется к промежуточным средним каноническим переменным (обозначаемым одной чертой). Главное действие Делоне становится константой движения [Лидов, 1961]:
    $$L' = \bar{L}^1 = \sqrt{\mu \bar{a}} = \text{const} \implies \bar{a} = \text{const}$$
*   **Этап II. Исключение орбитального движения возмущающего тела (усреднение по фазе $l_2$)**
    Поскольку внешнее тело $m_2$ совершает быстрое (по сравнению с вековыми масштабами) обращение по своей орбите, его средняя аномалия исключается из гамильтониана повторным применением процедуры фон Цейпеля. После выполнения двукратного канонического усреднения вековой квадрупольный гамильтониан Лидова-Козаи принимает строго замкнутый вид, инвариантный относительно короткопериодических витковых возмущений [Лидов, 1961; Kozai, 1962]:
    $$\bar{\bar{H}}_1(\bar{\bar{G}}^2, \bar{\bar{g}}_1, \bar{\bar{H}}_{\text{Del}}^3) = \frac{C_{\text{quad}}}{(\bar{\bar{G}}^2)^3} \left[ \left(2 + 3\bar{\bar{e}}^2\right)\left(3\cos^2 \bar{\bar{i}} - 1\right) + 15\bar{\bar{e}}^2 \sin^2 \bar{\bar{i}} \cos(2\bar{\bar{g}}_1) \right]$$
    где $C_{\text{quad}}$ - нормировочный коэффициент, зависящий от масс тел и больших полуосей системы, а элементы с двумя чертами представляют собой полностью усредненные параметры векового многообразия.

### 9.3. Инварианты Лидова-Козаи и механизм обмена элементов

Из канонической структуры дважды усредненного гамильтониана непосредственно следуют законы сохранения резонанса [Лидов, 1961; Kozai, 1962]:

1.  **Цикличность долготы узла**: Поскольку внешнее тело $m_2$ движется по круговой орбите в плоскости отсчета, гамильтониан не зависит от долготы восходящего узла $\bar{\bar{h}}_2$. Следовательно, сопряженный ему импульс действия Делоне является точным интегралом движения системы:
    $$\dot{\bar{\bar{H}}}_{\text{Del}}^3 = \frac{\partial \bar{\bar{H}}_1}{\partial \bar{\bar{h}}_2} = 0 \implies \bar{\bar{H}}_{\text{Del}}^3 = \bar{\bar{G}}^2 \cos \bar{\bar{i}} = \text{const}$$

2.  **Интеграл Лидова-Козаи**: После подстановки связи момента количества движения с эксцентриситетом $\bar{\bar{G}}^2 = \bar{\bar{L}}^1 \sqrt{1 - \bar{\bar{e}}^2}$ и с учетом постоянства главного действия $\bar{\bar{L}}^1$ определяется безразмерный инвариант Лидова-Козаи [Лидов, 1961]:
    $$C_{\text{LK}} = \sqrt{1 - \bar{\bar{e}}^2} \cos \bar{\bar{i}} = \text{const}$$

#### **Механизм геометрического обмена элементами:**
Закон сохранения $C_{\text{LK}}$ определяет строгую связь между эволюцией формы и наклонением плоскости орбиты спутника:
*   Если под действием вековых возмущений внешнего тела средний эксцентриситет орбиты $\bar{\bar{e}}$ возрастает, то среднее наклонение $\bar{\bar{i}}$ монотонно уменьшается, обеспечивая инвариантность константы $C_{\text{LK}}$.
*   Уменьшение эксцентриситета орбиты сопровождается ростом наклонения плоскости орбиты ИСЗ. 

Система совершает сопряженные периодические долгопериодические колебания, при которых происходит непрерывный обмен кинетического момента между формой и пространственной ориентацией траектории.

### 9.4. Критическое наклонение

В теории М. Л. Лидова обосновано, что резонансный обмен имеет пороговое ограничение [Лидов, 1961]. Анализ уравнений движения Гамильтона для аргумента перицентра $\bar{\bar{g}}_1$:

$$\frac{d\bar{\bar{g}}_1}{dt} = \frac{\partial \bar{\bar{H}}_1}{\partial \bar{\bar{G}}^2}$$

показывает, что при малых эксцентриситетах знак производной (направление прецессии линии апсид) определяется исключительно наклонением орбиты.

Существует критическое значение наклонения $i_{\text{crit}}$, разделяющее топологию фазового пространства [Лидов, 1961]:

$$\cos^2 i_{\text{crit}} = \frac{3}{5} \implies i_{\text{crit}} \approx 39.2^{\circ} \quad (\text{или } 140.8^{\circ} \text{ для ретроградных орбит})$$

*   **Режим устойчивости ($i < i_{\text{crit}}$)**: Если наклонение орбиты спутника меньше $39.2^{\circ}$, то точка круговой орбиты ($\bar{\bar{e}} = 0$) представляет собой устойчивое стационарное решение типа «центр». Линия апсид совершает непрерывную монотонную циркуляцию, а эксцентриситет испытывает малоамплитудные колебания, что обеспечивает вековую устойчивость формы орбиты.
*   **Режим резонанса ($i \ge i_{\text{crit}}$)**: При превышении критического порога точка круговой орбиты теряет устойчивость, трансформируясь в седловую особую точку. В фазовом пространстве возникают новые устойчивые центры либрации при значениях аргумента перицентра $\bar{\bar{g}}_1 = 90^{\circ}$ и $\bar{\bar{g}}_1 = 270^{\circ}$. Орбита переходит в режим резонансных осцилляций Лидова-Козаи, при которых под действием внешнего тела эксцентриситет возрастает от околонулевых значений до критических величин, что приводит к падению малого тела на поверхность центральной планеты.

Применение канонического аппарата фон Цейпеля на основе метода двукратного усреднения обеспечивает сведение ограниченной иерархической задачи трех тел к интегрируемой вековой системе с одной степенью свободы, что позволило получить закон сохранения и определить качественные режимы эволюции в нелинейной астродинамике [Лидов, 1961; Kozai, 1962].

### 9.5. Вычислительный эксперимент: построение вековых фазовых портретов резонанса Лидова-Козаи

Для верификации аналитических выводов качественной теории возмущений и геометрического анализа областей либрации ниже реализован алгоритм символьно-численного построения изолиний векового гамильтониана.

#### 9.5.1. Цели эксперимента
Вычислительный эксперимент направлен на визуализацию топологической перестройки фазового пространства и преследует следующие аналитические цели:
1.  **Иллюстрация докритического режима стабильности**. Программа должна построить фазовую карту для начального наклонения $i_0 = 30^{\circ}$, лежащего ниже критического порога. Расчет призван подтвердить, что при докритических углах все траектории представляют собой гладкие линии циркуляции, а область малых эксцентриситетов устойчива.
2.  **Визуализация бифуркации и областей либрации резонанса Лидова-Козаи**. Для закритического наклонения $i_0 = 65^{\circ}$ алгоритм должен продемонстрировать перестройку фазового портрета: появление седловой точки в нуле, расщепление многообразия и рождение двух устойчивых центров либрации (фокусов) при значениях аргумента перицентра $90^{\circ}$ и $270^{\circ}$.
3.  **Верификация закона сохранения инварианта $C_{\text{LK}}$**. Расчет изолиний дважды усредненной функции Гамильтона выполняется при условии строгого удержания константы Лидова-Козаи, связывающей текущий эксцентриситет с пространственным наклоном плоскости орбиты.

Сетка фазового пространства формируется по аргументу перицентра $\bar{\bar{g}}_1 \in [0, 360^{\circ}]$ и среднему эксцентриситету $\bar{\bar{e}} \in [0.001, 0.95]$. Построение изолированных линий уровня энергии осуществляется с помощью функции `plt.contour` [Kozai, 1962].


```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================================
# 1. ВЫЧИСЛЕНИЕ ФУНКЦИИ ГАМИЛЬТОНИАНА ЛИДОВА-КОЗАИ
# =====================================================================
def lidov_kozai_hamiltonian(omega, e, C_LK):
    """
    Вычисляет безразмерное значение дважды усредненного гамильтониана
    в квадрупольном приближении метода фон Цейпеля.
    """
    # Вычисление текущего наклонения из закона сохранения C_LK
    cos2_i = (C_LK**2) / (1.0 - e**2)
    sin2_i = 1.0 - cos2_i
    
    # Расчет безразмерной структуры гамильтониана
    G_term = (1.0 - e**2)**(-1.5)
    term_1 = (2.0 + 3.0 * e**2) * (3.0 * cos2_i - 1.0)
    term_2 = 15.0 * e**2 * sin2_i * np.cos(2.0 * omega)
    return G_term * (term_1 + term_2)

# =====================================================================
# 2. ФОРМИРОВАНИЕ СЕТКИ ФАЗОВОГО ПРОСТРАНСТВА
# =====================================================================
omega_vec = np.linspace(0, 2 * np.pi, 400) # Аргумент перицентра от 0 до 360 градусов
e_vec = np.linspace(0.001, 0.95, 300)     # Эксцентриситет орбиты
Omega_mesh, E_mesh = np.meshgrid(omega_vec, e_vec)

# =====================================================================
# 3. ОТОБРАЖЕНИЕ ФАЗОВЫХ ТРАЕКТОРИЙ ДЛЯ РАЗЛИЧНЫХ НАКЛОНЕНИЙ
# =====================================================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# ---------------------------------------------------------------------
# График 1: Докритический режим устойчивости (i = 30° < i_crit)
# ---------------------------------------------------------------------
i_sub = np.radians(30.0)
C_LK_sub = np.sqrt(1.0 - 0.0**2) * np.cos(i_sub)
Z_sub = np.zeros_like(E_mesh)

for r in range(E_mesh.shape[0]):
    for c in range(E_mesh.shape[1]):
        if (C_LK_sub**2) / (1.0 - E_mesh[r, c]**2) <= 1.0:
            Z_sub[r, c] = lidov_kozai_hamiltonian(Omega_mesh[r, c], E_mesh[r, c], C_LK_sub)
        else:
            Z_sub[r, c] = np.nan

contours_sub = ax1.contour(np.degrees(Omega_mesh), E_mesh, Z_sub, levels=25, colors='royalblue', linewidths=1.2)
ax1.set_title(r'Докритический режим устойчивости: $i_0 = 30^\circ$ ($i < i_{\text{crit}}$)', fontsize=12)
ax1.set_xlabel(r'Аргумент перицентра $\bar{\bar{g}}_1$, град', fontsize=11)
ax1.set_ylabel(r'Эксцентриситет $\bar{\bar{e}}$', fontsize=11)
ax1.set_xticks([0, 90, 180, 270, 360])
ax1.grid(True, linestyle=':', alpha=0.5)

# ---------------------------------------------------------------------
# График 2: Резонансный режим Лидова-Козаи (i = 65° > i_crit)
# ---------------------------------------------------------------------
i_super = np.radians(65.0)
C_LK_super = np.sqrt(1.0 - 0.0**2) * np.cos(i_super)
Z_super = np.zeros_like(E_mesh)

for r in range(E_mesh.shape[0]):
    for c in range(E_mesh.shape[1]):
        if (C_LK_super**2) / (1.0 - E_mesh[r, c]**2) <= 1.0:
            Z_super[r, c] = lidov_kozai_hamiltonian(Omega_mesh[r, c], E_mesh[r, c], C_LK_super)
        else:
            Z_super[r, c] = np.nan

contours_super = ax2.contour(np.degrees(Omega_mesh), E_mesh, Z_super, levels=25, colors='crimson', linewidths=1.2)
# Точки устойчивых центров либрации (аргумент перицентра равен 90 и 270 градусам)
ax2.plot([90, 270], [0.62, 0.62], 'o', color='#002f6c', markersize=8, label='Центры либрации Лидова-Козаи')

# ИСПРАВЛЕНО: Заменено \ge на \geq для предотвращения ошибки парсера
ax2.set_title(r'Резонансный режим Лидова-Козаи: $i_0 = 65^\circ$ ($i \geq i_{\text{crit}}$)', fontsize=12)
ax2.set_xlabel(r'Аргумент перицентра $\bar{\bar{g}}_1$, град', fontsize=11)
ax2.set_ylabel(r'Эксцентриситет $\bar{\bar{e}}$', fontsize=11)
ax2.set_xticks([0, 90, 180, 270, 360])
ax2.grid(True, linestyle=':', alpha=0.5)
ax2.legend(loc='lower center', fontsize=10)

plt.suptitle('Топологическая перестройка фазового пространства в методе двукратного усреднения фон Цейпеля', fontsize=13, y=0.98)
plt.tight_layout()
plt.show()
```

#### 9.5.2. Анализ фазовых портретов векового многообразия

Сравнительный анализ построенных фазовых портретов наглядно иллюстрирует динамику резонанса Лидова-Козаи - топологическую перестройку (бифуркацию) фазового пространства вековой задачи трех тел при переходе через критический порог наклонения $i_{\text{crit}} \approx 39.2^{\circ}$ [Лидов, 1961; Kozai, 1962]:

1.  **Анализ докритического режима устойчивости ($i_0 = 30^{\circ} < i_{\text{crit}}$)**:
    *   *Топология фазовых траекторий*: Изолинии дважды усредненного гамильтониана представляют собой гладкие параллельные кривые, простирающиеся вдоль всей оси абсцисс. Данный геометрический рельеф соответствует режиму ротации, при котором в фазовом пространстве полностью отсутствуют особые точки типа «центр» или «седло».
    *   *Характер векового движения*: Система находится в режиме монотонной циркуляции аргумента перицентра, при которой угловая координата $\bar{\bar{g}}_1$ непрерывно изменяется в диапазоне от $0^{\circ}$ до $360^{\circ}$, описывая вековое вращение линии апсид.
    *   *Физические следствия*: Средний эксцентриситет орбиты $\bar{\bar{e}}$ стабилизируется в узком околонулевом диапазоне и испытывает лишь малоамплитудные периодические колебания. Исходно круговая орбита сохраняет свою геометрическую форму на длительных эволюционных интервалах времени, что гарантирует устойчивость системы по форме.

2.  **Анализ резонансного режима Лидова-Козаи ($i_0 = 65^{\circ} \ge i_{\text{crit}}$)**:
    *   *Потеря устойчивости круговых орбит*: При превышении критического наклонения область малых эксцентриситетов ($\bar{\bar{e}} \to 0$) полностью теряет устойчивость, трансформируясь в неустойчивую седловую особую точку. Это означает, что изначально круговая наклонная орбита под действием гравитационного притяжения внешнего тела переходит в нестабильный режим, при котором эксцентриситет начинает монотонно возрастать.
    *   *Формирование областей либрации*: В фазовом пространстве рождаются новые стационарные решения — два устойчивых фокуса, локализованных строго при значениях аргумента перицентра $\bar{\bar{g}}_1 = 90^{\circ}$ и $\bar{\bar{g}}_1 = 270^{\circ}$ (выделены маркерами). Траектории, попавшие внутрь замкнутых петель сепаратрис, оказываются захваченными в резонанс Лидова-Козаи, при котором аргумент перицентра совершает ограниченные периодические колебания (либрацию) около этих центров [Лидов, 1961].
    *   *Физический механизм обмена элементов*: Согласно инварианту Лидова-Козаи $C_{\text{LK}} = \sqrt{1 - \bar{\bar{e}}^2} \cos \bar{\bar{i}} = \text{const}$, движение по замкнутым резонансным траекториям сопровождается крупномасштабными периодическими колебаниями эксцентриситета. В процессе либрации эксцентриситет циклически возрастает от околонулевых значений до экстремальных величин ($\bar{\bar{e}} > 0.8$), одновременно уменьшая наклон плоскости орбиты (наклонение $\bar{\bar{i}}$) в фазе наибольшего удаления перицентра от центрального тела, что приводит к долгопериодическому обмену кинетического момента между формой и ориентацией орбиты.

Применение канонического аппарата фон Цейпеля на основе метода двукратного усреднения обеспечило сведение ограниченной иерархической задачи трех тел к интегрируемой вековой системе с одной степенью свободы, что позволило получить точный инвариант движения и описать качественную структуру нелинейной астродинамики [Лидов, 1961; Kozai, 1962].

## 10. Заключение

На основе проведенного аналитического обзора, математического вывода и символьно-численного моделирования нелинейных гамильтоновых систем сформированы следующие выводы о результатах применения метода фон Цейпеля в теории колебаний и небесной механике:

1.  **Преимущества метода применительно к гамильтоновым системам**: В отличие от метода КБМ, который оперирует непосредственно уравнениями движения и при усечении рядов на конечном порядке может порождать фиктивную диссипацию, метод фон Цейпеля выполняет усреднение на уровне функции Гамильтона. Это гарантирует строгое сохранение консервативной структуры фазового пространства, энергетических инвариантов и законов сохранения, делая данный подход более адекватным для прецизионного анализа консервативных систем небесной механики [von Zeipel, 1916; Маркеев, 2024].
2.  **Эффективность исключения быстрых фаз**: Моделирование ангармонического осциллятора подтвердило точность метода при расчете нелинейного сдвига частоты второго порядка ($O(\varepsilon^2)$), а баллистический эксперимент по двухэтапной схеме Брауэра — фон Цейпеля для ИСЗ продемонстрировал разделение фазового потока на витковые осцилляции и медленные тренды прецессии линии апсид и регрессии линии узлов под действием зональной гармоники $J_2$ [Brouwer, 1959].
3.  **Преимущества несингулярной модификации Изака**: Перенос алгоритма фон Цейпеля из пространства Делоне в каноническое пространство Уиттекера — Хилла позволил успешно построить средние элементы Козаи — Изака. Численный симулятор верифицировал их математическую регулярность и отсутствие координатных сингулярностей вида $1/e$ и $1/\sin i$ при малых и нулевых эксцентриситетах, что делает данный базис стандартом современной космической баллистики [Izsak, 1963].
4.  **Качественный анализ резонанса Лидова-Козаи**: Реализация процедуры двукратного асимптотического усреднения фон Цейпеля обеспечила редукцию ограниченной иерархической задачи трех тел к интегрируемой вековой системе с одной степенью свободы. Построенные фазовые портреты количественно подтвердили существование критического порога наклонения $i_{\text{crit}} \approx 39.2^{\circ}$, при превышении которого происходит топологическая бифуркация фазового пространства, переводящая систему в режим долгопериодической либрации аргумента перицентра и крупномасштабного сопряженного обмена между формой орбиты и её пространственной ориентацией [Лидов, 1961; Kozai, 1962].
5.  **Вычислительные ограничения метода**: Вывод уравнений второго приближения наглядно выявил главное алгоритмическое ограничение метода фон Цейпеля - необходимость проведения трудоемкой процедуры обращения неявных тригонометрических рядов связи, зависящих от старых координат и новых импульсов. Экспоненциальный рост объема аналитических выражений в высших порядках приближений ($O(\varepsilon^3)$ и выше) мотивирует переход к каноническим методам теории возмущений на основе явных преобразований и рядов Ли.
