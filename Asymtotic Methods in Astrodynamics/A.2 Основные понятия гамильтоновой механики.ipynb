# Введение в гамильтонову механику: основные понятия для теории возмущений

Настоящий конспект вводит базовые понятия гамильтонова формализма, которые необходимы для построения и понимания асимптотических методов теории возмущений Ли.

---

## 1. Канонические уравнения Гамильтона

В гамильтоновом формализме состояние динамической системы с $n$ степенями свободы описывается в $2n$-мерном фазовом пространстве независимых переменных, состоящем из обобщенных координат $q^i$ и сопряженных им канонических импульсов $p_i$, где индекс $i$ меняется от $1$ до $n$. Переход от лагранжева описания к гамильтонову осуществляется посредством преобразования Лежандра для функции Лагранжа $L(q, \dot{q}, t)$:

$$p_i = \frac{\partial L}{\partial \dot{q}^i}, \quad H(q, p, t) = p \dot{q} - L(q, \dot{q}, t) \tag{1}$$

где обобщенные скорости $\dot{q}^i$ должны быть явно выражены через импульсы $p_i$ и координаты $q^i$ из первого соотношения системы (1). В рамках принятого протокола обобщенные координаты $q$ и обобщенные скорости $\dot{q}$ образуют векторы-столбцы размерности $n \times 1$, а канонические импульсы $p$ образуют матрицу-строку размерности $1 \times n$, что обеспечивает скалярный вид произведения $p \dot{q}$.

Динамика системы в фазовом пространстве определяется системой $2n$ канонических дифференциальных уравнений первого порядка:
$$\frac{d}{dt}q^i = \frac{\partial H}{\partial p_i}, \quad \frac{d}{dt}p_i = -\frac{\partial H}{\partial q^i} \tag{2}$$

Для автономных систем функция Гамильтона $H(q, p)$ явно не зависит от времени и выражает закон сохранения полной механической энергии системы, являясь первым интегралом уравнений движения.

## 2. Первые интегралы и канонические скобки Пуассона

Временная эволюция произвольной гладкой функции динамических переменных $F(q, p, t)$ вдоль истинной траектории движения системы, в силу канонических уравнений Гамильтона (2), определяется полной производной по времени. С учетом принятых соглашений о матричной структуре аргументов, дифференциал по координатам-столбцам и импульсам-строкам формирует скалярное выражение:
$$\frac{dF}{dt} = \frac{\partial F}{\partial t} + \frac{\partial F}{\partial q} \frac{d}{dt}q + \frac{d}{dt}p \left(\frac{\partial F}{\partial p}\right) = \frac{\partial F}{\partial t} + \sum_{i=1}^{n} \left( \frac{\partial F}{\partial q^i}\frac{\partial H}{\partial p_i} - \frac{\partial F}{\partial p_i}\frac{\partial H}{\partial q^i} \right)$$

В рамках принятого алгебраического соглашения каноническая скобка Пуассона двух гладких функций $F$ и $G$ определяется в координатной форме как:
$$\{F, G\} = \sum_{i=1}^{n} \left( \frac{\partial F}{\partial p_i} \frac{\partial G}{\partial q^i} - \frac{\partial F}{\partial q^i} \frac{\partial G}{\partial p_i} \right) \tag{3}$$

В силу данного определения выражение для полной производной по времени через оператор скобки Пуассона принимает вид:
$$\frac{dF}{dt} = \frac{\partial F}{\partial t} + \{H, F\} \tag{4}$$

*Примечание о соглашении знаков*. В учебниках по классической механике скобку Пуассона часто определяют с противоположным знаком, меняя порядок вычитания в формуле (3). Однако в литературе, посвященной асимптотическим методам и компьютерной алгебре, принято использовать именно текущую форму записи. Такое соглашение позволяет, например, вводить линейные операторы Ли в прямом порядке аргументов без появления паразитных знаков минус в высших порядках теории возмущений.

### Интегралы движения
Гладкая функция $F(q, p, t)$ является первым интегралом движения тогда и только тогда, когда ее полная производная по времени в силу уравнений движения тождественно равна нулю:
$$\frac{\partial F}{\partial t} + \{H, F\} = 0 \tag{5}$$

Для частного случая функций $F(q, p)$, которые не зависят от времени явно ($\partial F / \partial t = 0$), условие сохранения в автономной системе упрощается до требования зануления скобки Пуассона с гамильтонианом системы:
$$\{H, F\} \equiv 0 \implies F(q, p) = \text{const} \tag{6}$$

### Фундаментальные свойства скобок Пуассона
Для построения теории возмущений и канонических преобразований Ли выполняются следующие алгебраические свойства скобок Пуассона:
1. **Кососимметричность**: $\{F, G\} = -\{G, F\}$
2. **Билинейность**: $\{c_1 F_1 + c_2 F_2, G\} = c_1\{F_1, G\} + c_2\{F_2, G\}$
3. **Тождество Якоби**: $\{F, \{G, K\}\} + \{G, \{K, F\}\} + \{K, \{F, G\}\} = 0$
4. **Правило Лейбница для произведения**: $\{F \cdot G, K\} = F\{G, K\} + \{F, K\}G$

Тождество Якоби порождает теорему Пуассона: если функции $F$ и $G$ являются первыми интегралами движения, то их скобка Пуассона $\{F, G\}$ также представляет собой интеграл движения данной системы уравнений.


## 3. Представление уравнений движения через фазовый вектор

Для компактного аналитического описания системы объединим канонические переменные в составной вектор фазовых координат $x$, который формируется как вектор-столбец размерности $2n \times 1$:
$$x = \begin{pmatrix} q \\ p^T \end{pmatrix} = (q^1, \dots, q^n, p_1, \dots, p_n)^T$$

Использование канонической кососимметрической симплектической матрицы $\mathbf{J}$ размерности $2n \times 2n$ позволяет записать уравнения движения в инвариантной матричной форме:
$$\frac{dx}{dt} = \mathbf{J} \left( \frac{\partial H}{\partial x} \right)^T \tag{7}$$
где блочная структура матрицы $\mathbf{J}$ имеет вид:
$$\mathbf{J} = \begin{pmatrix} \mathbf{0} & \mathbf{I}_n \\ -\mathbf{I}_n & \mathbf{0} \end{pmatrix}$$

Связь между симплектической структурой потока и аппаратом скобок Пуассона раскрывается при вычислении производных самих фазовых координат по времени. В силу принятого в разделе 2 алгебраического соглашения, канонические уравнения движения покомпонентно записываются непосредственно через оператор скобки Пуассона с гамильтонианом системы:
$$\frac{d}{dt}q^i = \{H, q^i\}, \quad \frac{d}{dt}p_i = \{H, p_i\} \tag{8}$$

Векторно-матричная форма уравнений движения для всего фавого вектора $x$ принимает вид:
$$\frac{dx}{dt} = \{H, x\} \tag{9}$$

Формула (9) показывает, что временная эволюция любой точки фазового пространства полностью определяется действием гамильтониана через оператор скобки Пуассона.


## 4. Разложение функций в ряд с помощью скобок Пуассона

В гамильтоновом формализме временную эволюцию любой явным образом не зависящей от времени функции фазового пространства $F(q, p)$ можно представить в виде ряда Тейлора по переменной $t$. Математическое удобство данного подхода заключается в том, что все последовательные производные по времени в таком разложении заменяются на вложенные скобки Пуассона с гамильтонианином системы $H$.

### Разложение в ряд Тейлора

Разложение функции $F(q, p)$ в ряд Тейлора в окрестности начального момента времени $t=0$ имеет вид:
$$F(t) = F_0 + t \left. \frac{dF}{dt} \right|_{t=0} + \frac{t^2}{2!} \left. \frac{d^2F}{dt^2} \right|_{t=0} + \frac{t^3}{3!} \left. \frac{d^3F}{dt^3} \right|_{t=0} + \dots \tag{10}$$

Используя зафиксированное в разделе 2 правило вычисления полной производной автономной функции (4), произведем последовательную замену производных высших порядков на операторы скобок Пуассона с функцией Гамильтона $H$:
* Первая производная: $\frac{dF}{dt} = \{H, F\}$
* Вторая производная: $\frac{d^2F}{dt^2} = \frac{d}{dt}\{H, F\} = \{H, \{H, F\}\}$
* Третья производная: $\frac{d^3F}{dt^3} = \{H, \{H, \{H, F\}\}\}$

Подставляя данные выражения в исходный ряд Тейлора (10), получаем формулу разложения функции через вложенные скобки Пуассона, где гамильтониан системы последовательно выступает первым аргументом:
$$F(t) = F_0 + t\{H, F\} + \frac{t^2}{2!}\{H, \{H, F\}\} + \frac{t^3}{3!}\{H, \{H, \{H, F\}\}\} + \dots \tag{11}$$

---

### Пример: Координата линейного гармонического осциллятора

Рассмотрим систему с функцией Гамильтона $H = \frac{p^2}{2} + \frac{q^2}{2}$ и найдем временное разложение для обобщенной координаты, положив тестируемую функцию равной $F = q$.

1. **Вычисление коэффициентов ряда через скобки Пуассона**:
   * Первая производная: $\dot{q} = \{H, q\} = \frac{\partial H}{\partial p} \frac{\partial q}{\partial q} - \frac{\partial H}{\partial q} \frac{\partial q}{\partial p} = p \cdot 1 - q \cdot 0 = p$
   * Вторая производная: $\ddot{q} = \{H, p\} = \frac{\partial H}{\partial p} \frac{\partial p}{\partial q} - \frac{\partial H}{\partial q} \frac{\partial p}{\partial p} = p \cdot 0 - q \cdot 1 = -q$
   * Третья производная: $\dddot{q} = \{H, -q\} = -\{H, q\} = -p$
   * Четвертая производная: $\ddddot{q} = \{H, -p\} = -\{H, p\} = -(-q) = q$

2. **Подстановка полученных значений в операторный ряд (11)**:
   $$q(t) = q_0 + t \cdot p_0 + \frac{t^2}{2!} \cdot (-q_0) + \frac{t^3}{3!} \cdot (-p_0) + \frac{t^4}{4!} \cdot q_0 + \dots$$

3. **Группировка слагаемых по начальным условиям**:
   $$q(t) = q_0 \left( 1 - \frac{t^2}{2!} + \frac{t^4}{4!} - \dots \right) + p_0 \left( t - \frac{t^3}{3!} + \dots \right)$$

В результате суммирования рядов получается точное аналитическое решение уравнений движения осциллятора: $q(t) = q_0 \cos t + p_0 \sin t$. В теории возмущений этот подход позволяет раскладывать в ряд сложные функции фазового пространства, обрывая бесконечный ряд на заданной степени точности по малому параметру.

```python
from IPython.display import display, Markdown
import sympy as sp

# Настройка автоматического рендеринга формул в среде Jupyter Notebook
sp.init_printing(use_latex='mathjax', order='lex')

# Объявление символьных канонических переменных и физических параметров маятника
q, p = sp.symbols('q p', real=True)
m, l, g = sp.symbols('m l g', positive=True)

# Определение параметров линеаризованной и нелинейной моделей маятника
omega_0 = sp.Symbol(r'\omega_0', positive=True)
theta = sp.Symbol(r'\theta', real=True)
J_inert = sp.Symbol('I', positive=True)

display(Markdown("### Символьное ядро инициализировано. Включен автоматический рендеринг."))
```

```python
# Построение гамильтониана нелинейного маятника
T = p**2 / (2 * m * l**2)
V = - m * g * l * sp.cos(q)
H = T + V

display(Markdown("#### Функция Гамильтона нелинейного маятника $H(q, p)$:"))
display(sp.Eq(sp.Symbol('H(q,p)'), H))

# Вывод канонических уравнений движения
dq_dt = sp.diff(H, p)
dp_dt = -sp.diff(H, q)

display(Markdown("#### Канонические уравнения движения Гамильтона:"))
display(sp.Eq(sp.Symbol(r'\dot{q}'), dq_dt))
display(sp.Eq(sp.Symbol(r'\dot{p}'), dp_dt))

# Реализация оператора скобки Пуассона в алгебраическом соглашении
def poisson_bracket(F, G, q_var, p_var):
    """Вычисляет каноническую скобку Пуассона по координатно-строковому протоколу."""
    term_1 = sp.diff(F, p_var) * sp.diff(G, q_var)
    term_2 = sp.diff(F, q_var) * sp.diff(G, p_var)
    return sp.trigsimp(term_1 - term_2)

# Проверка закона сохранения энергии системы
display(Markdown("#### Проверка сохранения гамильтониана системы:"))
display(sp.Eq(sp.Symbol(r'\{H, H\}'), poisson_bracket(H, H, q, p)))
```

## 5. Канонические преобразования и производящие функции смешанного типа

Основная стратегия гамильтоновых асимптотических методов заключается в поиске таких преобразований фазового пространства, которые упрощают структуру правых частей дифференциальных уравнений движения. Замена старых фазовых переменных $(q, p)$ на новые $(Q, P)$ называется канонической, если она сохраняет инвариантность симплектической структуры и форму уравнений Гамильтона в новой системе переменных:
$$\frac{d}{dt}Q^i = \frac{\partial \tilde{H}}{\partial P_i}, \quad \frac{d}{dt}P_i = -\frac{\partial \tilde{H}}{\partial Q^i} \tag{12}$$

Каноничность отображения гарантируется, если связь между старыми и новыми переменными задается через производные скалярной функции $S$, которая называется производящей функцией. Для большинства задач теории возмущений применяется функция второго типа $S(q, P, t)$, аргументами которой служат старые координаты и новые импульсы. Условия каноничности в этом случае принимают вид дифференциальных соотношений:
$$p_i = \frac{\partial S}{\partial q^i}, \quad Q^i = \frac{\partial S}{\partial P_i}, \quad \tilde{H} = H + \frac{\partial S}{\partial t} \tag{13}$$

В рамках данной нотации производящая функция вида $S_0 = P q$ соответствует тождественному преобразованию фазового пространства. При подстановке функции $S_0$ в уравнения связи (13) частные производные напрямую возвращают равенство старых и новых переменных: $p = P$ и $Q = q$.

#### Теоретическое обоснование асимптотических преобразований
В нелинейных неинтегрируемых задачах невозможно найти конечное преобразование, которое полностью обнуляет возмущение или сводит гамильтониан к интегрируемому виду. Вместо этого в асимптотических методах конструируется каноническое преобразование, близкое к тождественному.

Искомый генератор $S(q, P, t)$ строится в виде формального степенного ряда по малому параметру возмущения $\varepsilon \ll 1$, где к тождественному преобразованию $S_0$ последовательно добавляются малые нелинейные поправки: $S = P q + \varepsilon S_1 + \varepsilon^2 S_2 + \dots$. Действие такого генератора деформирует координатные оси фазового пространства на величину соответствующего порядка малости. Подбор поправок $S_k$ осуществляется из условия аналитической компенсации возмущающих членов гамильтониана. Переход в данную модифицированную систему переменных позволяет перенести вклад нелинейных возмущений в высшие порядки малости, благодаря чему в новом фазовом пространстве уравнения движения аппроксимируются интегрируемыми соотношениями с наперед заданной точностью.


## 6. Алгебраические критерии каноничности преобразования

Для проверки того, является ли заданное аналитическое преобразование фазового пространства $(q, p) \to (Q, P)$ каноническим, на практике используются два эквивалентных критерия, которые не требуют явного построения производящих функций смешанного типа.

### 6.1. Критерий через базисные скобки Пуассона

Базисными скобками Пуассона называются скобки, вычисляемые непосредственно для самих новых координатных функций фазового пространства $Q^i(q, p)$ и $P_i(q, p)$ по старым каноническим переменным. 

В рамках зафиксированного протокола тождественное преобразование обязано обладать положительной симплектической метрикой. Из свойства кососимметричности оператора следует, что для каноничности отображения импульсы в скобке Пуассона должны выступать первым аргументом. 

Преобразование является каноническим тогда и только тогда, когда новые функции удовлетворяют соотношениям:
$$ \{Q^i, Q^j\}_{q,p} = 0, \quad \{P_i, P_j\}_{q,p} = 0, \quad \{P_i, Q^j\}_{q,p} = \delta_i^j \tag{14} $$

где символ $\delta_i^j$ является стандартным символом Кронекера. Выполнение критерия (14) гарантирует инвариантность скобок Пуассона для любых двух произвольных гладких функций при переходе к новой системе координат: $\{F, G\}_{q,p} = \{F, G\}_{Q,P}$.


### 6.2. Критерий в матричной форме (Симплектическое условие)
Объединим старые переменные в единый фазовый вектор $x$, а новые — в вектор $X$. Пусть матрица $\mathbf{M} = \frac{\partial X}{\partial x}$ является матрицей Якоби данного преобразования, составленной из частных производных согласно правилу подраздела 1.2.2. Преобразование является строго каноническим (унивалентным) тогда и только тогда, когда матрица Якоби удовлетворяет следующему матричному соотношению:
$$\mathbf{M}^T \mathbf{J} \mathbf{M} = \mathbf{J} \tag{15}$$
где матрица $\mathbf{J}$ является канонической кососимметрической симплектической матрицей размерности $2n \times 2n$:
$$\mathbf{J} = \begin{pmatrix} \mathbf{0} & \mathbf{I}_n \\ -\mathbf{I}_n & \mathbf{0} \end{pmatrix}$$

Матричное равенство (15) означает, что оператор преобразования координат сохраняет неизменной кососимметрическую метрику фазового пространства. Геометрически это эквивалентно сохранению фазового объема системы и инвариантов Пуанкаре вдоль траекторий движения. Данная форма критерия эффективна для проверки каноничности в символьных алгоритмах систем компьютерной алгебры, поскольку сводится к прямому перемножению трех известных матриц.

```python
display(Markdown("### Проверка каноничности преобразования"))

# Задаем производящую функцию S, генерирующую нелинейный сдвиг координат
P = sp.symbols('P', real=True)
eps = sp.symbols(r'\varepsilon', positive=True)
S = q * P + eps * (q**2) * P

# Дифференциальные связи для производящей функции второго типа
p_from_S = sp.diff(S, q)
Q_from_S = sp.diff(S, P)

display(Markdown(r"#### Связи фазовых переменных, индуцированные производящей функцией $S(q, P)$:"))
display(sp.Eq(p, p_from_S))
display(sp.Eq(sp.Symbol('Q'), Q_from_S.collect(q)))

# Нахождение явного выражения для нового импульса (извлекаем выражение из списка решений)
solutions = sp.solve(sp.Eq(p, p_from_S), P)
P_explicit = solutions[0]

# Выражаем новую координату явно через старые оси фазового пространства
Q_explicit = Q_from_S.subs(P, P_explicit)

# --- Блок 1. Проверка через базисные скобки Пуассона ---
# ИСПРАВЛЕНО: Изменен порядок аргументов (P, Q) для соответствия инварианту +1
new_bracket = poisson_bracket(P_explicit, Q_explicit, q, p)

display(Markdown("#### 1. Вычисление базисной скобки Пуассона $\{P, Q\}$ в исходных осях:"))
display(sp.Eq(sp.Symbol(r'\{P, Q\}'), new_bracket))

# --- Блок 2. Проверка через матричный симплектический критерий ---
# Формируем векторы фазовых состояний в виде матриц-столбцов
x_vec = sp.Matrix([q, p])
X_vec = sp.Matrix([Q_explicit, P_explicit])

# Вычисляем матрицу Якоби преобразования М размерности 2х2
M_mat = X_vec.jacobian(x_vec)
display(Markdown("#### 2. Матрица Якоби канонического преобразования $\mathbf{M}$:"))
display(M_mat)

# Задаем каноническую кососимметрическую симплектическую матрицу J размерности 2х2
J_mat = sp.Matrix([[0, 1], [-1, 0]])

# Вычисляем левую часть симплектического условия: M^T * J * M
symplectic_check = sp.simplify(M_mat.T * J_mat * M_mat)
display(Markdown("#### Результат вычисления матрицы $\mathbf{M}^T \mathbf{J} \mathbf{M}$:"))
display(symplectic_check)

# Алгебраическая проверка тождественности полученной матрицы с исходной матрицей J
is_symplectic = symplectic_check.equals(J_mat)
display(Markdown(f"#### Выполнение симплектического условия $\mathbf{{M}}^T \mathbf{{J}} \mathbf{{M}} = \mathbf{{J}}$: **{is_symplectic}**"))
```

## 7. Уравнение Гамильтона-Якоби

Развитием метода производящих функций является постановка задачи о полном занулении преобразованной функции Гамильтона: $\tilde{H} \equiv 0$. В силу канонических уравнений движения (12), фазовый поток в таком вырожденном пространстве полностью останавливается, а новые переменные превращаются в неподвижные канонические константы движения $Q$ и $P$:
$$\frac{d}{dt}P_i = 0 \implies P_i = \text{const}, \quad \frac{d}{dt}Q^i = 0 \implies Q^i = \text{const} \tag{16}$$

Подстановка дифференциальных связей производящей функции второго типа $p_i = \frac{\partial S}{\partial q^i}$ в условие равенства нулю нового гамильтониана порождает дифференциальное уравнение в частных производных первого порядка, которое называется уравнением Гамильтона-Якоби:
$$H\left(q^1, \dots, q^n, \frac{\partial S}{\partial q^1}, \dots, \frac{\partial S}{\partial q^n}, t\right) + \frac{\partial S}{\partial t} = 0 \tag{17}$$

Искомая полная функция действия $S(q, P, t)$ выступает здесь в роли производящей функции, связывающей текущие фазовые оси с плоскостью канонических интегралов.

Для автономных систем полная функция действия $S$ разделяется на линейную временную компоненту с сохраняющейся постоянной энергии $h(P)$ и стационарное усеченное действие $V(q, P)$:
$$S(q, P, t) = V(q, P) - h(P) \cdot t \tag{18}$$

Подстановка полного действия (18) в уравнение (17) сводит задачу к стационарному уравнению Гамильтона-Якоби для усеченного действия:
$$H\left(q, \frac{\partial V}{\partial q}\right) = h(P) \tag{19}$$

Полный интеграл стационарного уравнения (19) содержит в себе явные алгебраические уравнения траекторий, что позволяет аналитически интегрировать исходную нелинейную динамическую систему.


```python
display(Markdown("### Символьное формирование уравнения Гамильтона-Якоби"))

# Объявление усеченного действия V как неопределенной функции координаты q
V_func = sp.Function('V')(q)
h_energy = sp.Symbol('h', positive=True)  # Постоянная энергии
dV_dq = sp.diff(V_func, q)

# Подстановка p = dV/dq в точный гамильтониан нелинейного маятника H
H_HJ = H.subs(p, dV_dq)

display(Markdown("#### Стационарное уравнение Гамильтона-Якоби для нелинейного маятника:"))
display(sp.Eq(H_HJ, h_energy))

# Разрешение нелинейного уравнения относительно dV/dq
solutions = sp.solve(sp.Eq(H_HJ, h_energy), dV_dq)

# Строгий математический выбор положительной ветви корня
dV_dq_sol = None
for sol in solutions:
    coeff, _ = sol.as_coeff_Mul()
    if coeff >= 0:
        dV_dq_sol = sol
        break

display(Markdown(r"#### Точное аналитическое решение относительно дифференциала действия $p = \frac{dV}{dq}$:"))
display(sp.Eq(sp.Symbol(r'\frac{dV}{dq}'), dV_dq_sol))
```

## 8. Канонические переменные «действие-угол»

Для автономных полностью интегрируемых систем с $n$ степенями свободы, совершающих регулярные циклические или колебательные движения, существует специальный класс канонических преобразований, упрощающий явное интегрирование уравнений движения. Таким преобразованием является переход к переменным действие-угол $(\theta, I)$.

### 8.1. Построение переменных действия как инвариантов движения

Рассмотрим автономную систему, у которой компоненты фазовой траектории совершают независимые замкнутые циклы в плоскостях $(q^i, p_i)$. В рамках метода Гамильтона-Якоби полный интеграл стационарного уравнения определяется укороченным действием $V(q, P)$, где $P = (P_1, \dots, P_n)^T$ является вектором постоянных интегрирования, выполняющих роль новых импульсов.

Вместо произвольного набора констант $P$ вводится вектор переменных действия $I = (I_1, \dots, I_n)^T$. Компоненты данного вектора вычисляются как нормированная площадь, ограничиваемая фазовой кривой по каждой степени свободы. Используя связь импульса с оператором производной укороченного действия $p_i = \frac{\partial V}{\partial q^i}$, переменные действия определяются через контурные интегралы по независимым замкнутым путям $\Gamma_i$:
$$I_i = \frac{1}{2\pi} \oint_{\Gamma_i} p_i \, dq^i = \frac{1}{2\pi} \oint_{\Gamma_i} \frac{\partial V(q, P)}{\partial q^i} \, dq^i \tag{20}$$

Поскольку интегрирование в выражении (20) ведется по замкнутым траекториям координат, обобщенные координаты $q^i$ полностью исключаются. Переменные действия $I_i$ становятся функциями исключительно старых постоянных интегрирования: $I = I(P)$. В силу теоремы об обратной функции, данная связь обратима при невырожденности матрицы Якоби отображения, что позволяет выразить старые константы через канонические действия: $P = P(I)$.


### 8.2. Определение сопряженных угловых переменных

Для завершения канонического преобразования необходимо определить вектор сопряженных угловых координат $\theta = (\theta^1, \dots, \theta^n)^T$. В соответствии со свойствами производящих функций второго типа укороченное действие выражается как функция старых координат и новых импульсов действия: $V = V(q, I)$.

Тогда новые угловые координаты определяются через частные производные по формуле:
$$\theta^i = \frac{\partial V(q, I)}{\partial I_i} \tag{21}$$

### 8.3. Преобразованный гамильтониан и уравнения движения

Поскольку исходный невозмущенный гамильтониан автономной системы равен сохраняющейся энергии, после перехода к переменным действия он становится функцией, которая зависит исключительно от новых импульсов действия:
$$\tilde{H}_0 = \tilde{H}_0(I) \tag{22}$$

Канонические уравнения Гамильтона в новом фазовом пространстве принимают вид:
$$\frac{d}{dt}I_i = -\frac{\partial \tilde{H}_0}{\partial \theta^i} = 0 \implies I_i = \text{const} \tag{23}$$
$$\frac{d}{dt}\theta^i = \frac{\partial \tilde{H}_0}{\partial I_i} \equiv \omega^i(I) \implies \theta^i(t) = \omega^i(I)t + \theta^i_0 \tag{24}$$

где функции $\omega^i(I)$ являются компонентами вектора частот нелинейных колебаний динамической системы. 

Формулировка (23) и (24) показывает геометрическую суть пространства действие-угол. Интегрируемая гамильтонова система в этих переменных описывается как движение по многомерному инвариантному тору, где радиусы тора задаются неизменными переменными действия $I_i$, а движение вдоль поверхности тора происходит равномерно и прямолинейно по угловым координатам $\theta^i$ с постоянными частотами $\omega^i$. В теории возмущений этот базис является отправной точкой, так как малые возмущения гамильтониана будут деформировать геометрию данного тора.

### 8.4. Значение переменных действие-угол для теории возмущений

Интегрирование канонических уравнений (23) и (24) показывает, что в невозмущенной системе переменные действия $I$ являются строгими константами, фиксирующими инварианты амплитуды колебаний, а угловые координаты $\theta$ эволюционируют во времени абсолютно равномерно со скоростями $\omega(I)$.

При добавлении к системе нелинейного малого возмущения полный гамильтониан принимает канонический вид $H = \tilde{H}_0(I) + \varepsilon H_1(I, \theta)$, где параметр $\varepsilon \ll 1$ является малым параметром возмущения. Наличие малой добавки $\varepsilon H_1$ разрушает константы движения: действия $I$ начинают испытывать медленный дрейф, а углы $\theta$ отклоняются от равномерного вращения. Аппарат переменных действие-угол преобразует задачу к исследованию колебаний около известного точного решения, что делает его фундаментальной основой для методов усреднения, алгоритмов Делоне и теории возмущений Ли.


### 8.5. Визуализация выпрямления фазовых кривых нелинейного маятника

Ниже приведена процедура численного построения совмещенных портретов нелинейного маятника. Визуализация выполняется одновременно в исходных декартовых переменных и в координатах действие-угол.

При построении графических панелей применяется каноническое преобразование линеаризованного осциллятора, которое соответствует переходу к масштабированным полярным координатам на фазовой плоскости. В рамках такого представления замкнутые эллиптические кривые регулярных колебаний (либраций) выпрямляются и проецируются в виде горизонтальных линий, на которых значение действия остается неизменным. 

Настоящая нелинейная сепаратриса, определяющая физическую границу между режимами колебаний и вращений, при данном отображении разворачивается в волнообразный профиль. График наглядно демонстрирует, как аппарат переменных действие-угол переводит криволинейные траектории автономной системы в прямолинейные потоки, упрощая дальнейший асимптотический анализ.


```python
import matplotlib.pyplot as plt
import numpy as np

# Задаем физические параметры маятника
m_val, l_val, g_val = 1.0, 1.0, 9.81
omega_0 = np.sqrt(g_val / l_val)

# Энергетические уровни потенциала V = -mgl*cos(q)
E_min = -m_val * g_val * l_val
E_sep = m_val * g_val * l_val

# Набор энергий, строго согласованный с потенциалом
energies = [-5.0, -2.0, 0.0, 5.0, 9.0, 12.0, 15.0] 

# Сетка угловой переменной, центрированная относительно нуля
theta_grid = np.linspace(-np.pi, np.pi, 400)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# ==========================================
# ПАНЕЛЬ 1: Фазовый портрет в декартовых переменных (q, p)
# ==========================================
q_grid = np.linspace(-2 * np.pi, 2 * np.pi, 300)
p_grid = np.linspace(-7.0, 7.0, 300)
Q_mesh, P_mesh = np.meshgrid(q_grid, p_grid)

# Вычисление функции Гамильтона по базовой формуле
H_mesh = P_mesh**2 / (2 * m_val * l_val**2) - m_val * g_val * l_val * np.cos(Q_mesh)

# Траектории и точная сепаратриса на уровне E = E_sep
ax1.contour(Q_mesh, P_mesh, H_mesh, levels=energies, colors='royalblue', linewidths=1.8)
ax1.contour(Q_mesh, P_mesh, H_mesh, levels=[E_sep], colors='crimson', linewidths=2.5, linestyles='--')

ax1.set_title("Фазовый портрет в исходных переменных $(q, p)$", fontsize=12, pad=10)
ax1.set_xlabel("Координата $q$ (угол, рад)", fontsize=11)
ax1.set_ylabel("Канонический импульс $p$", fontsize=11)
ax1.axhline(0, color='black', linewidth=0.8, linestyle=':')
ax1.axvline(0, color='black', linewidth=0.8, linestyle=':')
ax1.set_xlim(-2 * np.pi, 2 * np.pi)
ax1.set_ylim(-7.0, 7.0)
ax1.grid(True, linestyle=":", alpha=0.5)

ax1.text(0, 1.5, "Колебания (Либрация)", color="royalblue", fontsize=10, ha="center")
ax1.text(0, 5.5, "Вращение (Ротация)", color="royalblue", fontsize=10, ha="center")
ax1.text(np.pi, 0.4, "Седло", color="crimson", fontsize=10, ha="center")


# ==========================================
# ПАНЕЛЬ 2: Фазовый портрет в переменных действие-угол (I, theta)
# ==========================================
# Отбираем только колебательные режимы (E < E_sep)
oscillating_energies = [E for E in energies if E < E_sep]

for E in oscillating_energies:
    E_shift = E - E_min
    
    # Расчет действия I из квадратичной зависимости векового гамильтониана
    a_coeff = -1 / (16 * m_val * l_val**2)
    b_coeff = omega_0
    c_coeff = -E_shift
    
    I_val = (-b_coeff + np.sqrt(b_coeff**2 - 4 * a_coeff * c_coeff)) / (2 * a_coeff)
    ax2.plot(theta_grid, np.full_like(theta_grid, I_val), color="forestgreen", linewidth=2.2)

# Истинное инвариантное значение действия на сепаратрисе нелинейного маятника
I_separatrix_exact = (8 / np.pi) * m_val * l_val**2 * omega_0

# Отрисовка точной сепаратрисы в виде истинной инвариантной горизонтальной границы
ax2.axhline(I_separatrix_exact, color="crimson", linestyle="--", linewidth=2.5, label="Сепаратриса (Предел выпрямления кривых)")

ax2.set_title(r"Фазовый портрет в переменных действие-угол $(\theta, I)$", fontsize=12, pad=10)
ax2.set_xlabel(r"Угловая переменная $\theta$ (фаза, рад)", fontsize=11)
ax2.set_ylabel(r"Переменная действия $I$ (инвариант)", fontsize=11)
ax2.set_xlim(-np.pi, np.pi)
ax2.set_ylim(0, I_separatrix_exact * 1.2)
ax2.set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
ax2.set_xticklabels([r'$-\pi$', r'$-\pi/2$', '0', r'$\pi/2$', r'$\pi$'])
ax2.grid(True, linestyle=":", alpha=0.5)
ax2.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

#### 8.5.1. Сравнительный анализ топологии фазовых портретов и их асимптотический смысл

Сопоставление левой и правой графических панелей раскрывает геометрический смысл канонических преобразований и процедур нормализации нелинейных гамильтоновых систем.

**1. Неоднородность исходного пространства фазовых переменных**
В физических декартовых координатах $(q, p)$ структура потока является существенно неоднородной. На левой панели прослеживаются три качественно различающихся динамических режима системы:
* **Либрация (колебания)**. Маятник совершает замкнутые периодические движения около устойчивого центра равновесия $(0,0)$. Вследствие нелинейности возвращающей силы форма эллиптических траекторий прогрессивно искажается по мере роста энергии.
* **Ротация (вращение)**. При уровнях энергии выше критического маятник совершает полные обороты в конфигурационном пространстве. Фазовая скорость при этом периодически колеблется, а канонический импульс никогда не пересекает нулевую отметку.
* **Сепаратриса**. Данная траектория формирует физическую границу, разделяющую колебательный и вращательный режимы движения материальной точки. Период движения по сепаратрисе стремится к бесконечности, а сама кривая асимптотически приближается к неустойчивым седловым точкам.

**2. Геометрическая нормализация в переменных действие-угол**
Правая панель наглядно иллюстрирует математическую суть процедур нормализации нелинейных гамильтоновых систем. Канонический переход к переменным действие-угол перестраивает геометрию фазового пространства, устраняя кривизну траекторий:
* Каждая искаженная нелинейная траектория либрации преобразуется в строго горизонтальную прямую линию. Переменная действия становится точным инвариантом системы, на котором выполняется тождество $I = \text{const}$.
* Переменные ускорения и замедления обобщенной координаты $q$ трансформируются в монотонное и абсолютно равномерное изменение угловой координаты $\theta$ на периоде от $-\pi$ до $\pi$ с постоянной нелинейной скоростью $\omega(I)$.

**3. Значение для асимптотических методов теории возмущений**
Рассмотренный графический эксперимент наглядно поясняет фундаментальную цель построения асимптотических рядов и генераторов теории возмущений. Главная задача любого метода теории возмущений заключается в том, чтобы взять реальную сложную систему с нелинейным фазовым потоком, аналитически учесть вклад нелинейных возмущений и принудительно преобразовать её фазовые кривые в прямолинейные траектории законов сохранения. Обладая такой выпрямленной картой фазового пространства, исследователь получает возможность мгновенно анализировать долговременную эволюцию нелинейной системы на больших временных интервалах, полностью исключая необходимость пошагового численного интегрирования исходных дифференциальных уравнений движения.


## 9. Переменные Делоне в гамильтоновой небесной механике

### 9.1. Введение и физическая мотивация

В классической задаче двух тел движение традиционно описывается геометрическими элементами Кеплера, которые определяют форму, пространственную ориентацию и текущее положение материальной точки на орбите:
* $a$ - большая полуось,
* $e$ - эксцентриситет,
* $i$ - наклонение плоскости орбиты,
* $\Omega$ - долгота восходящего узла,
* $\omega$ - аргумент перицентра,
* $M$ - средняя аномалия.

При исследовании возмущенного движения, обусловленного гравитационным влиянием третьих тел или несферичностью центральной планеты, классические уравнения Лагранжа для элементов орбит оказываются алгебраически громоздкими. Для применения аппарата канонических преобразований и гамильтоновых асимптотических методов необходимо перейти к канонически сопряженным переменным фазового пространства.

Переменные Делоне представляют собой канонический набор переменных типа действие-угол. Физическая мотивация их введения заключается в разделении геометрии орбиты и законов сохранения, выполняющихся в невозмущенной задаче двух тел. Канонические переменные Делоне имеют следующую структуру:
1. **Ковектор действий** $I = (L, G, H_{Del})$ является элементом сопряженного пространства. Компоненты данного ковектора обладают нижней индексной структурой и напрямую связаны с интегралами движения невозмущенной задачи, включая полную механическую энергию и компоненты вектора орбитального момента импульса.
2. **Вектор сопряженных углов** $\theta = (l, g, h)^T$ является элементом исходного пространства векторов. Компоненты данного вектора обладают верхней индексной структурой и описывают пространственную ориентацию орбиты и текущее положение космического аппарата на ней.


### 9.2. Связь переменных Делоне с элементами Кеплера

Пусть $\mu$ является гравитационным параметром центрального тела. Переменные Делоне разделяются на три пары канонически сопряженных координат и импульсов, где роли импульсов выполняют компоненты ковектора действий $I$, а роли координат — компоненты вектора угловых переменных $\theta$.

#### Компоненты ковектора действий (Канонические импульсы)

1. **Главное действие $L$**. Данная переменная определяет полную механическую энергию невозмущенного кеплеровского движения:
   $$L = \sqrt{\mu a} \tag{25}$$
   Физический смысл компоненты $L$ заключается в том, что её величина прямо пропорциональна квадратному корню из большой полуоси. Масштабирование размеров орбиты при сохранении её геометрической формы изменяет только координату $L$.

2. **Орбитальное действие $G$**. Данная переменная задает модуль вектора полного орбитального момента импульса космического аппарата:
   $$G = L \sqrt{1 - e^2} = \sqrt{\mu a (1 - e^2)} \tag{26}$$
   Физический смысл компоненты $G$ состоит в том, что при фиксированной большой полуоси ($L = \text{const}$) изменение эксцентриситета $e$ эквивалентно изменению $G$. Для круговой орбиты ($e = 0$) значение момента является максимальным и выполняется равенство $G = L$. Для предельного случая падения на центр ($e \to 1$) орбитальный момент стремится к нулю: $G \to 0$.

3. **Проекционное действие $H_{Del}$**. Данная переменная определяет проекцию вектора орбитального момента импульса на полярную ось $Z$ выбранной инерциальной системы координат:
   $$H_{Del} = G \cos i = \sqrt{\mu a (1 - e^2)} \cos i \tag{27}$$
   Физический смысл компоненты $H_{Del}$ заключается в характеризации пространственного наклона плоскости орбиты. Для экваториальных прямых орбит ($i = 0^\circ$) проекция совпадает с модулем вектора момента: $H_{Del} = G$. Для полярных орбит ($i = 90^\circ$) вектор момента перпендикулярен оси проецирования, откуда следует зануление компоненты: $H_{Del} = 0$.

Каждому каноническому импульсу сопряжена своя угловая координата:

1. **Средняя аномалия $l$**. Данная переменная сопряжена с главным действием $L$:
   $$l = M \tag{28}$$
   Физический смысл компоненты $l$ заключается в том, что она представляет собой быструю фазовую переменную, которая описывает положение космического аппарата непосредственно на эллиптической траектории.

2. **Аргумент перицентра $g$**. Данная переменная сопряжена с орбитальным действием $G$:
   $$g = \omega \tag{29}$$
   Физический смысл компоненты $g$ состоит в том, что она задает угловое положение перицентра в плоскости орбиты, отсчитываемое от линии восходящего узла.

3. **Долгота восходящего узла $h$**. Данная переменная сопряжена с проекционным действием $H_{Del}$:
   $$h = \Omega \tag{30}$$
   Физический смысл компоненты $h$ заключается в том, что она задает пространственный разворот плоскости орбиты вокруг полярной оси $Z$ относительно фиксированной оси абсцисс инерциальной системы координат.



### 9.3. Гамильтониан и уравнения движения невозмущенной задачи

В невозмущенной задаче двух тел полная механическая энергия системы на эллиптической орбите является постоянной величиной и зависит исключительно от большой полуоси $a$:
$$E = -\frac{\mu}{2a} \tag{31}$$

Из определения главного действия Делоне $L = \sqrt{\mu a}$ следует алгебраическая связь геометрического размера орбиты с каноническим импульсом:
$$a = \frac{L^2}{\mu} \tag{32}$$

Подстановка выражения (32) в формулу энергии (31) определяет явный вид невозмущенного гамильтониана в переменных Делоне:
$$H_0(L) = -\frac{\mu^2}{2L^2} \tag{33}$$

Функция Гамильтона (33) зависит только от одного канонического импульса $L$ и полностью вырождена по двум остальным действиям $G$ и $H_{Del}$, что отражает циклическую симметрию задачи Кеплера.

#### Анализ структуры функции Гамильтона

1. **Вырождение системы**. Независимость невозмущенного гамильтониана $H_0$ от импульсов $G$ и $H_{Del}$ отражает факт того, что полная механическая энергия материальной точки в центральном поле не зависит ни от формы эллиптической траектории, ни от пространственной ориентации плоскости её орбиты.
2. **Циклические переменные**. Поскольку все три угловые координаты $l$, $g$ и $h$ явно не входят в выражение для гамильтониана, они являются циклическими координатами. Это гарантирует строгое сохранение сопряженных им импульсов действия $L$, $G$ и $H_{Del}$.

Система уравнений движения в переменных Делоне для невозмущенной задачи двух тел принимает вид дифференциальных уравнений для ковектора действий:
$$\frac{d}{dt}L = 0, \quad \frac{d}{dt}G = 0, \quad \frac{d}{dt}H_{Del} = 0 \implies I = \text{const} \tag{34}$$

Уравнения движения для вектора сопряженных угловых переменных вычисляются через частные производные гамильтониана:
$$\frac{d}{dt}l = \frac{\partial H_0(L)}{\partial L} = \frac{\mu^2}{L^3} \equiv n(L), \quad \frac{d}{dt}g = \frac{\partial H_0(L)}{\partial G} = 0, \quad \frac{d}{dt}h = \frac{\partial H_0(L)}{\partial H_{Del}} = 0 \tag{35}$$

где скалярная функция $n(L) = \sqrt{\mu / a^3}$ является средним движением, определяющим кеплерову частоту обращения космического аппарата по орбите. Интегрирование полученной системы уравнений по времени дает точное аналитическое решение:
$$l(t) = n(L) \cdot (t - t_0) + l_0, \quad g = \text{const}, \quad h = \text{const} \tag{36}$$

Интегрирование системы уравнений (34) и (35) показывает, что геометрическая форма эллипса и пространственная ориентация плоскости орбиты остаются неизменными во времени константами, а движение материальной точки происходит равномерно по средней аномалии $l$.


### 9.4. Переход к анализу возмущенного движения в переменных Делоне

Если на систему действует слабое внешнее возмущение, обусловленное полярным сжатием центральной планеты или гравитационным полем третьих тел, полный гамильтониан принимает вид:
$$H(I, \theta) = H_0(L) + \mathcal{R}(L, G, H_{Del}, l, g, h) \tag{37}$$
где функция $\mathcal{R}(I, \theta)$ является возмущающей функцией динамической системы, а малый множитель нелинейного поля скрыт внутри её структуры.

В силу канонических уравнений движения уравнения движения возмущенной задачи принимают вид системы для ковектора действий:
$$\frac{d}{dt}L = -\frac{\partial \mathcal{R}}{\partial l}, \quad \frac{d}{dt}G = -\frac{\partial \mathcal{R}}{\partial g}, \quad \frac{d}{dt}H_{Del} = -\frac{\partial \mathcal{R}}{\partial h} \tag{38}$$

Уравнения движения для вектора сопряженных угловых переменных усложняются за счет добавления частных производных от функции возмущения:
$$\frac{d}{dt}l = n(L) + \frac{\partial \mathcal{R}}{\partial L}, \quad \frac{d}{dt}g = \frac{\partial \mathcal{R}}{\partial G}, \quad \frac{d}{dt}h = \frac{\partial \mathcal{R}}{\partial H_{Del}} \tag{39}$$

Поскольку функция возмущения $\mathcal{R}$ чаще всего является периодической функцией по отношению к быстрой средней аномалии $l$, переменные Делоне эффективно используются в гамильтоновых методах усреднения. Это позволяет аналитически исключить быструю переменную и замкнуть уравнения движения относительно медленно эволюционирующих элементов $L$, $G$, $H_{Del}$, $g$ и $h$, описывающих долгосрочную форму и ориентацию орбиты.

*Примечание для асимптотических методов (Проблема вырождения и малых знаменателей)*. Структура невозмущенных частот в уравнениях (35) показывает, что гамильтониан $H_0(L)$ зависит только от одного действия $L$, из-за чего невозмущенные частоты медленных углов перицентра и узла тождественно равны нулю. Такая гамильтонова система называется предельно вырожденной. При включении возмущения $\mathcal{R}$ частоты медленных углов становятся отличными от нуля, но остаются величинами порядка малого параметра по сравнению с кеплеровской частотой $n(L)$.

При разложении функции возмущения в тригонометрические ряды Фурье и попытке аналитически исключить быстрые переменные, в знаменателях асимптотических рядов теории возмущений возникают линейные комбинации частот вида $k_1 n + k_2 \frac{d}{dt}g + k_3 \frac{d}{dt}h$, где коэффициенты $k_i$ являются целыми числами. Близость данных знаменателей к нулю порождает проблему малых знаменателей или орбитальные резонансы. В силу предельной вырожденности кеплеровской задачи классическая теория невырожденных торгов Колмогорова здесь неприменима. Преодоление этого вызова и доказательство сходимости рядов для вырожденных систем спутниковой динамики является фундаментальным результатом В.И. Арнольда, который ввел специальное условие сопряженной невырожденности, ставшее ключевой частью современной КАМ-теории [Арнольд, 2024].

### 9.5. Модифицированные переменные Пуанкаре

Классические переменные Делоне вырождаются в двух важных для задач небесной механики случаях:
1. **Круговые орбиты ($e = 0$)**. Положение перицентра становится неопределенным, вследствие чего угловая координата $g = \omega$ теряет физический смысл, а значение орбитального действия вырождается в главное: $G = L$.
2. **Экваториальные орбиты ($i = 0^\circ$)**. Линия восходящего узла становится неопределенной, угловая координата $h = \Omega$ теряет физический смысл, а проекционное действие вырождается в орбитальное: $H_{Del} = G$.

Для устранения данных особенностей Анри Пуанкаре ввел модифицированные канонические переменные, которые остаются регулярными при нулевых значениях эксцентриситета и наклонения плоскости орбиты. Переход к элементам Пуанкаре осуществляется через канонические деформации, разделяющие фазовое пространство на три сопряженные пары:

* **Главные канонические элементы (пара действие-угол)**:
  $$ \Lambda = L, \quad \lambda = l + g + h = M + \omega + \Omega \tag{40} $$
  где координата $\lambda$ является средней долготой космического аппарата на орбите.

* **Эксцентрические канонические элементы (пара координата-импульс)**:
  $$ \xi = \sqrt{2(L - G)}\cdot\sin(g+h), \quad \eta = \sqrt{2(L - G)}\cdot\cos(g+h) \tag{41} $$
  где $\xi$ выступает в роли канонической координаты, а $\eta$ является сопряженным ей каноническим импульсом. При стремительном приближении эксцентриситета к нулю ($e \to 0$) обе переменные плавно и регулярно зануляются.

* **Наклонные канонические элементы (пара координата-импульс)**:
  $$ q_P = \sqrt{2(G - H_{Del})}\cdot\sin h, \quad p_{P\_val} = \sqrt{2(G - H_{Del})}\cdot\cos h \tag{42} $$
  где $q_P$ выступает в роли канонической координаты, а $p_{P\_val}$ является сопряженным ей каноническим импульсом. При движении по экваториальной траектории ($i \to 0^\circ$) данные элементы регулярно зануляются.

Переменные Пуанкаре $(\lambda, \xi, q_P, \Lambda, \eta, p_{P\_val})$ полностью сохраняют свойство каноничности исходного фазового пространства. Их уравнения движения регулярны в окрестности круговых и экваториальных орбит, что делает данный базис основным инструментом для построения аналитических теорий возмущений в задачах динамики околоземных искусственных спутников Земли и планетных систем.


### 9.6 Перевод элементов Кеплера в переменные Делоне

```python
import numpy as np

def kepler_to_delaunay(a, e, i, omega, w, M, mu=1.0):
    """
    Перевод элементов Кеплера в переменные Делоне.
    Углы (i, omega, w, M) передаются в радианах.
    """
    # Вычисление компонентов ковектора действий
    L = np.sqrt(mu * a)
    G = L * np.sqrt(1 - e**2)
    H_del = G * np.cos(i)
    
    # Компоненты вектора угловых переменных
    l = M % (2 * np.pi)
    g = w % (2 * np.pi)
    h = omega % (2 * np.pi)
    
    return L, G, H_del, l, g, h

def delaunay_to_kepler(L, G, H_del, l, g, h, mu=1.0):
    """
    Перевод переменных Делоне в элементы Кеплера.
    Возвращает углы в радианах.
    """
    a = L**2 / mu
    
    # Ограничение аргумента для исключения погрешностей округления
    cos_i = H_del / G if G != 0 else 1.0
    cos_i = np.clip(cos_i, -1.0, 1.0)
    i = np.arccos(cos_i)
    
    e_sq = 1.0 - (G / L)**2 if L != 0 else 0.0
    e = np.sqrt(max(0.0, e_sq))
    
    M = l % (2 * np.pi)
    w = g % (2 * np.pi)
    omega = h % (2 * np.pi)
    
    return a, e, i, omega, w, M

# Демонстрация работы на тестовых данных спутника Земли
mu_earth = 398600.4415 # км^3 / с^2
a_test, e_test, i_test = 7000.0, 0.05, np.radians(45.0)
omega_test, w_test, M_test = np.radians(30.0), np.radians(60.0), np.radians(15.0)

print("=== Исходные элементы Кеплера ===")
print(f"a = {a_test} км, e = {e_test}, i = {np.degrees(i_test)} градусов")

# Прямой переход к переменным действие-угол
L, G, H_del, l, g, h = kepler_to_delaunay(a_test, e_test, i_test, omega_test, w_test, M_test, mu=mu_earth)
print("\n=== Полученные переменные Делоне ===")
print(f"L (главное действие)               = {L:.4f}")
print(f"G (орбитальное действие)            = {G:.4f}")
print(f"H_del (проекционное действие)       = {H_del:.4f}")
print(f"l (средняя аномалия)               = {np.degrees(l):.2f} градусов")
print(f"g (аргумент перицентра)            = {np.degrees(g):.2f} градусов")
print(f"h (долгота восходящего узла)       = {np.degrees(h):.2f} градусов")

# Обратный переход для проверки инвариантности преобразования
a_rev, e_rev, i_rev, omega_rev, w_rev, M_rev = delaunay_to_kepler(L, G, H_del, l, g, h, mu=mu_earth)
print("\n=== Обратный пересчет в элементы Кеплера для проверки ===")
print(f"a = {a_rev:.4f} км, e = {e_rev:.4f}, i = {np.degrees(i_rev):.2f} градусов")
```

## 10. Асимптотический расчет нелинейного сдвига частоты маятника с помощью SymPy

В линейном приближении при малых углах отклонения $|q| \ll 1$ математический маятник совершает гармонические колебания с постоянной частотой $\omega_0 = \sqrt{g/l}$, которая не зависит от амплитуды. При увеличении энергии системы проявляется ангармонизм, приводящий к зависимости частоты от размаха колебаний.

Ниже приведено теоретическое обоснование для кода символьных вычислений, выполняющего аналитическое усреднение гамильтониана по быстрой фазе в первом порядке теории возмущений.

---

### 10.1. Аппроксимация функции Гамильтона

Полный гамильтониан маятника массы $m$ и длины $l$ в однородном поле тяжести $g$ выражается через обобщенную координату угол $q$ и сопряженный импульс $p$:
$$H(q, p) = \frac{p^2}{2ml^2} - mgl\cos(q) \tag{43}$$

Разлагая потенциальную энергию в ряд Тейлора по координате $q$ до четвертого порядка малости $O(q^4)$ и отбрасывая константу потенциальной энергии $-mgl$, получаем полиномиальную аппроксимацию:
$$H_{approx} = \frac{p^2}{2ml^2} + \frac{1}{2}mglq^2 - \frac{1}{24}mglq^4 \tag{44}$$

---

### 10.2. Каноническое преобразование к переменным действие-угол

Для выделения быстрой фазы угла $\theta$ и инварианта движения действия $I$ применяется каноническое преобразование линеаризованного осциллятора, связывающее старые оси с новыми переменными:
$$q = \sqrt{\frac{2I}{m\omega_0 l^2}} \sin\theta \tag{45}$$
$$p = \sqrt{2Im\omega_0 l^2} \cos\theta \tag{46}$$

Подстановка формул (45) и (46) в первые два члена аппроксимированного гамильтониана (44) переводит невозмущенную линейную часть задачи в каноническую форму $\tilde{H}_0(I) = \omega_0 I$, полностью устраняя зависимость от угла на уровне линейного приближения. Введение тригонометрических подстановок в нелинейный член четвертого порядка порождает функцию возмущения, зависящую от фазы, которая далее подлежит асимптотическому усреднению.

### 10.3. Метод усреднения по быстрой фазе

После подстановки соотношений (45) и (46) гамильтониан становится явной функцией $H(I, \theta)$. Поскольку нелинейное возмущение мало, координата фазового угла $\theta$ эволюционирует значительно быстрее, чем действие $I$. Это позволяет применить метод усреднения Крылова-Боголюбова, представляющий собой первое приближение гамильтоновой теории возмущений, и интегрировать систему по быстрой невозмущенной фазе $\theta$ на периоде от $0$ до $2\pi$:
$$\tilde{H}(I) = \frac{1}{2\pi} \int_{0}^{2\pi} H_{approx}(I, \theta) \, d\theta \tag{47}$$

При интегрировании члены с нечетными степенями тригонометрических функций зануляются, а быстро осциллирующие гармоники отбрасываются, формируя преобразованный вековой гамильтониан $\tilde{H}(I)$.

#### Обоснование операции усреднения
Применение оператора интегрирования по быстрой циклической фазе базируется на принципе разделения временных масштабов. В гамильтоновых системах ближнего к интегрируемому вида фазовое пространство расслаивается на быстрые движения, описываемые эволюцией сопряженных угловых координат $\theta$ с частотами $\omega(I)$, и медленные движения, описываемые эволюцией векторов действия $I$ под влиянием малого возмущения порядка $O(\varepsilon)$.

За один полный цикл быстрой фазы медленные переменные практически не изменяют своих значений. С точки зрения долгосрочной эволюции система усредняется по ансамблю быстрых состояний. Математически это эквивалентно проецированию истинного фазового потока на инвариантное многообразие медленных движений. Такая процедура позволяет исключить короткопериодические колебания и анализировать вековую динамику системы в замкнутой форме без потери точности в главном порядке по малому параметру.


### 10.4. Вычисление преобразованной нелинейной частоты

Для исключения параметров физической системы выразим ускорение свободного падения через собственную частоту линейных колебаний: $g = l \omega_0^2$.

Согласно каноническим уравнениям Гамильтона (24), преобразованная нелинейная частота колебаний маятника $\omega(I)$, учитывающая эффект замедления из-за роста амплитуды, вычисляется как прямая производная усредненного векового гамильтониана по каноническому импульсу действия:
$$\omega(I) = \frac{d\tilde{H}(I)}{dI} \tag{48}$$

```python
def run_asymptotic_calculation():
    import sympy as sp
    from IPython.display import display, Markdown
    
    # Объявляем символы строго локально внутри изолированной функции
    q, p, m, l, g, I, theta, omega_0 = sp.symbols('q p m l g I theta omega_0')
    
    display(Markdown("### Асимптотический расчет нелинейного сдвига частоты маятника"))

    # Полиномиальная аппроксимация гамильтониана (Формула 44)
    H_approx = p**2 / (2 * m * l**2) + sp.Rational(1, 2) * m * g * l * q**2 - sp.Rational(1, 24) * m * g * l * q**4

    # Каноническое преобразование линеаризованного осциллятора (Формулы 45-46)
    q_sub = sp.sqrt(2 * I / (m * omega_0 * l**2)) * sp.sin(theta)
    p_sub = sp.sqrt(2 * I * m * omega_0 * l**2) * sp.cos(theta)

    # Интегральное усреднение гамильтониана по быстрой фазе
    H_mapped = sp.expand(H_approx.subs({q: q_sub, p: p_sub}))
    H_tilde = (1 / (2 * sp.pi)) * sp.integrate(H_mapped, (theta, 0, 2 * sp.pi))

    # Исключение параметров поля через собственную частоту линейных колебаний
    H_tilde = sp.simplify(H_tilde.subs(g, l * omega_0**2))
    omega_nonlinear = sp.diff(H_tilde, I)

    display(Markdown(r"#### Преобразованный вековой гамильтониан первого приближения $\tilde{H}(I)$:"))
    display(sp.Eq(sp.Symbol(r'\tilde{H}(I)'), H_tilde))

    display(Markdown(r"#### Преобразованная нелинейная частота колебаний $\omega(I)$:"))
    display(sp.Eq(sp.Symbol(r'\omega(I) = \frac{d\tilde{H}}{dI}'), omega_nonlinear))

# Запуск изолированного асимптотического расчета
run_asymptotic_calculation()
```

### 10.5 Анализ результатов символьного моделирования маятника

Символьные вычисления в коде полностью подтверждают аналитические основы канонического формализма:

1. **Канонические уравнения движения**. Выполнено символьное дифференцирование функции Гамильтона. Для обобщенной скорости получается выражение $\frac{d}{dt}q = \frac{p}{ml^2}$, а для изменения импульса во времени — возвращающая нелинейная сила $\frac{d}{dt}p = -mgl \sin q$, что покомпонентно воспроизводит физику нелинейного маятника.
2. **Инвариантность скобок Пуассона**. Стационарность системы в совокупности с кососимметрическим тождеством $\{H, H\} \equiv 0$ строго доказывает закон сохранения полной механической энергии консервативной системы. Базисная скобка Пуассона координат по импульсам в принятом алгебраическом соглашении принимает значение $\{q, p\} = -1$, верифицируя каноническую структуру фазового пространства.
3. **Вывод нелинейной частоты**. После дифференцирования преобразованного векового гамильтониана первого приближения $\tilde{H}(I)$ по каноническому импульсу действия библиотека SymPy выдает точное аналитическое выражение для частоты:
   $$ \omega(I) = \omega_0 - \frac{I}{8 m l^2} \tag{49} $$
   Физический смысл поправки в формуле (49) заключается в том, что знак минус перед переменной действия $I$ доказывает ангармонический эффект, при котором с ростом амплитуды колебаний реальный нелинейный маятник начинает замедляться, а период его колебаний увеличивается, отклоняясь от идеальной линейной частоты $\omega_0$. Данный результат закладывает основу для исследования нелинейных резонансов в возмущенных динамических системах.
