# Введение в асимптотические методы аналитической динамики

Асимптотические методы теории возмущений представляют собой основной математический аппарат качественного и количественного анализа неинтегрируемых систем дифференциальных уравнений, возникающих в небесной механике и динамике искусственных спутников Земли (ИСЗ) [Дубошин, 1968]. Настоящий раздел посвящен сравнительному анализу трех подходов к построению приближенных решений: метода усреднения Крылова-Боголюбова-Митропольского, метода фон Цейпеля и метода Депри-Хори, основанного на рядах Ли.

---

## 1. Глоссарий теории возмущений

Приведем определения понятий качественной и асимптотической механики, описывающих структуру возмущенных фазовых потоков:

*   **Асимптотические поправки различных порядков малости ($O(\varepsilon^k)$)**: Скалярные потенциалы или вектор-функции, формирующие приближенное решение в виде степенного ряда по малым степеням параметра возмущения $\varepsilon$. Поправка первого порядка ($O(\varepsilon)$) описывает линейный отклик системы на возмущающее воздействие, поправка второго порядка ($O(\varepsilon^2)$) учитывает нелинейное взаимодействие поправок первого приближения [Пуанкаре, 1971].
*   **Вековые члены (типа $t^n \sin(k_i \theta^i + \alpha), \, t^n$)**: Функции времени, амплитуда которых монотонно возрастает на бесконечном интервале интегрирования ($t \to \infty$). Появление вековых членов в прямых асимптотических разложениях нелинейных систем обусловлено фиксацией частот невозмущенного движения и приводит к потере сходимости рядов на длительных интервалах времени, что нарушает законы сохранения в консервативных системах [Арнольд, 2024].
*   **Быстрые и медленные переменные**: Разделение фазового пространства по временным масштабам эволюции. Быстрые переменные (контравариантные углы: средняя аномалия $\theta^1$, аргумент широты $\theta^2$) совершают полный оборот за период кеплерова обращения тела. Медленные переменные (ковариантные импульсы действия $I_1$, эксцентриситет $e$, аргумент перицентра $g_1$) испытывают малые флуктуации порядка $O(\varepsilon)$ на одном витке и претерпевают изменения на больших интервалах времени [Брумберг, 1980].
*   **Оператор усреднения ($\langle \dots \rangle$)**: Линейный интегральный оператор, проецирующий полное фазовое пространство на подпространство медленных переменных путем исключения быстрых фаз на их периоде $[0, 2\pi]$ [von Zeipel, 1916]:
    $$\langle f \rangle = \frac{1}{2\pi}\int_0^{2\pi} f(I, \theta)\,d\theta^1$$
*   **Двухчастотное (многочастотное) усреднение**: Метод исключения нескольких независимых быстрых фаз (например, средних аномалий спутника $l_1$ и возмущающего тела $l_2$) на их фундаментальном тороидальном периоде $[0, 2\pi] \times [0, 2\pi]$ [Брумберг, 1980]. Процедура выполняется либо последовательно (поочередным интегрированием по каждой фазе), либо одновременно, если частоты несоизмеримы. Двухчастотное усреднение позволяет полностью изолировать вековую эволюцию системы от короткопериодических пространственных флуктуаций [Лидов, 1961].
*   **Усечение асимптотического ряда**: Процедура принудительного ограничения бесконечного асимптотического разложения членами заданного порядка малости (например, до $O(\varepsilon^2)$). Из-за расходимости рядов Пуанкаре усечение на оптимальном шаге является единственным способом получения содержательного численного прогноза, однако в неканонических методах усечение рядов на конечном шаге приводит к нарушению симплектической геометрии и появлению фиктивных диссипативных эффектов [Боголюбов, Митропольский, 1955; Арнольд, 2024].
*   **Гомологическое уравнение**: Линейное дифференциальное уравнение в частных производных, возникающее в каждом порядке теории возмущений. Оно связывает известную возмущающую функцию, неизвестную поправку к преобразованному гамильтониану и неизвестную производящую функцию (или генератор Ли) канонического преобразования [Пуанкаре, 1971; von Zeipel, 1916].
*   **Скобка Пуассона** - билинейная кососимметрическая операция, задающая структуру алгебры Ли на многообразии гладких функций фазового пространства. В канонических переменных, где обобщенные координаты пишутся с верхним индексом ($q^i$), а сопряженные импульсы - с нижним ($p_i$), скобка Пуассона двух функций $f$ и $g$ определяется как [Hori, 1966; Deprit, 1969]:
    $$\{f, g\} = \sum_{i=1}^n \left( \frac{\partial f}{\partial p_i}\frac{\partial g}{\partial q^i} - \frac{\partial f}{\partial q^i}\frac{\partial g}{\partial p_i} \right)$$
    Данный выбор знаков в определении скобки является стандартным для операторных подходов теории возмущений (методов Депри-Хори). Он обеспечивает наиболее удобную структуру рекуррентных соотношений при построении канонических преобразований с помощью рядов Ли, минимизируя появление паразитных отрицательных знаков в высших порядках приближения по малому параметру [Hori, 1966; Deprit, 1969]. В силу инвариантности относительно канонических замен, скобка сохраняет свой вид и при переходе к переменным «действие — угол» ($I_i, \theta^i$).
*   **Малые знаменатели**: Аналитическая проблема теории возмущений, возникающая в резонансных или близких к резонансу зонах фазового пространства. При интегрировании уравнений поправки в знаменателях поправочных членов явных рядов возникают линейные комбинации частот невозмущенного движения вида $(\mathbf{k} \cdot \mathbf{\omega}) = k_1 \omega^1 + k_2 \omega^2 + \dots$. Если частоты соизмеримы или близки к соизмеримости, этот знаменатель стремится к нулю, что приводит к неограниченному росту амплитуды поправок и потере сходимости асимптотических рядов [Пуанкаре, 1971; Новиков, Тайманов, 2005].
*   **Инвариантный тор**: Гладкое компактное подмногообразие в фазовом пространстве интегрируемых или близких к ним гамильтоновых систем, на поверхности которого фазовые траектории совершают обмотки, характеризующиеся независимыми частотами [Арнольд, 2024].
*   **Квазипериодическое движение**: Режим движения системы на инвариантном торе, при котором отношение частот является иррациональным числом. Траектория потока не замыкается, не пересекает себя и плотно (эргодически) заполняет поверхность тора при $t \to \infty$ [Чириков, 1979].
*   **Сечение Пуанкаре**: Метод редукции размерности, заключающийся в фиксации в многомерном фазовом пространстве плоскости среза и нанесении на плоскость точек последовательного сквозного пересечения этой плоскости траекториями движения [Арнольд, 2024].

## 2. Постановка задачи и исторические предпосылки

Рассматривается возмущенная система дифференциальных уравнений, правые части которой содержат малый параметр $\varepsilon \ll 1$. В канонических переменных «действие-угол» $(I, \theta)$ невозмущенная задача предполагается интегрируемой [Маркеев, 1999]:

$$\frac{dI}{dt} = \varepsilon \frac{\partial V}{\partial \theta}, \quad \frac{d\theta}{dt} = \omega(I) + \varepsilon \frac{\partial V}{\partial I}$$

где $V(I, \theta)$ — возмущающая функция. 

Применение классического метода Пуанкаре (разложения решения в ряды по степеням $\varepsilon$ при фиксированных частотах) приводит к появлению в высших приближениях вековых членов вида $t^n \sin(kt + \alpha)$. Наличие таких членов ограничивает область сходимости рядов интервалами времени $t \sim O(1/\varepsilon)$ и делает невозможным исследование вековой эволюции систем на временах порядка $t \sim O(1/\varepsilon^2)$ и более [Гребеников, 1986].

Для преодоления этой проблемы применяются методы усреднения, основанные на разделении движений на короткопериодические («быстрые») и вековые либо долгопериодические («медленные»).

### 2.1. Классический этап: метод Гаусса-Фату и метод Делоне
Историческим предшественником современных теорий усреднения в небесной механике является метод усреднения Гаусса-Фату. В его основе лежит геометрическое представление о распределении массы возмущающего тела по орбитальному эллипсу пропорционально времени нахождения в каждой точке. Полученные уравнения вековых возмущений позволяли качественно описать эволюцию орбит, однако метод не содержал регулярного алгоритма для построения приближений высших порядков [Гребеников, 1986].

Для аналитического описания движения Луны Ш. Делоне разработал метод, основанный на последовательном исключении периодических членов из функции Гамильтона [Delaunay, 1860; Дубошин, 1968]. В методе Делоне каждый конкретный тригонометрический член устранялся посредством отдельного канонического преобразования. Основным ограничением данного подхода являлась высокая вычислительная трудоемкость: для построения точной теории требовалось последовательное выполнение сотен промежуточных канонических замен переменных.

### 2.2. Метод Пуанкаре и возникновение метода фон Цейпеля
В конце XIX века А. Пуанкаре в фундаментальном труде *«Новые методы небесной механики»* формализовал теорию малого параметра и показал, что классические ряды теории возмущений являются расходящимися асимптотическими рядами. Пуанкаре указал на проблему малых знаменателей, возникающую в резонансных зонах, когда комбинация частот невозмущенного движения удовлетворяет условию соизмеримости: $\langle k, \omega \rangle \approx 0$ [Новиков, Тайманов, 2005].

В 1916 году Х. фон Цейпель модифицировал метод Гамильтона-Якоби для систем с малыми возмущениями [von Zeipel, 1916]. Вместо поочередного удаления каждого тригонометрического члена, фон Цейпель показал, как с помощью одной производящей функции смешанного типа $S(\bar{I}, \theta, \varepsilon)$ исключить класс короткопериодических слагаемых в заданном порядке приближения по малому параметру [Дубошин, 1968]. 

Метод фон Цейпеля стал стандартным инструментом небесной механики, однако при анализе динамики ИСЗ обнаружились его ограничения:
1. **Неканонические возмущения**: Метод применим только к консервативным системам и не позволяет учитывать диссипативные силы (например, аэродинамическое торможение).
2. **Проблема неявных замен**: Производящая функция смешанного типа задает связь старых и новых переменных неявно. Начиная со второго приближения ($O(\varepsilon^2)$), процедура аналитического обращения рядов сопряжена с экспоненциальным ростом объема вычислений [Гребеников, 1986].

### 2.3. Школа нелинейной механики: метод Крылова-Боголюбова-Митропольского
В 1930-х годах Н. М. Крылов и Н. Н. Боголюбов разработали методы анализа нелинейных колебательных систем общего (не обязательно гамильтонова) вида, обосновав подход Б. ван дер Поля. В 1945–1955 гг. Н. Н. Боголюбов и Ю. А. Митропольский обобщили этот подход на многочастотные системы, сформулировав метод Крылова-Боголюбова-Митропольского (КБМ) [Боголюбов, Митропольский, 1955]. 

Метод КБМ оперирует явными заменами переменных непосредственно в дифференциальных уравнениях движения, приведенных к стандартной форме Боголюбова. Н. Н. Боголюбов доказал фундаментальные теоремы, гарантирующие близость решений усредненной и точной систем на интервалах времени $t \sim O(1/\varepsilon)$. В динамике ИСЗ метод КБМ применяется для расчета эволюции орбит под действием диссипативных сил [Гребеников, 1986]. Однако при анализе строго консервативных задач усечение рядов КБМ на конечном шаге нарушает симплектическую структуру гамильтоновой системы, что может приводить к появлению фиктивной диссипации в высших приближениях [Маркеев, 1999].

### 2.4. Развитие систем аналитических вычислений: метод Депри-Хори
К середине 1960-х годов возникла необходимость построения аналитических теорий движения высоких порядков, что стимулировало создание автоматизированных алгоритмов.

В 1966 году Дж. Хори [Hori, 1966], а в 1969 году А. Депри [Deprit, 1969] независимо разработали аппарат построения канонических преобразований с помощью дифференциальных операторов и алгебр Ли [Новиков, Тайманов, 2005]. Метод Депри-Хори позволил генерировать канонические преобразования в явном виде. Уравнения для нахождения возмущений стали рекуррентными и свелись к последовательному вычислению скобок Пуассона от известных функций. Это устранило проблему неявных замен переменных, характерную для метода фон Цейпеля.

В 1970 году А. Камель [Kamel, 1970] расширил теорию преобразований Ли на неканонические системы (схема Камеля), что позволило объединить операторный подход Ли со способностью метода КБМ учитывать неконсервативные возмущения.

### 2.5. Теоретические основания сходимости: КАМ-теория
Вопрос о сходимости рядов теории возмущений при $t \to \infty$ был разрешен в рамках КАМ-теории (Колмогорова-Арнольда-Мозера). А. Н. Колмогоров предложил метод ускоренной сходимости (аналог метода Ньютона в функциональных пространствах) [Новиков, Тайманов, 2005]. КАМ-теория строго определила условия, при которых малые возмущения не разрушают условно-периодические движения на инвариантных торах фазового пространства, очертив математические границы применимости методов усреднения.

### 2.6. Анализ локального разрушения орбит: метод Мельникова
В окрестностях резонансов классические асимптотические методы предполагают регулярное поведение усредненных траекторий. Однако в неинтегрируемых системах вблизи сепаратрис невозмущенной задачи происходит расщепление устойчивых и неустойчивых инвариантных многообразий. В 1963 году В. К. Мельников разработал асимптотический метод оценки расстояния между этими многообразиями [Melnikov, 1963]. Метод Мельникова позволяет аналитически доказать факт их трансверсального пересечения, приводящего к возникновению гомоклинических структур Пуанкаре и локального динамического хаоса (стохастических слоев), где разделение движений на быстрые и медленные компоненты теряет физический смысл.

## 3. Сравнительный анализ методов

### Метод Крылова-Боголюбова-Митропольского
Метод оперирует непосредственно с дифференциальными уравнениями движения в стандартной форме Боголюбова [Боголюбов, Митропольский, 1955]:
$$\frac{dx}{dt} = \varepsilon X(x, \psi, \varepsilon), \quad \frac{d\psi}{dt} = \omega(x) + \varepsilon Y(x, \psi, \varepsilon)$$
Замена переменных ищется в явном виде: $x = \bar{x} + \varepsilon u_1(\bar{x}, \bar{\psi}) + \dots$.

### Метод фон Цейпеля
Усреднение осуществляется на уровне функции Гамильтона $H(I, \theta, \varepsilon) = H_0(I) + \varepsilon H_1(I, \theta) + \varepsilon^2 H_2(I, \theta) + \dots$. Переход к усредненным переменным $(\bar{I}, \bar{\theta})$ задается с помощью канонического преобразования посредством производящей функции смешанного типа $S(\bar{I}, \theta, \varepsilon)$ [Маркеев, 1999].

### Метод Депри-Хори (Ряды Ли)

Каноническое преобразование задается в явном виде с помощью оператора Ли $\mathcal{L}_W \equiv \{W, \cdot\}$, генерируемого функциями Хори [Hori, 1966] или Депри [Deprit, 1969]. Действие оператора на произвольную функцию фазового пространства $f$ определяется как:

$$\mathcal{L}_W f = \{W, f\} = \sum_{i=1}^n \left( \frac{\partial W}{\partial p_i}\frac{\partial f}{\partial q^i} - \frac{\partial W}{\partial q^i}\frac{\partial f}{\partial p_i} \right)$$

Связь между старыми и новыми переменными определяется экспоненциальным отображением под действием этого оператора, что полностью устраняет необходимость обращения рядов, характерную для метода фон Цейпеля [Новиков, Тайманов, 2005]:

$$\bar{x} = \exp(\varepsilon \mathcal{L}_W) x = x + \varepsilon \{W, x\} + \frac{\varepsilon^2}{2!} \{W, \{W, x\}\} + \dots$$

## 4. Границы применимости и особенности реализации в задачах астродинамики

### 4.1. Область применимости
* **Метод Крылова-Боголюбова-Митропольского**: Системы дифференциальных уравнений общего вида. Позволяет интегрировать уравнения движения ИСЗ при наличии диссипативных сил (например, аэродинамическое сопротивление верхней атмосферы) [Гребеников, 1986].
* **Метод фон Цейпеля и Ряды Ли**: Строго консервативные гамильтоновы системы [Маркеев, 1999]. Применяются для исследования возмущений от зональных, тессеральных и секторальных гармоник геопотенциала, а также лунно-солнечных возмущений [Дубошин, 1968].

### 4.2. Достоинства и недостатки
* **Метод Крылова-Боголюбова-Митропольского**: Универсален по отношению к классу правых частей уравнений. Ограничение заключается в нарушении симплектической структуры при анализе консервативных задач, что может приводить к появлению фиктивной диссипации в высших приближениях [Маркеев, 1999].
* **Метод фон Цейпеля**: Обеспечивает строгое выполнение законов сохранения благодаря каноничности преобразования. Ограничением является неявная структура замен переменных: начиная со второго приближения, аналитическое обращение рядов сопряжено со значительным ростом объема вычислений [Гребеников, 1986].
* **Метод Депри-Хори**: Сочетает инвариантность гамильтоновой структуры с явной формой преобразований на симплектическом многообразии [Новиков, Тайманов, 2005]. Все операции сводятся к вычислению скобок Пуассона от известных функций, что делает метод оптимальным для реализации в средах символьных вычислений.
