# Аналитическая теория возмущений Брауэра-Лиддена

## 1. Канонический базис Делоне и разложение возмущающей функции в тригонометрические ряды

Математическим фундаментом аналитической теории движения искусственных спутников Земли (ИСЗ) в гравитационном поле сплюснутой планеты является решение Дирка Брауэра [Brouwer, 1959]. Эта теория построена в рамках гамильтонова формализма с использованием канонических переменных Делоне.

### 1.1. Система уравнений Делоне

Состояние космического объекта описывается набором канонических элементов Делоне $(l, g, h, L, G, H)$, где угловые переменные (координаты) и сопряженные им импульсы связаны с классическими кеплеровыми элементами $(a, e, i, M, \omega, \Omega)$ следующими соотношениями [Brouwer, 1959; Валладо, 2006]:

$$
\begin{aligned}
l &= M && L = \sqrt{\mu a} \\
g &= \omega && G = L\sqrt{1 - e^2} \\
h &= \Omega && H = G\cos i
\end{aligned}
$$

где $\mu$ - геоцентрическая гравитационная постоянная. Дифференциальные уравнения возмущенного движения в канонической форме Гамильтона имеют вид:

$$
\frac{d(l, g, h)}{dt} = \frac{\partial \mathcal{H}}{\partial (L, G, H)}, \quad \frac{d(L, G, H)}{dt} = -\frac{\partial \mathcal{H}}{\partial (l, g, h)}
$$

Полный гамильтониан задачи $\mathcal{H}$ включает в себя невозмущенную кеплеровскую часть $\mathcal{H}_0$ и возмущающую функцию $\mathcal{R}$, обусловленную высшими зональными гармониками геопотенциала от $J_2$ до $J_5$ [Brouwer, 1959]:

$$
\mathcal{H}(l, g, -, L, G, H) = \mathcal{H}_0(L) + \mathcal{R}(l, g, L, G, H) = \frac{\mu^2}{2L^2} + \mathcal{R}_{J2} + \mathcal{R}_{J3} + \mathcal{R}_{J4} + \mathcal{R}_{J5}
$$

В канонических переменных Делоне долгота восходящего узла $h$ является циклической координатой гамильтониана ($\partial \mathcal{H} / \partial h = 0$), откуда следует интеграл площадей: $H = \text{const}$.

### 1.2. Разложение потенциала зональных гармоник

Для выполнения аналитического интегрирования методом фон Цейпеля исходное разложение возмущающей функции по полиномам Лежандра представляется в виде тригонометрических рядов по средней аномалии $l$ (быстрая переменная) и аргументу перицентра $g$ (медленная переменная).

Переход от геоцентрического расстояния $R$ и истинной аномалии $v$ к средним элементам осуществляется с использованием коэффициентов Ганзена $X_k^{n, m}(e)$, определяемых из разложений вида [Эльясберг, 1965]:

$$
\left(\frac{R}{a}\right)^n \cos m v = \sum_{k=-\infty}^{\infty} X_k^{n, m}(e) \cos k l
$$

После подстановки тригонометрических соотношений для синуса широты ИСЗ $\sin \varphi = \sin i \sin(v + g)$ возмущающая функция разлагается на составляющие группы членов.

#### 1. Главный член от второй зональной гармоники ($\mathcal{R}_{J2}$):
Член $\mathcal{R}_{J2}$ содержит постоянную (вековую) часть и короткопериодические слагаемые, зависящие от средней аномалии $l$ [Brouwer, 1959]:

$$
\mathcal{R}_{J2} = \frac{\mu^4 R_E^2 J_2}{2 L^3 G^3} \left[ \left(1 - \frac{3}{2}\sin^2 i\right) \left(\frac{a}{R}\right)^3 + \frac{3}{2}\sin^2 i \left(\frac{a}{R}\right)^3 \cos(2v + 2g) \right]
$$

С учетом аналитических свойств коэффициентов Ганзена для центрального члена $X_0^{-3, 0}(e) = (1-e^2)^{-3/2} = (L/G)^3$, выражение принимает вид:

$$
\mathcal{R}_{J2} = \frac{\mu^4 R_E^2 J_2}{2 L^6} \left[ \left(\frac{L}{G}\right)^3 \left(1 - \frac{3}{2}\sin^2 i\right) + \sum_{k \neq 0} X_k^{-3, 0}(e) \cos k l + \frac{3}{2}\sin^2 i \sum_{k=-\infty}^{\infty} X_k^{-3, 2}(e) \cos(kl + 2g) \right]
$$

#### 2. Члены от нечетных ($\mathcal{R}_{J3}, \mathcal{R}_{J5}$) и четвертой зональной ($\mathcal{R}_{J4}$) гармоник:
В теории Брауэра потенциалы высших зональных гармоник удерживаются в первом порядке малости. Разложение для первой нечетной гармоники $J_3$ имеет следующую структуру [Brouwer, 1959]:

$$
\mathcal{R}_{J3} = -\frac{\mu^5 R_E^3 J_3}{L^8} \left[ \sin i \left(1 - \frac{5}{4}\sin^2 i\right) \sum_{k=-\infty}^{\infty} X_k^{-4, 1}(e) \sin(kl + g) + \frac{5}{4}\sin^3 i \sum_{k=-\infty}^{\infty} X_k^{-4, 3}(e) \sin(kl + 3g) \right]
$$

Данная форма представления потенциала через явные тригонометрические аргументы $(kl + mg)$ служит основой для последующего разделения движений в методе фон Цейпеля.

## 2. Метод фон Цейпеля в теории Брауэра: уравнения первого и второго порядков

Метод фон Цейпеля основан на последовательном построении неявных канонических преобразований с помощью производящих функций, зависящих от старых координат и новых импульсов. Система дифференциальных уравнений в частных производных для определения этих функций и нового усредненного гамильтониана строится путем подстановки рядов в каноническое уравнение Гамильтона - Якоби и последующего группирования слагаемых по порядкам малости относительно малого параметра $J_2$ [Brouwer, 1959; Эльясберг, 1965].

### 2.1. Декомпозиция уравнения Гамильтона - Якоби

Пусть $S(l, g, -, L', G', H')$ - производящая функция первого этапа преобразования (исключение быстрой переменной $l$), где связь старых и новых переменных определяется соотношениями:

$$
(l', g', h') = \frac{\partial S}{\partial (L', G', H')}, \quad (L, G, H) = \frac{\partial S}{\partial (l, g, -)}
$$

Производящая функция $S$ и новый промежуточный гамильтониан $\mathcal{H}'$ разлагаются в ряды по степеням $J_2$:

$$
S = l L' + g G' + h H' + S_1(l, g, L', G', H') + S_2(l, g, L', G', H') + \dots
$$
$$
\mathcal{H}' = \mathcal{H}'_0 + \mathcal{H}'_1 + \mathcal{H}'_2 + \dots
$$

Подстановка данных рядов в каноническое уравнение Гамильтона - Якоби и разложение в ряд Тейлора в окрестности невозмущенного состояния $(L', G', H')$ приводит к рекуррентной системе уравнений для каждого порядка малости:

* **Уравнение нулевого порядка ($\mathcal{O}(1)$):**
  $$
  \mathcal{H}'_0 = \mathcal{H}_0(L') = \frac{\mu^2}{2 L'^2}
  $$

* **Уравнение первого порядка ($\mathcal{O}(J_2)$):**
  $$
  \frac{\partial \mathcal{H}_0}{\partial L'} \frac{\partial S_1}{\partial l} + \mathcal{R}_{J2}(l, g, L', G', H') = \mathcal{H}'_1(L', G', H')
  $$

* **Уравнение второго порядка ($\mathcal{O}(J_2^2)$):**
  $$
  \frac{\partial \mathcal{H}_0}{\partial L'} \frac{\partial S_2}{\partial l} + \frac{1}{2} \frac{\partial^2 \mathcal{H}_0}{\partial L'^2} \left( \frac{\partial S_1}{\partial l} \right)^2 + \frac{\partial \mathcal{R}_{J2}}{\partial L'} \frac{\partial S_1}{\partial l} + \frac{\partial \mathcal{R}_{J2}}{\partial G'} \frac{\partial S_1}{\partial g} + \mathcal{R}_{J4}(l, g, L', G', H') = \mathcal{H}'_2(g', -, L', G', H')
  $$

### 2.2. Разделение движений и определение функций первого порядка

В соответствии с принципом метода фон Цейпеля, новый гамильтониан первого порядка $\mathcal{H}'_1$ должен содержать только стационарные (усредненные) по быстрой переменной $l$ компоненты возмущающей функции. Интегрирование уравнения первого порядка по средней аномалии $l$ на интервале $[0; 2\pi]$ определяет промежуточный гамильтониан первого порядка, зависящий исключительно от канонических импульсов Делоне [Brouwer, 1959]:

$$
\mathcal{H}'_1 = \langle \mathcal{R}_{J2} \rangle_l = \frac{\mu^4 R_E^2 J_2}{2 L'^3 G'^3} \left( 1 - \frac{3}{2}\sin^2 i' \right)
$$

Периодическая часть уравнения служит для нахождения функции $S_1$ квадратурой:

$$
S_1 = - \left( \frac{\partial \mathcal{H}_0}{\partial L'} \right)^{-1} \int \left( \mathcal{R}_{J2} - \mathcal{H}'_1 \right) dl = \frac{L'^3}{\mu^2} \int \left( \mathcal{R}_{J2} - \mathcal{H}'_1 \right) dl
$$

Явное выражение для короткопериодической функции первого порядка $S_1$, полученное Д. Брауэром, имеет тригонометрическую структуру по углам истинной аномалии $v$ и аргумента перицентра $g$:

$$
S_1 = \frac{\mu^2 R_E^2 J_2}{4 G'^3} \left[ (2 - 3\sin^2 i') (v - l + e'\sin v) + 3\sin^2 i' \left( \frac{1}{2}\sin(2v + 2g) + \frac{e'}{2}\sin(v + 2g) + \frac{e'}{6}\sin(3v + 2g) \right) \right]
$$


### 2.3. Формирование векового гамильтониана второго порядка

Уравнение второго порядка содержит нелинейный член $\frac{1}{2} \frac{\partial^2 \mathcal{H}_0}{\partial L'^2} \left( \frac{\partial S_1}{\partial l} \right)^2$ и произведения производных потенциала первого порядка на производные функции $S_1$. Для построения полностью усредненного по $l$ и $g$ векового гамильтониана $\mathcal{H}''$ на втором этапе вводится второе каноническое преобразование по медленной переменной $g'$ с помощью функции $S^*$.

Полное двукратное усреднение уравнения второго порядка по быстрым и медленным фазам ($\langle \dots \rangle_{l, g}$) исключает производную $\partial S_2 / \partial l$ и определяет вековой гамильтониан второго порядка $\mathcal{H}''_2$, ответственный за нелинейный вековой тренд углов [Brouwer, 1959]:

$$
\mathcal{H}''_2 = \left\langle \frac{1}{2} \frac{\partial^2 \mathcal{H}_0}{\partial L'^2} \left( \frac{\partial S_1}{\partial l} \right)^2 + \frac{\partial \mathcal{R}_{J2}}{\partial L'} \frac{\partial S_1}{\partial l} + \frac{\partial \mathcal{R}_{J2}}{\partial G'} \frac{\partial S_1}{\partial g} \right\rangle_{l, g} + \langle \mathcal{R}_{J4} \rangle_{l, g}
$$

Дифференцированием результирующего векового гамильтониана $\mathcal{H}'' = \mathcal{H}'_0 + \mathcal{H}'_1 + \mathcal{H}''_2$ по импульсам Делоне определяются аналитические выражения для секундных приращений долготы узла $\dot{\Omega}_2$ и перицентра $\dot{\omega}_2$, удерживающие квадратичные эффекты сплюснутости $\mathcal{O}(J_2^2)$ и вклад высших зональных гармоник геопотенциала.

## 3. Математические сингулярности теории Брауэра и регуляризация Лиддена

Классическое аналитическое решение Брауэра, полученное в канонических переменных Делоне, обладает ограниченной областью применения из-за появления малых делителей в выражениях для периодических возмущений элементов орбиты [Lyddane, 1963]. Эта особенность ограничивает устойчивость алгоритмов при моделировании орбит с малыми эксцентриситетами ($e \to 0$) и наклонениями ($i \to 0$).

### 3.1. Природа появления малых знаменателей в элементах Делоне

Связь между производящей функцией фон Цейпеля $S_1$ и поправками к средним элементам устанавливается дифференциальными соотношениями канонического преобразования. Короткопериодическое возмущение средней аномалии $\delta l_1$ и аргумента перицентра $\delta g_1$ первого порядка вычисляется путем дифференцирования функции $S_1$ по сопряженным импульсам $L'$ и $G'$ [Brouwer, 1959]:

$$
\delta l_1 = \frac{\partial S_1}{\partial L'}, \quad \delta g_1 = \frac{\partial S_1}{\partial G'}
$$

Поскольку импульс Делоне $G' = L'\sqrt{1 - e'^2}$ содержит эксцентриситет, операция дифференцирования по $G'$ порождает оператор вида:

$$
\frac{\partial}{\partial G'} = \frac{\partial e'}{\partial G'} \frac{\partial}{\partial e'} + \frac{\partial i'}{\partial G'} \frac{\partial}{\partial i'} = -\frac{\sqrt{1 - e'^2}}{e' L'} \frac{\partial}{\partial e'} + \frac{\cot i'}{G'} \frac{\partial}{\partial i'}
$$

Наличие эксцентриситета $e'$ в знаменателе первого слагаемого и тригонометрического множителя $\cot i' = \frac{\cos i'}{\sin i'}$ во втором слагаемом приводит к тому, что в итоговых аналитических выражениях Брауэра для поправок к угловым элементам $\delta l_1$ и $\delta g_1$ возникают делители вида:

$$
\delta l_1 \propto \frac{1}{e'}, \quad \delta g_1 \propto \frac{1}{e'} \quad \text{и} \quad \delta g_1 \propto \frac{1}{\sin i'}
$$

При стремлении эксцентриситета или наклонения к нулю эти выражения теряют вычислительный смысл, хотя физическое положение космического объекта в пространстве остается непрерывным. Аналогичная проблема возникает в долгопериодических поправках первого порядка от нечетной гармоники $J_3$, где возмущения эксцентриситета и наклонения содержат малый знаменатель:

$$
\delta e_{lp} \propto \frac{J_3}{J_2 \sin i'}, \quad \delta i_{lp} \propto \frac{J_3 e'}{J_2 \sin i'}
$$

### 3.2. Переход к модифицированным переменным Пуанкаре

Для устранения указанных математических сингулярностей Р. Лидден предложил отказаться от раздельного вычисления поправок к каноническим элементам Делоне [Lyddane, 1963]. Вместо этого аналитические выражения Брауэра преобразуются в систему несингулярных параметров, эквивалентных модифицированным переменным Пуанкаре.

В алгоритме SGP4 для преодоления сингулярности околокруговых орбит вводятся две регулярные комбинации элементов Делоне, связывающие эксцентриситет и аргумент перицентра [Hoots, 1980; Lyddane, 1963]:

$$
\xi = e \sin(M + \omega) = e \sin(l + g), \quad \eta = e \cos(M + \omega) = e \cos(l + g)
$$

Для околоэкваториальных орбит аналогичным образом формируются несингулярные параметры, связывающие наклонение орбиты и долготу восходящего узла:

$$
\chi = \sin \frac{i}{2} \sin \Omega = \sin \frac{i}{2} \sin h, \quad \psi = \sin \frac{i}{2} \cos \Omega = \sin \frac{i}{2} \cos h
$$

### 3.3. Алгоритм алгебраической регуляризации

Метод Лиддена заключается в том, что полные поправки (вековые плюс долгопериодические) первого порядка рассчитываются непосредственно для введенных регулярных параметров $\xi, \eta, \chi, \psi$. Дифференциалы модифицированных переменных Пуанкаре раскладываются по поправкам элементов Делоне с помощью полных производных:

$$
\delta \xi = \delta e \sin(l_{df} + g_{df}) + e_{df} \cos(l_{df} + g_{df}) (\delta l + \delta g)
$$
$$
\delta \eta = \delta e \cos(l_{df} + g_{df}) - e_{df} \sin(l_{df} + g_{df}) (\delta l + \delta g)
$$

При подстановке в данные уравнения явных выражений Брауэра происходит взаимное алгебраическое сокращение сингулярных множителей. Член $\delta l$, содержащий делитель $1/e$, умножается на эксцентриситет $e_{df}$ во втором слагаемом, что полностью устраняет деление на ноль в результирующих расчетных формулах:

$$
e_{df} \cdot \delta l \propto e_{df} \cdot \frac{1}{e_{df}} \to 1
$$

Восстановление долгопериодических средних элементов на текущий шаг времени осуществляется через функции `arctan2` и извлечение квадратного корня от компонентов:

$$
e_{lp} = \sqrt{\xi_{lp}^2 + \eta_{lp}^2}, \quad (M + \omega)_{lp} = \text{atan2}(\xi_{lp}, \eta_{lp})
$$

Модификация Лиддена позволяет использовать аналитическую теорию Брауэра в качестве универсального вычислительного ядра для космических аппаратов с произвольным типом орбиты, обеспечивая инвариантность уравнений относительно малых значений геометрических параметров $e$ и $i$.

## 4. Автоматизированный аналитический вывод вековых скоростей первого порядка на SymPy

Для подтверждения выведенных Д. Брауэром аналитических соотношений первого порядка для скоростей вековой прецессии угловых элементов выполним символьное дифференцирование промежуточного гамильтониана $\mathcal{H}' = \mathcal{H}'_0 + \mathcal{H}'_1$. 

В соответствии с каноническими уравнениями Гамильтона, секунды приращения средней аномалии ($\dot{l}$), аргумента перицентра ($\dot{g}$) и долготы восходящего узла ($\dot{h}$) определяются как частные производные по сопряженным каноническим импульсам Делоне $L', G', H'$:

$$
\dot{l} = \frac{\partial (\mathcal{H}'_0 + \mathcal{H}'_1)}{\partial L'}, \quad \dot{g} = \frac{\partial \mathcal{H}'_1}{\partial G'}, \quad \dot{h} = \frac{\partial \mathcal{H}'_1}{\partial H'}
$$

Поскольку в формулу гамильтониана первого порядка $\mathcal{H}'_1$ традиционно входят геометрические элементы (наклонение $i$), символьный движок учитывает неявную каноническую связь параметров через импульсы Делоне: $\cos i = H'/G'$.


```python
import sympy as sp
from IPython.display import display, Math

# 1. Объявление символьных канонических переменных Делоне и констант
L, G, H = sp.symbols("L' G' H'", positive=True)
mu, J2, RE = sp.symbols("mu J2 R_E", positive=True)

# 2. Определение тригонометрических параметров через канонические импульсы Делоне
# В соответствии с теорией Брауэра: cos(i) = H' / G'
cos_i = H / G
sin_i_sq = 1 - cos_i**2

# 3. Формирование компонент векового гамильтониана Делоне первого порядка
H0 = mu**2 / (2 * L**2)
H1 = (mu**4 * RE**2 * J2) / (2 * L**3 * G**3) * (1 - 1.5 * sin_i_sq)

# Полный промежуточный гамильтониан первого порядка
H_total = H0 + H1

# 4. Вычисление частных производных (Канонические уравнения Гамильтона)
l_dot     = sp.diff(H_total, L)
g_dot     = sp.diff(H_total, G)
h_dot     = sp.diff(H_total, H)

# 5. Приведение полученных выражений к классическому баллистическому виду
p = sp.symbols("p_0", positive=True)
n0 = sp.symbols("n_0", positive=True)
cos_i_sym = sp.symbols(r"\cos(i_0)")

def simplify_to_classical_form(expr):
    # Заменяем комбинации импульсов на среднее движение n0 = mu^2 / L^3
    expr_sub = expr.subs(mu**2 / L**3, n0)
    # Заменяем комбинации импульсов на фокальный параметр p0 = G^2 / mu
    expr_sub = expr_sub.subs(G, sp.sqrt(mu * p))
    # Упрощаем и подставляем классические тригонометрические обозначения
    expr_sub = sp.simplify(expr_sub)
    expr_sub = expr_sub.subs(H / sp.sqrt(mu * p), cos_i_sym)
    expr_sub = expr_sub.subs(H**2 / (mu * p), cos_i_sym**2)
    return sp.factor(expr_sub)

# Вычисляем финальные классические выражения
h_dot_classical = simplify_to_classical_form(h_dot)
g_dot_classical = simplify_to_classical_form(g_dot)
l_dot_classical = simplify_to_classical_form(l_dot)

# 6. Вывод формул в формате LaTeX
print("=== РЕЗУЛЬТАТЫ АНАЛИТИЧЕСКОГО ВЫВОДА ВЕКОВЫХ СКОРОСТЕЙ БРАУЭРА ===")

display(Math(r"\text{1. Скорость изменения долготы восходящего узла: } \dot{\Omega}_1 = " + sp.latex(h_dot_classical)))
display(Math(r"\text{2. Скорость изменения аргумента перицентра: } \dot{\omega}_1 = " + sp.latex(g_dot_classical)))
display(Math(r"\text{3. Скорость изменения средней аномалии: } \dot{M}_1 = " + sp.latex(l_dot_classical)))
```

### 4.1. Анализ результатов символьного моделирования

Реализованный на базе библиотеки `sympy` алгоритм выполнил аналитическое дифференциарование векового потенциала Брауэра, подтвердив классические уравнения движения, заложенные в расчетный блок SGP4:

1. **Скорость прецессии узла ($\dot{h}$):** Полученное выражение пропорционально $-\cos i_0$. Это доказывает, что плоскость орбиты испытывает вековой разворот в направлении, противоположном вектору орбитального движения (ретроградное движение узлов для прямых орбит).
2. **Скорость движения перицентра ($\dot{g}$):** Выведенный сомножитель $(5\cos^2 i_0 - 1)$ подтверждает существование критического наклонения $i_{crit} = \arccos(1/\sqrt{5}) \approx 63.4^\circ$, при котором скорость апсидальной прецессии первого порядка обращается в ноль.
3. **Скорость изменения средней аномалии ($\dot{l}$):** Член первого порядка от $J_2$ вносит добавку к невозмущенному кеплеровскому среднему движению $n_0$. Данное приращение фазы обусловлено тем, что из-за сплюснутости планеты гравитационное притяжение вблизи экватора выше, что приводит к увеличению средней угловой скорости обращения спутника по сравнению с идеальной точечной моделью Ньютона.


## 4.2. Вычислительный эксперимент: Моделирование сингулярности околокруговых орбит и верификация численным методом

Для демонстрации вычислительной нестабильности уравнений Брауэра при малых эксцентриситетах и оценки точности регуляризации Лиддена выполнен сравнительный численный эксперимент.

### Постановка эксперимента
Рассматривается движение космического объекта на околокруговой средневысокой орбите ($h \approx 850$ км, эксцентриситет $e_0 = 10^{-4}$). Моделируется совместное влияние главного гравитационного возмущения от гармоники $J_2$ и долгопериодического возмущения от нечетной зональной гармоники $J_3$ на интервале времени в трое суток.

Выполняется независимое сопоставление трех расчетных траекторий:
1. **Классическая схема Брауэра (с сингулярностью):** Долгопериодическая поправка к аргументу перицентра рассчитывается напрямую через классическое выражение, содержащее критический делитель $e_0$ [Brouwer, 1959].
2. **Регуляризованная схема Лиддена:** Поправка вычисляется через несингулярные компоненты Пуанкаре $\xi, \eta$ (Раздел 3.3), полностью исключающие операцию деления на малую величину $e_0$ [Lyddane, 1963].
3. **Численный эталон (Интегратор уравнений движения):** Траектория формируется путем прямого численного интегрирования дифференциальных уравнений Ньютона в оскулирующих декартовых координатах под действием потенциала гармоник $J_2$ и $J_3$. На основе полученных прямоугольных векторов состояния аналитически вычисляется истинный (оскулирующий) аргумент перицентра на каждый шаг времени.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Гравитационные константы модели WGS-84
J2 = 0.00108262998905
J3 = -0.00000253215306
RE = 6378.137  # км
mu = 398600.5  # км^3/с^2

def gravity_j2_j3_rhs(t, state):
    """ Дифференциальные уравнения движения ИСЗ в декартовой системе координат """
    x, y, z, vx, vy, vz = state
    r = np.sqrt(x**2 + y**2 + z**2)
    ax_g, ay_g, az_g = -mu*x/r**3, -mu*y/r**3, -mu*z/r**3
    
    z_sq_fac = (z/r)**2
    j2_fac = 1.5 * J2 * (mu * RE**2 / r**5)
    ax_j2 = j2_fac * x * (5 * z_sq_fac - 1)
    ay_j2 = j2_fac * y * (5 * z_sq_fac - 1)
    az_j2 = j2_fac * z * (5 * z_sq_fac - 3)
    
    j3_fac = 0.5 * J3 * (mu * RE**3 / r**6)
    ax_j3 = j3_fac * x * (z / r) * (35 * z_sq_fac - 15)
    ay_j3 = j3_fac * y * (z / r) * (35 * z_sq_fac - 15)
    az_j3 = j3_fac * (35 * z_sq_fac * (z**2 / r) - 30 * (z**2 / r) + 3 * r)
    
    return [vx, vy, vz, ax_g + ax_j2 + ax_j3, ay_g + ay_j2 + ay_j3, az_g + az_j2 + az_j3]

def rv_to_poincare_elements_invariant(r_vec, v_vec):
    """ Строгий инвариантный перевод координат в элементы Пуанкаре через среднюю долготу """
    h_vec = np.cross(r_vec, v_vec)
    h_norm = np.linalg.norm(h_vec)
    r_norm = np.linalg.norm(r_vec)
    v_sq = np.sum(v_vec**2)
    
    # 1. Извлечение скалярного оскулирующего наклонения и долготы узла по компонентам вектора h
    i_osc = np.arccos(np.clip(h_vec[2] / h_norm, -1.0, 1.0))
    Omega_osc = np.arctan2(h_vec[0], -h_vec[1]) % (2 * np.pi)
    
    # 2. Вычисление аргумента широты u (угловое положение спутника от узла в плоскости орбиты)
    if np.sin(i_osc) != 0:
        u = np.arctan2(r_vec[2] / np.sin(i_osc), r_vec[0]*np.cos(Omega_osc) + r_vec[1]*np.sin(Omega_osc))
    else:
        u = np.arctan2(r_vec[1], r_vec[0])
        
    # 3. Извлечение скалярного вектора эксцентриситета Рунге - Ленца
    e_vec = ((v_sq - mu/r_norm)*r_vec - np.dot(r_vec, v_vec)*v_vec) / mu
    e_norm = np.linalg.norm(e_vec)
    
    if e_norm == 0:
        return 0.0, 0.0
        
    # 4. Построение линии узлов в декартовых координатах: n = k x h
    n_vec = np.array([-h_vec[1], h_vec[0], 0.0])
    n_norm = np.linalg.norm(n_vec)
    if n_norm == 0: 
        n_vec = np.array([1.0, 0.0, 0.0])
        n_norm = 1.0
        
    # Расчет аргумента перицентра omega
    cos_omega = np.dot(n_vec, e_vec) / (n_norm * e_norm)
    omega = np.arccos(np.clip(cos_omega, -1.0, 1.0))
    if e_vec[2] < 0:
        omega = 2 * np.pi - omega
        
    # 5. Вычисление истинной (v) и средней (M) аномалий
    v = (u - omega) % (2 * np.pi)
    M = v - 2.0*e_norm*np.sin(v) + 0.75*e_norm**2*np.sin(2.0*v)
    
    # Вычисление итоговой фазы элементов Пуанкаре lambda = M + omega
    lambda_param = (M + omega) % (2 * np.pi)
    return e_norm * np.sin(lambda_param), e_norm * np.cos(lambda_param)

def run_experiment(e_val):
    """ Функция моделирования и расчета траекторий для заданного эксцентриситета """
    a0, i0, omega0, M0, OMEGA0 = 7228.0, np.radians(51.64), np.radians(30.0), np.radians(0.0), np.radians(0.0)
    n0 = np.sqrt(mu / a0**3) * 60.0
    time_steps = np.linspace(0, 300, 600)  # ~3 витка ИСЗ
    
    p0 = a0 * (1.0 - e_val**2)
    factor = 1.5 * J2 * (RE / p0)**2 * (n0 / 60.0)
    omega_dot = 0.5 * factor * (5.0 * np.cos(i0)**2 - 1.0)
    M_dot     = (n0 / 60.0) + 0.5 * factor * np.sqrt(1.0 - e_val**2) * (3.0 * np.cos(i0)**2 - 1.0)
    
    # Перевод начальных условий Кеплера в декартовы координаты
    u0 = omega0 + M0
    r0_orb = a0 * (1 - e_val**2) / (1 + e_val * np.cos(M0))
    r_orb = [r0_orb * np.cos(u0), r0_orb * np.sin(u0), 0]
    v_orb = [-np.sin(u0)*np.sqrt(mu/p0), (e_val + np.cos(u0))*np.sqrt(mu/p0), 0]
    R_mat = np.array([[np.cos(OMEGA0), -np.sin(OMEGA0)*np.cos(i0), np.sin(OMEGA0)*np.sin(i0)],
                      [np.sin(OMEGA0), np.cos(OMEGA0)*np.cos(i0), -np.cos(OMEGA0)*np.sin(i0)],
                      [0, np.sin(i0), np.cos(i0)]])
    r_init, v_init = np.dot(R_mat, r_orb), np.dot(R_mat, v_orb)
    
    # Интегрирование методом DOP853
    sol = solve_ivp(gravity_j2_j3_rhs, [0, 300*60], list(r_init)+list(v_init), 
                    method='DOP853', t_eval=time_steps*60, rtol=1e-12, atol=1e-12)
    
    xi_l, eta_l, xi_b, eta_b, xi_n, eta_n = [], [], [], [], [], []
    for idx, t in enumerate(time_steps):
        t_sec = t * 60.0
        omega_df = (omega0 + omega_dot * t_sec) % (2 * np.pi)
        M_df = (M0 + M_dot * t_sec) % (2 * np.pi)
        amplitude_J3 = - (J3 * RE) / (2 * J2 * a0 * (1.0 - e_val**2)) * np.sin(i0)
        
        # Теория Лиддена
        xi_l.append(e_val * np.sin(M_df + omega_df) + amplitude_J3)
        eta_l.append(e_val * np.cos(M_df + omega_df))
        
        # Классическая теория Брауэра
        delta_omega_b = (amplitude_J3 / e_val) * np.cos(omega_df)
        omega_b = (omega_df + delta_omega_b) % (2 * np.pi)
        xi_b.append(e_val * np.sin(M_df + omega_b))
        eta_b.append(e_val * np.cos(M_df + omega_b))
        
        # Численный эталон (Истинное поведение системы)
        x_n, y_n = rv_to_poincare_elements_invariant(sol.y[0:3, idx], sol.y[3:6, idx])
        xi_n.append(x_n)
        eta_n.append(y_n)
        
    return eta_l, xi_l, eta_b, xi_b, eta_n, xi_n

print("Запуск высокоточного моделирования траекторий ИСЗ методом DOP853...")
e1_l, x1_l, e1_b, x1_b, e1_n, x1_n = run_experiment(0.05)   # Штатный эллиптический режим
e2_l, x2_l, e2_b, x2_b, e2_n, x2_n = run_experiment(0.001)  # Сингулярный околокруговой режим
print("Вычисления успешно завершены.")

# Построение сдвоенного фазового портрета
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 7.5))

# Левый график: Штатный режим (e = 0.05)
ax1.plot(e1_n, x1_n, color='blue', alpha=0.4, linewidth=1.5, label='Численный эталон DOP853 (Пульсации)')
ax1.plot(e1_l, x1_l, color='green', linewidth=3, label='Теория Лиддена')
ax1.plot(e1_b, x1_b, color='red', linestyle='--', linewidth=1.5, label='Теория Брауэра')
ax1.set_title(r'Штатный режим: эллиптическая орбита ($e = 0.05$)')
ax1.set_xlabel(r'Элемент $\eta = e \cdot \cos(M+\omega)$')
ax1.set_ylabel(r'Элемент $\xi = e \cdot \sin(M+\omega)$')
ax1.grid(True, linestyle='--')
ax1.axis('equal')
ax1.legend()

# Правый график: Сингулярный режим (e = 0.001)
ax2.plot(e2_n, x2_n, color='blue', alpha=0.4, linewidth=1.5, label='Численный эталон DOP853 (Истинные пульсации)')
ax2.plot(e2_l, x2_l, color='green', linewidth=3, label='Теория Лиддена (Регуляризованная)')
ax2.plot(e2_b, x2_b, color='red', linestyle='--', linewidth=1.5, label='Теория Брауэра (Сбоящая)')
ax2.set_title(r'Сингулярный режим: околокруговая орбита ($e = 0.001$)')
ax2.set_xlabel(r'Элемент $\eta = e \cdot \cos(M+\omega)$')
ax2.set_ylabel(r'Элемент $\xi = e \cdot \sin(M+\omega)$')
ax2.grid(True, linestyle='--')
ax2.axis('equal')
ax2.legend()

plt.tight_layout()
plt.show()
```

### 4.3. Анализ фазовых траекторий в элементах Пуанкаре

Построенная сдвоенная панель фазовых портретов в регулярных безразмерных координатах Пуанкаре $(\eta; \xi)$, верифицированная численным интегрированием уравнений движения методом 8-го порядка DOP853, определяет границы применимости исследуемых аналитических моделей:

1. **Согласованность моделей в штатном эллиптическом режиме ($e = 0.05$):** 
   На левом графике вековые траектории классической теории Брауэра (красный пунктир) и модификации Лиддена (зеленая линия) полностью совпадают, образуя идентичные концентрические окружности. Данный результат подтверждает тождественную эквивалентность двух аналитических подходов при удалении от математической сингулярности по эксцентриситету. 
   При этом обе средние аналитические кривые смещены по вертикальной оси относительно геометрического центра синего трека оскулирующих численных значений. Причина смещения заключается в том, что численный эталон DOP853 интегрирует полные уравнения Ньютона, содержащие весь спектр короткопериодических (витковых) возмущений первого и высших порядков от гармоники $J_2$. Аналитические же окружности Брауэра и Лиддена представляют собой **средние элементы**, из структуры которых короткопериодические составляющие исключены на первом этапе усреднения по методу фон Цейпеля.
2. **Искажение фазы в теории Брауэра в околокруговом режиме ($e = 0.001$):** 
   На правом графике уменьшение эксцентриситета орбиты активирует скрытую особенность уравнений Делоне. Присутствие сингулярного множителя $1/e$ в классической долгопериодической поправке первого порядка приводит к критическому искажению фазового угла аргумента перицентра в теории Брауэра. Красная окружность испытывает координатный сдвиг вниз по оси $\xi$, полностью теряя физическую связь с геометрическим центром реальной траектории ИСЗ.
3. **Инвариантность регуляризации Лиддена и профиль численного решения:** 
   Зеленый профиль Лиддена сохраняет геометрическую инвариантность в окрестности $e \to 0$ и прецизионно описывает средний долгопериодический дрейф параметров под действием нечетной гармоники $J_3$. Численный эталон DOP853 (синяя кривая) отражает истинную оскулирующую динамику системы. Ввиду двукратного прохождения спутника над экваториальным вздутием Земли за один виток, истинный эксцентриситет испытывает периодические колебания с частотой $2\nu$. На интервале моделирования в три витка это формирует устойчивую трехлепестковую эпициклоиду, намотанную вокруг среднего аналитического контура Лиддена.

### 5.1. Значение теории Брауэра

Разработанная Д. Брауэром общая теория возмущений движения ИСЗ [Brouwer, 1959] сформировала математический базис для полуаналитического прогнозирования орбит. Основные результаты его работы заключаются в следующем:

1. **Реализация канонического метода усреднения:** Брауэр применил метод фон Цейпеля к исследованию нецентрального гравитационного поля Земли, формализовав процедуру разделения движений в переменных Делоне. С помощью неявных производящих функций ему удалось разделить полный гамильтониан задачи на невозмущенную кеплеровскую часть, короткопериодические, долгопериодические и вековые компоненты (подробно математическая декомпозиция уравнений Гамильтона — Якоби приведена в [Разделе 2.1](#2.1.-Декомпозиция-уравнения-Гамильтона---Якоби), а процедура усреднения по быстрой переменной — в [Разделе 2.2](#2.2.-Разделение-движений-и-определение-функций-первого-порядка)).
2. **Вывод нелинейных вековых скоростей:** В работе были получены аналитические выражения для вековых изменений угловых элементов орбиты второго порядка малости относительно сплюснутости Земли ($\mathcal{O}(J_2^2)$) с одновременным удержанием вклада от зональных гармоник от $J_3$ до $J_5$ (механизм построения векового гамильтониана второго порядка описан в [Разделе 2.3](#2.3.-Формирование-векового-гамильтониана-второго-порядка), а его автоматизированная верификация представлена в [Разделе 4](#4.-Автоматизированный-аналитический-вывод-вековых-скоростей-первого-порядка-на-SymPy)).
3. **Определение структуры периодических поправок:** Аналитический вывод тригонометрических рядов для коротко- и долгопериодических возмущений (с использованием коэффициентов Ганзена, рассмотренных в [Разделе 1.2](#1.2.-Разложение-потенциала-зональных-гармоник)) позволил осуществлять оперативный аналитический переход от средних элементов к мгновенным оскулирующим параметрам траектории космического аппарата, минуя процедуру численного интегрирования.

### 5.2. Значение модификации Лайддена

Несмотря на теоретическую строгость, исходные уравнения Брауэра обладали ограничением — наличием математических сингулярностей при стремлении эксцентриситета орбиты или ее наклонения к нулю. Модификация, предложенная Р. Лайдденом [Lyddane, 1963], устранила этот недостаток и имела определяющее значение для прикладных расчетов:

1. **Алгебраическая регуляризация уравнений:** Лайдден доказал, что сингулярности вида $1/e$ и $1/\sin i$ являются следствием выбора координатного базиса переменных Делоне (как показано в [Разделе 3.1](#3.1.-Природа-появления-малых-знаменателей-в-элементах-Делоне)), а не физической природы возмущенного движения. Он разработал алгоритм перегруппировки тригонометрических рядов Брауэра, объединив поправки элементов в регулярные комбинации.
2. **Внедрение модифицированных переменных Пуанкаре:** Переход к несингулярным параметрам — компонентам эксцентриситета и наклонения плоскости орбиты (введенным в [Разделе 3.2](#3.2.-Переход-ко-модифицированным-переменным-Пуанкаре)) — позволил исключить операцию деления на малые величины в расчетных формулах за счет взаимного сокращения сингулярных множителей в дифференциалах (описано в [Разделе 3.3](#3.3.-Алгоритм-алгебраической-регуляризации)).
3. **Обеспечение вычислительной инвариантности:** Модификация Лайддена сделала аналитическую теорию возмущений инвариантной относительно малых эксцентриситетов и наклонений. Алгоритм приобрел устойчивость при моделировании околокруговых и околоэкваториальных орбит, которые составляют большинство среди искусственных спутников Земли (включая низкоорбитальные аппараты ДЗЗ и геостационарные спутники связи).
