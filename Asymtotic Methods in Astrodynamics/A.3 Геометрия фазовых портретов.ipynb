# Геометрия фазовых портретов нелинейных динамических систем

В настоящем конспекте рассматриваются геометрические понятия качественной теории дифференциальных уравнений, необходимые для топологического анализа нелинейных консервативных и диссипативных систем. Качественный подход позволяет определять глобальные свойства фазовых потоков и находить законы долгосрочной эволюции систем без вычисления точных аналитических решений уравнений движения [Арнольд, 2024].

## 1. Фазовое пространство, фазовые траектории и фазовый поток

Состояние динамической системы с $n$ степенями свободы однозначно задается точкой в $2n$-мерном гладком многообразии, называемом фазовым пространством. Для одномерной механической системы ($n=1$) фазовым пространством является плоскость, осями которой служат обобщенная координата $q^1$ и сопряженный канонический импульс $p_1$ [Маркеев, 2024].

Временная эволюция системы описывается автономной системой обыкновенных дифференциальных уравнений первого порядка, правые части которой задают векторное поле скоростей в каждой точке пространства:

$$\frac{dq^1}{dt} = f^1(q^1, p_1), \quad \frac{dp_1}{dt} = f_1(q^1, p_1)$$

*   **Фазовая траектория** - гладкая кривая на фазовой плоскости, параметризованная временем $t$ и описывающая движение изображающей точки. В силу теоремы Коши — Липшица о существовании и единственности решений, через каждую точку фазового пространства гладкого неколлинеарного поля проходит одна траектория. Из этого следует, что фазовые траектории автономной системы не пересекаются и не разветвляются [Арнольд, 2024].
*   **Фазовый поток** - однопараметрическая группа преобразований фазового пространства, сдвигающая каждую начальную точку вдоль её фазовой траектории на интервал времени $t$. Направление фазового потока на портретах указывается стрелками, сонаправленными с вектором скорости $(\dot{q}^1, \dot{p}_1)$.

### 1.1. Базовая топология фазового портрета: линейный гармонический осциллятор

Моделью для изучения геометрии консервативного фазового пространства служит линейный гармонический осциллятор. Функция Гамильтона одномерной системы в безразмерных канонических переменных имеет вид [Арнольд, 2024]:

$$H(q^1, p_1) = \frac{p_1^2 + (q^1)^2}{2}$$

где $q^1$ - обобщенная координата (вектор положения), а $p_1$ - сопряженный обобщенный импульс (количественная мера скорости). Канонические уравнения Гамильтона определяют векторное поле скоростей потока [Маркеев, 2024]:

$$\frac{dq^1}{dt} = \frac{\partial H}{\partial p_1} = p_1, \quad \frac{dp_1}{dt} = -\frac{\partial H}{\partial q^1} = -q^1$$

Поскольку полная механическая энергия сохраняется во времени ($H = E = \text{const}$), траектории изображающей точки на фазовой плоскости представляют собой строгие концентрические окружности фиксированного радиуса $R = \sqrt{2E}$. 

#### ** Правила чтения базового фазового портрета:**
При анализе и чтении любой фазовой карты необходимо придерживаться следующих фундаментальных геометрических правил [Арнольд, 2024]:
1.  **Направление фазового потока (направление стрелок)**: 
    *   В верхнем полупространстве ($\mathbf{p_1 > 0}$) импульс положителен, следовательно, координата обязана возрастать ($\dot{q}^1 = p_1 > 0$). Фазовый поток всегда движется **слева направо**.
    *   В нижнем полупространстве ($\mathbf{p_1 < 0}$) импульс отрицателен, координата убывает ($\dot{q}^1 < 0$). Поток всегда движется **справа налево**.
    *   В силу этих двух топологических законов, вращение фазового потока вокруг устойчивых положений равновесия в декартовой системе координат $(q^1, p_1)$ **всегда происходит строго по часовой стрелке**.
2.  **Пересечение оси абсцисс**: Фазовые траектории пересекают ось координат $q^1$ (где импульс $p_1 = 0$) строго ортогонально (под углом $90^{\circ}$). В этих точках скорость изменения координаты равна нулю ($\dot{q}^1 = 0$), что соответствует моментам остановки системы при достижении максимальной амплитуды колебаний (крайние точки отклонения).
3.  **Пересечение оси ординат**: Траектории пересекают ось импульсов $p_1$ (где координата $q^1 = 0$) также строго ортогонально. В этих точках потенциальная энергия равна нулю, а импульс и скорость системы достигают своих экстремальных (максимальных или минимальных) значений.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Формирование координатной сетки фазовой плоскости
q_vec = np.linspace(-2.0, 2.0, 400)
p_vec = np.linspace(-2.0, 2.0, 400)
Q_mesh, P_mesh = np.meshgrid(q_vec, p_vec)

# 2. Расчет энергетической поверхности (Гамильтониана) линейного осциллятора
H_mesh = (P_mesh**2 + Q_mesh**2) / 2.0

plt.figure(figsize=(9, 9))

# 3. Отрисовка траекторий либрации (концентрических окружностей энергии)
energies = [0.2, 0.5, 1.0, 1.6]
contours = plt.contour(Q_mesh, P_mesh, H_mesh, levels=energies, colors='blue', linewidths=1.2)
plt.clabel(contours, inline=True, fmt='E = %.1f', fontsize=9)

# 4. Расчет и нанесение стрелок направления фазового потока (по часовой стрелке)
# Для наглядности рассчитываем векторное поле на разреженной сетке
q_arrow, p_arrow = np.meshgrid(np.linspace(-1.8, 1.8, 11), np.linspace(-1.8, 1.8, 11))
dq1_dt = p_arrow
dp1_dt = -q_arrow

# Нормализация векторов для одинаковой длины стрелок направления
norm = np.hypot(dq1_dt, dp1_dt)
norm[norm == 0] = 1.0  # Избегаем деления на ноль в начале координат
plt.quiver(q_arrow, p_arrow, dq1_dt/norm, dp1_dt/norm, color='navy', alpha=0.4, scale=25, headwidth=4)

# 5. Маркировка устойчивого положения равновесия типа "Центр"
plt.plot(0, 0, 'ko', markersize=8, label='Положение равновесия (Центр)')

# Академическое оформление фазовой карты
plt.title('Фазовый портрет линейного гармонического осциллятора', fontsize=12, pad=15)
plt.xlabel(r'Каноническая координата $q^1$ (вектор положения)', fontsize=11)
plt.ylabel(r'Канонический импульс $p_1$ (ковектор состояния)', fontsize=11)
plt.xlim(-2.0, 2.0)
plt.ylim(-2.0, 2.0)

# Выделение главных осей координат
plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.axvline(0, color='black', linewidth=0.8, linestyle='-')
plt.grid(True, linestyle=':', alpha=0.5)
plt.gca().set_box_aspect(1) # Строго квадратные пропорции

plt.legend(loc='upper right', frameon=True, fontsize=10)
plt.tight_layout()
plt.show()
```

## 2. Положения равновесия и локальная классификация особых точек

Положения равновесия (особые, стационарные или критические точки) динамической системы соответствуют координатам фазового пространства, в которых вектор скорости потока тождественно обращается в ноль, а изображающая точка остается неподвижной во времени [Арнольд, 2024]:

$$f^1(q^1_*, p_{1,*}) = 0, \quad f_1(q^1_*, p_{1,*}) = 0$$

Для исследования топологии фазового потока в окрестности равновесия нелинейное поле раскладывается в ряд Тейлора. Ограничиваясь линейными членами, формируют матрицу линеаризации (матрицу Якоби) $\mathbf{M}$ размерности $2 \times 2$ [Маркеев, 2024]:

$$\mathbf{M} = \begin{pmatrix} \frac{\partial f^1}{\partial q^1} & \frac{\partial f^1}{\partial p_1} \\ \frac{\partial f_1}{\partial q^1} & \frac{\partial f_1}{\partial p_1} \end{pmatrix}_{(q^1_*, p_{1,*})}$$

Качественный характер локальных фазовых траекторий полностью определяется собственными значениями $\lambda_{1,2}$ матрицы Якоби, находящимися из характеристического уравнения $\det(\mathbf{M} - \lambda \mathbf{E}) = 0$:

1.  **Центр** ($\lambda_{1,2} = \pm i\beta$, чисто мнимые сопряженные числа): Положение устойчивого равновесия. Окрестность заполняется концентрическими замкнутыми эллиптическими траекториями, соответствующими незатухающим периодическим колебаниям. Характерен для строго консервативных гамильтоновых систем в точках локального минимума потенциальной энергии [Арнольд, 2024].
2.  **Седло** ($\lambda_{1,2} = \pm \alpha$, действительные числа противоположных знаков): Неустойчивое положение равновесия. Существуют две входящие асимптотические траектории (устойчивые многообразия) и две выходящие траектории (неустойчивые многообразия), называемые сепаратрисами седла. Характерно для гамильтоновых систем в точках локального максимума или перевальных точках потенциала [Маркеев, 2024].
3.  **Узел** ($\lambda_{1,2}$ - действительные числа одного знака): Траектории имеют вид параболических кривых, монотонно входящих в критическую точку (устойчивый узел при $\lambda < 0$) или выходящих из неё (неустойчивый узел при $\lambda > 0$). Описывает сильную апериодическую диссипацию, нарушающую каноническую структуру уравнений.
4.  **Фокус** ($\lambda_{1,2} = \alpha \pm i\beta$, комплексные сопряженные числа с ненулевой действительной частью): Траектории представляют собой логарифмические спирали, асимптотически скручивающиеся к равновесию (устойчивый фокус при $\alpha < 0$) или раскручивающиеся от него (неустойчивый фокус при $\alpha > 0$). Описывает затухающие или раскачивающиеся колебания в неконсервативных системах.

### 2.1. Визуализация и качественный анализ локальных типов особых точек

Для сопоставления аналитической классификации собственных значений $\lambda_{1,2}$ с геометрической структурой фазовых траекторий ниже приведена программная реализация симулятора линейных систем вида $\dot{\mathbf{x}} = \mathbf{M}\mathbf{x}$ в окрестности изолированных положений равновесия $(0,0)$. 

Набор графических панелей иллюстрирует поведение фазового потока для четырех канонических типов критических точек:
1.  **Центр**: Система $\dot{q}^1 = p_1, \dot{p}_1 = -q^1$. Характеристические числа $\lambda_{1,2} = \pm i$ определяют консервативное вращение изображающей точки по замкнутым круговым орбитам.
2.  **Седло**: Система $\dot{q}^1 = p_1, \dot{p}_1 = q^1$. Характеристические числа $\lambda_{1,2} = \pm 1$ формируют гиперболическую структуру. Четко прослеживаются входящие (устойчивые) и выходящие (неустойчивые) прямые сепаратрисы, направленные вдоль собственных векторов матрицы линеаризации.
3.  **Устойчивый узел**: Система $\dot{q}^1 = -2q^1, \dot{p}_1 = -0.5p_1$. Характеристические числа $\lambda_1 = -2, \lambda_2 = -0.5$ обуславливают монотонное апериодическое притяжение траекторий к началу координат без осцилляций.
4.  **Устойчивый фокус**: Система $\dot{q}^1 = -0.2q^1 + p_1, \dot{p}_1 = -q^1 - 0.2p_1$. Характеристические числа $\lambda_{1,2} = -0.2 \pm i$ задают спиралевидное скручивание траекторий, отражающее колебательный процесс с затуханием амплитуды.

Расчет векторных полей скоростей выполняется с помощью функции `np.streamplot`, которая интегрирует линии тока и визуализирует направление фазового потока.


```python
import numpy as np
import matplotlib.pyplot as plt

# Формирование координатной сетки для расчета векторных полей скоростей
fig, axs = plt.subplots(2, 2, figsize=(14, 12))
q_grid, p_grid = np.meshgrid(np.linspace(-2.0, 2.0, 20), np.linspace(-2.0, 2.0, 20))

# ---------------------------------------------------------------------
# ПАНЕЛЬ 1: ЦЕНТР (Чисто мнимые собственные значения)
# ---------------------------------------------------------------------
# dq1/dt = p1, dp1/dt = -q1
dq1_center = p_grid
dp1_center = -q_grid
axs[0, 0].streamplot(q_grid, p_grid, dq1_center, dp1_center, color='blue', linewidth=1.2, density=1.2)
axs[0, 0].plot(0, 0, 'ko', markersize=8)
axs[0, 0].set_title(r'1. Центр ($\lambda_{1,2} = \pm i$)', fontsize=12)

# ---------------------------------------------------------------------
# ПАНЕЛЬ 2: СЕДЛО (Действительные числа противоположных знаков)
# ---------------------------------------------------------------------
# dq1/dt = p1, dp1/dt = q1
dq1_saddle = p_grid
dp1_saddle = q_grid
axs[0, 1].streamplot(q_grid, p_grid, dq1_saddle, dp1_saddle, color='crimson', linewidth=1.2, density=1.2)
# Отрисовка явных линейных сепаратрис седла по направлениям собственных векторов
axs[0, 1].plot([-2, 2], [-2, 2], color='black', linestyle=':', alpha=0.7, label='Сепаратрисы')
axs[0, 1].plot([-2, 2], [2, -2], color='black', linestyle=':', alpha=0.7)
axs[0, 1].plot(0, 0, 'ko', markersize=8)
axs[0, 1].set_title(r'2. Седло ($\lambda_{1,2} = \pm 1$)', fontsize=12)
axs[0, 1].legend(loc='upper right')

# ---------------------------------------------------------------------
# ПАНЕЛЬ 3: УСТОЙЧИВЫЙ УЗЕЛ (Действительные отрицательные числа)
# ---------------------------------------------------------------------
# dq1/dt = -2*q1, dp1/dt = -0.5*p1
dq1_node = -2.0 * q_grid
dp1_node = -0.5 * p_grid
axs[1, 0].streamplot(q_grid, p_grid, dq1_node, dp1_node, color='darkgreen', linewidth=1.2, density=1.2)
axs[1, 0].plot(0, 0, 'ko', markersize=8)
axs[1, 0].set_title(r'3. Устойчивый узел ($\lambda_1 = -2, \lambda_2 = -0.5$)', fontsize=12)

# ---------------------------------------------------------------------
# ПАНЕЛЬ 4: УСТОЙЧИВЫЙ ФОКУС (Комплексные сопряженные с Re < 0)
# ---------------------------------------------------------------------
# dq1/dt = -0.2*q1 + p1, dp1/dt = -q1 - 0.2*p1
dq1_focus = -0.2 * q_grid + p_grid
dp1_focus = -q_grid - 0.2 * p_grid
axs[1, 1].streamplot(q_grid, p_grid, dq1_focus, dp1_focus, color='purple', linewidth=1.2, density=1.2)
axs[1, 1].plot(0, 0, 'ko', markersize=8)
axs[1, 1].set_title(r'4. Устойчивый фокус ($\lambda_{1,2} = -0.2 \pm i$)', fontsize=12)

# Унификация оформления всех четырех координатных панелей
for ax in axs.flat:
    ax.set_xlabel(r'Координата $q^1$', fontsize=11)
    ax.set_ylabel(r'Импульс $p_1$', fontsize=11)
    ax.set_xlim(-2.0, 2.0)
    ax.set_ylim(-2.0, 2.0)
    ax.axhline(0, color='black', linewidth=0.5, linestyle='--')
    ax.axvline(0, color='black', linewidth=0.5, linestyle='--')
    ax.grid(True, linestyle=':', alpha=0.5)

plt.suptitle('Локальная топология фазового потока линейных систем', fontsize=14, y=0.98)
plt.tight_layout()
plt.show()
```

## 3. Макроструктура фазового портрета и глобальные топологические режимы

Локальный анализ особых точек в окрестности равновесий позволяет описать поведение системы лишь «в малом». Глобальная топология фазового пространства (его макроструктура) определяется взаимным расположением критических точек, замкнутых кривых и разделяющих инвариантных многообразий [Арнольд, 2024].

Для корректной классификации глобальных режимов фазовые потоки разделяют по физическому признаку на два класса:
*   **Консервативные системы**: Системы, в которых фазовый объем сохраняется во времени (теорема Лиувилля), а полная механическая энергия системы является интегралом движения ($H = \text{const}$). В них отсутствуют источники и стоки фазовых траекторий.
*   **Диссипативные системы**: Системы с рассеянием энергии, в которых фазовый объем монотонно сжимается со временем. Траектории таких систем стремятся к предельным множествам - **аттракторам** (притягивающим подпространствам фазового объема) [Маркеев, 2024].

В нелинейных полях различной физической природы выделяют следующие макроскопические структуры и режимы движения:

1.  **Либрация (Периодические колебания)**: Глобальный режим движения, при котором фазовые траектории являются замкнутыми кривыми, окружающими устойчивую особую точку типа «центр». В процессе движения изображающей точки каноническая координата $q^1$ и сопряженный импульс $p_1$ испытывают периодические флуктуации, оставаясь строго в ограниченной области фазового пространства. Фазовый объем, заполненный такими замкнутыми кривыми, называется областью либрации [Маркеев, 2024].
2.  **Ротация (Монотонное вращение)**: Динамический режим, при котором координата или угловая фаза $q^1$ монотонно возрастает либо убывает со временем, не меняя знака своей производной ($\dot{q}^1 \neq 0$). На фазовом портрете траектории ротации представляют собой незамкнутые волнообразные линии, простирающиеся вдоль всей оси координат. Сопряженный импульс $p_1$ испытывает периодические колебания, но никогда не пересекает нулевое значение [Арнольд, 2024].
3.  **Инвариантный тор**: Геометрическое многообразие в многомерных консервативных системах, на поверхности которого фазовые траектории совершают сложные обмотки, характеризующиеся несколькими независимыми частотами. Если отношение этих частот иррационально, движение называется **квазипериодическим** - траектория никогда не замыкается и плотно заполняет всю поверхность тора, не пересекая себя [Арнольд, 2024].
4.  **Сепаратриса**: Инвариантная фазовая траектория, разделяющая фазовое пространство на изолированные области с качественно различной топологией потока (например, отделяющая область либрации от области ротации). В консервативных системах сепаратрисы всегда выходят из неустойчивых особых точек типа «седло» и входят в них же. Время движения изображающей точки вдоль сепаратрисы от одного седла до другого стремится к бесконечности ($t \to \infty$) в силу зануления скорости векторного поля в самих критических точках [Маркеев, 2024].
5.  **Предельный цикл**: Изолированная замкнутая фазовая траектория в диссипативных автоколебательных системах, являющаяся простейшим типом аттрактора. К ней асимптотически сходятся при $t \to \infty$ (устойчивый предельный цикл) или от неё расходятся (неустойчивый предельный цикл) соседние спиральные траектории. Наличие предельного цикла указывает на протекание стационарного периодического процесса, амплитуда и частота которого определяются внутренними свойствами нелинейного поля, а не начальными условиями [Арнольд, 2024].
6.  **Сечение Пуанкаре**: Геометрический метод редукции размерности, применяемый для визуализации многомерных фазовых пространств (размерности 3 и выше). Метод заключается в фиксации в пространстве плоскости среза и нанесении на плоскость точек последовательного сквозного пересечения этого многообразия исследуемыми траекториями движения. Регулярные периодические движения отображаются на сечении в виде изолированных точек, квазипериодические — в виде замкнутых кривых, а хаотические режимы (хаотическое море) формируют диффузные облака случайных точек [Арнольд, 2024].

## 4. Глобальная топология нелинейного маятника

Классической физической моделью для исследования сопряженных режимов либрации, ротации и структуры сепаратрис в консервативных системах является математический маятник в однородном поле тяжести. Функция Гамильтона системы в безразмерных канонических переменных имеет вид [Арнольд, 2024]:

$$H(q^1, p_1) = \frac{p_1^2}{2} - \cos q^1$$

где $q^1 \in [-2\pi, 2\pi]$ - угловая координата маятника, а $p_1$ - сопряженный канонический импульс. Канонические уравнения движения Гамильтона определяются дифференциальными соотношениями [Маркеев, 2024]:

$$\frac{dq^1}{dt} = \frac{\partial H}{\partial p_1} = p_1, \quad \frac{dp_1}{dt} = -\frac{\partial H}{\partial q^1} = -\sin q^1$$

Поскольку гамильтониан не зависит от времени явно, полная механическая энергия является интегралом движения: $H(q^1, p_1) = E = \text{const}$. Топологическая структура фазового пространства качественно разделяется на три изолированные области в зависимости от величины энергетического уровня $E$:

1.  **Область либрации ($E < 1$)**: Фазовые траектории представляют собой замкнутые изолинии контура энергии, окружающие устойчивые положения равновесия типа «центр» с координатами $(2\pi k, 0), \, k \in \mathbb{Z}$. Изображающая точка совершает периодические колебания в ограниченной области фазового пространства. Из-за нелинейности возвращающей силы форма траекторий искажается по мере роста энергии, отклоняясь от гармонических эллипсов.
2.  **Область ротации ($E > 1$)**: Энергия превосходит потенциальный барьер, маятник совершает непрерывные полные обороты. Координата $q^1$ монотонно возрастает или убывает. Траектории на фазовой плоскости превращаются в незамкнутые волнообразные линии, расположенные в верхнем ($p_1 > 0$) или нижнем ($p_1 < 0$) полупространствах фазовой плоскости.
3.  **Сепаратриса ($E = 1$)**: Критическая фазовая траектория, формирующая геометрическую границу между либрацией и ротацией. Она соединяет неустойчивые положения равновесия типа «седло» с координатами $(\pi + 2\pi k, 0), \, k \in \mathbb{Z}$, образуя гетероклинические траектории. Скорость фазового потока стремится к нулю при приближении к седловым точкам.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Формирование сетки фазового пространства
q_vec = np.linspace(-2 * np.pi, 2 * np.pi, 500)
p_vec = np.linspace(-3.0, 3.0, 500)
Q_mesh, P_mesh = np.meshgrid(q_vec, p_vec)

# 2. Расчет энергетической поверхности (Гамильтониана)
H_mesh = (P_mesh**2) / 2.0 - np.cos(Q_mesh)

# 3. Визуализация фазового портрета
plt.figure(figsize=(12, 6))

# Отрисовка траекторий либрации и ротации через изолинии энергии
energies_lib = np.linspace(-0.9, 0.9, 8)
energies_rot = np.linspace(1.1, 4.0, 8)
all_energies = np.concatenate([energies_lib, energies_rot])

plt.contour(Q_mesh, P_mesh, H_mesh, levels=all_energies, colors='royalblue', linewidths=1.0)

# Явное выделение критического уровня энергии E = 1 (Сепаратриса)
plt.contour(Q_mesh, P_mesh, H_mesh, levels=[1.0], colors='crimson', linewidths=2.5, linestyles='-')

# Нанесение особых точек (положений равновесия)
plt.plot([-2*np.pi, 0, 2*np.pi], [0, 0, 0], 'go', markersize=8, label='Устойчивое равновесие (Центр)')
plt.plot([-np.pi, np.pi], [0, 0], 'ro', markersize=8, label='Неустойчивое равновесие (Седло)')

# Оформление графического окна по академическим стандартам
plt.title('Глобальная топология фазового пространства нелинейного маятника', fontsize=13, pad=15)
plt.xlabel(r'Каноническая координата $q^1$ (угол, рад)', fontsize=11)
plt.ylabel(r'Канонический импульс $p_1$', fontsize=11)
plt.xlim(-2 * np.pi, 2 * np.pi)
plt.ylim(-3.0, 3.0)
plt.xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi], [r'$-2\pi$', r'$-\pi$', '0', r'$\pi$', r'$2\pi$'])
plt.axhline(0, color='black', linewidth=0.6, linestyle=':')
plt.grid(True, linestyle=':', alpha=0.5)
plt.legend(loc='upper right', frameon=True)
plt.tight_layout()
plt.show()
```

### 4.1. Качественный анализ фазового портрета маятника

Численная визуализация подтверждает аналитические свойства нелинейного маятника и наглядно иллюстрирует геометрию консервативного фазового потока:

1.  **Искажение фазовых кривых в области либрации**. Вблизи центров $(0, 0)$ и $(\pm 2\pi, 0)$ замкнутые траектории близки к эллиптическим, что соответствует линейным гармоническим колебаниям. По мере увеличения энергии изолированные замкнутые контуры деформируются, приобретая выраженную угловатую форму, обусловленную ангармонизмом возвращающей силы потенциала.
2.  **Асимптотическое торможение на сепаратрисе**. Красный гетероклинический контур ($E=1$) строго разграничивает внутреннее колебательное и внешнее вращательное движения. При приближении к седловым точкам $(\pm \pi, 0)$ расстояние между линиями уровня расширяется, что численно отражает падение скорости изменения фазы ($\dot{q}^1 \to 0$).
3.  **Спектр траекторий ротации**. Вне сепаратрисы волнообразные изолинии энергии не замкнуты по координате $q^1$. С ростом начального импульса $p_1$ амплитуда пульсаций кинетической энергии падает, и траектории асимптотически стремятся к прямым линиям идеального равномерного вращения.


### 4.2. Выпрямление инвариантов и геометрия многомерных пространств

Переход от исходных декартовых фазовых координат $(q^1, p_1)$ к полярным каноническим переменным «действие-угол» $(\theta^1, I_1)$ представляет собой нормализующее отображение фазового пространства. Физический и геометрический смысл этого преобразования заключается в выпрямлении инвариантных многообразий консервативной системы [Арнольд, 2024].

Для двухстепенной системы ($n=2$), состоящей из двух независимых линейных осцилляторов, функция Гамильтона равна сумме энергий каждой из степеней свободы [Маркеев, 2024]:

$$H(q^1, q^2, p_1, p_2) = \frac{p_1^2 + \omega_1^2 (q^1)^2}{2} + \frac{p_2^2 + \omega_2^2 (q^2)^2}{2}$$

Каждая изолированная степень свободы обладает собственным двумерным подпространством, траекториями в котором служат вложенные замкнутые эллиптические контуры. Применение канонического преобразования фон Цейпеля трансформирует макроструктуру фазового объема [Арнольд, 2024]:

1.  **В декартовых координатах $(q^1, p_1)$ и $(q^2, p_2)$**: Первому меридиональному осциллятору соответствует семейство красных концентрических эллипсов, а второму экваториальному — семейство голубых окружностей. Скорость движения изображающих точек по траекториям непостоянна.
2.  **В канонических переменных $(\theta^i, I_i)$**: Поскольку преобразованный гамильтониан зависит исключительно от импульсов действия ($\bar{H} = \bar{H}_0(\bar{I})$), уравнения движения принимают линейный вид: $\dot{\bar{I}}_1 = 0, \, \dot{\bar{I}}_2 = 0$. Семейства траекторий выпрямляются, превращаясь в горизонтальные параллельные линии красного и голубого цвета. Переменные действий определяют радиусы, а фазы $\theta^1, \theta^2$ равномерно набегают от $0$ до $2\pi$.

Объединение двух плоских подпространств формирует четырехмерный фазовый объем ($\text{dim} = 4$). Фиксация двух индивидуальных интегралов энергии (или сопряженных импульсов действия $I_1 = \text{const}, \, I_2 = \text{const}$) выделяет в этом пространстве компактное инвариантное подмногообразие - **инвариантный тор** [Арнольд, 2024]. Геометрическая поверхность тора строится в декартовом пространстве проекций как отображение канонического произведения индивидуальных фазовых плоскостей посредством тригонометрического перехода $q^1 = \sqrt{2I_1}\cos\theta^1$ [Маркеев, 2024]. Топология намотки фазовой траектории на поверхность тора определяется отношением частот $\gamma = \omega^1 / \omega^2$ [Чириков, 1979]. Ниже приведен численный эксперимент, сопоставляющий неэргодический (резонансный) и эргодический (квазипериодический) режимы.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# =====================================================================
# 1. ФОРМИРОВАНИЕ ГЕОМЕТРИЧЕСКОГО БАЗИСА ИНВАРИАНТНОГО ТОРА
# =====================================================================
t_vec = np.linspace(0, 100, 3000)
theta_grid = np.linspace(0, 2 * np.pi, 40)

R_large, r_small = 3.0, 1.0
q_mesh_grid, p_mesh_grid = np.meshgrid(np.linspace(-4.5, 4.5, 200), np.linspace(-4.5, 4.5, 200))
max_range = 4.2

# =====================================================================
# ГРАФИК I: РАЦИОНАЛЬНЫЙ РЕЗОНАНС (omega_1 = 3.0, omega_2 = 1.0)
# =====================================================================
w1_res, w2_res = 3.0, 1.0
q1_res = r_small * np.cos(w1_res * t_vec); p1_res = -w1_res * r_small * np.sin(w1_res * t_vec)
q2_res = R_large * np.cos(w2_res * t_vec); p2_res = -w2_res * R_large * np.sin(w2_res * t_vec)
I1_res_const = 0.5 * w1_res**2 * r_small**2
I2_res_const = 0.5 * w2_res**2 * R_large**2

x_tr_res = (R_large + r_small * np.cos(w1_res * t_vec)) * np.cos(w2_res * t_vec)
y_tr_res = (R_large + r_small * np.cos(w1_res * t_vec)) * np.sin(w2_res * t_vec)
z_tr_res = r_small * np.sin(w1_res * t_vec)

fig1 = plt.figure(figsize=(20, 7.5))
# ИСПРАВЛЕНО: width_ratios обернут в np.array для предотвращения ошибки from_bounds()
gs1 = gridspec.GridSpec(2, 3, width_ratios=np.array([1, 1, 1.3]))

# Осциллятор 1 — Красная палитра (crimson)
ax_c1_res = fig1.add_subplot(gs1[0, 0])
H1_mesh = (p_mesh_grid**2 + w1_res**2 * q_mesh_grid**2) / 2.0
ax_c1_res.contour(q_mesh_grid, p_mesh_grid, H1_mesh, levels=[1.0, 2.5, 4.5, 7.0], colors='crimson', alpha=0.3, linewidths=1.2)
ax_c1_res.plot(q1_res, p1_res, color='crimson', linewidth=3.0)
ax_c1_res.set_ylabel(r'Импульс $p_1$', fontsize=10)
ax_c1_res.set_title(r'Осциллятор 1: $(q^1, p_1)$', fontsize=10, color='crimson')

ax_a1_res = fig1.add_subplot(gs1[0, 1])
# Отрисовка фоновых уровней через сплошные контрастные линии
for I_level in [1.0, 2.5, 7.0]: ax_a1_res.axhline(I_level, color='crimson', linewidth=1.2, alpha=0.35, linestyle='-')
ax_a1_res.axhline(I1_res_const, color='crimson', linewidth=3.5)
ax_a1_res.set_ylabel(r'Действие $I^1$', fontsize=10)
ax_a1_res.set_title(r'Осциллятор 1: $(\theta^1, I_1)$', fontsize=10, color='crimson')

# Осциллятор 2 — Голубая палитра (deepskyblue)
ax_c2_res = fig1.add_subplot(gs1[1, 0])
H2_mesh = (p_mesh_grid**2 + w2_res**2 * q_mesh_grid**2) / 2.0
ax_c2_res.contour(q_mesh_grid, p_mesh_grid, H2_mesh, levels=[1.0, 2.5, 4.5, 7.0], colors='deepskyblue', alpha=0.3, linewidths=1.2)
ax_c2_res.plot(q2_res, p2_res, color='deepskyblue', linewidth=3.0)
ax_c2_res.set_xlabel(r'Координата $q$', fontsize=10)
ax_c2_res.set_ylabel(r'Импульс $p_2$', fontsize=10)
ax_c2_res.set_title(r'Осциллятор 2: $(q^2, p_2)$', fontsize=10, color='deepskyblue')

ax_a2_res = fig1.add_subplot(gs1[1, 1])
# Отрисовка фоновых уровней через сплошные контрастные линии
for I_level in [1.0, 2.5, 7.0]: ax_a2_res.axhline(I_level, color='deepskyblue', linewidth=1.2, alpha=0.35, linestyle='-')
ax_a2_res.axhline(I2_res_const, color='deepskyblue', linewidth=3.5)
ax_a2_res.set_xlabel(r'Фаза $\theta$, рад', fontsize=10)
ax_a2_res.set_ylabel(r'Действие $I_2$', fontsize=10)
ax_a2_res.set_title(r'Осциллятор 2: $(\theta^2, I_2)$', fontsize=10, color='deepskyblue')

# 3D-Тор (Случай I) — Объемный серый металлический каркас
ax_3d_res = fig1.add_subplot(gs1[:, 2], projection='3d')
for t2 in theta_grid[::2]:
    x = (R_large + r_small * np.cos(theta_grid)) * np.cos(t2)
    y = (R_large + r_small * np.cos(theta_grid)) * np.sin(t2)
    z = r_small * np.sin(theta_grid)
    ax_3d_res.plot(x, y, z, color='gray', alpha=0.45, linewidth=1.0)
for t1 in theta_grid[::2]:
    x = (R_large + r_small * np.cos(t1)) * np.cos(theta_grid)
    y = (R_large + r_small * np.cos(t1)) * np.sin(theta_grid)
    z = r_small * np.sin(t1) * np.ones_like(theta_grid)
    ax_3d_res.plot(x, y, z, color='gray', alpha=0.45, linewidth=1.0)

ax_3d_res.plot(x_tr_res, y_tr_res, z_tr_res, color='indigo', linewidth=2.5, label='Замкнутая орбита резонанса (3:1)')

# Оформление графика I
for ax in [ax_c1_res, ax_c2_res, ax_a1_res, ax_a2_res]: ax.grid(True, linestyle=':', alpha=0.4); ax.set_box_aspect(1)
for ax in [ax_c1_res, ax_c2_res]: ax.set_xlim(-4.5, 4.5); ax.set_ylim(-4.5, 4.5)
for ax in [ax_a1_res, ax_a2_res]:
    ax.set_xlim(0, 2 * np.pi); ax.set_xticks([0, np.pi, 2 * np.pi], ['0', r'$\pi$', r'$2\pi$'])
ax_a1_res.set_ylim(0, 8.0); ax_a2_res.set_ylim(0, 8.0)
ax_3d_res.set_xlim(-max_range, max_range); ax_3d_res.set_ylim(-max_range, max_range); ax_3d_res.set_zlim(-max_range, max_range)
ax_3d_res.view_init(elev=30, azim=45); ax_3d_res.grid(True, linestyle=':', alpha=0.3); ax_3d_res.legend(loc='upper left', fontsize=10)
ax_3d_res.set_title(r'Неэргодический режим ($\omega^1 / \omega^2 = 3 / 1$ — рациональный резонанс)', fontsize=12, pad=10, weight='bold')
plt.tight_layout()

# =====================================================================
# ГРАФИК II: ИРРАЦИОНАЛЬНАЯ КВАЗИПЕРИОДИЧНОСТЬ (omega_1 = sqrt(2), omega_2 = 1.0)
# =====================================================================
w1_erg, w2_erg = np.sqrt(2.0), 1.0
q1_erg = r_small * np.cos(w1_erg * t_vec); p1_erg = -w1_erg * r_small * np.sin(w1_erg * t_vec)
q2_erg = R_large * np.cos(w2_erg * t_vec); p2_erg = -w2_erg * R_large * np.sin(w2_erg * t_vec)
I1_erg_const = 0.5 * w1_erg**2 * r_small**2
I2_erg_const = 0.5 * w2_erg**2 * R_large**2

x_tr_erg = (R_large + r_small * np.cos(w1_erg * t_vec)) * np.cos(w2_erg * t_vec)
y_tr_erg = (R_large + r_small * np.cos(w1_erg * t_vec)) * np.sin(w2_erg * t_vec)
z_tr_erg = r_small * np.sin(w1_erg * t_vec)

fig2 = plt.figure(figsize=(20, 7.5))

gs2 = gridspec.GridSpec(2, 3, width_ratios=np.array([1, 1, 1.3]))

# Прорисовка плоских проекций (Случай II)
ax_c1_erg = fig2.add_subplot(gs2[0, 0])
H1_mesh_erg = (p_mesh_grid**2 + w1_erg**2 * q_mesh_grid**2) / 2.0
ax_c1_erg.contour(q_mesh_grid, p_mesh_grid, H1_mesh_erg, levels=[0.3, 1.0, 2.2, 3.5], colors='crimson', alpha=0.3, linewidths=1.2)
ax_c1_erg.plot(q1_erg, p1_erg, color='crimson', linewidth=3.0)
ax_c1_erg.set_ylabel(r'Импульс $p_1$', fontsize=10)
ax_c1_erg.set_title(r'Осциллятор 1: $(q^1, p_1)$', fontsize=10, color='crimson')

ax_a1_erg = fig2.add_subplot(gs2[0, 1])
# Отрисовка фоновых уровней через сплошные контрастные линии
for I_level in [0.3, 1.5, 3.5]: ax_a1_erg.axhline(I_level, color='crimson', linewidth=1.2, alpha=0.35, linestyle='-')
ax_a1_erg.axhline(I1_erg_const, color='crimson', linewidth=3.0)
ax_a1_erg.set_ylabel(r'Действие $I^1$', fontsize=10)
ax_a1_erg.set_title(r'Осциллятор 1: $(\theta^1, I_1)$', fontsize=10, color='crimson')

ax_c2_erg = fig2.add_subplot(gs2[1, 0])
H2_mesh_erg = (p_mesh_grid**2 + w2_erg**2 * q_mesh_grid**2) / 2.0
ax_c2_erg.contour(q_mesh_grid, p_mesh_grid, H2_mesh_erg, levels=[1.0, 2.5, 4.5, 7.0], colors='deepskyblue', alpha=0.3, linewidths=1.2)
ax_c2_erg.plot(q2_erg, p2_erg, color='deepskyblue', linewidth=3.0)
ax_c2_erg.set_xlabel(r'Координата $q$', fontsize=10)
ax_c2_erg.set_ylabel(r'Импульс $p_2$', fontsize=10)
ax_c2_erg.set_title(r'Осциллятор 2: $(q^2, p_2)$', fontsize=10, color='deepskyblue')

ax_a2_erg = fig2.add_subplot(gs2[1, 1])
# Отрисовка фоновых уровней через сплошные контрастные линии
for I_level in [1.0, 2.5, 7.0]: ax_a2_erg.axhline(I_level, color='deepskyblue', linewidth=1.2, alpha=0.35, linestyle='-')
ax_a2_erg.axhline(I2_erg_const, color='deepskyblue', linewidth=3.0)
ax_a2_erg.set_xlabel(r'Фаза $\theta$, рад', fontsize=10)
ax_a2_erg.set_ylabel(r'Действие $I^2$', fontsize=10)
ax_a2_erg.set_title(r'Осциллятор 2: $(\theta^2, I_2)$', fontsize=10, color='deepskyblue')

# Построение 3D-Тора (Случай II)
ax_3d_erg = fig2.add_subplot(gs2[:, 2], projection='3d')
for t2 in theta_grid[::2]:
    x = (R_large + r_small * np.cos(theta_grid)) * np.cos(t2)
    y = (R_large + r_small * np.cos(theta_grid)) * np.sin(t2)
    z = r_small * np.sin(theta_grid)
    ax_3d_erg.plot(x, y, z, color='gray', alpha=0.45, linewidth=1.0)
for t1 in theta_grid[::2]:
    x = (R_large + r_small * np.cos(t1)) * np.cos(theta_grid)
    y = (R_large + r_small * np.cos(t1)) * np.sin(theta_grid)
    z = r_small * np.sin(t1) * np.ones_like(theta_grid)
    ax_3d_erg.plot(x, y, z, color='gray', alpha=0.45, linewidth=1.0)

ax_3d_erg.plot(x_tr_erg, y_tr_erg, z_tr_erg, color='indigo', linewidth=1.5, alpha=0.9, label=r'Квазипериодический трек ($\sqrt{2}:1$)')

# Оформление графика II
for ax in [ax_c1_erg, ax_c2_erg, ax_a1_erg, ax_a2_erg]: ax.grid(True, linestyle=':', alpha=0.4); ax.set_box_aspect(1)
for ax in [ax_c1_erg, ax_c2_erg]: ax.set_xlim(-4.5, 4.5); ax.set_ylim(-4.5, 4.5)
for ax in [ax_a1_erg, ax_a2_erg]:
    ax.set_xlim(0, 2 * np.pi); ax.set_xticks([0, np.pi, 2 * np.pi], ['0', r'$\pi$', r'$2\pi$'])
ax_a1_erg.set_ylim(0, 4.0); ax_a2_erg.set_ylim(0, 8.0)
ax_3d_erg.set_xlim(-max_range, max_range); ax_3d_erg.set_ylim(-max_range, max_range); ax_3d_erg.set_zlim(-max_range, max_range)
ax_3d_erg.view_init(elev=30, azim=45); ax_3d_erg.grid(True, linestyle=':', alpha=0.3); ax_3d_erg.legend(loc='upper left', fontsize=10)
ax_3d_erg.set_title(r'Эргодический квазипериодический режим ($\omega^1 / \omega^2 = \sqrt{2} / 1$ — иррациональная обмотка тора)', fontsize=12, pad=10, weight='bold')
plt.tight_layout()

plt.show()
```

### 4.3. Качественный анализ геометрии инвариантного многообразия

Сравнительный анализ плоских сечений и сопряженных трехмерных конфигураций двухчастотного фазового потока позволяет сделать следующие качественные выводы [Арнольд, 2024; Чириков, 1979]:

1.  **Верификация эффекта выпрямления**: Плоские декартовы панели наглядно иллюстрируют исходную колебательную геометрию: траектории первого осциллятора (красные линии) и второго осциллятора (голубые линии) представляют собой вложенные замкнутые контуры. Переход в пространство «действие — угол» (второй столбец матриц) подтверждает закон сохранения импульсов действия невозмущенной системы: кривые полностью выпрямляются в строго горизонтальные линии, высота которых определяется начальной энергией, а фаза равномерно набегает на периоде $[0, 2\pi]$ [Маркеев, 2024].
2.  **Феномен неэргодического замыкания (График I)**: При соизмеримых частотах ($\omega^1 / \omega^2 = 3/1$) индивидуальные периодические движения синхронизируются. На поверхности трехмерного тора, прорисованного серым цветом, фазовая траектория совершает три полных витка по меридиональному радиусу за один обход по экваториальному, после чего строго замыкается в начальной точке. Темно-индиговый трек образует периодическую орбиту, которая занимает нулевую площадь на поверхности многообразия, оставляя остальное пространство тора полностью недоступным для фазового потока [Чириков, 1979].
3.  **Феномен эргодической обмотки (График II)**: При иррациональном отношении частот ($\omega^1 / \omega^2 = \sqrt{2}/1$) индивидуальные колебания рассинхронизированы. На поверхности серого тора траектория совершает бесконечную незамкнутую намотку. Витки темно-индиговой кривой никогда не пересекаются и не повторяют свой путь. В результате длительного движения траектория эргодически заполняет всю доступную двухмерную площадь тороидальной поверхности, экспериментально верифицируя фундаментальные положения КАМ-теории о квазипериодических многообразиях интегрируемых гамильтоновых систем [Арнольд, 2024].

## 5. Перестройка фазовых портретов при бифуркациях

Качественное изменение топологической структуры фазового портрета при непрерывном изменении параметров динамической системы называется бифуркацией. Значения параметров, при которых происходит перестройка фазового потока (изменение числа состояний равновесия, появление или исчезновение периодических движений, смена типа устойчивости), называются критическими или бифуркационными [Арнольд, 2024].

Классическим примером перестройки фазового пространства, описывающей рождение автоколебательных режимов, является суперкритическая бифуркация Андронова-Хопфа. В канонической форме нормальной системы дифференциальных уравнений первого порядка эволюция системы в окрестности начала координат задается соотношениями [Маркеев, 2024]:

$$\frac{dq^1}{dt} = \mu q^1 + p_1 - q^1 \left( (q^1)^2 + p_1^2 \right)$$
$$\frac{dp_1}{dt} = -q^1 + \mu p_1 - p_1 \left( (q^1)^2 + p_1^2 \right)$$

где $\mu$ — управляющий бифуркационный параметр. 

Для наглядной визуализации топологической перестройки индивидуальные фазовые плоскости $(q^1, p_1)$ объединяются в трехмерное фазово-параметрическое пространство, где вертикальной осью аппликат служит непрерывно изменяющийся параметр $\mu$. В рамках геометрического анализа исследуется эволюция фазового потока в трех режимах:

1.  **Докритическое подпространство ($\mu < 0$)**: Изолированная особая точка в начале координат $(0, 0, \mu)$ является асимптотически устойчивым фокусом. Характеристические числа матрицы линеаризации имеют отрицательную действительную часть. Все близлежащие фазовые траектории представляют собой скручивающиеся спирали, стремящиеся к оси состояний равновесия при $t \to \infty$. Периодические движения в этой области пространства отсутствуют.
2.  **Бифуркационная плоскость ($\mu = 0$)**: Точка $(0, 0, 0)$ теряет линейную устойчивость, превращаясь в слабый фокус. Скорость скручивания траекторий существенно падает, асимптотическое приближение к равновесию определяется нелинейными членами высших порядков.
3.  **Закритическое подпространство ($\mu > 0$)**: Положение равновесия $(0, 0, \mu)$ становится полностью неустойчивым фокусом, а траектории начинают спирально откручиваться от центральной оси. Одновременно с потерей устойчивости из точки равновесия рождается устойчивый предельный цикл радиуса $R = \sqrt{\mu}$. В трехмерном пространстве совокупность этих циклов образует гладкую параболическую поверхность — бифуркационный рупор Хопфа, на который асимптотически наматываются все близлежащие траектории фазового потока при $t \to \infty$, формируя стационарный автоколебательный режим.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

# 1. Определение нелинейного поля системы Андронова — Хопфа
def hopf_system(t, y, mu):
    q1, p1 = y
    r2 = q1**2 + p1**2
    dq1_dt = mu * q1 + p1 - q1 * r2
    dp1_dt = -q1 + mu * p1 - p1 * r2
    return [dq1_dt, dp1_dt]

# 2. Инициализация комплексного графического окна
fig = plt.figure(figsize=(18, 9))
q_grid, p_grid = np.meshgrid(np.linspace(-1.5, 1.5, 25), np.linspace(-1.5, 1.5, 25))
r2_mesh = q_grid**2 + p_grid**2

# ---------------------------------------------------------------------
# ПАНЕЛЬ 1: ПЛОСКОЕ СЕЧЕНИЕ — ДОКРИТИЧЕСКИЙ РЕЖИМ (mu = -0.3)
# ---------------------------------------------------------------------
ax1 = fig.add_subplot(2, 2, 1)
mu_sub = -0.3
dq1_sub = mu_sub * q_grid + p_grid - q_grid * r2_mesh
dp1_sub = -q_grid + mu_sub * p_grid - p_grid * r2_mesh

ax1.streamplot(q_grid, p_grid, dq1_sub, dp1_sub, color='darkcyan', linewidth=1.1, density=1.2)
ax1.plot(0, 0, 'ko', markersize=8, label='Устойчивый фокус')
ax1.set_title(r'Сечение 1: Докритический режим ($\mu = -0.3$)', fontsize=11)
ax1.set_xlabel(r'Координата $q^1$', fontsize=10)
ax1.set_ylabel(r'Импульс $p_1$', fontsize=10)
ax1.set_box_aspect(1) # Квадратные пропорции

# ---------------------------------------------------------------------
# ПАНЕЛЬ 2: ПЛОСКОЕ СЕЧЕНИЕ — ЗАКРИТИЧЕСКИЙ РЕЖИМ (mu = 0.5)
# ---------------------------------------------------------------------
ax2 = fig.add_subplot(2, 2, 3)
mu_super = 0.5
dq1_super = mu_super * q_grid + p_grid - q_grid * r2_mesh
dp1_super = -q_grid + mu_super * p_grid - p_grid * r2_mesh

ax2.streamplot(q_grid, p_grid, dq1_super, dp1_super, color='purple', linewidth=1.1, density=1.2)

theta_vec = np.linspace(0, 2 * np.pi, 200)
R_cycle = np.sqrt(mu_super)
ax2.plot(R_cycle * np.cos(theta_vec), R_cycle * np.sin(theta_vec), color='crimson', linewidth=2.0, linestyle='-', label='Предельный цикл')
ax2.plot(0, 0, 'ko', markersize=8, markerfacecolor='none', label='Неустойчивый фокус')
ax2.set_title(r'Сечение 2: Закритический режим ($\mu = 0.5$)', fontsize=11)
ax2.set_xlabel(r'Координата $q^1$', fontsize=10)
ax2.set_ylabel(r'Импульс $p_1$', fontsize=10)
ax2.set_box_aspect(1) # Квадратные пропорции

# Унификация оформления плоских панелей
for ax in [ax1, ax2]:
    ax.set_xlim(-1.5, 1.5); ax.set_ylim(-1.5, 1.5)
    ax.axhline(0, color='black', linewidth=0.5, linestyle=':')
    ax.axvline(0, color='black', linewidth=0.5, linestyle=':')
    ax.grid(True, linestyle=':', alpha=0.5)
    ax.legend(loc='upper right', fontsize=9)

# ---------------------------------------------------------------------
# ПАНЕЛЬ 3: ТРЕХМЕРНОЕ ПРОСТРАНСТВЕННОЕ ПРЕДСТАВЛЕНИЕ (РУПОР ХОПФА)
# ---------------------------------------------------------------------
ax3 = fig.add_subplot(1, 2, 2, projection='3d')

# Построение полупрозрачного бифуркационного рупора
mu_surface_vec = np.linspace(0.0, 1.0, 100)
Mu_mesh, Theta_mesh = np.meshgrid(mu_surface_vec, theta_vec)
Q_surface = np.sqrt(Mu_mesh) * np.cos(Theta_mesh)
P_surface = np.sqrt(Mu_mesh) * np.sin(Theta_mesh)
ax3.plot_surface(Q_surface, P_surface, Mu_mesh, color='crimson', alpha=0.25, rstride=6, cstride=6)

# Построение маркировочных плоскостей сечений в 3D-пространстве
q_plane, p_plane = np.meshgrid(np.linspace(-1.5, 1.5, 10), np.linspace(-1.5, 1.5, 10))
ax3.plot_surface(q_plane, p_plane, np.full_like(q_plane, mu_sub), color='darkcyan', alpha=0.1, shade=False)
ax3.plot_surface(q_plane, p_plane, np.full_like(q_plane, mu_super), color='purple', alpha=0.1, shade=False)

# Интегрирование пространственных траекторий для дискретных уровней mu
mu_levels = [-0.3, 0.0, 0.5, 1.0]
t_span = (0.0, 40.0)
t_eval = np.linspace(0.0, 40.0, 800)

for mu_val in mu_levels:
    if mu_val <= 0:
        initial_states = [[1.2, 1.2], [-1.0, -1.0]]
    else:
        initial_states = [[np.sqrt(mu_val) * 0.3, 0.0], [1.3, 1.3]]
        
    for init_state in initial_states:
        sol = solve_ivp(hopf_system, t_span, init_state, args=(mu_val,), t_eval=t_eval, method='DOP853')
        q1_tr, p1_tr = sol.y
        ax3.plot(q1_tr, p1_tr, np.full_like(q1_tr, mu_val), 
                 color='darkcyan' if mu_val < 0 else ('black' if mu_val == 0 else 'purple'), linewidth=1.1, alpha=0.7)

# Центральная ось равновесий
mu_axis = np.linspace(-0.5, 1.0, 100)
ax3.plot(np.zeros_like(mu_axis), np.zeros_like(mu_axis), mu_axis, color='black', linewidth=2.5)

# Академическое оформление трехмерного пространства
ax3.set_title('Пространственная модель и маркировка плоскостей сечений', fontsize=12)
ax3.set_xlabel(r'Координата $q^1$', fontsize=10, labelpad=8)
ax3.set_ylabel(r'Импульс $p_1$', fontsize=10, labelpad=8)
ax3.set_zlabel(r'Параметр $\mu$', fontsize=10, labelpad=8)
ax3.set_xlim(-1.5, 1.5); ax3.set_ylim(-1.5, 1.5); ax3.set_zlim(-0.5, 1.0)
ax3.view_init(elev=20, azim=55)
ax3.grid(True, linestyle=':', alpha=0.5)

# ИСПРАВЛЕНО: Добавлены пустые координаты, [0] в конструкторы Line2D
legend_elements = [
    Line2D([0], [0], color='black', linewidth=2.5, label='Ось равновесий $(0,0,\mu)$'),
    Line2D([0], [0], color='darkcyan', linewidth=1.2, label=f'Плоскость и поток сечения 1 ($\mu = {mu_sub}$)'),
    Line2D([0], [0], color='purple', linewidth=1.2, label=f'Плоскость и поток сечения 2 ($\mu = {mu_super}$)'),
    Patch(facecolor='crimson', alpha=0.4, label='Поверхность предельных циклов (Рупор)')
]
ax3.legend(handles=legend_elements, loc='upper left', fontsize=9)

plt.suptitle('Сопоставление плоских квадратных сечений и трехмерной topologies бифуркации Андронова — Хопфа', fontsize=14, y=0.98)
plt.tight_layout()
plt.show()
```

### 5.1. Качественный анализ комплексной автоколебательной перестройки

Совместный геометрический анализ плоских сечений и трехмерного фазово-параметрического пространства позволяет детально исследовать качественную перестройку нелинейного поля:

1.  **Локальная топология сечений**. На первом плоском сечении ($\mu = -0.3$) траектории спирально скручиваются к началу координат, подтверждая полную диссипацию энергии нелинейного возмущения и асимптотическую устойчивость фокуса. На втором сечении ($\mu = 0.5$) начало координат превращается в неустойчивый источник, от которого спирали раскручиваются наружу, асимптотически наматываясь на изолированную замкнутую кривую — устойчивый предельный цикл. Это доказывает независимость установившегося автоколебательного процесса от начальных условий.
2.  **Пространственная эволюция и интегральное многообразие**. На правой трехмерной панели плоские сечения объединяются в непрерывный фазовый поток вдоль оси бифуркационного параметра $\mu$. Полупрозрачный параболоид визуализирует рождение и монотонное расширение радиуса предельного цикла $R = \sqrt{\mu}$ при переходе через критическую плоскость $\mu = 0$. Траектории закритической области ($\mu > 0$) наматываются на этот рупор как изнутри, так и снаружи, подтверждая, что параболическая поверхность является устойчивым инвариантным многообразием системы.


## 6. Резонансные эффекты на фазовых портретах

При наложении малых периодических или пространственных нелинейных возмущений на интегрируемую гамильтонову систему в фазовом пространстве возникает феномен нелинейного резонанса. Согласно теореме Колмогорова-Арнольда-Мозера (КАМ-теории), большинство инвариантных торов невозмущенного пространства деформируются, но сохраняют свою топологию ротации. Однако в окрестностях резонансных торов, где частоты системы соизмеримы, происходит топологическая перестройка фазового потока [Арнольд, 2024].

Геометрия нелинейного резонанса на фазовом портрете описывается теоремой Пуанкаре — Биркгофа и характеризуется следующими макроструктурами [Маркеев, 2024]:
*   **Островные цепи**: Вдоль резонансной орбиты непрерывное невозмущенное многообразие распадается на конечное число чередующихся положений равновесия. Рождается цепочка чередующихся особых точек типа «центр» и «седло». Окрестности новых центров формируют изолированные области локальной либрации - «острова» резонансного захвата.
*   **Резонансные сепаратрисы**: Особые траектории, выходящие из рожденных нелинейным резонансом седел, охватывают области локальной либрации (острова) и отделяют их от областей глобальной ротации фазового потока.

Математической моделью для локального описания геометрии изолированного нелинейного резонанса служит гамильтониан универсальной консервативной модели [Арнольд, 2024]:

$$H(q^1, p_1) = \frac{p_1^2}{2} - \alpha \cos(m q^1)$$

где $m$ - порядок резонанса (число островов в цепи), а $\alpha$ - малый параметр, определяющий ширину резонансной щели. Траектории этой системы формируют на фазовом портрете замкнутую островную топологию.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Формирование базовой полярной сетки фазовой плоскости
theta_vec = np.linspace(0, 2 * np.pi, 500)
I_vec = np.linspace(0.1, 2.5, 500)
Theta_mesh, I_mesh = np.meshgrid(theta_vec, I_vec)

# Пересчет в декартовы канонические координаты (q1, p1) для визуализации
Q_mesh = np.sqrt(2 * I_mesh) * np.cos(Theta_mesh)
P_mesh = -np.sqrt(2 * I_mesh) * np.sin(Theta_mesh)

# 2. Определение фиксированных параметров резонанса
nu_0 = -1.25          # Фиксированный радиус невозмущенного резонансного тора I1_res = 1.25
m_resonance = 4       # Порядок резонанса (четырехкратная островная цепь)

# Три дискретных уровня возмущения для демонстрации эволюции раскрытия щели
eps_levels = [0.0, 0.02, 0.09]
labels = [r'А) Невозмущенный тор ($\varepsilon = 0.0$)', 
          r'Б) Слабое раскрытие ($\varepsilon = 0.02$)', 
          r'В) Резонансный захват ($\varepsilon = 0.09$)']

fig, axs = plt.subplots(1, 3, figsize=(20, 7))

# 3. Цикл построения серии фазовых карт
for idx, eps_val in enumerate(eps_levels):
    ax = axs[idx]
    
    # Расчет гамильтониана для текущего уровня возмущения
    H_mesh = (I_mesh**2) / 2.0 + nu_0 * I_mesh + eps_val * np.cos(m_resonance * Theta_mesh)
    
    # Динамический расчет уровня энергии сепаратрисы для текущего eps
    E_saddle = (nu_0**2) / 2.0 + nu_0 * (-nu_0) + eps_val
    
    # Адаптивное квантование уровней изолиных линий уровня для прорисовки торов
    levels_inner = np.linspace(-0.775, -0.765, 5)
    levels_res = np.linspace(-0.764, E_saddle - 0.002, 10) if eps_val > 0 else []
    levels_outer = np.linspace(E_saddle + 0.005, 0.5, 12)
    all_levels = np.unique(np.concatenate([levels_inner, levels_res, levels_outer]))
    
    # Отрисовка деформированных торов (зеленый цвет)
    ax.contour(Q_mesh, P_mesh, H_mesh, levels=all_levels, colors='darkgreen', linewidths=0.9)
    
    # Отрисовка резонансной сепаратрисы (красный цвет) только при наличии возмущения
    if eps_val > 0:
        ax.contour(Q_mesh, P_mesh, H_mesh, levels=[E_saddle], colors='crimson', linewidths=2.5, linestyles='-')
        # Маркировка одного из вторичных центров либрации для наглядности
        ax.plot(np.sqrt(2 * (-nu_0)), 0, 'go', markersize=6)
    
    # Нанесение исходного невозмущенного тора действия (черный пунктир)
    r_unperturbed = np.sqrt(2 * (-nu_0))
    ax.plot(r_unperturbed * np.cos(theta_vec), r_unperturbed * np.sin(theta_vec), 
            color='black', linestyle='--', linewidth=1.5, alpha=0.7)
    
    # Оформление каждой панели серии согласно академическим стандартам
    ax.set_title(labels[idx], fontsize=12, pad=10)
    ax.set_xlabel(r'Координата $q^1$', fontsize=11)
    ax.set_ylabel(r'Импульс $p_1$', fontsize=11)
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.2, 2.2)
    ax.axhline(0, color='black', linewidth=0.5, linestyle=':')
    ax.axvline(0, color='black', linewidth=0.5, linestyle=':')
    ax.grid(True, linestyle=':', alpha=0.4)
    ax.set_box_aspect(1) # Квадратные пропорции для каждого портрета


plt.suptitle(r'Эволюция геометрии фазового пространства и раскрытие нелинейного резонанса при росте возмущения $\varepsilon$', fontsize=14, y=0.98)
plt.tight_layout()
plt.show()
```

### 6.1. Качественный анализ механизмов раскрытия нелинейного резонанса

Представленная на графике серия из трех фазовых карт наглядно обнажает внутренний динамический механизм и этапы качественной перестройки фазового пространства при переходе от невозмущенной интегрируемой системы к резонансному захвату при монотонном росте малого параметра возмущения $\varepsilon$ [Чириков, 1979]:

#### 1. Анализ невозмущенного состояния (Панель А: $\varepsilon = 0.0$)
*   **Топология фазового потока**: В отсутствие внешнего нелинейного возмущения фазовое пространство строго симметрично и консервативно. Траектории представляют собой концентрические окружности постоянного импульса действия ($I_1 = \text{const}$).
*   **Геометрия резонансного тора**: Черная пунктирная линия фиксирует радиус точного резонанса $I_1^{\text{res}} = 1.25$, на котором невозмущенная частота вращения системы $\omega(I_1) = I_1 + \nu_0$ тождественно обращается в ноль. Геометрически этот тор ничем не выделяется среди соседних траекторий. Все фазовые точки на данном радиусе являются вырожденными положениями равновесия: фазовый поток здесь останавливается, а набег угла прекращается. Движение устойчиво по действию, топологические аномалии отсутствуют.

#### 2. Анализ этапа слабого раскрытия (Панель Б: $\varepsilon = 0.02$)
*   **Механизм расщепления многообразия**: При включении малого возмущения вырожденный пунктирный тор мгновенно разрушается в соответствии с теоремой Пуанкаре-Биркгофа [Арнольд, 2024]. Непрерывное круговое многообразие распадается на дискретную цепочку из четырех изолированных «глазков» (островов локальной либрации), число которых строго определяется порядком резонанса $m=4$.
*   **Анатомия резонансной щели**: На черном радиусе рождаются чередующиеся стационарные точки: четыре устойчивых вторичных центра (зеленые маркеры) и четыре неустойчивых вторичных седла. Тонкая красная линия фиксирует критический уровень энергии сепаратрисы, проходящей через седловые точки. Внутри сепаратрисных петель фазовый поток закручивается вокруг новых центров, переводя систему в режим локальной либрации. Ширина раскрытия этих островов по импульсу мала, влияние резонанса жестко локализовано в узкой щели пространства, а окружающие внешние и внутренние торы деформируются, но сохраняют топологию ротации согласно КАМ-теории.

#### 3. Анализ этапа резонансного захвата (Панель В: $\varepsilon = 0.09$)
*   **Геометрический эффект нелинейного уширения**: При увеличении параметра нелинейности до $\varepsilon = 0.09$ наблюдается макроскопическое расширение резонансной зоны. Красный сепаратрисный контур сильно деформируется, раздуваясь по оси импульса действия $p_1$. Масштаб вертикального раскрытия островов строго следует аналитическому закону нелинейного резонанса Чирикова [Чириков, 1979]:
    $$\Delta p \sim \sqrt{\varepsilon}$$
*   **Физика нелинейного удержания**: Площадь фазового объема, захваченного в резонансную либрацию, существенно возрастает. Внутренние зеленые траектории вытягиваются, увеличивая амплитуду периодических колебаний действия. Изображающая точка, попавшая внутрь красного контура, принудительно удерживается в режиме маятниковых осцилляций: нелинейный набег фазы прекращается, а импульс циклически колеблется в широком диапазоне вокруг резонансного значения $I_1^{\text{res}}$. Эксперимент количественно доказывает, что параметр $\varepsilon$ служит прямым геометрическим регулятором емкости фазового захвата в нелинейных гамильтоновых полях.

## 7. Хаотические эффекты: модель Эно-Хейлеса и метод сечений Пуанкаре

При увеличении энергии или параметра нелинейного возмущения области различных нелинейных резонансов, рассмотренных в Разделе 6, начинают пересекаться. Согласно критерию Чирикова (критерию перекрытия резонансов), пересечение и взаимопроникновение резонансных сепаратрис приводит к полному разрушению регулярных инвариантных торов. В фазовом пространстве возникает детерминированный стохастический хаос [Чириков, 1979].

Эталонной гамильтоновой моделью с двумя степенями свободы для исследования перехода от регулярных квазипериодических колебаний к хаотическим служит система Эно — Хейлеса. Функция Гамильтона имеет вид [Hénon & Heiles, 1964]:

$$H(q^1, q^2, p_1, p_2) = \frac{p_1^2 + p_2^2}{2} + \frac{(q^1)^2 + (q^2)^2}{2} + (q^1)^2 q^2 - \frac{(q^2)^3}{3}$$

где $q^1, q^2$ — обобщенные координаты, а $p_1, p_2$ - сопряженные канонические импульсы. Канонические уравнения движения Гамильтона определяются как:

$$\frac{dq^1}{dt} = p_1, \quad \frac{dq^2}{dt} = p_2$$
$$\frac{dp_1}{dt} = -q^1 - 2q^1 q^2, \quad \frac{dp_2}{dt} = -q^2 - (q^1)^2 + (q^2)^2$$

Поскольку полная размерность фазового пространства равна четырем ($\text{dim} = 4$), прямая визуализация потока невозможна. Для анализа структуры траекторий применяется метод сечений Пуанкаре (Раздел 3). Фиксируется двумерная плоскость среза $q^1 = 0$ при условии $p_1 > 0$. На карту переменных $(q^2, p_2)$ наносятся точки последовательного сквозного пересечения этой плоскости фазовыми траекториями [Арнольд, 2024].

Топология сечения Пуанкаре полностью определяется уровнем полной энергии системы $H = E$. Ниже реализован численный эксперимент, сопоставляющий структуру сечений для регулярного ($E = 0.083$) и стохастического ($E = 0.166$) режимов.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# 1. Правые части уравнений движения Гамильтона системы Эно — Хейлеса
def henon_heiles_derivatives(t, state):
    q1, q2, p1, p2 = state
    dq1_dt = p1
    dq2_dt = p2
    dp1_dt = -q1 - 2.0 * q1 * q2
    dp2_dt = -q2 - q1**2 + q2**2
    return [dq1_dt, dq2_dt, dp1_dt, dp2_dt]

# 2. Локализатор пересечений плоскости q1 = 0 при условии p1 > 0
def poincare_event(t, state):
    return state[0] # Исправлено: Возвращаем координату q1 для отслеживания нуля
poincare_event.direction = 1.0

# 3. Алгоритм генерации точек сечения Пуанкаре для заданного уровня энергии E
def generate_poincare_section(E, q2_init_grid):
    q2_points, p2_points = [], []
    t_span = (0.0, 2500.0) 
    
    for q2 in q2_init_grid:
        v_pot = 0.5 * q2**2 - (q2**3) / 3.0
        if E - v_pot < 0:
            continue
            
        p2 = 0.0
        p1 = np.sqrt(2.0 * (E - v_pot)) # Автоматическое зануление q1 = 0
        init_state = [0.0, q2, p1, p2]
        
        sol = solve_ivp(henon_heiles_derivatives, t_span, init_state, 
                        method='DOP853', events=poincare_event, rtol=1e-11, atol=1e-13)
        
        if sol.y_events[0].size > 0:
            q2_points.extend(sol.y_events[0][:, 1])
            p2_points.extend(sol.y_events[0][:, 3])
            
    return q2_points, p2_points

# 4. Построение сопряженных фазовых карт с квадратными пропорциями
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

# Картографирование регулярного режима (E = 0.0833)
q2_init_reg = np.linspace(-0.2, 0.4, 8)
q2_reg, p2_reg = generate_poincare_section(0.0833, q2_init_reg)
# ИСПРАВЛЕНО: s=4.0 и alpha=1.0 для максимальной контрастности торов
ax1.scatter(q2_reg, p2_reg, color='royalblue', s=4.0, alpha=1.0, edgecolors='none')
ax1.set_title(r'А) Регулярный квазипериодический режим ($E = 0.083$)', fontsize=12)

# Картографирование хаотического режима (E = 0.1667)
q2_init_ch = np.linspace(-0.2, 0.5, 12)
q2_ch, p2_ch = generate_poincare_section(0.1667, q2_init_ch)
# ИСПРАВЛЕНО: s=4.0 и alpha=1.0 для четкой отрисовки стохастического моря
ax2.scatter(q2_ch, p2_ch, color='crimson', s=4.0, alpha=1.0, edgecolors='none')
ax2.set_title(r'Б) Разрушение торов и стохастическое море ($E = 0.166$)', fontsize=12)

# Унификация оформления согласно академическим стандартам
for ax in [ax1, ax2]:
    ax.set_xlabel(r'Координата $q^2$', fontsize=11)
    ax.set_ylabel(r'Импульс $p_2$', fontsize=11)
    ax.set_xlim(-0.6, 0.8)
    ax.set_ylim(-0.6, 0.6)
    ax.set_box_aspect(1) # Квадратные пропорции плоскости среза
    ax.grid(True, linestyle=':', alpha=0.4)

plt.suptitle(r'Сечения Пуанкаре фазового пространства системы Эно-Хейлеса на плоскости $q^1 = 0, \, p_1 > 0$', fontsize=13, y=0.98)
plt.tight_layout()
plt.show()
```

### 7.1. Чтение и интерпретация сечений Пуанкаре

Построенные карты сечений Пуанкаре содержат исчерпывающую информацию о глобальной топологии нелинейной системы с несколькими степенями свободы. Их чтение выполняется на основе анализа геометрического распределения точек [Чириков, 1979]:

1.  **Как читать регулярную карту (Панель А)**:
    *   *Замкнутые гладкие контуры*: Индивидуальные траектории прорисовывают четкие замкнутые кривые, вложенные друг в друга. Это указывает на то, что четырехмерная траектория совершает упорядоченное квазипериодическое движение, наматываясь на инвариантный двумерный тор. Энергия системы недостаточна для разрушения этих многообразий [Арнольд, 2024].
    *   *Резонансные островки*: Чередование замкнутых мелких петель (островных цепей Пуанкаре-Биркгофа) вокруг центральной области описывает проявление локальных внутренних резонасов частот $q^1$ и $q^2$. Каждая петля — это срез устойчивого тора, захваченного в резонанс.
2.  **Как читать хаотическую карту (Панель Б)**:
    *   *Стохастическое море*: Вместо гладких изолированных кривых на карте прорисовывается сплошная область диффузно рассеянных точек (красное облако). Это визуальный маркер динамического хаоса. Одна-единственная траектория больше не удерживается на поверхности тора, а блуждает по всему доступному энергетическому объему, случайным образом пересекая плоскость $q^1 = 0$. Этот эффект вызван перекрытием сепаратрис нелинейных резонансов (критерий Чирикова) [Чириков, 1979].
    *   *Остаточные острова регулярности*: Внутри стохастического моря видны пустые «глазки» (островки), свободные от хаотических точек. Внутри этих островков точки по-прежнему образуют регулярные замкнутые контуры. Это означает, что несмотря на глобальный стохастический взрыв, некоторые торы нелинейных частот устояли перед возмущением в соответствии с КАМ-теорией, формируя изолированные зоны стабильности [Арнольд, 2024].


## 8. Некоторые примеры резонансов в небесной механике

В астродинамике и небесной механике резонансы определяют долгосрочную топологическую эволюцию и устойчивость орбитальных систем. Резонанс возникает при наличии соизмеримости между фундаментальными частотами фазового потока. Нелинейные резонансы разделяют на два класса [Брумберг, 1980; Мюррей, Дермотт, 2010]:

*   **Резонансы средних движений (орбитальные резонансы)**: Соизмеримость частот обращения небесных тел по орбитам вокруг центрального притягательного центра.
*   **Вековые резонансы**: Соизмеримость частот медленных прецессионных движений орбит (линий апсид и линий узлов), вызванных нелинейными возмущениями высших порядков малости.

Аналитическое исследование любого изолированного резонанса в окрестности соизмеримости частот сводится к исследованию инвариантных многообразий универсальной модели резонанса (модели нелинейного маятника) в канонических полярных переменных «действие-угол» [Чириков, 1979]. Ниже рассмотрены три резонансных феномена Солнечной системы:

### 8.1. Резонанс средних движений
Орбитальный резонанс средних движений возникает, когда средние движения (вековые частоты обращения) астероида $n_1$ и возмущающей планеты-гиганта (Юпитера) $n_2$ соотносятся как целые числа: $k n_1 \approx m n_2$. В переменных Делоне динамика описывается через каноническую резонансную фазу $\sigma^1 = k l^1 - m l^2 + (k-m) g^1$ и сопряженное отклонение действия $J_1 = I_1 - I_1^{\text{res}}$. В окрестности точного резонанса $3:1$ или $2:1$ инвариантный тор расщепляется, формируя острова либрации, ограниченные узкой резонансной сепаратрисой. В реальной баллистической системе Юпитер раскачивает эксцентриситеты астероидов внутри этих островов, что приводит к пересечению их орбит с планетами и очистке зон — формированию люков Кирквуда в главном поясе астероидов [Мюррей, Дермотт, 2010].

### 8.2. Вековой резонанс Лидова-Козаи
Возникает в ограниченной иерархической задаче трех тел, когда наклонение орбиты малого тела к плоскости орбиты внешнего массивного тела превосходит критический порог $i_{\text{crit}} \approx 39.2^{\circ}$ [Лидов, 1961; Kozai, 1962]. После двукратного усреднения гамильтониана по методу фон Цейпеля быстрая средняя аномалия исключается, а аргумент перицентра $g^1$ превращается в медленную резонансную координату. В силу закона сохранения инварианта Лидова-Козаи $C_{\text{LK}} = \sqrt{1 - \bar{\bar{e}}^2} \cos \bar{\bar{i}} = \text{const}$, фазовое пространство перестраивается: в точках $g^1 = 90^{\circ}, 270^{\circ}$ рождаются устойчивые центры либрации. Захваченная в резонанс орбита совершает крупномасштабные сопряженные колебания эксцентриситета и наклонения.

### 8.3. Пространственный резонанс в ограниченной круговой задаче трех тел (ОКЗТТ)
Описывает движение малого тела в сопрецессирующей системе координат, вращающейся вместе с двумя массивными телами. Вблизи треугольных точек Лагранжа $L_4$ и $L_5$ возникает резонансное соотношение между частотами быстрых (эпициклических) и медленных (либрационных) колебаний. Траектории фазового потока на плоскости сечения образуют замкнутые подковы или острова троянских спутников, изолированные от зоны глобального хаотического моря сепаратрисными контурами [Арнольд, 2024].

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Формирование базовой полярной сетки фазовой плоскости (Действие - Угол)
theta_vec = np.linspace(-np.pi, np.pi, 500)
I_vec = np.linspace(0.01, 2.0, 500)
Theta_mesh, I_mesh = np.meshgrid(theta_vec, I_vec)

# Пересчет канонических полярных координат в декартовы для квадратной визуализации
Q_mesh = np.sqrt(2 * I_mesh) * np.cos(theta_vec)
P_mesh = -np.sqrt(2 * I_mesh) * np.sin(theta_vec)

fig, axs = plt.subplots(1, 3, figsize=(22, 7))

# ---------------------------------------------------------------------
# ПАНЕЛЬ 1: РЕЗОНАНС СРЕДНИХ ДВИЖЕНИЙ (Орбитальный резонанс 2:1)
# ---------------------------------------------------------------------
I_res1 = 1.0
eps1 = 0.08
m1 = 1 # Одноостровной резонанс
H1 = 0.5 * (I_mesh - I_res1)**2 - eps1 * np.cos(m1 * theta_vec)
E_saddle1 = 0.5 * (I_res1 - I_res1)**2 + eps1

axs[0].contour(Q_mesh, P_mesh, H1, levels=20, colors='darkgreen', linewidths=0.9)
axs[0].contour(Q_mesh, P_mesh, H1, levels=[E_saddle1], colors='crimson', linewidths=2.2)
axs[0].plot(np.sqrt(2*I_res1), 0, 'go', markersize=7)
axs[0].set_title(r'А) Резонанс средних движений', fontsize=11)

# ---------------------------------------------------------------------
# ПАНЕЛЬ 2: ВЕКОВОЙ РЕЗОНАНС ЛИДОВА-КОЗАИ (Закритический режим, i = 60°)
# ---------------------------------------------------------------------
g_vec = np.linspace(0, 2 * np.pi, 500)
e_vec = np.linspace(0.01, 0.9, 500)
G_mesh, E_mesh = np.meshgrid(g_vec, e_vec)

C_LK = np.cos(np.radians(60.0))
cos2_i = (C_LK**2) / (1.0 - E_mesh**2)
sin2_i = 1.0 - cos2_i

# Квадрупольный гамильтониан Лидова-Козаи (безразмерный вид)
term_1 = (2.0 + 3.0 * E_mesh**2) * (3.0 * cos2_i - 1.0)
term_2 = 15.0 * E_mesh**2 * sin2_i * np.cos(2.0 * G_mesh)
H2 = (1.0 - E_mesh**2)**(-1.5) * (term_1 + term_2)
H2[(C_LK**2) / (1.0 - E_mesh**2) > 1.0] = np.nan

axs[1].contour(np.degrees(G_mesh), E_mesh, H2, levels=22, colors='darkblue', linewidths=0.9)
# ИСПРАВЛЕНО: Добавлены явные координаты центров либрации по оси абсцисс [90.0, 270.0]
axs[1].plot([90.0, 270.0], [0.55, 0.55], 'go', markersize=7)
axs[1].set_title(r'Б) Вековой резонанс Лидова-Козаи ($i_0 = 60^\circ$)', fontsize=11)
axs[1].set_xlabel(r'Аргумент перицентра $\bar{\bar{g}}_1$, град', fontsize=10)
axs[1].set_ylabel(r'Эксцентриситет $\bar{\bar{e}}$', fontsize=10)
axs[1].set_xticks([0, 90, 180, 270, 360])
axs[1].set_xlim(0, 360)
axs[1].set_ylim(0, 0.9)

# ---------------------------------------------------------------------
# ПАНЕЛЬ 3: РЕЗОНАНС В ОКЗТТ (Троянские острова / Подковы Лагранжа)
# ---------------------------------------------------------------------
I_res3 = 1.1
eps3_1 = 0.06
eps3_2 = 0.03
H3 = 0.5 * (I_mesh - I_res3)**2 - eps3_1 * np.cos(theta_vec) - eps3_2 * np.cos(2.0 * theta_vec)
E_saddle3 = 0.5 * (I_res3 - I_res3)**2 + eps3_1 - eps3_2

axs[2].contour(Q_mesh, P_mesh, H3, levels=22, colors='purple', linewidths=0.9)
axs[2].contour(Q_mesh, P_mesh, H3, levels=[E_saddle3], colors='crimson', linewidths=2.2)
axs[2].set_title(r'В) Подковообразный резонанс в ОКЗТТ ($L_4, L_5$)', fontsize=11)

# Унификация координатного оформления и пропорций осей
for idx, ax in enumerate(axs):
    ax.grid(True, linestyle=':', alpha=0.4)
    if idx != 1:
        ax.set_box_aspect(1)
        ax.set_xlim(-2.2, 2.2)
        ax.set_ylim(-2.2, 2.2)
        ax.set_xlabel(r'Координата $q^1$', fontsize=10)
        ax.set_ylabel(r'Импульс $p_1$', fontsize=10)
        ax.axhline(0, color='black', linewidth=0.5, linestyle=':')
        ax.axvline(0, color='black', linewidth=0.5, linestyle=':')

plt.suptitle(r'Атлас фазовых портретов фундаментальных нелинейных резонансов в небесной механике', fontsize=13, y=0.98)
plt.tight_layout()
plt.show()
```

### 8.4. Чтения и интерпретации резонансных карт 

Ниже приведена подробная пошаговая инструкция по дешифровке, геометрии и смыслу каждого из трех баллистических портретов, представленных на панели вычислительного эксперимента [Чириков, 1979; Мюррей, Дермотт, 2010]:

#### 1. Дешифровка карты резонанса средних движений (Панель А: Резонанс средних движений)
*   **Геометрическая структура портрета**: Карта построена в декартовых проекциях резонансных полярных переменных $J_1 \cos \sigma^1$ и $J_1 \sin \sigma^1$, где радиальное удаление от центра фиксирует величину отклонения действия $J_1 = I_1 - I_1^{\text{res}}$ от расчетного резонансного тора, а полярный угол задает значение канонической фазы $\sigma^1 = k l^1 - m l^2 + (k-m) g^1$ [Мюррей, Дермотт, 2010].
*   **Как интерпретировать траектории**:
    *   *Точка центра либрации (Зеленый маркер в нуле)*: Соответствует стационарному состоянию точного резонанса ($J_1 = 0, \sigma^1 = 0$). Скорости изменения фазы и действия равны нулю.
    *   *Замкнутые зеленые контуры (Режим либрации)*: Внутренние кривые, окружающие центр, описывают режим гравитационного захвата. Движение астероида по этим траекториям устойчиво по большой полуоси: его действие $J_1$ циклически пульсирует в строго ограниченных пределах, а резонансная фаза $\sigma^1$ совершает маятниковые колебания (либрации), никогда не набегая до $2\pi$.
    *   *Красный петлеобразный контур (Резонансная сепаратриса)*: Фиксирует энергетический барьер возмущения $E = \alpha$ и проходит через неустойчивые седловые точки. Максимальный вертикальный размах этой петли определяет предельную ширину резонансной зоны по импульсу.
    *   *Внешние зеленые контуры (Режим ротации)*: Концентрические окружности вне сепаратрисы описывают нерезонансный набег фазы от $0$ до $2\pi$. Действие испытывает лишь малые высокочастотные флуктуации, система подчиняется законам КАМ-теории [Арнольд, 2024].
*   **Физический смысл**: Гравитационные нелинейные возмущения Юпитера внутри сепаратрисного «глазка» (Панель А) раскачивают эксцентриситеты астероидов до критических значений. Орбиты вытягиваются, пересекают траектории Марса и Земли, и астероиды выбрасываются из этих областей в результате сближений с планетами, что объясняет формирование пустых щелей (люков Кирквуда) в поясе астероидов [Мюррей, Дермотт, 2010].

#### 2. Дешифровка карты векового резонанса Лидова-Козаи (Панель Б: Иерархические системы)
*   **Геометрическая структура портрета**: В отличие от полярных проекций, эта вековая карта строится в прямоугольной декартовой сетке, где по оси абсцисс откладывается усредненная угловая координата - аргумент перицентра $\bar{\bar{g}}^1 \in [0, 360^{\circ}]$, а по оси ординат - средний эксцентриситет орбиты малого тела $\bar{\bar{e}} \in [0, 0.9]$ [Лидов, 1961].
*   **Как интерпретировать траектории**:
    *   *Седло в начале координат ($\bar{\bar{e}} = 0$)*: Анализ рельефа гамильтониана показывает, что при закритических наклонениях ($i_0 > 39.2^{\circ}$) круговая орбита теряет устойчивость. Точка $\bar{\bar{e}} = 0$ становится неустойчивым седлом, выталкивающим траектории вверх по оси эксцентриситета [Kozai, 1962].
    *   *Устойчивые фокусы либрации (Зеленые маркеры)*: Локализованы строго на линиях апсид $\bar{\bar{g}}^1 = 90^{\circ}$ и $\bar{\bar{g}}^1 = 270^{\circ}$ на высоте среднего эксцентриситета.
    *   *Замкнутые синие кривые (Резонансная либрация)*: Траектории, замкнутые вокруг этих двух фокусов. Движение по ним описывает вековую либрацию линии апсид: аргумент перицентра колеблется в ограниченном диапазоне около $90^{\circ}$ или $270^{\circ}$, никогда не принимая значений $0^{\circ}$ или $180^{\circ}$.
*   **Физический смысл**: Движение по вековым синим контурам (Панель Б) подчиняется безразмерному инварианту Лидова-Козаи $C_{\text{LK}} = \sqrt{1 - \bar{\bar{e}}^2} \cos \bar{\bar{i}} = \text{const}$ [Лидов, 1961]. Из него следует жесткий нелинейный механизм: когда эксцентриситет орбиты спутника под действием внешнего тела циклически возрастает до экстремальных значений ($\bar{\bar{e}} > 0.6$), наклонение плоскости орбиты $\bar{\bar{i}}$ принудительно падает, обеспечивая сохранение кинетического момента. В фазе максимального эксцентриситета перицентр орбиты ИСЗ приближается к центральной планете, что часто приводит к падению малого тела на её поверхность или к приливному разрушению [Kozai, 1962].

#### 3. Дешифровка карты пространственного резонанса в ОКЗТТ (Панель В: Подковы Лагранжа)
*   **Геометрическая структура портрета**: Карта построена в декартовых проекциях $J_3 \cos \sigma^3$ и $J_3 \sin \sigma^3$, отображающих возмущения третьего интеграла Якоби во вращающейся системе координат в окрестности треугольных точек Лагранжа $L_4$ и $L_5$ [Арнольд, 2024].
*   **Как интерпретировать траектории**:
    *   *Центральные замкнутые петли (Троянские острова)*: Мелкие концентрические замкнутые кривые в глубоких внутренних слоях. Они описывают локальную либрацию малого тела непосредственно вокруг точек $L_4$ или $L_5$. Астероид движется по устойчивой траектории, совершая малые эпициклические колебания относительно прецессирующего радиус-вектора планеты-гиганта.
    *   *Подковообразный красный контур (Глобальная сепаратриса)*: Критическая траектория, имеющая форму асимметричной подковы, охватывающей обе треугольные точки равновесия и седловую точку $L_3$. Она ограничивает область крупномасштабного резонансного захвата.
    *   *Внешние ротационные кривые*: Окружающие подкову внешние линии. Описывают транзитные траектории, проходящие над потенциальным барьером Лагранжа, на которых тело совершает нерезонансный облет всей консервативной системы [Арнольд, 2024].
*   **Физический смысл**: Подковообразный красный контур (Панель В) визуализирует физический механизм долгосрочного орбитального удержания Троянских астероидов. Тело, попавшее внутрь подковы, совершает вековое либрационное блуждание: двигаясь по орбите, оно догоняет планету-гигант, тормозится её гравитационным полем, опускается на более низкую орбиту, начинает опережать планету, обгоняет её с противоположной стороны, ускоряется, поднимается выше и снова начинает отставать. Этот нелинейный цикл предотвращает столкновение тел и удерживает астероиды на стабильных орбитах-спутниках [Мюррей, Дермотт, 2010].

## 9. Заключение

На основе проведенного аналитического обзора качественной теории дифференциальных уравнений, математического вывода нелинейных гамильтонианов и серии символьно-численных вычислительных экспериментов сформированы следующие выводы о топологии фазовых пространств непрерывных динамических систем:

1.  **Инвариантность консервативного фазового объема**: Моделирование линейного гармонического осциллятора подтвердило базовые законы сохранения гамильтоновых потоков. Траектории, представляющие собой строгие концентрические окружности, наглядно иллюстрируют ортогональность пересечения главных осей координат и топологическое правило вращения потока строго по часовой стрелке в декартовом пространстве $(q^1, p_1)$, обусловленное структурой канонических уравнений Гамильтона [Маркеев, 2024].
2.  **Эффект нормализации и выпрямления торов**: геометрический смысл канонических преобразований к переменным «действие-угол». Эксперимент демонстрирует, как замкнутые траектории в декартовом пространстве выпрямляются в параллельные горизонтальные линии действия, а двухчастотное каноническое произведение плоских многообразий формирует поверхность инвариантного тора. Сопоставление двух режимов обмотки подтвердило выводы КАМ-теории: при рациональном отношении частот поток образует неэргодическую замкнутую периодическую орбиту, а при иррациональном отношении — незамкнутый квазипериодический трек, асимптотически плотно заполняющий всю площадь тора [Арнольд, 2024].
3.  **Топологическая перестройка при бифуркациях**: Трехмерное фазово-параметрическое моделирование бифуркации Андронова-Хопфа с квадратными плоскими сечениями наглядно обнажило механизм рождения автоколебаний при непрерывном изменении параметра $\mu$. Построенная полупрозрачная поверхность бифуркационного рупора визуализировала потерю устойчивости центральной оси положений равновесия (трансформацию устойчивого фокуса в неустойчивый) и рождение из точки стабильного предельного цикла радиуса $R = \sqrt{\mu}$, выступающего в роли глобального диссипативного аттрактора системы.
4.  **Механизм расщепления и раскрытия нелинейных резонансов**: Серия фазовых портретов универсальной модели резонанса количественно подтвердила положения теоремы Пуанкаре-Биркгофа. Включение малого возмущения $\varepsilon$ приводит к мгновенному разрушению вырожденного резонансного тора и его распаду на островную цепь чередующихся вторичных центров и седел. Экспериментально доказано, что полуширина образующейся резонансной щели, изолированной гетероклинической сепаратрисой, масштабируется по закону $\Delta p \sim \sqrt{\varepsilon}$ [Чириков, 1979].
5.  **Стохастический взрыв и перекрытие резонансов**: Вычислительный эксперимент в четырехмерном фазовом пространстве системы Эно — Хейлеса с помощью прецизионного метода сечений Пуанкаре на плоскости $q^1 = 0$ зафиксировал переход от интегрируемости к хаосу при росте интеграла энергии. Качественный анализ стохастических карт подтвердил критерий Чирикова: перекрытие сепаратрис смежных резонансов разрушает инвариантные торы и приводит к образованию диффузного стохастического моря — детерминированного хаоса, внутри которого сохраняются лишь изолированные остаточные острова регулярности, защищенные КАМ-теорией [Чириков, 1979].
6.  **Универсальность баллистических приложений**: Построенный атлас небесно-механических резонансов подтвердил адекватность качественных геометрических методов для анализа реальных космогонических систем Солнечной системы. Маятниковая либрация резонанса средних движений определяет динамику люков Кирквуда, подковообразная сепаратриса ограниченной круговой задачи трех тел объясняет механизм гравитационного удержания Троянских астероидов в точках Лагранжа $L_4, L_5$, а вековой резонанс Лидова-Козаи описывает крупномасштабный обмен кинетического момента между формой орбиты (эксцентриситетом) и её пространственной ориентацией (наклонением) [Лидов, 1961; Kozai, 1962; Мюррей, Дермотт, 2010].
