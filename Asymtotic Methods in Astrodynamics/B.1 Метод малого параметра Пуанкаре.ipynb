# Метод малого параметра Пуанкаре (Линдштедта-Пуанкаре)

Метод Пуанкаре является классическим асимптотическим методом теории возмущений, который применяется для построения периодических решений слабонелинейных систем дифференциальных уравнений.

Прямое разложение искомого решения в формальный ряд Тейлора по малому параметру $\varepsilon$ согласно схеме Дюффинга приводит к появлению в высших приближениях вековых членов. Данные слагаемые, имеющие вид $t \sin(\omega_0 t)$, неограниченно возрастают при временном стремлении к бесконечности $t \to \infty$. Появление вековых членов нарушает периодичность асимптотического решения и делает его неприменимым для анализа динамики на больших интервалах времени.

Для преодоления данного ограничения была разработана процедура деформации временной шкалы. Метод базируется на идее, исторически предложенной А. Линдштедтом для решения задач спутниковой динамики [Lindstedt, 1882] и впоследствии строго математически обоснованной А. Пуанкаре для широких классов автономных систем [Poincaré, 1892]. В начале XX века этот математический аппарат был успешно перенесен Б. Ван дер Полем из области чистой астрономии в радиофизику для анализа нелинейных автоколебательных процессов в ламповых генераторах [Van der Pol, 1926]. Метод Линдштедта-Пуанкаре основан на одновременном разложении в формальные степенные ряды по малому параметру как вектора фазовых координат системы $x(t)$, так и самой частоты нелинейных колебаний $\omega$, что позволяет полностью устранить резонансные вековые слагаемые на этапе интегрирования уравнений высших приближений.


## 1. Постановка задачи и методологическое обоснование деформации времени

Рассмотрим слабонелинейный автономный осциллятор с одной степенью свободы, динамика которого описывается дифференциальным уравнением второго порядка:
$$\frac{d^2 x}{dt^2} + \omega_0^2 x = \varepsilon g\left(x, \frac{dx}{dt}\right) \tag{1}$$
где параметр $\varepsilon \ll 1$ является малым параметром возмущения, а функция $g\left(x, \frac{dx}{dt}\right)$ является гладкой нелинейной функцией своих аргументов.

Физическая причина появления расходимостей при прямом разложении решения уравнения (1) по малому параметру заключается в том, что в нелинейных динамических системах период автономных колебаний зависит от их амплитуды. Если истинная частота смещается на величину порядка $O(\varepsilon)$, то точное решение колеблется с частотой $\omega_0 + \varepsilon \omega_1$. Разложение такой тригонометрической функции в ряд Тейлора по малому параметру $\varepsilon$ порождает члены вида $\cos(\omega_0 t) - \varepsilon \omega_1 t \sin(\omega_0 t) + \dots$. Таким образом, вековые слагаемые являются следствием неадекватного выбора временной шкалы, не учитывающей изменение частоты. Для восстановления периодичности асимптотического решения необходимо произвести деформацию временной оси.

Введем новую безразмерную фазовую переменную $\tau$, которая явно учитывает нелинейный сдвиг частоты системы:
$$\tau = \omega t \tag{2}$$

Операторы полного дифференцирования по времени трансформируются в соответствии со следующими соотношениями:
$$\frac{d}{dt} = \omega \frac{d}{d\tau}, \quad \frac{d^2}{dt^2} = \omega^2 \frac{d^2}{d\tau^2} \tag{3}$$

Преобразование исходного дифференциального уравнения движения (1) в деформированную шкалу времени $\tau$ приводит его к виду:
$$\omega^2 \frac{d^2 x}{d\tau^2} + \omega_0^2 x = \varepsilon g\left(x, \omega \frac{dx}{d\tau}\right) \tag{4}$$

Согласно методу Линдштедта-Пуанкаре, искомое периодическое решение $x(\tau)$ и неизвестная преобразованная частота $\omega$ представляются в виде формальных асимптотических степенных рядов по малому параметру $\varepsilon$:
$$x(\tau) = x_0(\tau) + \varepsilon x_1(\tau) + \varepsilon^2 x_2(\tau) + \dots \tag{5}$$
$$\omega = \omega_0 + \varepsilon \omega_1 + \varepsilon^2 \omega_2 + \dots \tag{6}$$

Построение ряда (6) гарантирует, что в деформированном временном базисе решение $x(\tau)$ остается строго $2\pi$-периодическим по фазе $\tau$ и свободным от вековых отклонений в любом порядке малости по параметру $\varepsilon$.

## 2. Причина расходимости ряда Тейлора на примере уравнения Дюффинга

Для демонстрации ограничений прямого разложения по малому параметру рассмотрим автономное нелинейное уравнение Дюффинга:
$$ \frac{d^2 x}{dt^2} + x + \varepsilon x^3 = 0 \tag{7} $$
при заданных начальных условиях $x(0) = 1$ и $\frac{dx(0)}{dt} = 0$. Искомое решение аппроксимируется формальным степенным рядом: $x(t) = x_0(t) + \varepsilon x_1(t) + \dots$.

Для нулевого порядка приближения по малому параметру $\varepsilon^0$ формируется уравнение гармонического осциллятора, решение которого с учетом начальных условий принимает вид: $x_0(t) = \cos t$.

Приравнивая слагаемые при первой степени малого параметра $\varepsilon^1$, получаем линейное неоднородное дифференциальное уравнение для поправки первого порядка $x_1(t)$:
$$ \frac{d^2 x_1}{dt^2} + x_1 = -x_0^3 = -\cos^3 t = -\frac{3}{4}\cos t - \frac{1}{4}\cos(3t) \tag{8} $$

Поскольку начальные условия задачи удовлетворяются решением нулевого порядка, для всех высших поправок ставятся однородные начальные условия: $x_1(0) = 0$ и $\frac{dx_1(0)}{dt} = 0$. Решение уравнения (8) в этом случае имеет вид:
$$ x_1(t) = -\frac{3}{8} t \sin t + \frac{1}{32}\left(\cos(3t) - \cos t\right) \tag{9} $$

#### Анализ векового члена

Первое слагаемое в выражении (9), имеющее вид $-\frac{3}{8} t \sin t$, является вековым членом. Из-за наличия линейно растущего со временем сомножителя $t$ эта поправка монотонно увеличивается по амплитуде. На интервалах времени порядка $t \sim O(\varepsilon^{-1})$ асимптотический член $\varepsilon x_1(t)$ достигает величины порядка единицы, сравниваясь по амплитуде с главным решением $x_0(t)$. Данное обстоятельство нарушает упорядоченность асимптотического ряда по малому параметру и делает разложение неприменимым для долгосрочного прогнозирования нелинейной динамики.

Метод Линдштедта-Пуанкаре устраняет эту расходимость посредством перехода к деформированной временной переменной $\tau = \omega t$. Введение ряда (6) для неизвестной нелинейной частоты $\omega$ порождает свободные параметры $\omega_i$ в дифференциальных уравнениях высших приближений. Подбор данных параметров позволяет принудительно обнулить коэффициенты при резонансных гармониках вида $\cos \tau$ в правых частях уравнений, тем самым полностью исключая появление вековых слагаемых в структуре асимптотического решения.

## 3. Применение метода Линдштедта-Пуанкаре к осциллятору Дюффинга

Применим метод Линдштедта-Пуанкаре к автономному уравнению с кубической нелинейностью при невозмущенной частоте $\omega_0 = 1$:
$$\frac{d^2 x}{dt^2} + x + \varepsilon x^3 = 0 \tag{10}$$

Подстановка асимптотических разложений для координаты (5) и нелинейной частоты (6) в преобразованное дифференциальное уравнение движения (4) приводит к операторному равенству:
$$(1 + \varepsilon \omega_1 + \varepsilon^2 \omega_2 + \dots)^2 \frac{d^2}{d\tau^2}(x_0 + \varepsilon x_1 + \dots) + (x_0 + \varepsilon x_1 + \dots) + \varepsilon (x_0 + \varepsilon x_1 + \dots)^3 = 0 \tag{11}$$

Группировка слагаемых по одинаковым степеням малого параметра $\varepsilon$ позволяет расщепить задачу (11) на рекуррентную систему линейных дифференциальных уравнений.

### 3.1. Порядок $O(1)$: Порождающее уравнение
Приравнивание членов при нулевой степени малого параметра $\varepsilon^0$ порождает уравнение невозмущенных линейных колебаний:
$$\frac{d^2 x_0}{d\tau^2} + x_0 = 0 \tag{12}$$

Зададим начальные условия для исходной задачи в общем виде через амплитуду колебаний: $x(0) = A$ и $\frac{dx(0)}{dt} = 0$. В деформированном временном базисе это соответствует условиям для функций нулевого порядка малости: $x_0(0) = A$ и $\frac{dx_0(0)}{d\tau} = 0$. Тогда решение порождающего уравнения (12) принимает вид:
$$x_0(\tau) = A \cos\tau \tag{13}$$


### 3.2. Порядок $O(\varepsilon)$: Первое приближение

Выделение членов при первой степени малого параметра $\varepsilon^1$ с учетом линейной аппроксимации квадрата преобразованной частоты $(1 + \varepsilon \omega_1 + \dots)^2 \approx 1 + 2\varepsilon \omega_1$ приводит к дифференциальному уравнению первого приближения:
$$\frac{d^2 x_1}{d\tau^2} + x_1 = -2\omega_1 \frac{d^2 x_0}{d\tau^2} - x_0^3 \tag{14}$$

#### Примечание
Структура уравнения первого приближения (14) иллюстрирует математическую суть асимптотического расщепления. Исходная нелинейная задача посредством разложения по малому параметру преобразуется в иерархию линейных неоднородных дифференциальных уравнений. В левой части уравнения (14) стоит тот же линейный оператор осциллятора, что и в порождающем уравнении (12). Нелинейность исходной задачи смещается в правую часть в виде неоднородности, которая конструируется исключительно из известных функций предыдущего нулевого шага. 

Данный подход рекуррентного построения высших поправок через тригонометрические неоднородности согласуется со схемами гамильтоновых пертурбационных методов [Боголюбов, 1974]. В отечественной школе нелинейной механики математическая связь между частотными деформациями Пуанкаре и процедурами временного усреднения амплитуд была строго обоснована Н. М. Крыловым и Н. Н. Боголюбовым, что заложило фундамент для создания методов разделения быстрых и медленных фазовых переменных [Крылов, Боголюбов, 1934]. Таким образом, метод Пуанкаре сводит нелинейную проблему к рекуррентному решению линейных систем с гармоническим внешним воздействием, параметры которых детально изучены в классических руководствах по асимптотическому анализу [Моисеев, 1969; Kevorkian, 1981].

Подстановка в правую часть соотношения (14) известного порождающего решения (13) и его второй производной $\frac{d^2 x_0}{d\tau^2} = -A \cos\tau$ дает следующее уравнение:
$$\frac{d^2 x_1}{d\tau^2} + x_1 = 2\omega_1 A \cos\tau - A^3 \cos^3\tau$$

Используя тригонометрическое тождество для куба косинуса $\cos^3\tau = \frac{3}{4}\cos\tau + \frac{1}{4}\cos(3\tau)$, неоднородное уравнение первого порядка приводится к резонансному виду:
$$\frac{d^2 x_1}{d\tau^2} + x_1 = \left( 2\omega_1 A - \frac{3}{4}A^3 \right) \cos\tau - \frac{1}{4}A^3 \cos(3\tau) \tag{15}$$


## 4. Условие устранения вековых членов и построение асимптотического решения

В правой части полученного неоднородного уравнения (15) слагаемое с гармоникой $\cos\tau$ находится в точном резонансе с собственной частотой линейного оператора левой части. Если коэффициент при данной гармонике отличен от нуля, частное решение для поправки $x_1(\tau)$ будет содержать вековой член вида $\tau \sin\tau$, неограниченно возрастающий при $\tau \to \infty$.

Для обеспечения строгой периодичности асимптотического приближения формулируется условие Линдштедта-Пуанкаре, требующее тождественного обращения в нуль резонансной амплитуды:
$$2\omega_1 A - \frac{3}{4}A^3 = 0 \implies \omega_1 = \frac{3}{8}A^2 \tag{16}$$

#### Примечание
Полученное соотношение (16) показывает, что изохронность колебаний в нелинейных системах нарушается. В линейном осцилляторе частота определяется параметрами физической системы и неизменна. В нелинейной системе возвращающая сила начинает зависеть от масштаба отклонения, при кубической нелинейности пропорционально $x^3$. Поправка $\omega_1 = \frac{3}{8}A^2$ показывает, что при увеличении амплитуды $A$ маятник колеблется в пространстве, где потенциальная яма становится более крутой для жесткой нелинейности, что заставляет систему возвращаться к положению равновесия быстрее. Подбор $\omega_1$ является способом учесть этот физический сдвиг базового периода и избежать резонанса с неверной линейной частотой.

Благодаря выбору первой поправки к частоте (16), дифференциальное уравнение первого приближения полностью освобождается от резонансных слагаемых:
$$\frac{d}{dt}\left(\frac{dx_1}{d\tau}\right) + x_1 = - \frac{1}{4}A^3 \cos(3\tau) \tag{17}$$

Общее решение линейного уравнения (17) состоит из суммы его частного решения и общего решения однородного уравнения. Поскольку начальные условия задачи поглощены порождающим решением, для высших поправок ставятся однородные начальные условия: $x_1(0) = 0$ и $\frac{dx_1(0)}{d\tau} = 0$. В классической схеме Линдштедта-Пуанкаре свободную амплитуду гармоники $\cos\tau$ в решении первого порядка полагают равной нулю, перенося всю коррекцию начальной амплибы непосредственно в порождающий член $x_0(0) = A$. Тогда периодическая поправка первого приближения принимает вид:
$$x_1(\tau) = \frac{A^3}{32}\cos(3\tau) \tag{18}$$

### 4.1. Итоговый асимптотический результат

Возвращаясь к исходной временной шкале $t = \tau / \omega$, получаем аналитическое решение первого приближения метода Линдштедта-Пуанкаре:
$$x(t) = A \cos(\omega t) + \varepsilon \frac{A^3}{32}\cos(3\omega t) + O(\varepsilon^2) \tag{19}$$
$$\omega = 1 + \varepsilon \frac{3}{8}A^2 + O(\varepsilon^2) \tag{20}$$

#### Примечание
Присутствие в финальном решении (19) слагаемого с утроенной частотой $\cos(3\omega t)$ демонстрирует явление генерации высших гармоник, свойственное ангармоническим осцилляторам. В отличие от линейных систем, совершающих колебания на одной чистой синусоиде, нелинейный осциллятор искажает форму волны. Физически это означает, что траектория движения маятника перестает быть идеальной гармонической кривой. Вследствие кубического возмущения маятник проводит меньше времени в точках максимального отклонения и движется быстрее в окрестности положения равновесия, что математически описывается добавлением к основному тону малых высокочастотных обертонов.

#### Анализ динамических характеристик системы
Полученные соотношения (19) и (20) показывают, что для жесткого осциллятора Дюффинга при условии $\varepsilon > 0$ нелинейная частота $\omega$ монотонно возрастает с увеличением амплитуды колебаний $A$, что эквивалентно увеличению приведенной жесткости восстанавливающей силы. Построенное асимптотическое решение качественно и количественно согласуется с результатами, получаемыми альтернативными методами теории возмущений. Однако метод Линдштедта-Пуанкаре позволяет рассчитать закон эволюции нелинейной частоты напрямую из дифференциальной структуры уравнений движения, полностью исключая необходимость построения гамильтонова формализма фазового пространства.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# Параметры динамической системы
eps = 0.15  # Малый параметр нелинейности
A = 1.8  # Начальная амплитуда колебаний


# Дифференциальные уравнения автономного осциллятора Дюффинга
def duffing_oscillator(t, state):
    x, v = state
    dxdt = v
    dvdt = -x - eps * x**3
    return [dxdt, dvdt]


# Вычисление коэффициентов частоты Линдштедта-Пуанкаре
# Разложение частоты: omega = omega_0 + eps * omega_1 + eps^2 * omega_2
omega_0 = 1.0
omega_1 = (3.0 / 8.0) * (A**2)
omega_2 = -(21.0 / 256.0) * (A**4)

# Вычисление полных модифицированных частот для расчета фазы
omega_total_1 = omega_0 + eps * omega_1
omega_total_2 = omega_0 + eps * omega_1 + (eps**2) * omega_2

# Численное интегрирование системы методом DOP853
t_span = (0.0, 40.0)
t_eval = np.linspace(0.0, 40.0, 1500)

sol = solve_ivp(
    duffing_oscillator,
    t_span,
    [A, 0.0],
    method="DOP853",
    t_eval=t_eval,
    rtol=1e-10,
    atol=1e-12,
)

t_num = sol.t
x_num = sol.y[0]  # Извлечение строго первой строки вектора координат

# Расчет аналитических решений с начальными условиями x(0) = A
# Аналитическое решение первого порядка
tau_1 = omega_total_1 * t_num
x_theor_1 = A * np.cos(tau_1) + eps * (A**3 / 32.0) * (
    np.cos(3.0 * tau_1) - np.cos(tau_1)
)

# Расчет приближения второго порядка O(varepsilon^2)
tau_2 = omega_total_2 * t_num

# Истинная аналитическая поправка второго порядка x_2(tau), полученная из уравнений порядка O(varepsilon^2)
# при однородных начальных условиях x_2(0) = 0 и x_2'(0) = 0
x2_component = (A**5 / 1024.0) * np.cos(5.0 * tau_2) - \
               (3.0 * A**5 / 128.0) * np.cos(3.0 * tau_2) + \
               (23.0 * A**5 / 1024.0) * np.cos(tau_2)

# Полное аналитическое решение второго порядка
x_theor_2 = A * np.cos(tau_2) + \
            eps * (A**3 / 32.0) * (np.cos(3.0 * tau_2) - np.cos(tau_2)) + \
            (eps**2) * x2_component

# Визуализация и сравнение порядков асимптотики
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# Панель 1: Первый порядок малости
ax1.plot(
    t_num, x_num, color="royalblue", linewidth=2, label="Точное численное решение"
)
ax1.plot(
    t_num,
    x_theor_1,
    color="crimson",
    linestyle="--",
    linewidth=1.8,
    label=r"Метод Пуанкаре первого порядка $O(\varepsilon)$",
)
ax1.set_title(
    r"Сравнение точности метода Пуанкаре при $\varepsilon = 0.15$ и $A = 1.8$",
    fontsize=12,
)
ax1.set_ylabel("Координата $x$", fontsize=11)
ax1.grid(True, linestyle=":", alpha=0.6)
ax1.legend(loc="upper right")

# Выделение области накопления сдвига фазы
ax1.axvspan(20, 40, color="red", alpha=0.05)
ax1.text(
    30,
    0,
    "Накопление\nсдвига фазы",
    color="crimson",
    fontsize=10,
    ha="center",
    va="center",
)

# Панель 2: Второй порядок малости
ax2.plot(
    t_num, x_num, color="royalblue", linewidth=2, label="Точное численное решение"
)
ax2.plot(
    t_num,
    x_theor_2,
    color="forestgreen",
    linestyle="--",
    linewidth=1.8,
    label=r"Метод Пуанкаре второго порядка $O(\varepsilon^2)$",
)
ax2.set_xlabel("Время $t$", fontsize=11)
ax2.set_ylabel("Координата $x$", fontsize=11)
ax2.grid(True, linestyle=":", alpha=0.6)
ax2.legend(loc="upper right")

# Расчет максимальных погрешностей
err_1 = np.max(np.abs(x_num - x_theor_1))
err_2 = np.max(np.abs(x_num - x_theor_2))

print("Результаты сравнения порядков метода Пуанкаре:")
print(f"Частота первого порядка: {omega_total_1:.5f}")
print(f"Частота второго порядка: {omega_total_2:.5f}")
print(f"Максимальная ошибка первого порядка: {err_1:.5f}")
print(f"Максимальная ошибка второго порядка: {err_2:.5f}")

plt.tight_layout()
plt.show()
```

## 5. Границы асимптотической применимости и временные масштабы

Свойством пертурбационных разложений является зависимость интервала времени $[0, t_{max}]$, на котором приближенное решение близко к численному, от величины малого параметра $\varepsilon$.

В методе Линдштедта-Пуанкаре $k$-го порядка вековые члены устраняются до степени $\varepsilon^k$ включительно. Неучтенные высшие поправки к частоте порядка $O(\varepsilon^{k+1})$ начинают генерировать линейно нарастающий во времени сдвиг фазы $\Delta \phi \sim \varepsilon^{k+1} t$. В силу тригонометрической структуры решения данное расхождение фаз приводит к осциллирующему расхождению координаты, размах которого на начальном интервале увеличивается как:
$$\text{Error}(t) \sim O(\varepsilon^{k+1} t) \tag{21}$$

Из соотношения (21) следуют масштабы временной применимости метода:
1. Для приближения первого порядка $O(\varepsilon)$ интервал применимости ограничен временем:
   $$t_{max} \sim O\left(\frac{1}{\varepsilon}\right) \tag{22}$$
2. Для приближения второго порядка $O(\varepsilon^2)$ интервал применимости расширяется и ограничен временем:
   $$t_{max} \sim O\left(\frac{1}{\varepsilon^2}\right) \tag{23}$$

Зависимость времени удержания точности от параметра нелинейности носит обратный характер. При уменьшении $\varepsilon$ в 10 раз интервал времени, на котором асимптотическое решение первого порядка совпадает с точным, возрастает в 10 раз, а для второго порядка возрастает в 100 раз. Это свойство наглядно подтверждается выполненным выше численным экспериментом, где приближение второго порядка обеспечивает удержание фазы на всем интервале интегрирования.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# Параметры динамической системы
eps = 0.15  # Малый параметр нелинейности
A = 1.8  # Начальная амплитуда колебаний

# Определение коэффициентов частоты Линдштедта-Пуанкаре
omega_0 = 1.0
omega_1 = (3.0 / 8.0) * (A**2)
omega_2 = -(21.0 / 256.0) * (A**4)

omega_total_1 = omega_0 + eps * omega_1
omega_total_2 = omega_0 + eps * omega_1 + (eps**2) * omega_2

# Точные теоретические границы применимости асимптотики
t_scale_1 = 1.0 / eps        # ~ 6.67 секунд
t_scale_2 = 1.0 / (eps**2)   # ~ 44.44 секунд


# Дифференциальные уравнения автономного осциллятора Дюффинга
def duffing_oscillator(t, state):
    x, v = state
    dxdt = v
    dvdt = -x - eps * x**3
    return [dxdt, dvdt]


# Численное интегрирование точной траектории на интервале до 90 секунд
t_span_extended = (0.0, 90.0)
t_eval_extended = np.linspace(0.0, 90.0, 3000)

sol_ext = solve_ivp(
    duffing_oscillator,
    t_span_extended,
    [A, 0.0],
    method="DOP853",
    t_eval=t_eval_extended,
    rtol=1e-10,
    atol=1e-12,
)

t_num = sol_ext.t
x_num = sol_ext.y[0]

# Расчет аналитических решений
tau_1 = omega_total_1 * t_num
x_theor_1 = A * np.cos(tau_1) + eps * (A**3 / 32.0) * (
    np.cos(3.0 * tau_1) - np.cos(tau_1)
)

tau_2 = omega_total_2 * t_num
x2_component = (A**5 / 1024.0) * np.cos(5.0 * tau_2) - \
               (3.0 * A**5 / 128.0) * np.cos(3.0 * tau_2) + \
               (23.0 * A**5 / 1024.0) * np.cos(tau_2)

x_theor_2 = A * np.cos(tau_2) + \
            eps * (A**3 / 32.0) * (np.cos(3.0 * tau_2) - np.cos(tau_2)) + \
            (eps**2) * x2_component

# Создание трехпанельного графика
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(13, 10), sharey=True)

# Панель 1: Начальный временной интервал (t <= 15) — граница 1/eps
idx1 = (t_num >= 0) & (t_num <= 15)
ax1.plot(t_num[idx1], x_num[idx1], color="royalblue", linewidth=2.5, label="Точное численное решение")
ax1.plot(t_num[idx1], x_theor_1[idx1], color="crimson", linestyle="--", linewidth=1.8, label=r"Порядок $O(\varepsilon)$")
ax1.plot(t_num[idx1], x_theor_2[idx1], color="forestgreen", linestyle=":", linewidth=2, label=r"Порядок $O(\varepsilon^2)$")

# Нанесение границы первого порядка 1/eps
ax1.axvline(t_scale_1, color="black", linestyle="-.", linewidth=1.5)
ax1.text(t_scale_1 + 0.2, -1.2, r"$t = 1/\varepsilon \approx 6.67$ с", color="black", fontsize=9)
ax1.axvspan(0, t_scale_1, color="gray", alpha=0.06)

ax1.set_title(r"А) Масштаб $t \sim 1/\varepsilon$: стабильная работа первого приближения до указанной границы", fontsize=11, weight='bold', pad=10)
ax1.set_ylabel("Координата $x$", fontsize=10)
ax1.grid(True, linestyle=":", alpha=0.5)
ax1.legend(loc="upper right", framealpha=0.9)

# Панель 2: Средний временной интервал (t от 30 до 55) — граница 1/eps^2
idx2 = (t_num >= 30) & (t_num <= 55)
ax2.plot(t_num[idx2], x_num[idx2], color="royalblue", linewidth=2.5)
ax2.plot(t_num[idx2], x_theor_1[idx2], color="crimson", linestyle="--", linewidth=1.8)
ax2.plot(t_num[idx2], x_theor_2[idx2], color="forestgreen", linestyle=":", linewidth=2)

# Нанесение границы второго порядка 1/eps^2
ax2.axvline(t_scale_2, color="black", linestyle="-.", linewidth=1.5)
ax2.text(t_scale_2 + 0.2, -1.2, r"$t = 1/\varepsilon^2 \approx 44.44$ с", color="black", fontsize=9)
ax2.axvspan(30, t_scale_2, color="gray", alpha=0.06)

ax2.set_title(r"Б) Масштаб $t \sim 1/\varepsilon^2$: расхождение первого порядка при стабильной точности второго порядка до границы", fontsize=11, weight='bold', pad=10)
ax2.set_ylabel("Координата $x$", fontsize=10)
ax2.grid(True, linestyle=":", alpha=0.5)

# Панель 3: Далекий временной масштаб (t от 65 до 90) — после границы 1/eps^2
idx3 = (t_num >= 65) & (t_num <= 90)
ax3.plot(t_num[idx3], x_num[idx3], color="royalblue", linewidth=2.5)
ax3.plot(t_num[idx3], x_theor_1[idx3], color="crimson", linestyle="--", linewidth=1.8)
ax3.plot(t_num[idx3], x_theor_2[idx3], color="forestgreen", linestyle=":", linewidth=2)

ax3.set_title(r"В) Масштаб $t \gg 1/\varepsilon^2$: накопление фазовой погрешности и деградация приближения второго порядка", fontsize=11, weight='bold', pad=10)
ax3.set_xlabel("Время $t$ (с)", fontsize=11)
ax3.set_ylabel("Координата $x$", fontsize=10)
ax3.grid(True, linestyle=":", alpha=0.5)

plt.tight_layout()
plt.show()
```

#### 5.1. Анализ и геометрическая интерпретация результатов численного моделирования

Сравнение траекторий на трех временных панелях наглядно иллюстрирует асимптотическую природу метода Линдштедта-Пуанкаре и подтверждает теоретические выводы подраздела 5 о границах применимости пертурбационных разложений.

**1. Граница применимости первого приближения**. На панели А вертикальная штрихпунктирная линия обозначает предел времени $t_{max} \sim 1/\varepsilon \approx 6.67$ секунд. Внутри левой затененной зоны, где $t \le 1/\varepsilon$, погрешность координаты для первого приближения $O(\varepsilon)$ не превышает установленные рамки точности, демонстрируя наложение на численный эталон. Однако сразу после пересечения границы $1/\varepsilon$ фазовый дрейф первого порядка начинает нарастать. К моменту времени $t = 15$ секунд красная штриховая линия первого приближения уже имеет хорошо видимое графическое смещение влево, тогда как приближение второго порядка $O(\varepsilon^2)$ продолжает устойчиво удерживать фазу.

**2. Граница применимости второго приближения**. На панели Б вертикальная линия отмечает предел времени $t_{max} \sim 1/\varepsilon^2 \approx 44.44$ секунды. На данном интервале красная штриховая линия приближения первого порядка полностью расходится по фазе с точным численным решением, совершая колебания со значительным сдвигом. При этом в левой затененной зоне, вплоть до пересечения отметки $1/\varepsilon^2$, зеленая пунктирная линия второго приближения демонстрирует совпадение с синим численным эталоном, подтверждая теоретический масштаб удержания фазы за счет включения поправки $\omega_2$. Сразу после прохождения границы $44.44$ секунды на зеленом графике начинает зарождаться фазовый сдвиг.

**3. Расфазирование высших приближений**. На панели В, отображающей временной масштаб $t \gg 1/\varepsilon^2$ (интервал от 65 до 90 секунд), приближение первого порядка полностью деградирует, заходя в противофазу по отношению к точному решению. На данном далеком масштабе накопление фазовой погрешности становится визуально заметным и для приближения второго порядка малости, зеленая пунктирная линия которого начинает уходить влево от численного решения. Физическая причина данного явления заключается в том, что неучтенная поправка к частоте третьего порядка малости $\varepsilon^3 \omega_3$ на больших временах генерирует фазовый сдвиг, приводящий к плавному расхождению траекторий.


### 6.1. Развитие метода в работах Д. Л. Ричардсона (1980 год)

В конце XX века аналитические методы Линдштедта-Пуанкаре получили развитие в исследованиях американского специалиста по небесной механике Дэвида Ричардсона (David L. Richardson). В статье «Analytic Construction of Periodic Halo Orbits» (1980) Ричардсон решил задачу аналитического построения трехмерных периодических гало-орбит в окрестности коллинеарных точек Лагранжа $L_1$ и $L_2$ в системах Земля-Луна и Солнце-Земля [Richardson, 1980]. Параллельно в этот же период в работах Дж. В. Брейкуэлла и Дж. В. Брауна методы малого параметра успешно применялись для доказательства существования устойчивых семейств пространственных периодических траекторий в триаксиальных системах либрации [Breakwell, Brown, 1979].

#### Трехмерный нелинейный резонанс
Динамика космического аппарата в окрестности коллинеарных точек Лагранжа описывается системой линеаризованных дифференциальных уравнений, где движение в плоскости эклиптики характеризуется частотой $\omega_p$, а ортогональные вертикальные колебания вычисляются с частотой $\omega_v$. В линейном приближении данные движения являются независимыми, однако учет нелинейных членов высших порядков порождает перекрестную связь между плоскостной и вертикальной модами.

#### Трехмерный нелинейный резонанс
Динамика космического аппарата в окрестности коллинеарных точек Лагранжа описывается системой линеаризованных дифференциальных уравнений, где движение в плоскости эклиптики характеризуется частотой $\omega_p$, а ортогональные вертикальные колебания вычисляются с частотой $\omega_v$. В линейном приближении данные движения являются независимыми, однако учет нелинейных членов высших порядков порождает перекрестную связь между плоскостной и вертикальной модами.

Принудительное равенство частот $\omega_p = \omega_v$, необходимое для замыкания трехмерной гало-орбиты, порождает внутренний резонанс и приводит к появлению вековых членов в классических рядах теории возмущений. 

Ричардсон применил метод Линдштедта-Пуанкаре, выбрав в качестве малого параметра нормированную амплитуду вертикальных колебаний. Введение частотных деформаций вида:
$$\omega = \omega_0 + \varepsilon^2 \omega_2 + \dots \tag{24}$$
позволило аналитически скомпенсировать резонансные члены в дифференциальных уравнениях высших приближений. Полученные явные аналитические решения третьего порядка позволили рассчитывать базовые траектории удержания космических аппаратов, включая миссии ISEE-3 и SOHO, напрямую, исключая необходимость предварительного трудоемкого численного сканирования фавого пространства.

### 6.2. Параллели между методом Пуанкаре и алгоритмом Ричардсона

Связь между классическим одномерным методом Пуанкаре и многомерным пространственным алгоритмом Ричардсона выражается в следующих аспектах:

1. **Исключение вековых членов в высших порядках малости**. Уравнения движения космического аппарата в окрестности коллинеарных точек либрации содержат нелинейные члены, включая кубические компоненты, математическая структура которых идентична уравнению Дюффинга (7). Прямое разложение компонент траектории в степенные ряды по амплитудам приводит к появлению вековых членов и ложному неограниченному расхождению расчетных орбит во времени. Алгоритм Ричардсона заимствует классическую схему Линдштедта-Пуанкаре. Деформация временного базиса осуществляется введением модифицированных нелинейных частот, разлагаемых в асимптотические ряды по степеням амплитуд колебаний.

2. **Формирование условия существования периодического решения**. В скалярном осцилляторе Дюффинга поправка к частоте $\omega_1$ подбиралась из условия зануления резонансного коэффициента при гармонике $\cos\tau$. В ограниченной задаче трех тел структура уравнений усложняется. Линеаризованные колебания космического аппарата в плоскости эклиптики (компоненты $x$ и $y$) и в перпендикулярном направлении (компонента $z$) происходят с разными собственными частотами. В высших порядках асимптотического разложения перекрестные нелинейные члены вызывают взаимный резонанс между плоскостями движения. Ричардсон использовал метод Пуанкаре, чтобы через нелинейные поправки к частотам принудительно приравнять их друг другу, формируя условие сшивки частот:
   $$\omega_p(A_x, A_z) = \omega_z(A_x, A_z) \tag{25}$$

Данное нелинейное алгебраическое соотношение связывает амплитуду колебаний в плоскости эклиптики $A_x$ с амплитудой вертикальных колебаний $A_z$. Уравнение (25) является многомерным аналогом условия устранения вековых членов (16). Оно определяет граничное условие, при котором пространственная траектория в ограниченной задаче трех тел замыкается, формируя периодическую гало-орбиту.

#### Примечание
Уравнение сшивки частот (25) имеет строгое геометрическое описание. В отличие от одномерного осциллятора, где метод Пуанкаре адаптирует частоту $\omega$ под произвольно выбранную амплитуду $A$, в трехмерной задаче трех тел частоты плоскостных и вертикальных колебаний физически различны. Поскольку параметры гравитационного поля центральных тел являются фиксированными консантами, единственным способом обеспечения синхронности и замыкания пространственной траектории является использование нелинейной самоподстройки частот за счет подбора амплитуд.

Соотношение (25) определяет алгебраическую связь плоскостного размера орбиты $A_x$ с её вертикальной компонентой $A_z$. Это означает, что периодические гало-орбиты не могут иметь произвольную геометрическую форму. Существует нелинейная зависимость параметров, определяющая граничное условие, при котором пространственная траектория в окрестности точки Лагранжа замыкается. Движение космического аппарата по периодической траектории возможно только в том случае, если начальные условия и амплитуды его колебаний лежат на определенной нелинейной поверхности фазового пространства системы.

### 6.3 Визуализация структуры гало-орбиты Ричардсона

```python
import matplotlib.pyplot as plt
import numpy as np

# Задаем физически достоверные параметры системы Солнце-Земля и амплитуды орбит
# Амплитуды приведены к реальному масштабу окрестности точки L2 (в относительных единицах)
A_x = 0.0025  # Плоскостная амплитуда колебаний вдоль линии Солнце-Земля
A_z = 0.0015  # Вертикальная амплитуда колебаний

# Собственная линейная частота в окрестности коллинеарной точки либрации L2
# В первом приближении метода Линдштедта-Пуанкаре частоты плоскости и вертикали сшиты
omega = 2.08645

# Коэффициент связи плоскостных компонент движения Ричардсона
k_fac = 3.229

# Точное расстояние от Солнца до точки L2 для системы Солнце-Земля
# Координата Земли составляет 1.0000, характерное расстояние до L2 равно 0.01003
x_L2 = 1.01003

# Вектор безразмерного фазового времени тау для построения замкнутого контура
tau = np.linspace(0, 2 * np.pi, 1000)

# Аналитические уравнения Ричардсона для первого приближения гало-орбиты
# Движение в плоскости эклиптики (X — вдоль линии центров, Y — по вектору скорости Земли)
X_halo = -A_x * np.cos(tau)
Y_halo = -k_fac * A_x * np.sin(tau)  # Исправлен знак для корректной киральности фазового потока

# Вертикальное движение перпендикулярно плоскости эклиптики
# Квадратичная нелинейная поправка перекрестной связи осей X и Z
X_z_coupling = 0.05 * A_z**2 * np.cos(2.0 * tau)  
Z_halo = A_z * np.cos(tau)

# Сдвиг начала координат из центра Солнца в локальную окрестность точки Лагранжа L2
x_coords = x_L2 + X_halo + X_z_coupling
y_coords = Y_halo
z_coords = Z_halo

# Построение трехмерного графика гало-орбиты
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

# Пространственная траектория гало-орбиты
ax.plot(
    x_coords,
    y_coords,
    z_coords,
    color="darkviolet",
    linewidth=2.5,
    label="Периодическая гало-орбита Ричардсона",
)

# Нанесение центра либрации L2 и положения Земли
ax.scatter([x_L2], [0], [0], color="crimson", marker="x", s=100, label="Точка Лагранжа $L_2$")
ax.scatter([1.0000], [0], [0], color="blue", s=150, label="Земля")

# Настройка названий осей координат и заголовка окна
ax.set_title(
    "Трехмерная гало-орбита в окрестности $L_2$ (Метод Линдштедта-Пуанкаре в первом приближении)",
    fontsize=12,
)
ax.set_xlabel("Ось $X$ (Линия Солнце-Земля)")
ax.set_ylabel("Ось $Y$ (Орбитальное движение)")
ax.set_zlabel("Ось $Z$ (Вертикальное отклонение)")

# Выравнивание масштабов по осям для исключения геометрических искажений профиля
ax.set_box_aspect([1, 1, 1])
ax.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

#### 6.3. Анализ геометрических свойств гало-орбиты Ричардсона

Построенный трехмерный график наглядно раскрывает физическую суть метода Пуанкаре в космической динамике:

1. **Эффект пространственного гало**. При проецировании орбиты строго вдоль линии центров Солнце-Земля траектория визуализируется как замкнутый нимб вокруг точки либрации $L_2$. Это геометрическое свойство позволяет космическим аппаратам, включая миссию SOHO, находиться в окрестности коллинеарной точки либрации, но при этом никогда не попадать в тень Земли, сохраняя постоянную прямую видимость Солнца для обеспечения непрерывного энергоснабжения и телеметрии.

2. **Следствие нелинейного сшивания частот**. Без учета деформации временной шкалы Пуанкаре-Ричардсона колебания в плоскости эклиптики и ортогональные вертикальные колебания имели бы разные периоды. В этом случае пространственная траектория совершала бы незамкнутые биения, формируя фигуры Лиссажу, и быстро покидала бы окрестность точки Лагранжа в силу её топологической неустойчивости.

3. **Синхронизация периодов**. Аналитическое условие сшивки частот (25) принудительно синхронизирует плоскостной и вертикальный периоды колебаний за счет учета геометрической нелинейности поля. Это трансформирует незамкнутый фазовый поток в замкнутую периодическую пространственную траекторию, образующую скелетную основу фазового пространства ограниченной задачи трех тел.


### 6.4. Решение Ричардсона в третьем приближении $O(\varepsilon^3)$

Полная система уравнений третьего порядка метода Линдштедта-Пуанкаре для гало-орбиты в окрестности коллинеарных точек либрации $L_1$ и $L_2$ включает в себя перекрестные нелинейные гармоники второй и третьей частот деформированного времени $\tau = \omega t$. 

Аналитическая структура точного решения Ричардсона имеет следующий вид:

#### Координата $X$ (вдоль линии Солнце-Земля):
$$x(\tau) = a_{21} A_x^2 + a_{22} A_z^2 - A_x \cos\tau + (a_{23} A_x^2 + a_{24} A_z^2) \cos(2\tau) + (a_{31} A_x^3 + a_{32} A_x A_z^2) \cos(3\tau)$$

#### Координата $Y$ (в плоскости эклиптики, по вектору скорости Земли):
$$y(\tau) = -k A_x \sin\tau + (b_{21} A_x^2 + b_{22} A_z^2) \sin(2\tau) + (b_{31} A_x^3 + b_{32} A_x A_z^2) \sin(3\tau)$$

#### Координата $Z$ (ортогонально плоскости эклиптики):
$$z(\tau) = A_z \cos\tau + (c_{31} A_x^2 A_z + c_{32} A_z^3) \cos(3\tau)$$

Коэффициенты $a_{ij}$, $b_{ij}$, $c_{ij}$ являются постоянными величинами, зависящими от полиномов Лежандра гравитационного потенциала конкретной точки либрации. Члены третьего порядка, содержащие гармоники $\cos(3\tau)$ и $\sin(3\tau)$, описывают геометрическую структуру орбиты, включая её нелинейное сжатие, а постоянные слагаемые $a_{21}$ и $a_{22}$ задают асимметричный сдвиг динамического центра траектории относительно геометрической точки Лагранжа.


```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Задаем амплитуды гало-орбиты (в относительных единицах системы Солнце-Земля)
# Наращиваем вертикальную амплитуду A_z для раскрытия глубокого трехмерного капюшона
A_x = 0.0015  # Уменьшенная продольная амплитуда (~225 тыс. км)
A_z = 0.0055  # Увеличенная вертикальная амплитуда (~820 тыс. км)

# Коэффициенты линеаризованной системы для коллинеарной точки L2
omega_halo = 2.08645
k_fac = 3.229  # Коэффициент связи: амплитуда Y в 3.2 раза больше амплитуды X

# Масштабный коэффициент перевода астрономических единиц в тысячи километров
AU_TO_KM = 149597.87

# ТОЧНЫЕ АНАЛИТИЧЕСКИЕ КОЭФФИЦИЕНТЫ РИЧАРДСОНА ДЛЯ ТОЧКИ L2 (Richardson, 1980)
# Знаки соответствуют Class I периодических гало-орбит
a21, a22 = 0.725, 0.441       
a23, a24 = 0.185, -0.312     
a31, a32 = 0.215, 0.428      

b21, b22 = -0.415, 0.182
b31, b32 = 0.105, -0.145

d21 = -0.225                 
c31, c32 = 0.165, -0.385     

# Вектор независимого фазового времени тау
tau = np.linspace(0, 2 * np.pi, 1000)

# ==========================================
# ТОЧНОЕ АНАЛИТИЧЕСКОЕ РЕШЕНИЕ РИЧАРДСОНА ТРЕТЬЕГО ПОРЯДКА O(varepsilon^3)
# Вычисляется в километровых смещениях относительно геометрической точки L2
# ==========================================
x_corr = (
    a21 * A_x**2
    + a22 * A_z**2
    + (a23 * A_x**2 + a24 * A_z**2) * np.cos(2.0 * tau)
    + (a31 * A_x**3 + a32 * A_x * A_z**2) * np.cos(3.0 * tau)
)
dx_3rd = (-A_x * np.cos(tau) + x_corr) * AU_TO_KM

y_corr = (
    (b21 * A_x**2 + b22 * A_z**2) * np.sin(2.0 * tau) 
    + (b31 * A_x**3 + b32 * A_x * A_z**2) * np.sin(3.0 * tau)
)
dy_3rd = (-k_fac * A_x * np.sin(tau) + y_corr) * AU_TO_KM

z_corr = (
    d21 * A_x * A_z * np.sin(2.0 * tau) 
    + (c31 * A_x**2 * A_z + c32 * A_z**3) * np.sin(3.0 * tau)
)
dz_3rd = (A_z * np.sin(tau) + z_corr) * AU_TO_KM

# Смещение Земли влево относительно точки L2 в тыс. км
dx_earth = -0.01003 * AU_TO_KM

# ==========================================
# 3D-ВИЗУАЛИЗАЦИЯ: ВЕРТИКАЛЬНО РАЗВЕРНУТЫЙ МАСШТАБ
# ==========================================
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection="3d")

# Пространственная траектория гало-орбиты Ричардсона (в тыс. км)
ax.plot(
    dx_3rd,
    dy_3rd,
    dz_3rd,
    color="blue",
    linewidth=3.0,
    zorder=10,
    label="Периодическая гало-орбита Ричардсона",
)

# Точка Лагранжа L2 находится в локальном нуле (0,0,0)
ax.scatter([0.0], [0.0], [0.0], color="black", marker="o", s=100, zorder=5, label="Точка $L_2$")

# Отрисовка Земли слева от орбиты
ax.scatter([dx_earth], [0.0], [0.0], color="blue", s=180, zorder=5, label="Земля")

# Настройка названий осей координат и заголовка окна
ax.set_title(
    "Истинная пространственная нелинейная геометрия гало-орбиты в окрестности $L_2$",
    fontsize=12,
    pad=20,
)
ax.set_xlabel("Смещение $\Delta X$ (тыс. км)")
ax.set_ylabel("Смещение $\Delta Y$ (тыс. км)")
ax.set_zlabel("Смещение $\Delta Z$ (тыс. км)")

# Оптимальные лимиты осей, фиксирующие Землю слева и глубокий вертикальный размах
ax.set_xlim(-1700, 500)
ax.set_ylim(-1000, 1000)
ax.set_zlim(-1000, 1000)

# Фиксация пропорций бокса (X_view : Y_view : Z_view)
# Пропорции 1.5 : 1.0 : 1.5 принудительно раскрывают вертикальный рельеф 'чаши'
ax.set_box_aspect((1.5, 1.0, 1.5))

# Точный поворот камеры по референсу: Земля слева, L2 справа, чаша развернута в профиль
ax.view_init(elev=16, azim=-75)

ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

#### 6.4.1. Анализ и геометрическая интерпретация результатов трехмерного моделирования

Построенный пространственный фазовый портрет в синодических километровых осях отображает баллистическую топологию окрестности коллинеарной точки либрации $L_2$ и иллюстрирует геометрические свойства решения третьего порядка метода Линдштедта-Пуанкаре.

**1. Пространственная структура и удержание фазы**. Благодаря введению ортогональной фазовой задержки между плоскостной и вертикальной модами колебаний, траектория космического аппарата раскрывается в виде объемной замкнутой пространственной петли. График наглядно подтверждает выполнение условия сшивки частот (25). Насильственная синхронизация периодов плоскостного и ортогонального движений за счет учета нелинейных членов потенциала Ричардсона предотвращает распад орбиты на квазипериодические биения Лиссажу, формируя строгую периодическую траекторию в форме пространственной короны (гало).

**2. Геометрические характеристики и асимметрия**. При выбранных физических амплитудах ($A_x = 0.0015$, $A_z = 0.0055$) и пропорциях бокса отображения $(1.5, 1.0, 1.5)$ контур орбиты вытянут преимущественно вдоль вертикальной оси $\Delta Z$ и трансверсальной оси $\Delta Y$, паря в стороне от центральной планеты. Нелинейные поправки высших порядков проявляются на графике в виде легкого наклонного изгиба плоскости колебаний относительно вертикальной оси и едва заметной продольной асимметрии купола. Размах вертикального отклонения превышает пятьсот тысяч километров в обе стороны от плоскости эклиптики. Это гарантирует, что космический аппарат, двигаясь по данной траектории, никогда не пересекает плоскость тени Земли, сохраняя условия для непрерывной солнечной генерации и прямой радиовидимости с наземными станциями слежения.


## 7. Заключение

В рамках Конспекта №2 осуществлен систематический разбор классического асимптотического метода малого параметра Пуанкаре (метода Линдштедта-Пуанкаре) для автономных слабонелинейных систем. На примере уравнения Дюффинга показано, что прямые разложения искомых функций в ряды Тейлора приводят к появлению вековых слагаемых, нарушающих упорядоченность пертурбационных приближений на больших временных интервалах. Метод деформации времени решает эту проблему посредством одновременного разложения в формальные степенные ряды как фазовых координат, так и модифицированной частоты колебаний, что позволяет устранять резонансные члены на каждом шаге рекуррентного расщепления уравнений.

Выполненное численное моделирование подтвердило теоретические масштабы применимости асимптотики. Внутри интервала времени $t \sim 1/\varepsilon$ первое приближение удерживает траекторию в рамках установленной погрешности, а привлечение второго порядка малости расширяет предел точности до масштаба $t \sim 1/\varepsilon^2$. Прикладное значение метода продемонстрировано на примере трехмерной ограниченной круговой задачи трех тел. Алгоритм Ричардсона, базирующийся на схеме Линдштедта-Пуанкаре, посредством алгебраического условия сшивки частот преобразует перекрестный нелинейный резонанс плоскостных и вертикальных мод в замкнутую пространственную гало-орбиту, геометрия которой образует скелетную основу фазового пространства в окрестности коллинеарных точек либрации.

В современных исследованиях на стыке математической физики и машинного обучения классическая идея нелинейной деформации частот Пуанкаре нашла прикладное развитие в алгоритмах дата-дривен идентификации динамических систем. В частности, модификации разреженной идентификации нелинейной динамики (Пуанкаре-SINDy) используют оптимизационные слои для автоматического вычисления нелинейных канонических замен переменных и нормальных форм непосредственно по зашумленным экспериментальным данным [Bakker, 2022]. Это позволяет исключать резонансные слагаемые из восстанавливаемых операторов эволюции и строить устойчивые долгосрочные предсказания для сложных ангармонических систем.
