# 4. Метод генераторов Ли в теории возмущений гамильтоновых систем

### Генераторы Ли и экспоненциальный оператор

Классические методы теории возмущений (например, метод Фон Цейпеля) опираются на смешанные производящие функции вида $S(\bar{I}, \theta)$, связывающие старые и новые переменные неявными уравнениями:
$$I = \frac{\partial S}{\partial \theta}, \quad \bar{\theta} = \frac{\partial S}{\partial \bar{I}}$$

В высших порядках малости ($O(\varepsilon^2), O(\varepsilon^3)$ и далее) это требует обращения нелинейных функций и решения сложных алгебраических систем уравнений. Чтобы избежать неявных связей, Григорий Хори (1966) и Андре Депри (1969) предложили строить канонические преобразования на основе **потоков векторных полей Ли**.

### 1. Оператор Ли и скобка Пуассона

Пусть $W(q, p)$ — некоторая гладкая функция в фазовом пространстве, называемая **генератором Ли** (или производящим гамильтонианом преобразования). 

Каждому такому генератору можно сопоставить линейный дифференциальный **оператор Ли** $L_W$, действие которого на любую фазовую функцию $F(q, p)$ определяется через каноническую скобку Пуассона:

$$L_W F = \{ F, W \} = \sum_{i=1}^{n} \left( \frac{\partial F}{\partial q_i} \frac{\partial W}{\partial p_i} - \frac{\partial F}{\partial p_i} \frac{\partial W}{\partial q_i} \right) \tag{1}$$

В переменных «действие–угол» $(I, \theta)$ для одномерной системы оператор Ли принимает вид:
$$L_W F = \{ F, W \} = \frac{\partial F}{\partial \theta} \frac{\partial W}{\partial I} - \frac{\partial F}{\partial I} \frac{\partial W}{\partial \theta} \tag{2}$$

Свойства оператора Ли наследуют свойства скобок Пуассона:
1. **Линейность:** $L_W (c_1 F_1 + c_2 F_2) = c_1 L_W F_1 + c_2 L_W F_2$
2. **Правило Лейбница:** $L_W (FG) = (L_W F)G + F(L_W G)$
3. **Антисимметричность:** $L_W F = -L_F W$


### 2. Экспоненциальный ряд Ли

Каноническое преобразование от старых координат $z = (q, p)$ к новым $\bar{z} = (\bar{q}, \bar{p})$ задается в явном виде как решение фиктивных уравнений Гамильтона, где роль «времени» играет малый параметр $\varepsilon$:

$$\frac{dz}{d\varepsilon} = \{ z, W \} = L_W z \tag{3}$$

Решая это дифференциальное уравнение операторным методом с начальным условием $z(0) = \bar{z}$, мы получаем **формальный ряд Ли** (или экспоненциальный оператор Ли):

$$z(\varepsilon) = \exp(\varepsilon L_W) \bar{z} = \sum_{k=0}^{\infty} \frac{\varepsilon^k}{k!} L_W^k \bar{z} \tag{4}$$

Разворачивая экспоненту в ряд Тейлора, получаем **явные формулы** связи старых переменных с новыми:
$$z = \bar{z} + \varepsilon \{ \bar{z}, W \} + \frac{\varepsilon^2}{2!} \{ \{ \bar{z}, W \}, W \} + \frac{\varepsilon^3}{3!} \{ \{ \{ \bar{z}, W \}, W \}, W \} + \dots \tag{5}$$

### 3. Преобразование Гамильтониана (Теорема Ли)

Особенность метода Ли заключается в том, что все скалярные функции преобразуется по одинаковым законам. Если исходная система описывалась гамильтонианом $H(z)$, то в новых переменных $\bar{z}$ её динамика задается новым эффективным гамильтонианом $\bar{H}(\bar{z})$, который рассчитывается по формуле:

$$\bar{H}(\bar{z}) = \exp(\varepsilon L_W) H(\bar{z}) = H(\bar{z}) + \varepsilon \{ H(\bar{z}), W \} + \frac{\varepsilon^2}{2!} \{ \{ H(\bar{z}), W \}, W \} + \dots \tag{6}$$

Если исходный гамильтониан и генератор сами зависят от $\varepsilon$ в виде рядов (как это обычно бывает в теории возмущений):
$$H = H_0 + \varepsilon H_1 + \varepsilon^2 H_2 + \dots$$
$$W = W_1 + \varepsilon W_2 + \dots$$

то подстановка этих рядов в (6) порождает классическую **рекуррентную схему Хори–Депри**, позволяющую шаг за шагом находить поправки к энергии $\bar{H}_k$ и функции генераторов $W_k$. Схема сводится к чисто алгебраическим операциям над дифференциальными полиномами, что делает её идеальной мишенью для автоматизации на компьютере.


### 4. Простой пример действия оператора Ли

Чтобы понять, как генератор Ли в явном виде связывает старые и новые переменные фазового пространства, рассмотрим первый порядок разложения (формула 5) в переменных «действие–угол» $(I, \theta)$:

$$I = \bar{I} + \varepsilon \{ \bar{I}, W_1 \} = \bar{I} - \varepsilon \frac{\partial W_1}{\partial \bar{\theta}} \tag{7}$$

$$\theta = \bar{\theta} + \varepsilon \{ \bar{\theta}, W_1 \} = \bar{\theta} + \varepsilon \frac{\partial W_1}{\partial \bar{I}} \tag{8}$$

Пусть в ходе анализа ангармонического осциллятора мы уже нашли генератор Ли первого порядка (как в формуле 29):
$$W_1(\bar{I}, \bar{\theta}) = -(2\bar{I})^{3/2} \left( \frac{3}{4}\sin\bar{\theta} + \frac{1}{12}\sin(3\bar{\theta}) \right)$$

Покажем, как оператор Ли рассчитывает явную замену переменных:

#### 1. Нахождение старого действия $I$ через новое $\bar{I}$
Возьмем частную производную от генератора $W_1$ по новой фазе $\bar{\theta}$:
$$\frac{\partial W_1}{\partial \bar{\theta}} = -(2\bar{I})^{3/2} \left( \frac{3}{4}\cos\bar{\theta} + \frac{3}{12}\cos(3\bar{\theta}) \right) = -(2\bar{I})^{3/2} \left( \frac{3}{4}\cos\bar{\theta} + \frac{1}{4}\cos(3\bar{\theta}) \right)$$

Подставим полученную производную в формулу (7):
$$I = \bar{I} + \varepsilon (2\bar{I})^{3/2} \left( \frac{3}{4}\cos\bar{\theta} + \frac{1}{4}\cos(3\bar{\theta}) \right)$$

#### 2. Нахождение старой фазы $\theta$ через новую $\bar{\theta}$
Возьмем частную производную от генератора $W_1$ по новому действию $\bar{I}$, используя правило дифференцирования сложной функции $\frac{\partial}{\partial \bar{I}}(2\bar{I})^{3/2} = \frac{3}{2}(2\bar{I})^{1/2} \cdot 2 = 3(2\bar{I})^{1/2}$:
$$\frac{\partial W_1}{\partial \bar{I}} = -3(2\bar{I})^{1/2} \left( \frac{3}{4}\sin\bar{\theta} + \frac{1}{12}\sin(3\bar{\theta}) \right)$$

Подставим полученную производную в формулу (8):
$$\theta = \bar{\theta} - \varepsilon 3(2\bar{I})^{1/2} \left( \frac{3}{4}\sin\bar{\theta} + \frac{1}{12}\sin(3\bar{\theta}) \right)$$

### 📌 Главный вывод примера
Мы получили **прямые и явные функции** связи переменных: $I = f(\bar{I}, \bar{\theta})$ и $\theta = g(\bar{I}, \bar{\theta})$. 

В отличие от метода Фон Цейпеля, нам **не требовалось** решать нелинейные тригонометрические уравнения, чтобы выразить старые координаты через новые. Оператор Ли выполнил это автоматически за счёт канонических свойств скобки Пуассона.


### 5. Ряды Хори — Депри: Математический мост к компьютерной алгебре

Как обсуждалось ранее, классическая двухэтапная схема **Брауэра - фон Цейпеля** имела существенный недостаток — использование *смешанных производящих функций* $S(\bar{I}, \theta)$, — требующий  ручного обращения нелинейных тригонометрических рядов Тейлора. С ростом порядка возмущений формулы разрастались на сотни страниц, делая вычисления вручную невозможными. Чтобы переложить эту рутину на компьютеры, японский астроном Генъитиро Хори (1966) и бельгийский математик Андре Депри (1969) независимо разработали метод **рядов Хори — Депри (Hori–Deprit series)**. Они переписали теорию возмущений гамильтоновых систем, полностью заменив неявные функции фон Цейпеля на явный экспоненциальный оператор Ли:

$$\bar{z} = \exp(\varepsilon L_W) z$$

#### Рекуррентные уравнения Хори — Депри
Пусть исходный гамильтониан системы и искомый генератор Ли (функция Депри) разложены в асимптотические ряды по степеням малого параметра $\varepsilon$:

$$H(z, \varepsilon) = H_0(z) + \varepsilon H_1(z) + \varepsilon^2 H_2(z) + \dots$$
$$W(z, \varepsilon) = W_1(z) + \varepsilon W_2(z) + \varepsilon^2 W_3(z) + \dots$$

Мы ищем преобразованный, усредненный гамильтониан $\bar{H}$, который также представляется в виде ряда $\bar{H} = \bar{H}_0 + \varepsilon \bar{H}_1 + \varepsilon^2 \bar{H}_2 + \dots$. Подстановка оператора Ли и группировка членов при одинаковых степенях $\varepsilon$ порождает строго последовательный рекуррентный алгоритм:

* **Порядок $O(1)$:**
  $$ \bar{H}_0 = H_0 $$

* **Порядок $O(\varepsilon)$:**
  $$ \bar{H}_1 = H_1 + \{H_0, W_1\} $$
  *Усреднением $H_1$ по быстрой фазе находится $\bar{H}_1$, а интегрированием остатка — первый генератор Ли $W_1$.*

* **Порядок $O(\varepsilon^2)$:**
  Благодаря свойствам преобразований Ли, Депри упростил громоздкие вложенные скобки до компактного канонического вида:
  $$ \bar{H}_2 = H_2 + \frac{1}{2}\{H_1 + \bar{H}_1, W_1\} + \{H_0, W_2\} $$

* **Порядок $O(\varepsilon^3)$:**
  $$ \bar{H}_3 = H_3 + \{H_2, W_1\} + \frac{1}{2}\{H_1, W_2\} + \frac{1}{2}\{\bar{H}_1, W_2\} + \frac{1}{12}\{\{\{H_0, W_1\}, W_1\}, W_1\} + \{H_0, W_3\} $$

#### Историческое значение
Андре Депри с помощью этого алгоритма и ЭВМ первого поколения полностью пересчитал классическую *главную задачу теории движения Луны*, разложив её гамильтониан до **13-го порядка малости**. Вручную методом фон Цейпеля такой расчет занял бы всю человеческую жизнь и неизбежно содержал бы арифметические ошибки. Сегодня явный рекуррентный аппарат Хори — Депри является стандартом автоматизации в компьютерной алгебре.

## Автоматизация высших порядков метода Ли с помощью SymPy

В высших порядках метода Ли преобразование гамильтониана задается рекуррентными уравнениями Хори–Депри. Для эффективного гамильтониана $\bar{H} = \bar{H}_0 + \varepsilon \bar{H}_1 + \varepsilon^2 \bar{H}_2 + \varepsilon^3 \bar{H}_3 + \dots$ уравнения имеют вид:

* $O(1): \bar{H}_0 = H_0$
* $O(\varepsilon): \bar{H}_1 = \langle H_1 \rangle = 0$, а генератор $W_1$ находится из уравнения $\frac{\partial W_1}{\partial \theta} = -H_1$
* $O(\varepsilon^2): \bar{H}_2 = \frac{1}{2} \langle \{H_1, W_1\} \rangle$, а генератор $W_2$ находится из уравнения $\frac{\partial W_2}{\partial \theta} = \bar{H}_2 - \frac{1}{2}\{H_1, W_1\}$
* $O(\varepsilon^3): \bar{H}_3 = \frac{1}{3} \langle \{\{H_1, W_1\}, W_1\} \rangle + \frac{1}{2} \langle \{H_1, W_2\} \rangle$

В переменных действие-угол каноническая скобка Пуассона двух функций $F(I, \theta)$ и $G(I, \theta)$ вычисляется по формуле:
$$ \{F, G\} = \frac{\partial F}{\partial \theta}\frac{\partial G}{\partial I} - \frac{\partial F}{\partial I}\frac{\partial G}{\partial \theta} $$

Реализуем автоматический алгоритм, который самостоятельно вычисляет скобки Пуассона, выполняет тригонометрическое усреднение $\langle \cdot \rangle = \frac{1}{2\pi}\int_0^{2\pi} \cdot \, d\theta$ и находит поправки к гамильтониану и сдвигу частоты до **третьего порядка малости включительно**.


```python
import sympy as sp

# 1. Объявляем символьные переменные
# I — действие, theta — угол (фаза)
I, theta = sp.symbols("I theta", positive=True)


# 2. Функция для автоматического расчета скобок Пуассона
# {F, G} = (dF/d_theta)*(dG/d_I) - (dF/d_I)*(dG/d_theta)
def poisson_bracket(F, G, q_var=theta, p_var=I):
    bracket = sp.diff(F, q_var) * sp.diff(G, p_var) - sp.diff(
        F, p_var
    ) * sp.diff(G, q_var)
    return sp.trigsimp(bracket)


# 3. Функция для усреднения по быстрой фазе theta от 0 до 2*pi
def average_over_phase(expr, phase_var=theta):
    # Интегрируем по фазе и делим на длину периода (2*pi)
    avg = (1 / (2 * sp.pi)) * sp.integrate(expr, (phase_var, 0, 2 * sp.pi))
    return sp.expand(sp.trigsimp(avg))


# 4. Постановка задачи для ангармонического осциллятора
# H0 = I, H1 = q^3 = (2I)^(3/2) * cos(theta)^3
H0 = I
H1 = (2 * I) ** sp.Rational(3, 2) * sp.cos(theta) ** 3

print("=== СИМВОЛЬНЫЙ РАСЧЕТ МЕТОДА ЛИ (SymPy) ===")

# --- Порядок O(1) ---
H_bar_0 = H0
print(f"H_bar_0 = {H_bar_0}")

# --- Порядок O(eps) ---
H_bar_1 = average_over_phase(H1)
# Уравнение для W1: dW1/d_theta = H_bar_1 - H1 = -H1 (так как dH0/dI = 1)
dW1_dtheta = H_bar_1 - H1
W1 = sp.integrate(dW1_dtheta, theta)
W1 = sp.trigsimp(W1)
print(f"H_bar_1 = {H_bar_1}")
print(f"Генератор Ли W1 = {W1}")

# --- Порядок O(eps^2) ---
# Находим первую скобку Пуассона {H1, W1}
bracket_1 = poisson_bracket(H1, W1)
H_bar_2 = sp.Rational(1, 2) * average_over_phase(bracket_1)
# Уравнение для W2: dW2/d_theta = H_bar_2 - 1/2 * {H1, W1}
dW2_dtheta = H_bar_2 - sp.Rational(1, 2) * bracket_1
W2 = sp.integrate(dW2_dtheta, theta)
W2 = sp.trigsimp(W2)
print(f"H_bar_2 = {H_bar_2}")
print(f"Генератор Ли W2 = {W2}")

# --- Порядок O(eps^3) ---
# Вычисляем двойную скобку Пуассона {{H1, W1}, W1}
bracket_2 = poisson_bracket(bracket_1, W1)
# Вычисляем скобку Пуассона {H1, W2}
bracket_3 = poisson_bracket(H1, W2)

# Собираем гамильтониан третьего порядка по формуле Хори-Депри
H_bar_3_expr = sp.Rational(1, 3) * bracket_2 + sp.Rational(1, 2) * bracket_3
H_bar_3 = average_over_phase(H_bar_3_expr)
print(f"H_bar_3 = {H_bar_3}\n")


# ==========================================
# ИТОГОВЫЙ РЕЗУЛЬТАТ И СДВИГ ЧАСТОТЫ
# ==========================================
eps = sp.symbols("varepsilon")
H_bar_total = H_bar_0 + eps * H_bar_1 + eps**2 * H_bar_2 + eps**3 * H_bar_3

# Находим модифицированную частоту как производную dH_bar/dI
omega_perturbed = sp.diff(H_bar_total, I)

print("=== ИТОГОВЫЕ УРАВНЕНИЯ СИСТЕМЫ ===")
print(f"Преобразованный Гамильтониан H_bar(I):")
sp.pprint(H_bar_total)
print(f"\nМодифицированная частота осциллятора omega(I):")
sp.pprint(omega_perturbed)
```

### Анализ результатов символьных вычислений

С помощью разработанного алгоритма компьютерной алгебры мы получили точные аналитические выражения без рутинных тригонометрических преобразований:

1. **Порядок $O(\varepsilon^2)$:** Программа подтвердила наш предыдущий ручной вывод. Эффективный гамильтониан второго порядка равен $-\frac{15}{4}\varepsilon^2 I^2$, что дает сдвиг частоты на $-\frac{15}{2}\varepsilon^2 I$.
2. **Порядок $O(\varepsilon^3)$:** Автоматический расчет показал, что $H_{bar, 3} = 0$. Это важный теоретический результат: для нелинейных потенциалов нечетной степени (таких как кубическое возмущение $\varepsilon q^3$) все нечетные поправки к усредненному гамильтониану ($\bar{H}_1, \bar{H}_3, \bar{H}_5 \dots$) тождественно обращаются в нуль. Симметричный сдвиг траекторий полностью компенсируется при интегрировании по замкнутому периоду.

Этот скрипт является универсальным. Чтобы рассчитать высшие поправки для любой другой системы (например, для маятника или осциллятора Дуффинга с четвертой степенью возмущения $\varepsilon q^4$), достаточно просто изменить выражение для функции `H1` в четвертом блоке кода ячейки.


## Метод генераторов Ли в теории возмущений гамильтоновых систем

**Метод Ли** основывается на использовании генераторов Ли (производящих векторных полей) вместо классических производящих функций. Каноническое преобразование от старых переменных $z = (q, p)$ к новым $\bar{z} = (\bar{q}, \bar{p})$ задается в явном виде как решение вспомогательной гамильтоновой системы с гамильтонианом $W(z, \varepsilon)$, называемым *генератором Ли*:

$$z = \exp(\varepsilon L_W) \bar{z} = \bar{z} + \varepsilon \{ \bar{z}, W \} + \frac{\varepsilon^2}{2!} \{ \{ \bar{z}, W \}, W \} + \dots \tag{9}$$

где $L_W = \{\cdot, W\}$ — оператор, сопоставляющий функции её скобку Пуассона с генератором $W$. Любая функция (включая Гамильтониан) преобразуется по аналогичному явному закону:

$$\bar{H} = \exp(\varepsilon L_W) H = H + \varepsilon \{ H, W \} + \frac{\varepsilon^2}{2} \{ \{ H, W \}, W \} + \dots \tag{10}$$

Рассмотрим этот метод на примере ангармонического осциллятора с кубическим возмущением в переменных «действие–угол»:

$$H(I, \theta) = H_0(I) + \varepsilon H_1(I, \theta) = I + \varepsilon (2I)^{3/2} \cos^3\theta \tag{11}$$

Разложим генератор Ли $W$ и новый гамильтониан $\bar{H}$ в ряды по степеням $\varepsilon$:
$$W = W_1 + \varepsilon W_2 + \dots$$
$$\bar{H} = \bar{H}_0 + \varepsilon \bar{H}_1 + \varepsilon^2 \bar{H}_2 + \dots$$

Подставив эти разложения в (10) и собирая члены при одинаковых степенях $\varepsilon$, получаем систему уравнений Хори–Депри:

* **Порядок $O(1)$:**
  $$\bar{H}_0(\bar{I}) = H_0(\bar{I}) = \bar{I} \tag{12}$$

* **Порядок $O(\varepsilon)$:**
  $$\bar{H}_1(\bar{I}) = H_1(\bar{I}, \theta) + \{ H_0, W_1 \} \tag{13}$$

В переменных действие-угол скобка Пуассона с невозмущенным гамильтонианом имеет простой вид: 
$$\{ H_0, W_1 \} = \frac{\partial H_0}{\partial \bar{I}}\frac{\partial W_1}{\partial \bar{\theta}} - \frac{\partial H_0}{\partial \bar{\theta}}\frac{\partial W_1}{\partial \bar{I}} = \frac{\partial W_1}{\partial \bar{\theta}}$$. 

Тогда уравнение (11) перепишется как:
$$\bar{H}_1(\bar{I}) = H_1(\bar{I}, \bar{\theta}) + \frac{\partial W_1}{\partial \bar{\theta}} \tag{14}$$

Усредняя уравнение (12) по быстрой фазе $\bar{\theta}$, находим:
$$\bar{H}_1(\bar{I}) = \langle H_1(\bar{I}, \bar{\theta}) \rangle = 0 \tag{15}$$
$$\frac{\partial W_1}{\partial \bar{\theta}} = -H_1(\bar{I}, \bar{\theta}) = -(2\bar{I})^{3/2} \left( \frac{3}{4}\cos\bar{\theta} + \frac{1}{4}\cos(3\bar{\theta}) \right) \tag{16}$$

Интегрируя по фазе, получаем генератор Ли первого порядка:
$$W_1(\bar{I}, \bar{\theta}) = -(2\bar{I})^{3/2} \left( \frac{3}{4}\sin\bar{\theta} + \frac{1}{12}\sin(3\bar{\theta}) \right) \tag{17}$$

* **Порядок $O(\varepsilon^2)$:**
  Уравнение второго порядка в методе Ли выглядит так:
  $$\bar{H}_2(\bar{I}) = \langle H_2 \rangle + \frac{1}{2} \langle \{ H_1 + \bar{H}_1, W_1 \} \rangle \tag{18}$$
  Поскольку исходного $H_2 = 0$ и $\bar{H}_1 = 0$, уравнение сводится к среднему от скобки Пуассона:
  $$\bar{H}_2(\bar{I}) = \frac{1}{2} \left\langle \left\{ H_1, W_1 \right\} \right\rangle = \frac{1}{2} \left\langle \frac{\partial H_1}{\partial \bar{I}}\frac{\partial W_1}{\partial \bar{\theta}} - \frac{\partial H_1}{\partial \bar{\theta}}\frac{\partial W_1}{\partial \bar{I}} \right\rangle \tag{19}$$

Вычисление этого среднего дает тот же инвариантный результат, что и метод Фон Цейпеля: $\bar{H}_2(\bar{I}) = -\frac{15}{4}\bar{I}^2$. Однако здесь связь между старыми и новыми переменными рассчитывается **напрямую по формуле (9)** без необходимости неявного дифференцирования и обращения функций:
$$I = \bar{I} + \varepsilon \{ \bar{I}, W_1 \} = \bar{I} - \varepsilon \frac{\partial W_1}{\partial \bar{\theta}} \tag{20}$$

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# 1. Параметры системы
eps = 0.08
I_target = 1.2  # Целевое значение инварианта Ли


# Гамильтоновы уравнения движения в декартовых координатах
def duffing_equations(t, state):
    q, p = state
    return [p, -q - 3 * eps * q**2]


# Начальные условия при theta = 0, вычисленные ПРЯМЫМ методом Ли
theta_init = 0.0
I_init = I_target - eps * (2 * I_target) ** 1.5 * (
    0.75 * np.cos(theta_init) + 0.25 * np.cos(3 * theta_init)
)

q_init = np.sqrt(2 * I_init) * np.cos(theta_init)
p_init = -np.sqrt(2 * I_init) * np.sin(theta_init)

# Симуляция
t_span = (0.0, 20.0)
t_eval = np.linspace(0.0, 20.0, 800)
sol = solve_ivp(duffing_equations, t_span, [q_init, p_init], t_eval=t_eval)

q_arr = sol.y[0]  # Координата (размерность 800,)
p_arr = sol.y[1]  # Скорость (размерность 800,)

# 2. Прямой пересчет траектории в новые переменные Ли
I_old = 0.5 * (p_arr**2 + q_arr**2)
theta_old = np.arctan2(-p_arr, q_arr)

# Явное преобразование Ли для нахождения инварианта в любой точке:
bar_I_calculated = I_old + eps * (2 * I_old) ** 1.5 * (
    0.75 * np.cos(theta_old) + 0.25 * np.cos(3 * theta_old)
)


# 3. Визуализация выпрямления фазовых траекторий оператором Ли
plt.figure(figsize=(11, 5))

plt.plot(
    sol.t,
    I_old,
    color="royalblue",
    alpha=0.5,
    label="Старое действие $I(t)$ (до преобразования Ли)",
)
plt.plot(
    sol.t,
    bar_I_calculated,
    color="forestgreen",
    linewidth=2.5,
    label="Новое действие $\\bar{I}(t)$ (после преобразования Ли)",
)
plt.axhline(
    I_target,
    color="black",
    linestyle="--",
    alpha=0.7,
    label="Теоретический инвариант",
)

plt.title("Действие оператора преобразования Ли на переменную действия", fontsize=12)
plt.xlabel("Время $t$", fontsize=11)
plt.ylabel("Значение действия", fontsize=11)
plt.ylim(I_target - 0.4, I_target + 0.4)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

### Анализ эффективности преобразования Ли

График наглядно демонстрирует фундаментальное преимущество методов преобразования Ли:
* **Синяя линия ($I(t)$):** Отражает изменение старой физической переменной действия, которая колеблется из-за нелинейных членов.
* **Зеленая гладкая линия ($\bar{I}(t)$):** Показывает результат применения экспоненциального оператора Ли $\exp(\varepsilon L_W)$ непосредственно к точкам траектории. Быстрые гармоники полностью взаимоуничтожаются скобками Пуассона, превращая колеблющуюся кривую в математически строгую константу. 

Поскольку нам не потребовалось обращать нелинейные уравнения (как в методе фон Цейпеля), этот алгоритм легко масштабируется на компьютере до 10-го порядка малости и выше с помощью систем компьютерной алгебры (SymPy/Mathematica).


## Нелинейный резонанс и фазовый портрет в резонансной полосе

Рассмотрим нелинейный осциллятор под внешним гармоническим воздействием в условиях основного резонанса ($\Omega \approx \omega_0$). Перейдем к медленно меняющимся амплитуде $a(t)$ и фазовому сдвигу $\psi(t) = \Omega t - \varphi(t)$ с помощью метода усреднения Ван-дер-Поля. 

В отсутствие затухания ($\delta = 0$) уравнения для медленных переменных можно представить в гамильтоновой форме относительно канонического действия $I \propto a^2$ и резонансной фазы $\psi$. Эффективный резонансный гамильтониан имеет вид:

$$ \mathcal{H}(I, \psi) = \Delta \omega \cdot I + \frac{3\beta}{8\omega_0} I^2 - \varepsilon \sqrt{2I} \cos\psi \tag{33} $$

где $\Delta \omega = \Omega - \omega_0$ — частотная расстройка.

Линии уровня этого гамильтониана $\mathcal{H}(I, \psi) = \text{const}$ задают фазовые траектории системы в резонансной полосе. Чтобы избавиться от сингулярности в нуле, перейдем к декартовым медленным переменным (синфазной и квадратурной компонентам):
$$ u = \sqrt{2I} \cos\psi, \quad v = \sqrt{2I} \sin\psi $$

Тогда резонансный гамильтониан (33) перепишется как:
$$ \mathcal{H}(u, v) = \frac{\Delta \omega}{2} (u^2 + v^2) + \frac{3\beta}{32\omega_0} (u^2 + v^2)^2 - \varepsilon u \tag{34} $$

Построим этот фазовый портрет в плоскости медленных переменных $(u, v)$, чтобы визуализировать области захвата в резонанс, седловые точки и сепаратрисы.


```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Задаем параметры нелинейного резонанса
omega_0 = 1.0  # Собственная частота
beta = 0.4  # Кубическая нелинейность (жесткий осциллятор)
eps = 0.15  # Амплитуда внешнего возмущения
delta_omega = -0.35  # Расстройка частоты (Omega - omega_0)

# 2. Создаем сетку на плоскости медленных переменных (u, v)
u_range = np.linspace(-2.5, 2.5, 400)
v_range = np.linspace(-2.5, 2.5, 400)
U, V = np.meshgrid(u_range, v_range)

# 3. Вычисляем значение резонансного гамильтониана (формула 34) в каждой точке
# H = (delta_omega / 2) * (u^2 + v^2) + (3 * beta / (32 * omega_0)) * (u^2 + v^2)^2 - eps * u
I_coords = 0.5 * (U**2 + V**2)
H = delta_omega * I_coords + (3 * beta / (8 * omega_0)) * (I_coords**2) - eps * U

# 4. Визуализация фазового портрета
plt.figure(figsize=(9, 8))

# Рисуем изолинии гамильтониана (траектории движения медленных переменных)
contour = plt.contour(
    U, V, H, levels=50, colors="royalblue", linewidths=1.0, alpha=0.8
)
plt.clabel(contour, inline=True, fontsize=8, fmt="%.2f")

# Находим и отмечаем стационарные точки численно (где градиент H равен 0 при v=0)
# dH/du = delta_omega * u + (3 * beta / 8) * u^3 - eps = 0
poly_coeffs = [3 * beta / 8, 0, delta_omega, -eps]
roots = np.roots(poly_coeffs)

# Фильтруем только вещественные корни
real_roots = roots[np.isreal(roots)].real

for root in real_roots:
    # Проверяем тип точки по знаку второй производной d2H/du2
    d2H_du2 = delta_omega + (9 * beta / 8) * root**2
    if d2H_du2 > 0:
        # Минимум H -> Устойчивый фокус / центр (Резонансный режим)
        plt.scatter(
            root,
            0,
            color="forestgreen",
            edgecolor="black",
            s=70,
            zorder=5,
            label="Устойчивый центр (Амплитудный режим)"
            if "Устойчивый центр" not in plt.gca().get_legend_handles_labels()[1]
            else "",
        )
    else:
        # Максимум H -> Седловая точка (Неустойчивый режим, порождающий сепаратрису)
        plt.scatter(
            root,
            0,
            color="crimson",
            marker="x",
            s=90,
            linewidths=2.5,
            zorder=5,
            label="Седловая точка (Сепаратриса)"
            if "Седловая точка" not in plt.gca().get_legend_handles_labels()[1]
            else "",
        )

# Оформление графика
plt.title(
    f"Фазовый портрет нелинейного резонанса в медленных переменных\n"
    f"(Расстройка $\Delta\omega = {delta_omega}$, Возмущение $\epsilon = {eps}$)",
    fontsize=12,
    pad=15,
)
plt.xlabel("Синфазная компонента $u = a \cos\psi$", fontsize=11)
plt.ylabel("Квадратурная компонента $v = a \sin\psi$", fontsize=11)

plt.axhline(0, color="black", linewidth=0.8, linestyle="--", alpha=0.5)
plt.axvline(0, color="black", linewidth=0.8, linestyle="--", alpha=0.5)
plt.xlim(-2.5, 2.5)
plt.ylim(-2.5, 2.5)
plt.gca().set_aspect("equal")
plt.grid(True, linestyle=":", alpha=0.4)
plt.legend(loc="upper right", fontsize=10)

plt.tight_layout()
plt.show()
```

### Анализ фазовой структуры резонансной полосы

Полученный фазовый портрет наглядно иллюстрирует явление нелинейного резонанса и объясняет физику бифуркации:

1. **Множественность стационарных режимов:** При выбранной отрицательной расстройке $\Delta\omega$ кубическое уравнение дает три вещественных корня. Это означает, что в системе одновременно существуют **три стационарные точки**: два устойчивых центра (зеленые маркеры) и одна седловая точка (красный крест).
2. **Физический смысл центров:** Правый и левый центры соответствуют двум различным амплитудам установившихся вынужденных колебаний осциллятора (одно решение с большой амплитудой в фазе с силой, другое — с меньшей амплитудой в противофазе). Какой именно режим займет система, зависит исключительно от начальных условий.
3. **Резонансная петля (Сепаратриса):** Седловая точка порождает замкнутую фазовую петлю — сепаратрису (аналог петли маятника). Траектории внутри этой петли совершают колебания вокруг резонансного центра, описывая медленные биения амплитуды и фазы в режиме фазового захвата.
