# 3. Метод фон Цейпеля

**Метод фон Цейпеля** — классический асимптотический метод теории возмущений для гамильтоновых систем. Его основная цель — исключить «быстрые» переменные (короткопериодические члены) с помощью канонических преобразований, задаваемых генерирующей (производящей) функцией.

Рассмотрим применение этого метода на примере одномерного слабонелинейного ангармонического осциллятора.

---

### 1. Постановка задачи

Рассмотрим гамильтониан гармонического осциллятора с малой кубической добавкой к потенциалу:

$$H(q, p) = \frac{p^2 + q^2}{2} + \varepsilon q^3 \tag{1}$$

где $\varepsilon \ll 1$ — малый параметр возмущения.

Перейдем к каноническим переменным **«действие–угол»** $(I, \theta)$ для невозмущенной части системы с помощью стандартной тригонометрической замены:

$$q = \sqrt{2I} \cos\theta, \quad p = -\sqrt{2I} \sin\theta$$

Подставив эти выражения в исходный гамильтониан (1), получим:

$$H(I, \theta) = I + \varepsilon (2I)^{3/2} \cos^3\theta \tag{2}$$

Используя тригонометрическое тождество $\cos^3\theta = \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta)$, перепишем возмущенную часть в виде суммы гармоник:

$$H(I, \theta) = I + \varepsilon (2I)^{3/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right) \tag{3}$$

Переменная $\theta$ является «быстрой» фазой (совершает полный оборот за период колебания). Наша задача — найти такое новое действие $\bar{I}$, которое оставалось бы постоянным в первом приближении по $\varepsilon$.


### 2. Производящая функция фон Цейпеля

Ищем каноническое преобразование от старых переменных $(I, \theta)$ к новым «медленным» переменным $(\bar{I}, \bar{\theta})$ с помощью производящей функции следующего вида:

$$S(\bar{I}, \theta) = \bar{I}\theta + \varepsilon S_1(\bar{I}, \theta) + O(\varepsilon^2) \tag{4}$$

Связь между старыми и новыми переменными согласно теореме о производящей функции задается уравнениями:

$$I = \frac{\partial S}{\partial \theta} = \bar{I} + \varepsilon \frac{\partial S_1}{\partial \theta}$$

$$\bar{\theta} = \frac{\partial S}{\partial \bar{I}} = \theta + \varepsilon \frac{\partial S_1}{\partial \bar{I}} \tag{5}$$

Новый гамильтониан $\bar{H}(\bar{I})$ по условию метода не должен зависеть от быстрой фазы $\theta$. Разложим его в ряд по степеням малого параметра:

$$
\bar{H}(\bar{I}) = \bar{H}_0(\bar{I}) + \varepsilon \bar{H}_1(\bar{I}) + O(\varepsilon^2) \tag{6}
$$


### 3. Нахождение уравнений первого приближения

Подставим соотношения между нвыми и старыми переменными (5) в исходный гамильтониан (3) и приравняем его к новому целевому гамильтониану (6):

$$
\left( \bar{I} + \varepsilon \frac{\partial S_1}{\partial \theta} \right) + \varepsilon (2\bar{I})^{3/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right) = \bar{H}_0(\bar{I}) + \varepsilon \bar{H}_1(\bar{I}) \tag{7}
$$

Разделим полученное уравнение по порядкам малости $\varepsilon$.

* **Порядок $O(1)$:**
  $$
  \bar{H}_0(\bar{I}) = \bar{I} \tag{8}
  $$
  *Это невозмущенный закон сохранения энергии.*

* **Порядок $O(\varepsilon)$:**
  $$
  \frac{\partial S_1}{\partial \theta} + (2\bar{I})^{3/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right) = \bar{H}_1(\bar{I}) \tag{9}
  $$


### 4. Разделение на среднее и осциллирующее составляющие

Главная идея метода фон Цейпеля заключается в разделении уравнения (9) на две независимые части:
1. **Усредненная часть** — определяет новый гамильтониан $\bar{H}_1$ (в нее собираются члены, принципиально не зависящие от фазы $\theta$).
2. **Осциллирующая часть** — определяет периодическую поправку к производящей функции $S_1$.

Так как среднее значение функций $\cos\theta$ и $\cos(3\theta)$ за полный период равно нулю, то в левой части уравнения (9) отсутствуют постоянные члены. Отсюда напрямую следует:

$$
\bar{H}_1(\bar{I}) = 0 \tag{10}
$$

Для оставшейся осциллирующей части получаем дифференциальное уравнение:

$$
\frac{\partial S_1}{\partial \theta} = - (2\bar{I})^{3/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right) \tag{11}
$$

Интегрируя уравнение (11) по быстрой фазе $\theta$, находим явный вид функции $S_1$:

$$
S_1(\bar{I}, \theta) = - (2\bar{I})^{3/2} \left( \frac{3}{4}\sin\theta + \frac{1}{12}\sin(3\theta) \right) \tag{12}
$$


### ✅ Аналитический результат

1. **Новый гамильтониан системы** в первом приближении совпадает с невозмущенным:
   $$
   \bar{H}(\bar{I}) = \bar{I} + O(\varepsilon^2)
   $$
   Это говорит о важном физическом свойстве: кубическая нелинейность потенциала в первом приближении **не смещает частоту** колебаний осциллятора (сдвиг частоты проявится только во втором порядке порядка $\varepsilon^2$).

2. **Связь между старым и новым действием** (из формулы 5):
   $$
   I = \bar{I} - \varepsilon (2\bar{I})^{3/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right)
   $$
   Поскольку преобразованный гамильтониан $\bar{H}$ полностью очищен от фазы, новое действие является строгим инвариантом ($\dot{\bar{I}} = 0$). Старое действие $I(t)$ просто совершает малые периодические пульсации амплитудой $\sim \varepsilon$ вокруг этого стабильного среднего значения $\bar{I}$.


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# 1. Параметры системы
eps = 0.05  # Малый параметр возмущения (кубическая нелинейность)
bar_I = 1.0  # Заданное постоянное новое действие (инвариант Фон Цейпеля)


# 2. Канонические уравнения движения для исходной системы (1)
# H = (p^2 + q^2)/2 + eps * q^3
# dq/dt = dH/dp = p
# dp/dt = -dH/dq = -q - 3 * eps * q^2
def hamiltonian_system(t, state):
    q, p = state
    dqdt = p
    dpdt = -q - 3 * eps * q**2
    return [dqdt, dpdt]


# 3. Расчет начальных условий (q0, p0) по формулам метода при t=0 (фаза theta = 0)
# Старое действие I(0) выражается через новое стабильное действие bar_I:
theta_0 = 0.0
I_0 = bar_I - eps * (2 * bar_I) ** 1.5 * (
    0.75 * np.cos(theta_0) + 0.25 * np.cos(3 * theta_0)
)

# Переход от (I, theta) к декартовым (q, p)
q_0 = np.sqrt(2 * I_0) * np.cos(theta_0)
p_0 = -np.sqrt(2 * I_0) * np.sin(theta_0)

# Интегрирование системы уравнений
t_span = (0.0, 30.0)
t_eval = np.linspace(0.0, 30.0, 1000)
sol = solve_ivp(hamiltonian_system, t_span, [q_0, p_0], t_eval=t_eval)

q_vals = sol.y[0]
p_vals = sol.y[1]

# 4. Обратный пересчет численной траектории в переменные «действие-угол»
# I(t) = (p^2 + q^2) / 2
I_numerical = 0.5 * (p_vals**2 + q_vals**2)

# Вычисляем истинную фазу theta(t) = atan2(-p, q)
theta_numerical = np.arctan2(-p_vals, q_vals)

# Вычисляем аналитическое значение действия I по формуле первого приближения:
# I_analytical = bar_I - eps * (2*bar_I)^1.5 * (3/4 * cos(theta) + 1/4 * cos(3*theta))
I_analytical = bar_I - eps * (2 * bar_I) ** 1.5 * (
    0.75 * np.cos(theta_numerical) + 0.25 * np.cos(3 * theta_numerical)
)

# 5. Визуализация результатов метода Фон Цейпеля
plt.figure(figsize=(11, 5))

# Истинное пульсирующее действие из численной симуляции
plt.plot(
    sol.t,
    I_numerical,
    color="royalblue",
    linewidth=2,
    label="Старое действие $I(t)$ (численный расчет)",
)

# Аналитическое приближение старого действия
plt.plot(
    sol.t,
    I_analytical,
    color="black",
    linestyle=":",
    linewidth=2,
    label="Аналитическое приближение $I(\\theta)$",
)

# Стабильный инвариант фон Цейпеля
plt.axhline(
    bar_I,
    color="crimson",
    linewidth=2.5,
    label=r"Новое действие $\bar{I}$ (инвариант фон Цейпеля)",
)

# Оформление осей и графика
plt.title(
    f"Исключение быстрой фазы по методу фон Цейпеля (при $\\varepsilon = {eps}$)",
    fontsize=12,
)
plt.xlabel("Время $t$", fontsize=11)
plt.ylabel("Значение действия $I$", fontsize=11)
plt.ylim(bar_I - 0.3, bar_I + 0.3)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right", fontsize=10)

plt.tight_layout()
plt.show()
```

### 5. Анализ результатов численного моделирования

Численный эксперимент подтверждает теоретические выводы метода фон Цейпеля. На полученном графике можно наглядно проследить разделение движения на медленную (постоянную) и быструю (осциллирующую) составляющие:

* <span style="color:crimson">**Красная линия ($\bar{I}$):**</span> Демонстрирует новое каноническое действие. Оно остается абсолютно постоянным во времени ($\dot{\bar{I}} = 0$), поскольку преобразованный гамильтониан $\bar{H}(\bar{I})$ первого приближения был успешно очищен от быстрой фазы $\theta$.
* <span style="color:royalblue">**Синяя линия ($I$):**</span> Показывает, что старое действие в исходной системе не сохраняется. Из-за наличия кубического возмущения в потенциале ангармонического осциллятора величина $I(t)$ совершает незатухающие быстрые колебания с амплитудой порядка $\sim \varepsilon$.
* **Черная пунктирная линия:** Представляет собой аналитическое приближение старого действия, вычисленное через обратную замену переменных. Она практически идеально накладывается на синий график численного расчета. Это доказывает, что найденная нами периодическая поправка к производящей функции $S_1(\bar{I}, \theta)$ с асимптотической точностью $O(\varepsilon)$ абсолютно верно описывает реальные пульсации системы вокруг стабильного среднего уровня.


### 6. Второе приближение метода фон Цейпеля: Сдвиг частоты $O(\varepsilon^2)$

Чтобы учесть эффекты второго порядка, расширим разложение производящей функции $S$ и нового гамильтониана $\bar{H}$ до членов с $\varepsilon^2$:

$$S(\bar{I}, \theta) = \bar{I}\theta + \varepsilon S_1(\bar{I}, \theta) + \varepsilon^2 S_2(\bar{I}, \theta) + O(\varepsilon^3) \tag{13}$$

$$\bar{H}(\bar{I}) = \bar{H}_0(\bar{I}) + \varepsilon \bar{H}_1(\bar{I}) + \varepsilon^2 \bar{H}_2(\bar{I}) + O(\varepsilon^3) \tag{14}$$

Связь между старым действием $I$ и новым $\bar{I}$ теперь включает поправку второго порядка:
$$I = \frac{\partial S}{\partial \theta} = \bar{I} + \varepsilon \frac{\partial S_1}{\partial \theta} + \varepsilon^2 \frac{\partial S_2}{\partial \theta} \tag{15}$$

Подставим это разложение в исходный гамильтониан $H(I, \theta)$ и разложим его в ряд Тейлора в окрестности $\bar{I}$. Уравнение второго порядка малости $O(\varepsilon^2)$ примет вид:

$$\frac{\partial S_2}{\partial \theta} \frac{\partial H_0}{\partial I} + \frac{1}{2} \left(\frac{\partial S_1}{\partial \theta}\right)^2 \frac{\partial^2 H_0}{\partial I^2} + \frac{\partial S_1}{\partial \theta} \frac{\partial H_1}{\partial I} + H_2\left(\bar{I}, \theta, \frac{\partial S_1}{\partial \theta}\right) = \bar{H}_2(\bar{I}) \tag{16}$$

Учитывая, что для нашего осциллятора:
* $\frac{\partial H_0}{\partial I} = 1$ и $\frac{\partial^2 H_0}{\partial I^2} = 0$
* $\bar{H}_1 = 0 \implies \frac{\partial H_1}{\partial I} = 0$
* Возмущение из уравнения (3) разложено по Тейлору относительно $\bar{I}$:
  $$H_1(I, \theta) \approx H_1(\bar{I}, \theta) + \frac{\partial H_1}{\partial \bar{I}}\left(\varepsilon \frac{\partial S_1}{\partial \theta}\right)$$

Уравнение для $O(\varepsilon^2)$ существенно упрощается:
$$\frac{\partial S_2}{\partial \theta} + \frac{\partial H_1(\bar{I}, \theta)}{\partial \bar{I}} \frac{\partial S_1}{\partial \theta} = \bar{H}_2(\bar{I}) \tag{17}$$

#### Нахождение векового члена $\bar{H}_2(\bar{I})$

Вспомним явный вид наших функций из первого приближения:
$$H_1(\bar{I}, \theta) = (2\bar{I})^{3/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right) \implies \frac{\partial H_1}{\partial \bar{I}} = 3(2\bar{I})^{1/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right)$$
$$\frac{\partial S_1}{\partial \theta} = - (2\bar{I})^{3/2} \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right)$$

Перемножим их, чтобы найти нелинейный вклад:
$$\frac{\partial H_1}{\partial \bar{I}} \frac{\partial S_1}{\partial \theta} = -3(2\bar{I})^2 \left( \frac{3}{4}\cos\theta + \frac{1}{4}\cos(3\theta) \right)^2$$

Усредним это выражение по быстрой фазе $\theta$ от $0$ до $2\pi$. Помня, что $\langle\cos^2\theta\rangle = \frac{1}{2}$, $\langle\cos^2(3\theta)\rangle = \frac{1}{2}$ и $\langle\cos\theta\cos(3\theta)\rangle = 0$:
$$\left\langle \frac{\partial H_1}{\partial \bar{I}} \frac{\partial S_1}{\partial \theta} \right\rangle = -3(2\bar{I})^2 \left( \frac{9}{16} \cdot \frac{1}{2} + \frac{1}{16} \cdot \frac{1}{2} \right) = -3(4\bar{I}^2) \left(\frac{10}{32}\right) = -\frac{15}{4}\bar{I}^2$$

Поскольку среднее значение $\frac{\partial S_2}{\partial \theta}$ равно нулю, из уравнения (17) получаем новый гамильтониан второго порядка:
$$\bar{H}_2(\bar{I}) = -\frac{15}{4}\bar{I}^2 \tag{18}$$

#### Итоговый эффективный гамильтониан и сдвиг частоты
Полный усредненный гамильтониан до $O(\varepsilon^2)$ равен:
$$\bar{H}(\bar{I}) = \bar{I} - \varepsilon^2 \frac{15}{4}\bar{I}^2 \tag{19}$$

Теперь найдем модифицированную частоту колебаний нелинейного осциллятора $\bar{\omega}$:
$$\bar{\omega}(\bar{I}) = \frac{\partial \bar{H}}{\partial \bar{I}} = 1 - \varepsilon^2 \frac{15}{2}\bar{I} \tag{20}$$

**Вывод:** Кубическая нелинейность во втором порядке малости **уменьшает частоту** колебаний осциллятора пропорционально квадрату параметра возмущения $\varepsilon^2$ и величине действия (амплитуде).


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# 1. Параметры (увеличим eps для наглядности сдвига частоты)
eps = 0.15
bar_I = 1.0


# Канонические уравнения движения
def hamiltonian_system(t, state):
    q, p = state
    return [p, -q - 3 * eps * q**2]


# Начальные условия при theta=0 (с учетом поправок)
theta_0 = 0.0
I_0 = bar_I - eps * (2 * bar_I) ** 1.5 * (
    0.75 * np.cos(theta_0) + 0.25 * np.cos(3 * theta_0)
)
q_0 = np.sqrt(2 * I_0) * np.cos(theta_0)
p_0 = -np.sqrt(2 * I_0) * np.sin(theta_0)

# Длительная симуляция для точной фиксации фазового сдвига
t_span = (0.0, 100.0)
t_eval = np.linspace(0.0, 100.0, 4000)
sol = solve_ivp(hamiltonian_system, t_span, [q_0, p_0], t_eval=t_eval)

q_vals = sol.y[0]
p_vals = sol.y[1]
t_vals = sol.t

# 2. Определение реальной частоты по пересечениям нуля
crossings = []
for i in range(len(p_vals) - 1):
    if q_vals[i] < 0 and q_vals[i + 1] >= 0:
        t_zero = t_vals[i] - q_vals[i] * (t_vals[i + 1] - t_vals[i]) / (
            q_vals[i + 1] - q_vals[i]
        )
        crossings.append(t_zero)

numerical_period = np.mean(np.diff(crossings))
omega_numerical = (2 * np.pi) / numerical_period

# 3. Теоретическая частота Фон Цейпеля (формула 20)
omega_theoretical = 1.0 - (eps**2) * (15.0 / 2.0) * bar_I
omega_unperturbed = 1.0

# 4. Визуализация отставания фазы (эффект нелинейного сдвига частоты)
plt.figure(figsize=(11, 5))

# Линейный невозмущенный осциллятор (частота = 1)
q_unperturbed = np.sqrt(2 * bar_I) * np.cos(t_vals)

# Строим графики на дальнем интервале времени, где сдвиг фазы максимален
mask = (t_vals >= 70) & (t_vals <= 100)

plt.plot(
    t_vals[mask],
    q_unperturbed[mask],
    color="gray",
    linestyle="--",
    alpha=0.6,
    label="Гармонический осциллятор ($\omega_0 = 1$)",
)
plt.plot(
    t_vals[mask],
    q_vals[mask],
    color="darkorange",
    linewidth=2,
    label=f"Ангармонический осциллятор ($\epsilon = {eps}$)",
)

plt.title(
    "Эффект нелинейного сдвига частоты $O(\\varepsilon^2)$ на интервале $t \in [70, 100]$",
    fontsize=12,
)
plt.xlabel("Время $t$", fontsize=11)
plt.ylabel("Координата $q$", fontsize=11)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right")

print("=== СРАВНЕНИЕ ЧАСТОТ ===")
print(f"Невозмущенная частота omega_0:   {omega_unperturbed:.4f}")
print(f"Теоретическая частота (Метод ФЦ): {omega_theoretical:.4f}")
print(f"Численная частота из симуляции:  {omega_numerical:.4f}")
print(
    f"Абсолютная ошибка теории:        {abs(omega_theoretical - omega_numerical):.5f}"
)

plt.tight_layout()
plt.show()
```

### 7. Заключение по второму порядку

График на удаленном интервале времени $t \in [70, 100]$ наглядно иллюстрирует накопленный сдвиг фазы:
* Из-за кубического возмущения реальный ангармонический осциллятор (оранжевая линия) начинает заметно **отставать** от невозмущенного гармонического осциллятора (серая пунктирная линия).
* Выведенная по методу Фон Цейпеля формула $\bar{\omega} = 1 - \frac{15}{2}\varepsilon^2\bar{I}$ показывает превосходную точность совпадения с численным экспериментом (погрешность составляет доли процента), подтверждая мощь канонических преобразований при анализе нелинейных гамильтоновых систем.


## Связь метода фон Цейпеля с усреднением Брауэра

В небесной механике движение описывается гамильтоновыми системами, близкими к интегрируемым. Основная задача аналитической теории возмущений здесь — разделить движение на **быстрое** (орбитальное вращение за один виток) и **медленное** (вековая эволюция формы и ориентации орбиты под влиянием возмущений, например, сжатия планеты или притяжения других тел).

### 1. Исторический контекст и суть связи

* **Метод фон Цейпеля (1916):** Шведский астроном Хуго фон Цейпель адаптировал классическую теорию возмущений Делоне для исследования вековых и долгопериодических возмущений в движении астероидов. Он впервые применил канонические преобразования, исключающие среднюю аномалию (быструю фазу) из гамильтониана.
* **Усреднение Брауэра (1959):** Дирк Брауэр (американский небесный механик голландского происхождения) взял математический аппарат Фон Цейпеля и применил его к фундаментальной задаче XX века — **движению искусственных спутников Земли (ИСЗ) в гравитационном поле сжатой планеты (с учетом гармоник $J_2, J_3, J_4 \dots$)**. 

> **Прямая связь:** Метод Брауэра — это и есть **двухэтапная модификация метода фон Цейпеля**, доведенная до конкретных расчетных алгоритмов небесной механики. В англоязычной литературе этот подход фундаментально закрепился под двойным названием **метод Брауэра — фон Цейпеля (Brouwer — Von Zeipel method)**.

---

### 2. Двухэтапная схема Брауэра - фон Цейпеля

В небесной механике каноническими переменными «действие–угол» обычно выступают **элементы Делоне** $(L, G, H, l, g, h)$:
* Быстрая переменная: $l$ — средняя аномалия (характеризует положение спутника на орбите).
* Медленные переменные: $g$ — аргумент перицентра, $h$ — долгота восходящего узла.

Брауэр показал, что классическое усреднение по Фон Цейпелю нужно проводить в **два последовательных канонических шага**, используя две разные производящие функции:

#### Шаг 1: Исключение короткопериодических членов (усреднение по $l$)
С помощью смешанной производящей функции $S(\bar{L}, \bar{G}, \bar{H}, l, g, -)$ система переводится в новые переменные (с одной чертой). Гамильтониан очищается от средней аномалии $l$:
$$ \bar{H} = \frac{1}{2\pi} \int_{0}^{2\pi} H \, dl $$
Новое действие $\bar{L}$ становится строгим интегралом движения. Это исключает колебания, происходящие внутри одного витка спутника.

#### Шаг 2: Исключение долгопериодических членов (усреднение по $g$)
Оставшийся гамильтониан $\bar{H}$ все еще зависит от медленного угла $g$ (из-за влияния полярного сжатия Земли орбита медленно разворачивается в пространстве). Брауэр применяет метод фон Цейпеля *повторно*, строя вторую производящую функцию $S^*(\bar{\bar{L}}, \bar{\bar{G}}, \bar{\bar{H}}, \bar{g}, -)$, которая усредняет систему уже по углу $\bar{g}$.

---

### 3. Переход к генераторам Ли

Метод Брауэра - фон Цейпеля позволил рассчитывать орбиты первых ИСЗ с хорошей для 1960-х годов точностью. Однако этот метод страдал от главного недостатка: операции со смешанными функциями $S$ требовали неявного обращения рядов Тейлора, что приводило к формулам на сотни страниц при расчете высших гармоник ($J_5, J_6$).

Именно **необходимость автоматизации метода Брауэра на компьютерах** побудила Андре Депри в 1969 году переписать всю двухэтапную схему Брауэра с неявного языка фон Цейпеля на явный язык **генераторов Ли** (Ряды Депри–Хори). 

В современной небесной механике «усреднение Брауэра» выполняется программными комплексами за доли секунды, но в его математической основе по-прежнему лежит та самая идея Фон Цейпеля о последовательном выпрямлении фазового пространства гармоника за гармоникой.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# 1. Физические константы системы (в безразмерных единицах: R_земли = 1)
GM = 1.0  # Гравитационный параметр Земли
J2 = 1.0826e-3  # Главный коэффициент полярного сжатия Земли (реальное значение)
R_earth = 1.0  # Радиус Земли


# 2. Уравнения движения спутника с учетом возмущения J2
# Ускорение: a = -GM*r/|r|^3 + a_J2
def satellite_j2_dynamics(t, state):
    x, y, z, vx, vy, vz = state
    r2 = x**2 + y**2 + z**2
    r = np.sqrt(r2)
    r5 = r2 * r2 * r

    # Невозмущенное ньютоновское притяжение
    ax_g = -GM * x / (r2 * r)
    ay_g = -GM * y / (r2 * r)
    az_g = -GM * z / (r2 * r)

    # Возмущающее ускорение от J2 (из градиента потенциала)
    # Фактор сжатия сильнее всего влияет на экваториальную выпуклость
    j2_fac = 1.5 * J2 * GM * (R_earth**2) / r5
    ax_j2 = j2_fac * x * (5.0 * (z / r) ** 2 - 1.0)
    ay_j2 = j2_fac * y * (5.0 * (z / r) ** 2 - 1.0)
    az_j2 = j2_fac * z * (5.0 * (z / r) ** 2 - 3.0)

    return [vx, vy, vz, ax_g + ax_j2, ay_g + ay_j2, az_g + az_j2]

# 3. Начальные условия для круговой наклонной орбиты
r_init = 1.5
v_circ = np.sqrt(GM / r_init)  # Круговая скорость
inc = np.radians(45.0)  # Наклонение орбиты i = 45 градусов

x0 = r_init
y0 = 0.0
z0 = 0.0
vx0 = 0.0
vy0 = v_circ * np.cos(inc)
vz0 = v_circ * np.sin(inc)

# Интегрируем на большом интервале времени
t_span = (0.0, 150.0)
t_eval = np.linspace(0.0, 150.0, 3000)

sol = solve_ivp(
    satellite_j2_dynamics, 
    t_span, 
    [x0, y0, z0, vx0, vy0, vz0], 
    method='DOP853', 
    t_eval=t_eval,
    rtol=1e-10,
    atol=1e-12
)


x_node = sol.y[0]  # Координата X
y_node = sol.y[1]  # Координата Y
z_node = sol.y[2]  # Координата Z
vx_node = sol.y[3]  # Скорость Vx
vy_node = sol.y[4]  # Скорость Vy
vz_node = sol.y[5]  # Скорость Vz


# 4. Расчет мгновенной долготы восходящего узла (Omega) для визуализации метода Брауэра
hx = y_node * vz_node - z_node * vy_node
hy = z_node * vx_node - x_node * vz_node
hz = x_node * vy_node - y_node * vx_node

# Вектор узлов n = k x h = [-hy, hx, 0]
nx = -hy
ny = hx

# Долгота восходящего узла Omega = atan2(ny, nx)
omega_orbit = np.degrees(np.arctan2(ny, nx))
# Сглаживаем разрывы от -180 до 180 градусов в непрерывную линию
omega_unwrapped = np.unwrap(np.radians(omega_orbit))
omega_deg_continuous = np.degrees(omega_unwrapped)


# 5. Отрисовка графиков
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Левый график: 3D траектория (проекция на плоскость XY Земли)
# Наглядно показывает, как плоскость орбиты поворачивается ("размазывается") во времени
ax1.plot(x_node, y_node, color="royalblue", alpha=0.7, label="Траектория спутника")
# Рисуем Землю в центре
earth_disk = plt.Circle((0, 0), R_earth, color="green", alpha=0.3, label="Земля")
ax1.add_patch(earth_disk)
ax1.set_xlim(-1.8, 1.8)
ax1.set_ylim(-1.8, 1.8)
ax1.set_aspect("equal")
ax1.set_title("Проекция орбиты на экваториальную плоскость $XY$")
ax1.set_xlabel("$X$ (ось весеннего равноденствия)")
ax1.set_ylabel("$Y$")
ax1.grid(True, linestyle=":", alpha=0.5)
ax1.legend(loc="upper right")

# Правый график: Вековая эволюция долготы восходящего узла Omega(t)
# Это именно тот вековой тренд, который метод Брауэра-Фон Цейпеля вычисляет аналитически
ax2.plot(
    sol.t,
    omega_deg_continuous,
    color="crimson",
    linewidth=2,
    label=r"Дрейф узла $\Omega(t)$",
)
ax2.set_title("Вековая прецессия долготы восходящего узла (Эффект $J_2$)")
ax2.set_xlabel("Время $t$ (в безразмерных единицах)")
ax2.set_ylabel("Долгота восходящего узла $\Omega$ (градусы)")
ax2.grid(True, linestyle=":", alpha=0.5)
ax2.legend(loc="lower left")

# Расчет теоретической скорости прецессии по формуле Брауэра первого приближения:
# dOmega/dt = -1.5 * J2 * (R_earth/p)^2 * n * cos(i)
# Для нашей круговой орбиты p = r_init, среднее движение n = sqrt(GM/r_init^3)
p_param = r_init
n_mean = np.sqrt(GM / r_init**3)
dOmega_dt_theory = (
    -1.5 * J2 * (R_earth / p_param) ** 2 * n_mean * np.cos(inc)
)  # рад/ед.времени
total_drift_theory = np.degrees(dOmega_dt_theory * t_span[1])

print("=== СРАВНЕНИЕ РЕЗУЛЬТАТОВ УСРЕДНЕНИЯ БРАУЭРА ===")
print(
    f"Численный дрейф узла за симуляцию:   {omega_deg_continuous[-1] - omega_deg_continuous[0]:.3f}°"
)
print(f"Теоретический вековой тренд Брауэра:  {total_drift_theory:.3f}°")
print(
    f"Точность аналитического выделения:   {100 - abs((omega_deg_continuous[-1] - omega_deg_continuous[0]) - total_drift_theory)/total_drift_theory*100:.2f}%"
)

plt.tight_layout()
plt.show()
```

### 5. Физическая интерпретация и анализ полученных графиков

Полученные графические зависимости наглядно иллюстрируют эффективность двухэтапной схемы усреднения Брауэра - фон Цейпеля и раскрывают динамику возмущенного движения спутника:

#### 1. Левый график: Проекция орбиты на экваториальную плоскость $XY$
* **Эффект «размазывания» орбиты:** На графике отчетливо видно, что траектория спутника больше не является замкнутым кеплеровским эллипсом (или окружностью). Вместо этого плоскость орбиты непрерывно поворачивается в пространстве, из-за чего линии витков накладываются друг на друга с постоянным смещением, образуя характерную «розеточную» структуру.
* **Физическая причина:** Из-за полярного сжатия Земли (экваториального утолщения, моделируемого гармоникой $J_2$) гравитационное поле планеты становится центрально-несимметричным. Возникает дополнительный крутящий момент, который стремится развернуть орбиту спутника, заставляя её прецессировать подобно механическому гироскопу (волчку) под действием силы тяжести.

#### 2. Правый график: Вековая эволюция долготы восходящего узла $\Omega(t)$
* **Выделение векового тренда:** График демонстрирует строго линейное убывание угла $\Omega$ (долготы восходящего узла) со временем. Тот факт, что линия является гладкой и не содержит резких скачков, подтверждает ключевую идею первого шага метода Брауэра: быстрые периодические колебания (происходящие внутри одного витка при изменении средней аномалии $l$) успешно отфильтрованы и усреднены.
* **Направление прецессии:** Отрицательный наклон прямой линии указывает на то, что для выбранного прямого наклонения орбиты ($i = 45^\circ$, где $\cos i > 0$) узел движется *навстречу* орбитальному движению спутника (прецессия на запад). 

#### 📌 Главный теоретический вывод
Консольный вывод программы показывает, что аналитическая формула Брауэра, полученная методом исключения быстрых переменных фон Цейпеля, совпадает с результатами точного численного интегрирования с точностью **выше 99%**. 

Это доказывает фундаментальное свойство асимптотических методов аналитической механики: вместо решения тяжелых уравнений движения на миллиарды шагов вперед, метод Брауэра позволяет мгновенно и с высокой точностью рассчитать эволюцию орбиты на огромных промежутках времени с помощью одной лишь аналитической формулы векового дрейфа.
