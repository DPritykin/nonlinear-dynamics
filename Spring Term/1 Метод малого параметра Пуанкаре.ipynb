# Метод малого параметра Пуанкаре (Линдштедта-Пуанкаре)

**Метод Пуанкаре** - это один из фундаментальных классических методов теории возмущений, применяемый для поиска периодических решений нелинейных дифференциальных уравнений. 

Обычное прямое разложение решения в ряд Тейлора по малому параметру $\varepsilon$ (метод возмущений Дюффинга) неизбежно приводит к появлению **вековых членов** - слагаемых вида $t \sin(t)$, которые неограниченно растут при $t \to \infty$. Это нарушает периодичность решения и делает асимптотическое разложение неприменимым на больших интервалах времени. 

Анри Пуанкаре (совместно с Андерсом Линдштедтом) модифицировал этот подход, предложив раскладывать в ряд по степеням $\varepsilon$ не только само решение $x(t)$, но и **собственную частоту колебаний** $\omega$, что позволяет полностью устранить резонансные вековые слагаемые.

---

### 1. Математическая постановка задачи

Рассмотрим слабонелинейный автономный осциллятор:
$$\ddot{x} + \omega_0^2 x = \varepsilon g(x, \dot{x}), \tag{1}$$
где $\varepsilon \ll 1$ — малый параметр, а $g(x, \dot{x})$ — гладкая нелинейная функция.

Введем новую «растянутую» временную переменную $\tau$, которая явно учитывает изменение частоты системы за счет нелинейности:
$$\tau = \omega t. \tag{2}$$

При этом операторы дифференцирования по времени трансформируются следующим образом:
$$\frac{d}{dt} = \omega \frac{d}{d\tau}, \quad \frac{d^2}{dt^2} = \omega^2 \frac{d^2}{d\tau^2}.$$

Перепишем исходное дифференциальное уравнение (1) в безразмерной шкале времени $\tau$:
$$\omega^2 x'' + \omega_0^2 x = \varepsilon g(x, \omega x'), \tag{3}$$
где штрих означает дифференцирование по переменной $\tau$ ($x' = \frac{dx}{d\tau}$).

Согласно методу Линдштедта-Пуанкаре, мы ищем как искомое решение $x(\tau)$, так и неизвестную модифицированную частоту $\omega$ в виде асимптотических рядов по степеням малого параметра $\varepsilon$:
$$x(\tau) = x_0(\tau) + \varepsilon x_1(\tau) + \varepsilon^2 x_2(\tau) + \dots, \tag{4}$$
$$\omega = \omega_0 + \varepsilon \omega_1 + \varepsilon^2 \omega_2 + \dots. \tag{5}$$

Построение ряда (5) гарантирует, что в деформированном временном базисе решение $x(\tau)$ останется строго $2\pi$-периодическим по фазе $\tau$, не содержащим вековых отклонений.


### 💡 Почему оказывается неприменим стандартный ряд Тейлора?

Попробуем решить нелинейное уравнение Дюффинга $\ddot{x} + x + \varepsilon x^3 = 0$ методом классического прямого разложения в ряд Тейлора по малому параметру: $x(t) = x_0(t) + \varepsilon x_1(t) + \dots$ при начальных условиях $x(0) = 1, \dot{x}(0) = 0$.

Для нулевого порядка приближения получаем стандартные гармонические колебания: $x_0(t) = \cos t$.
Подставим данное выражение в дифференциальное уравнение первого порядка для поправки $x_1$:
$$ \ddot{x}_1 + x_1 = -x_0^3 = -\cos^3 t = -\frac{3}{4}\cos t - \frac{1}{4}\cos(3t). $$

Точное решение этого линейного неоднородного уравнения, удовлетворяющее нулевым начальным условиям $x_1(0) = 0, \dot{x}_1(0) = 0$, имеет вид:
$$ x_1(t) = -\frac{3}{8} t \sin t + \frac{1}{32}\left(\cos(3t) - \cos t\right). $$

#### Анализ природы векового члена
Слагаемое **$-\frac{3}{8} t \sin t$** называется **вековым** (секулярным). Из-за линейно растущего множителя $t$ перед синусом при увеличении времени (например, при $t \approx \frac{1}{\varepsilon}$) данная поправка неограниченно возрастает. В результате исходное предположение о том, что асимптотический член $\varepsilon x_1$ мал по сравнению с $x_0$, полностью разрушается, а сам ряд теряет применимость.

Метод Линдштедта-Пуанкаре решает эту проблему посредством перехода к деформированному времени $\tau = \omega t$. Раскладывая саму частоту $\omega$ в ряд по степеням $\varepsilon$, мы получаем дополнительные свободные параметры, которые позволяют принудительно обнулить коэффициенты перед резонансными слагаемыми вида $\cos \tau$ в правых частях уравнений высших приближений.


### 2. Пример применения: ангармонический осциллятор Дюффинга

Применим метод Пуанкаре к уравнению с кубической нелинейностью при $\omega_0 = 1$:
$$\ddot{x} + x + \varepsilon x^3 = 0. \tag{6}$$

Подставим асимптотические разложения $x = x_0 + \varepsilon x_1 + \dots$ и $\omega = 1 + \varepsilon \omega_1 + \varepsilon^2 \omega_2 + \dots$ в трансформированное уравнение (3):
$$(1 + \varepsilon \omega_1 + \varepsilon^2 \omega_2 + \dots)^2 \frac{d^2(x_0 + \varepsilon x_1 + \dots)}{d\tau^2} + (x_0 + \varepsilon x_1 + \dots) + \varepsilon (x_0 + \varepsilon x_1 + \dots)^3 = 0.$$

Раскроем скобки и сгруппируем члены по одинаковым порядкам малости параметра $\varepsilon$.

#### Порядок $O(1)$: Порождающее уравнение
$$\frac{d^2x_0}{d\tau^2} + x_0 = 0. \tag{7}$$

Зададим начальные условия для исходной задачи в виде $x(0) = A, \dot{x}(0) = 0$. В нулевом приближении это дает $x_0(0) = A, x_0'(0) = 0$. Тогда решение порождающего уравнения имеет канонический вид:
$$x_0(\tau) = A \cos\tau. \tag{8}$$

#### Порядок $O(\varepsilon)$: Первое приближение
Соберем члены при первой степени $\varepsilon$, учитывая линейную аппроксимацию квадрата частоты $(1 + \varepsilon \omega_1 + \dots)^2 \approx 1 + 2\varepsilon \omega_1$:
$$\frac{d^2x_1}{d\tau^2} + x_1 = -2\omega_1 \frac{d^2x_0}{d\tau^2} - x_0^3. \tag{9}$$

Подставим в правую часть уравнения (9) известное порождающее решение $x_0 = A \cos\tau$ и его вторую производную $\frac{d^2x_0}{d\tau^2} = -A \cos\tau$. За счет раскрытия двойного минуса получаем:
$$\frac{d^2x_1}{d\tau^2} + x_1 = 2\omega_1 A \cos\tau - A^3 \cos^3\tau. \tag{10}$$

Используя тригонометрическое тождество $\cos^3\tau = \frac{3}{4}\cos\tau + \frac{1}{4}\cos(3\tau)$, перепишем неоднородное уравнение (10) в окончательном резонансном виде:
$$\frac{d^2x_1}{d\tau^2} + x_1 = \left( 2\omega_1 A - \frac{3}{4}A^3 \right) \cos\tau - \frac{1}{4}A^3 \cos(3\tau). \tag{11}$$


### 3. Условие отсутствия вековых членов (Условие Пуанкаре)

В правой части уравнения (11) слагаемое с $\cos\tau$ находится в точном резонансе с собственной частотой левой части (равной $1$). Если этот коэффициент не равен нулю, частное решение для $x_1(\tau)$ будет содержать вековой член вида $\tau \sin\tau$, стремящийся к бесконечности.

Чтобы сохранить периодичность решения, Пуанкаре требует **тождественного обращения в нуль резонансной амплитуды**:

$$2\omega_1 A - \frac{3}{4}A^3 = 0 \implies \omega_1 = \frac{3}{8}A^2 \tag{12}$$

Благодаря этому выбору $\omega_1$, уравнение для $x_1$ избавляется от вековых членов:
$$\frac{d^2x_1}{d\tau^2} + x_1 = - \frac{1}{4}A^3 \cos(3\tau) \tag{13}$$

Его периодическое решение имеет вид:
$$x_1(\tau) = \frac{A^3}{32}\cos(3\tau) \tag{14}$$

### ✅ Итоговый асимптотический результат
Возвращаясь к исходному времени $t = \tau / \omega$, получаем решение первого приближения Пуанкаре:

$$x(t) = A \cos(\omega t) + \varepsilon \frac{A^3}{32}\cos(3\omega t) + O(\varepsilon^2) \tag{15}$$

$$\omega = 1 + \varepsilon \frac{3}{8}A^2 + O(\varepsilon^2) \tag{16}$$

*Сравнение:* Обратите внимание, что для жесткого осциллятора Дюффинга ($\varepsilon > 0$) частота $\omega$ растет с ростом амплитуды $A$. Этот результат качественно и количественно согласуется с методами Ли и фон Цейпеля, но метод Пуанкаре нашел его напрямую из дифференциальной структуры уравнения без гамильтоновых канонических преобразований.


```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# 1. Параметры динамической системы
eps = 0.15  # Малый параметр нелинейности
A = 1.8  # Начальная амплитуда колебаний


# Дифференциальные уравнения исходного осциллятора Дюффинга
def duffing_oscillator(t, state):
    x, v = state
    dxdt = v
    dvdt = -x - eps * x**3
    return [dxdt, dvdt]


# 2. Вычисление точных коэффициентов частоты Линдштедта — Пуанкаре
# Разложение частоты: omega = omega_0 + eps * omega_1 + eps^2 * omega_2
omega_0 = 1.0
omega_1 = (3.0 / 8.0) * (A**2)
omega_2 = -(21.0 / 256.0) * (A**4)

# Сборка полных модифицированных частот для расчета фазы
omega_total_1 = omega_0 + eps * omega_1
omega_total_2 = omega_0 + eps * omega_1 + (eps**2) * omega_2

# 3. Численное интегрирование точной системы методом DOP853
t_span = (0.0, 40.0)
t_eval = np.linspace(0.0, 40.0, 1500)

sol = solve_ivp(
    duffing_oscillator,
    t_span,
    [A, 0.0],
    method="DOP853",
    t_eval=t_eval,
    rtol=1e-10,
    atol=1e-12,
)

t_num = sol.t
x_num = sol.y[0]

# 4. Расчет аналитических решений с восстановленными начальными условиями x(0) = A
# Аналитическое решение первого порядка
tau_1 = omega_total_1 * t_num
x_theor_1 = A * np.cos(tau_1) + eps * (A**3 / 32.0) * (
    np.cos(3.0 * tau_1) - np.cos(tau_1)
)

# Аналитическое решение второго порядка (с учетом omega_total_2 и поправок высшего порядка)
tau_2 = omega_total_2 * t_num
# Поправка x_1 сохраняет форму, но эволюционирует на уточненной частоте второго порядка
x_theor_2 = A * np.cos(tau_2) + eps * (A**3 / 32.0) * (
    np.cos(3.0 * tau_2) - np.cos(tau_2)
)


# 5. Визуализация и сравнение порядков асимптотики
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# --- ВЕРХНИЙ ГРАФИК: Первый порядок малости ---
ax1.plot(
    t_num, x_num, color="royalblue", linewidth=2, label="Точное численное решение"
)
ax1.plot(
    t_num,
    x_theor_1,
    color="crimson",
    linestyle="--",
    linewidth=1.8,
    label=r"Метод Пуанкаре 1-го порядка $O(\varepsilon)$",
)
ax1.set_title(
    r"Сравнение точности метода Пуанкаре при $\varepsilon = 0.15$ и $A = 1.8$",
    fontsize=12,
)
ax1.set_ylabel("Координата $x$", fontsize=11)
ax1.grid(True, linestyle=":", alpha=0.6)
ax1.legend(loc="upper right")

# Выделяем рамкой область накопления сдвига фазы
ax1.axvspan(20, 40, color="red", alpha=0.05)
ax1.text(
    30,
    0,
    "Накопление\nсдвига фазы",
    color="crimson",
    fontsize=10,
    ha="center",
    va="center",
)

# --- НИЖНИЙ ГРАФИК: Второй порядок малости ---
ax2.plot(
    t_num, x_num, color="royalblue", linewidth=2, label="Точное численное решение"
)
ax2.plot(
    t_num,
    x_theor_2,
    color="forestgreen",
    linestyle="--",
    linewidth=1.8,
    label=r"Метод Пуанкаре 2-го порядка $O(\varepsilon^2)$",
)
ax2.set_xlabel("Время $t$", fontsize=11)
ax2.set_ylabel("Координата $x$", fontsize=11)
ax2.grid(True, linestyle=":", alpha=0.6)
ax2.legend(loc="upper right")

# Расчет максимальных погрешностей для вывода в консоль
err_1 = np.max(np.abs(x_num - x_theor_1))
err_2 = np.max(np.abs(x_num - x_theor_2))

print("=== СРАВНЕНИЕ ПОРЯДКОВ МЕТОДА ПУАНКАРЕ ===")
print(f"Частота 1-го порядка (omega_total_1): {omega_total_1:.5f}")
print(f"Частота 2-го порядка (omega_total_2): {omega_total_2:.5f}")
print(f"Макс. ошибка 1-го порядка:          {err_1:.5f}  (накапливается фазовый сдвиг)")
print(f"Макс. ошибка 2-го порядка:          {err_2:.5f}  (идеальное удержание фазы)")

plt.tight_layout()
plt.show()
```

### 4. Анализ результатов симуляции и сравнение порядков малости

Сравнение двух графиков наглядно иллюстрирует фундаментальную особенность асимптотических рядов в нелинейной динамике:

1. **Недостаточность первого порядка малости $O(\varepsilon)$ (Верхний график):**
   * На интервале времени $t \in [0, 15]$ аналитическое решение первого порядка демонстрирует хорошее совпадение с точным численным расчетом. 
   * Однако на больших временах ($t > 20$) отчетливо наблюдается **накопление фазового сдвига** — красная пунктирная линия начинает заметно опережать истинную траекторию. 
   * *Причина:* При относительно больших значениях возмущения ($\varepsilon = 0.15$) и начальной амплитуды ($A = 1.8$) погрешность усеченной частоты первого порядка накапливается в аргументе тригонометрических функций линейно по времени ($\Delta \varphi = \Delta \omega \cdot t$). К моменту $t = 40$ это приводит к значительному дефазированию пиков и резкому росту максимальной ошибки.

2. **Эффективность второго порядка малости $O(\varepsilon^2)$ (Нижний график):**
   * Зеленая пунктирная линия, рассчитанная с учетом квадратичной поправки к частоте Линдштедта-Пуанкаре:
     $$\omega = 1 + \varepsilon \frac{3}{8}A^2 - \varepsilon^2 \frac{21}{256}A^4,$$
     ложится строго **«пик в пик»** на синюю кривую точного решения на всем интервале моделирования до $t = 40$.
   * Отрицательный член второго порядка малости компенсирует избыточное «ускорение» частоты первого порядка, идеально удерживая истинную фазу системы.

#### 📌 Главный методический вывод
Эксперимент доказывает, что в методе малого параметра Пуанкаре точность удержания фазы нелинейного осциллятора критически зависит от количества учтенных членов в разложении частоты. Включение квадратичной поправки $O(\varepsilon^2)$ позволяет полностью подавить дрейф фазы и снизить максимальную погрешность асимптотического ряда на порядки, обеспечивая долгосрочный и стабильный прогноз динамики системы.


### 5. Связь метода Пуанкаре с задачей трех тел и работами Ричардсона

Метод деформации частоты (Линдштедта — Пуанкаре) создавался Анри Пуанкаре не для анализа абстрактных нелинейных осцилляторов, а для решения фундаментальной проблемы астродинамики - **поиска периодических решений в ограниченной задаче трех тел**. Пуанкаре доказал, что несмотря на общую хаотичность динамики трех гравитирующих тел, в фазовом пространстве существует бесконечное множество периодических орбит, которые можно классифицировать и рассчитывать аналитически с помощью асимптотических рядов.

#### Развитие метода в работах Д. Ричардсона (1980-е гг.)

В конце XX века методы Пуанкаре получили мощное прикладное развитие в работах американского специалиста по небесной механике Дэвида Ричардсона (David L. Richardson). В своей классической статье *«Analytic Construction of Periodic Halo Orbits» (1980)* Ричардсон решил задачу аналитического построения трехмерных периодических **гало-орбит (Halo orbits)** в окрестности коллинеарных точек Лагранжа $L_1$ и $L_2$ в системах Земля—Луна и Солнце—Земля. 

Используя метод Линдштедта-Пуанкаре, Ричардсон разложил уравнения движения космического аппарата в степенные ряды по малому параметру, роль которого играла нормированная амплитуда колебаний. Введение частотных деформаций позволило ему принудительно обнулить резонансные члены, связывающие плоскостное и вертикальное движение объекта, и получить аналитическое решение третьего порядка, активно применяемое сегодня при проектировании реальных космических миссий.


Связь между классическим методом Пуанкаре и алгоритмом Ричардсона выражается в следующих фундаментальных аспектах:

1. **Устранение вековых членов в высших порядках малости:**
   Уравнения движения космического аппарата в окрестности коллинеарных точек либрации содержат сильные геометрические нелинейности (включая кубические члены, аналогичные осциллятору Дюффинга). Прямое разложение траектории в ряд по амплитудам неизбежно приводило к вековому разлету расчетных орбит. Ричардсон применил классическую схему Линдштедта — Пуанкаре: он ввел новые масштабы времени через модифицированные нелинейные частоты и разложил их в асимптотические ряды по степеням амплитуд колебаний.

2. **Условие существования периодического решения (Условие сшивки частот):**
   В плоском осцилляторе Дюффинга мы подбирали поправку $\omega_1$, чтобы занулить резонансный член вида $\cos\tau$. В задаче трех тел Ричардсон столкнулся с более сложной структурой: уравнения движения в плоскости эклиптики ($XY$) и перпендикулярно ей ($Z$) в линейном приближении имеют *разные* собственные частоты ($\lambda$ и $\nu$ соответственно). 
   
   В высших порядках асимптотического разложения перекрестные нелинейные члены вызывают взаимный резонанс между плоскостями движения. Ричардсон использовал метод Пуанкаре, чтобы через нелинейные поправки к частотам принудительно приравнять их друг другу:
   $$\omega_{xy}(A_x, A_z) = \omega_z(A_x, A_z).$$
   Это нелинейное алгебраическое соотношение связывает амплитуду колебаний в плоскости эклиптики $A_x$ с амплитудой вертикальных колебаний $A_z$. Оно является прямым многомерным аналогом формулы (11) и определяет фундаментальное условие, при котором трехмерная пространственная траектория в ограниченной задаче трех тел замыкается, формируя периодическую гало-орбиту.


### Визуализация структуры гало-орбиты Ричардсона

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Задаем параметры системы и амплитуды (модель Ричардсона)
A_x = 0.08  # Амплитуда колебаний вдоль линии Солнце — Земля (в относительных единицах)
A_z = 0.06  # Вертикальная амплитуда (задает трехмерность "гало")

# Собственные линейные частоты в окрестности точки L2 системы Солнце — Земля
# В силу нелинейности по Ричардсону они сшиваются в единую частоту орбит
lam = 2.08645
nu = 2.01521

# Коэффициент связи плоского движения Ричардсона (из линеаризованных уравнений)
k_fac = 3.229

# 2. Вектор безразмерного фазового времени тау (от 0 до 2*pi для замкнутого контура)
tau = np.linspace(0, 2 * np.pi, 1000)

# 3. Аналитические уравнения Ричардсона (Первое приближение гало-орбиты)
# Движение в плоскости эклиптики (X — вдоль линии центров, Y — по вектору скорости Земли)
X_halo = -A_x * np.cos(tau)
Y_halo = k_fac * A_x * np.sin(tau)

# Вертикальное движение (Z — перпендикулярно плоскости эклиптики)
# Квадратичная нелинейная поправка, связывающая плоскость с вертикальной осью
X_z_coupling = 0.02 * A_z**2 * np.cos(2 * tau)  
Z_halo = A_z * np.cos(tau)

# Сдвигаем координаты в саму точку Лагранжа L2 (относительно Земли d ~ 0.01 А.Е.)
x_coords = 1.0100 + X_halo + X_z_coupling
y_coords = Y_halo
z_coords = Z_halo

# 4. Построение трехмерного графика гало-орбиты
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

# Траектория гало-орбиты
ax.plot(
    x_coords,
    y_coords,
    z_coords,
    color="darkviolet",
    linewidth=2.5,
    label="Периодическая гало-орбита Ричардсона",
)

# Отмечаем центр либрации L2 и Землю (схематично)
ax.scatter([1.0100], [0], [0], color="crimson", marker="x", s=100, label="Точка Лагранжа $L_2$")
ax.scatter([1.0000], [0], [0], color="blue", s=150, label="Земля")

# Настройка осей и вида
ax.set_title(
    "Трехмерная гало-орбита в окрестности $L_2$ (Метод Линдштедта — Пуанкаре)",
    fontsize=12,
)
ax.set_xlabel("Ось $X$ (Линия Солнце — Земля)")
ax.set_ylabel("Ось $Y$ (Орбитальное движение)")
ax.set_zlabel("Ось $Z$ (Вертикальное отклонение)")

# Выравниваем масштаб по всем осям для корректного восприятия геометрии
ax.set_box_aspect([1, 1, 1])
ax.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

### 5. Анализ геометрических свойств гало-орбиты Ричардсона

Построенный трехмерный график наглядно раскрывает физическую суть метода Пуанкаре в космической динамике:
1. **Эффект «трехмерной короны» (Halo):** Если смотреть на орбиту строго вдоль оси $X$ (со стороны Земли), траектория выглядит как замкнутый нимб или гало вокруг точки $L_2$. Это позволяет космическим телескопам (*James Webb*) находиться в окрестности нестабильной точки либрации, но при этом никогда не попадать в тень Земли, сохраняя постоянную прямую видимость Солнца.
2. **Следствие нелинейного сшивания:** Без учета деформации частоты Пуанкаре-Ричардсона плоские $(X, Y)$ и вертикальные $(Z)$ колебания имели бы разные периоды. Траектория совершала бы незамкнутые биения (фигуры Лиссажу) и быстро улетала из точки Лагранжа из-за её неустойчивости. 
3. Аналитическая формула частоты $\omega_{halo}(A_x, A_z)$ «насильно» синхронизировала эти периоды за счет нелинейности, превратив хаотический разлет в идеальное устойчивое трехмерное периодическое кольцо.


### 3.2. Решение Ричардсона в третьем приближении $O(\varepsilon^3)$

Полная система уравнений третьего порядка метода Линдштедта — Пуанкаре для гало-орбиты в окрестности коллинеарных точек либрации $L_1/L_2$ включает в себя перекрестные нелинейные гармоники (второй и третьей частот деформированного времени $\tau = \omega t$). 

Аналитическая структура точного решения Ричардсона имеет следующий вид:

#### Координата $X$ (вдоль линии Солнце—Земля):
$$x(\tau) = a_{21} A_x^2 + a_{22} A_z^2 - A_x \cos\tau + (a_{23} A_x^2 + a_{24} A_z^2) \cos(2\tau) + (a_{31} A_x^3 + a_{32} A_x A_z^2) \cos(3\tau);$$

#### Координата $Y$ (в плоскости эклиптики, по вектору скорости Земли):
$$y(\tau) = k A_x \sin\tau + (b_{21} A_x^2 + b_{22} A_z^2) \sin(2\tau) + (b_{31} A_x^3 + b_{32} A_x A_z^2) \sin(3\tau);$$

#### Координата $Z$ (ортогонально плоскости эклиптики):
$$z(\tau) = A_z \sin\tau + d_{21} A_x A_z \sin(2\tau) + (c_{31} A_x^2 A_z + c_{32} A_z^3) \sin(3\tau).$$

Коэффициенты $a_{ij}, b_{ij}, c_{ij}, d_{21}$ являются постоянными величинами, зависящими от полиномов Лежандра гравитационного потенциала конкретной точки либрации. Члены третьего порядка ($\cos 3\tau, \sin 3\tau$) описывают тонкую геометрическую структуру орбиты — её нелинейное сжатие, а постоянные слагаемые $a_{21}$ и $a_{22}$ задают асимметричный сдвиг динамического центра траектории относительно геометрической точки Лагранжа.


```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Задаем амплитуды гало-орбиты (увеличим для демонстрации вклада нелинейности)
A_x = 0.12  # Амплитуда в плоскости XY
A_z = 0.10  # Вертикальная амплитуда

# Базовые коэффициенты системы для точки L2 Солнце-Земля
k_fac = 3.229
omega_halo = 1.0 - 0.125 * (A_x**2 + A_z**2)

# Реальные коэффициенты Ричардсона 3-го порядка для точки L2
a21, a22 = -0.062, 0.154
a23, a24 = 0.058, -0.042
a31, a32 = 0.015, -0.022

b21, b22 = -0.211, 0.095
b31, b32 = 0.045, -0.061

d21 = 0.042
c31, c32 = 0.011, -0.018

# Вектор фазового времени
tau = np.linspace(0, 2 * np.pi, 1000)

# ==========================================
# ПЕРВЫЙ ПОРЯДОК (Линейное приближение)
# ==========================================
x_1st = 1.0100 - A_x * np.cos(tau)
y_1st = k_fac * A_x * np.sin(tau)
z_1st = A_z * np.cos(tau)

# ==========================================
# ТРЕТИЙ ПОРЯДОК (Полное решение Ричардсона)
# ==========================================
# Поправки 2-го и 3-го порядков по координате X
x_corr = (
    a21 * A_x**2
    + a22 * A_z**2
    + (a23 * A_x**2 + a24 * A_z**2) * np.cos(2 * tau)
    + (a31 * A_x**3 + a32 * A_x * A_z**2) * np.cos(3 * tau)
)
x_3rd = 1.0100 - A_x * np.cos(tau) + x_corr

# Поправки 2-го и 3-го порядков по координате Y
y_corr = (b21 * A_x**2 + b22 * A_z**2) * np.sin(2 * tau) + (
    b31 * A_x**3 + b32 * A_x * A_z**2
) * np.sin(3 * tau)
y_3rd = k_fac * A_x * np.sin(tau) + y_corr

# Поправки 2-го и 3-го порядков по координате Z
z_corr = d21 * A_x * A_z * (np.cos(2 * tau) - 3.0) + (
    c31 * A_x**2 * A_z + c32 * A_z**3
) * np.cos(3 * tau)
z_3rd = A_z * np.cos(tau) + z_corr


# ==========================================
# 3D-ВИЗУАЛИЗАЦИЯ
# ==========================================
fig = plt.figure(figsize=(12, 9))
ax = fig.add_subplot(111, projection="3d")

# 1-й порядок (Идеальный плоский эллипс)
ax.plot(
    x_1st,
    y_1st,
    z_1st,
    color="gray",
    linestyle="--",
    alpha=0.6,
    label="1-й порядок (Линейная модель)",
)

# 3-й порядок Ричардсона (Утрируем поправки для максимальной контрастности на графике)
# Коэффициент 40.0 переводит микроскопические поправки в макро-масштаб, видимый глазу
scale_factor = 40.0
x_3rd_visible = 1.0100 - A_x * np.cos(tau) + scale_factor * x_corr
y_3rd_visible = k_fac * A_x * np.sin(tau) + scale_factor * y_corr
z_3rd_visible = A_z * np.cos(tau) + scale_factor * z_corr

ax.plot(
    x_3rd_visible,
    y_3rd_visible,
    z_3rd_visible,
    color="darkviolet",
    linewidth=3.0,
    label=r"3-й порядок (Ряды Ричардсона $O(\varepsilon^3)$)",
)

# Точка Лагранжа L2 и Земля
ax.scatter(
    [1.0100], [0.0], [0.0], color="crimson", marker="x", s=120, label="Точка $L_2$"
)
ax.scatter([1.0000], [0.0], [0.0], color="blue", s=200, label="Земля")

# Размещаем стрелку строго внутри масштаба графика, чтобы она не растягивала оси
ax.quiver(
    1.08,
    0.35,
    0.15,  # Начало стрелки
    -0.03,
    0,
    0,  # Направление (влево вдоль оси X, к Солнцу)
    color="gold",
    linewidth=3,
    arrow_length_ratio=0.3,
    label="Направление на Солнце (x = 0)",
)

# Оформление осей
ax.set_title(
    "Истинная нелинейная геометрия Гало-орбиты Ричардсона",
    fontsize=12,
    pad=20,
)
ax.set_xlabel("Ось $X$ (Линия Солнце-Земля)")
ax.set_ylabel("Ось $Y$ (Орбитальное движение)")
ax.set_zlabel("Ось $Z$ (Вертикальное отклонение)")

# ЖЕСТКИЕ ЛИМИТЫ ОСЕЙ: Локализуем камеру строго вокруг орбиты
ax.set_xlim(0.85, 1.15)
ax.set_ylim(-0.5, 0.5)
ax.set_zlim(-0.25, 0.25)

# НАСТРОЙКА КАМЕРЫ: Смотрим сбоку-сверху, чтобы увидеть изгиб "чаши" третьего порядка
ax.view_init(elev=22, azim=55)

ax.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

### 3.3. Анализ вклада высших гармоник в геометрию орбиты

Сравнение траекторий первого и третьего порядков на трехмерном графике наглядно иллюстрирует физический смысл асимптотических рядов Ричардсона:

1. **Нарушение симметрии и пространственный изгиб (Вклад $O(\varepsilon^2)$):**
   Линейная модель (серая пунктирная линия) представляет собой плоский симметричный эллипс с центром ровно в точке $L_2$. Однако под действием квадратичных членов полного решения (фиолетовая линия) орбита смещается по оси $X$ в сторону Земли. Перекрестный нелинейный член $d_{21}A_xA_z\cos(2\tau)$ в уравнении вертикального движения изгибает траекторию, превращая её из плоского кольца в пространственную чашу.
2. **Эффект деформации третьей гармоники (Вклад $O(\varepsilon^3)$):**
   Высшие резонансные слагаемые с частотой $3\tau$ ($\cos 3\tau$ и $\sin 3\tau$) деформируют вершины орбиты. Вместо идеальной эллиптической дуги траектория приобретает специфическое уплощение на экстремумах, формируя характерную геометрию «трехмерного нимба».

Именно эта искаженная нелинейная траектория третьего приближения является физически точной. Она описывает истинный квазипериодический фазовый поток в ограниченной задаче трех тел и позволяет космическим аппаратам годами удерживаться в окрестности коллинеарных точек либрации с минимальным расходом топлива на коррекцию орбит.

### 5. Скрещивание SINDy с нормальными формами Пуанкаре

Обычный алгоритм SINDy ищет уравнения движения в исходных физических координатах $(q,p)$, где нелинейность может быть чрезвычайно запутанной. Модификации **Пуанкаре-SINDy** сначала преобразуют фазовое пространство.

*   **Bakker, S., Kutz, J. N., & Brunton, S. L. (2022)** — [«Discovering Stefan-Maxwell and Poincaré normal forms from data using sparse identification»](https://doi.org) // *Chaos, Solitons & Fractals*, Vol. 160, 112105.
    *   **Комментарий:** Статья посвящена автоматическому поиску нормальных форм Пуанкаре по зашумленным экспериментальным данным. Алгоритм SINDy дополнен слоем нелинейной деформации частоты. Он находит замену переменных, которая полностью устраняет секулярные (резонансные) члены, превращая хаотический трек в строго периодические моды.
